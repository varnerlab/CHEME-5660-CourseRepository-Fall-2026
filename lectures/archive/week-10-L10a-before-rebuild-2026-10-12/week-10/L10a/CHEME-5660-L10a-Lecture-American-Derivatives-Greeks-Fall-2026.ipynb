## Concept Review: Intrinsic and Extrinsic Value of Options
In the last lecture, we did a deep dive into American options premium and its components: intrinsic value and extrinsic value.

* __Intrinsic Value__: The intrinsic value of an American option is the immediate exercise value. For a call option, it's calculated as $\max(0, S_{t} - K)$, where $S_{t}$ is the current stock price and $K$ is the strike price. For a put option, it's $\max(0, K - S_{t})$.
* __Extrinsic Value__: The extrinsic value, also known as time value, represents the additional value of the option based on the potential for future price movements before expiration. It is calculated as the difference between the total option price and its intrinsic value.

The extrinsic value $E_{V}(t)$ can be expressed mathematically as:
$$
\boxed{
\begin{aligned}
E_{V}(t) = \text{Option Price} - \text{Intrinsic Value}\quad\blacksquare
\end{aligned}}
$$
Using the immediate exercise values, the two cases are:
$$
\boxed{
\begin{aligned}
E_{V}(t) &= \begin{cases}
\mathcal{P}_{c} - \max(0, S_{t} - K) & \text{(for Call Options)} \\
\mathcal{P}_{p} - \max(0, K - S_{t}) & \text{(for Put Options)}
\end{cases}
\end{aligned}}
$$
This creates an interesting edge case: when the option is __out of the money__, i.e., the option has no intrinsic value, the entire option price is extrinsic value. Why would we be willing to pay for an option that has no intrinsic value?

## Options Premiums and Market Conditions
The central question we will address in this lecture is: "How do American options premiums change with market conditions?" We will explore the sensitivity to the underlying asset price, volatility, time to expiration, and risk-free interest rate. We hold dividends fixed; the formulas in this lecture assume a non-dividend-paying underlying asset.

This question is answered locally by computing sensitivity coefficients of the options premium, collectively known as the __Greeks__.

> __The Greeks:__
>
> Let $\mathcal{P}_{\kappa}(S,\sigma,\tau,r)$ be the premium in USD/share for a call or put, with contract type $\kappa\in\{c,p\}$. The current share price is $S$, implied volatility is the annualized decimal $\sigma$, and the continuously compounded risk-free rate is $r$. Calendar time $t$ and expiration $T$ are measured in years, so remaining time is $\tau=T-t$. At the current market conditions, define the raw derivatives by:
>
> $$
> \begin{align*}
> \Delta &= \frac{\partial\mathcal{P}_{\kappa}}{\partial S}, &
> \Gamma &= \frac{\partial^2\mathcal{P}_{\kappa}}{\partial S^2}, &
> \Theta &= \frac{\partial\mathcal{P}_{\kappa}}{\partial t}=-\frac{\partial\mathcal{P}_{\kappa}}{\partial\tau},\\
> \nu &= \frac{\partial\mathcal{P}_{\kappa}}{\partial\sigma}, &
> \rho &= \frac{\partial\mathcal{P}_{\kappa}}{\partial r}.
> \end{align*}
> $$
>
> * **Delta:** The local slope with respect to the share price supplies the share-equivalent exposure and hedge ratio.
> * **Gamma:** The curvature measures how Delta changes as the share price moves.
> * **Theta:** The calendar-time derivative measures the price change as time passes. It is commonly negative for long options, although American exercise, dividends, and interest rates can change its sign.
> * **Vega:** The volatility derivative measures how the premium responds to a change in implied volatility. Long vanilla options generally have positive Vega.
> * **Rho:** The rate derivative measures the response to a change in the risk-free rate. Calls and puts can respond in opposite directions.

Trading platforms commonly display rescaled Greeks. A change in volatility from 20% to 21% is one percentage point, or a decimal change of 0.01. With a 365-day year, the conversions are:

| Greek | Raw derivative | Common displayed quote |
| :--- | :--- | :--- |
| Delta | Share-equivalent exposure per option share | Same value |
| Gamma | Change in Delta per USD/share | Same value |
| Theta | USD/share per year of calendar time | $\Theta_{\mathrm{day}}=\Theta/365$ |
| Vega | USD/share per unit decimal volatility | $\nu_{\mathrm{point}}=\nu/100$ |
| Rho | USD/share per unit decimal interest rate | $\rho_{\mathrm{point}}=\rho/100$ |

Multiply a per-share price change by the contract multiplier, usually 100 shares per contract, and the signed number of contracts to obtain the position's dollar change. The sign is positive for a long position and negative for a short position.

### Derivation: A local approximation of the price change
For a small share-price move $\Delta S$, volatility change $\Delta\sigma$, elapsed calendar time $\Delta t>0$, and rate change $\Delta r$, a local Taylor approximation gives:

$$
\begin{align*}
\Delta\mathcal{P}_{\kappa}
&\approx \underbrace{\Delta_{\kappa}\Delta S}_{\text{share-price slope}}
+\underbrace{\frac{1}{2}\Gamma_{\kappa}(\Delta S)^2}_{\text{share-price curvature}}
+\underbrace{\Theta_{\kappa}\Delta t}_{\text{time passing}}
+\underbrace{\nu_{\kappa}\Delta\sigma}_{\text{volatility change}}
+\underbrace{\rho_{\kappa}\Delta r}_{\text{rate change}}.
\end{align*}
$$

All derivatives are evaluated at the current market conditions. We retain second order in the share price and first order in the other inputs, while omitting cross derivatives and higher-order terms. This approximation becomes less accurate for large moves or when an American option crosses an exercise boundary.

__What does Gamma add?__ Suppose a long call has Delta 0.50 and Gamma 0.02 per USD/share. A five-dollar increase gives a delta-only price change of 2.50 USD/share and a delta–gamma change of 2.75 USD/share. For a five-dollar decrease, the corresponding estimates are −2.50 and −2.25 USD/share. Curvature contributes 0.25 USD/share, or 25 USD per standard contract, in either direction. The NVDA example compares these estimates with actual repricing.

## Computing the Greeks: CRR and Leisen–Reimer
The CRR model from L9b gives us an American-option price by working backward through a binomial tree and comparing continuation with immediate exercise at each node. To calculate a sensitivity, we can perturb one market input, rebuild the tree, and compare the prices.

> __Central differences:__
>
> Write $\mathcal{P}=\mathcal{P}_{\kappa}(S,\sigma,\tau,r)$ for the current per-share value. Let $b_S$, $b_\sigma$, $b_\tau$, and $b_r$ be positive perturbation sizes in USD/share, decimal volatility, years, and inverse years, respectively. Central differences estimate the derivatives by:
>
> $$
> \begin{align*}
> \widehat\Delta &= \frac{\mathcal{P}(S+b_S)-\mathcal{P}(S-b_S)}{2b_S},\\
> \widehat\Gamma &= \frac{\mathcal{P}(S+b_S)-2\mathcal{P}(S)+\mathcal{P}(S-b_S)}{b_S^2},\\
> \widehat\Theta &= \frac{\mathcal{P}(\tau-b_\tau)-\mathcal{P}(\tau+b_\tau)}{2b_\tau},\\
> \widehat\nu &= \frac{\mathcal{P}(\sigma+b_\sigma)-\mathcal{P}(\sigma-b_\sigma)}{2b_\sigma}, &
> \widehat\rho &= \frac{\mathcal{P}(r+b_r)-\mathcal{P}(r-b_r)}{2b_r}.
> \end{align*}
> $$
>
> Inputs omitted from a price argument are held fixed. Every valuation uses the same number of steps and full precision. We inspect both the perturbation size and tree depth because a smaller perturbation can expose tree discretization error.

The existing example functions use a one-dollar forward change for Delta, a full one-day repricing for Theta, and one-percentage-point forward changes for Vega and Rho. Gamma uses differences of slopes near the tree's root. We retain those calculations for comparison and introduce central differences to study convergence. A full one-day price change can differ from the local daily Theta quote.

### Why introduce another binomial tree?
The location of the strike relative to the terminal CRR nodes changes as we refine the tree. This causes option-price errors to oscillate, and subtracting nearby prices can make those errors visible in the Greeks. The [Leisen–Reimer model](https://www.wiwi.uni-bonn.de/sfb303/papers/abstracts/bonnsfb309.html) changes the tree factors to obtain smoother European-price convergence toward the Black–Scholes solution.

Let $n$ be an odd number of steps and $\delta\tau=\tau/n$ the time per step. With the non-dividend BSM quantities $d_1$ and $d_2$, define the Peizer–Pratt inversion $H$ and the two probabilities by:

$$
\begin{align*}
d_1 &= \frac{\ln(S/K)+(r+\sigma^2/2)\tau}{\sigma\sqrt\tau}, & d_2&=d_1-\sigma\sqrt\tau,\\
H(z;n)&=\frac12+\frac{\operatorname{sgn}(z)}{2}\sqrt{1-\exp\left[-\left(\frac{z}{n+1/3+0.1/(n+1)}\right)^2(n+1/6)\right]},\\
p&=H(d_2;n), & p'&=H(d_1;n).
\end{align*}
$$

Here, $K$ is the strike price, $p$ is the risk-neutral up probability, and $p'$ is an auxiliary probability used to construct the up and down factors. The resulting Leisen–Reimer factors are:

$$
u=e^{r\delta\tau}\frac{p'}{p},\qquad
d=e^{r\delta\tau}\frac{1-p'}{1-p}.
$$

These factors satisfy the same risk-neutral growth condition as CRR. The discounted backward calculation and the American exercise decision stay the same. Because the factors depend on the strike and market inputs, each perturbed valuation requires a new tree. We will compare the computed Greeks across tree depths and perturbation sizes; an American exercise boundary can still affect their smoothness.

### A European benchmark
For a non-dividend-paying underlying asset, differentiating the L9a BSM formulas gives an analytic benchmark. Let $N$ and $\phi$ be the standard Normal CDF and density, respectively. The call and put Greeks are:

$$
\begin{align*}
\Delta_c&=N(d_1), & \Delta_p&=N(d_1)-1,\\
\Gamma_c&=\Gamma_p=\frac{\phi(d_1)}{S\sigma\sqrt\tau}, &
\nu_c&=\nu_p=S\phi(d_1)\sqrt\tau,\\
\Theta_c&=-\frac{S\phi(d_1)\sigma}{2\sqrt\tau}-rKe^{-r\tau}N(d_2), &
\Theta_p&=-\frac{S\phi(d_1)\sigma}{2\sqrt\tau}+rKe^{-r\tau}N(-d_2),\\
\rho_c&=K\tau e^{-r\tau}N(d_2), &
\rho_p&=-K\tau e^{-r\tau}N(-d_2).
\end{align*}
$$

Under these assumptions and nonnegative interest rates, early exercise adds no value to a call, so the American call also supplies a BSM check. For the American put, we use a refined tree calculation and report the difference between successive refinements. We compare broker quotes separately because their dividend, rate, and numerical conventions can differ.

## The Secret Lives of Delta $\Delta$
The $\Delta$ of an option measures how much the price of the option changes locally for a `+1 USD/share` change in the price of the underlying asset. It also determines the local hedge ratio; a separate probability interpretation requires care:

> __Delta Interpretations:__
> 
> * __Hedging Ratio:__ Delta can be interpreted as the number of shares of the underlying asset needed to hedge one option contract. For example, if an option contract has $\Delta$, it means that for a signed position of $n$ contracts, with $n>0$ for long and $n<0$ for short, an investor would need to hold $-{100}\cdot{n}\cdot\Delta$ shares of the underlying asset to be delta-neutral. A delta-neutral position is one where the overall portfolio value is insensitive to small changes in the price of the underlying asset.
> * __Probability heuristic:__ Traders commonly use call Delta as an approximate implied probability of expiring ITM and $1-\Delta$ as the approximate probability of expiring OTM. Keep the event labels straight and remember that this is a heuristic, not an exact probability identity.

We've already seen the hedge-ratio interpretation of $\Delta$. We now compute terminal ITM probability directly and compare it with Delta to see why the two quantities should not be identified.

> __Terminal In-the-Money Probability:__
>
> The probability that a call or put finishes in the money at expiration $T$ is given by:
> $$
\boxed{
\begin{aligned}
\mathbb{P}(\text{terminal ITM}) & = \begin{cases}
\mathbb{P}\left(S_{T} > K + 0.01\right) & \text{(for Call Options)} \\
\mathbb{P}\left(S_{T} < K - 0.01\right) & \text{(for Put Options)}
\end{cases}
\end{aligned}}
$$
> where $S_{T}$ is the share price of the underlying asset at expiration and $K$ is the strike price. Actual exercise or assignment—especially before expiration for an American option—also depends on holder behavior, dividends, borrowing constraints, and exercise rules.

For a GBM projection over remaining time $\tau$, the endpoint distribution is specified by:

$$
\ln S(T)\sim\mathcal{N}\!\left(\ln S+(\mu-\sigma^2/2)\tau,\;\sigma^2\tau\right).
$$

The drift $\mu=r$ gives a risk-neutral distribution under our no-dividend assumptions. A real-world projection requires a separately justified drift and volatility. The TSLA example checks simulated endpoints against this analytic distribution and compares probabilities under several stated assumptions.

We can also compute the __Probability of Profit (POP)__ for an option contract, which is closely related to the probability of exercise.

> __Probability of Profit (POP):__ 
>
> The probability of profit (POP) is the probability that the contract profit is positive at expiration $T$. It is given by:
> $$
\boxed{
\begin{aligned}
\text{POP} & = \mathbb{P}\left(\hat{P}_{\mathcal{C}} > 0\right)
\end{aligned}}
$$
> where $\hat{P}_{\mathcal{C}}$ is the profit of the (single or composite) contract at expiration.

> __Key Takeaways:__
>
