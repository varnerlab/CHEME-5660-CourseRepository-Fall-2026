## Many Assets: The Sector Factor Model
Previously, we studied __factor models__ which describe the growth rate in terms of factors. For stock $i$ on day $t$, a general factor model is given by:

$$
g_{i}(t) = \underbrace{\alpha_{i}}_{\text{baseline}} + \underbrace{\beta_{i}\,g_{m}(t)}_{\text{market}} + \underbrace{\sum_{j}\delta_{ij}\,f_{j}(t)}_{\text{other factors}} + \underbrace{\varepsilon_{i}(t)}_{\text{residual}}
$$

Here $\alpha_{i}$ is the stock-specific baseline (intercept), $g_{m}(t)$ is market growth, and $f_{j}(t)$ are other factors, such as interest rates or inflation. The coefficients $\beta_{i}$ and $\delta_{ij}$ measure sensitivity to these factors. The __residual__ $\varepsilon_{i}(t)$ is the growth left after subtracting the baseline and factor contributions.

> __Definition (sector factor model):__ Each stock has one other factor besides the market, the factor $f_{s}(t)$ of its sector $s$. This factor captures co-movement left unexplained by the market.
>
> The daily growth rate of stock $i$ in sector $s$ is given by:
>
> $$
> g_{i}(t) = \underbrace{\alpha_{i}}_{\text{baseline}} + \underbrace{\beta_{i}\,g_{m}(t)}_{\text{market}} + \underbrace{\delta_{i}\,f_{s}(t)}_{\text{sector}} + \underbrace{\psi_{i}\,\tilde{\varepsilon}_{i}(t)}_{\text{stock residual}}
> $$
>
> The four components are:
> * __Baseline:__ The stock-specific baseline $\alpha_{i}$ is the intercept from the stock's single index fit. We hold $\alpha_{i}$ and the market sensitivity $\beta_{i}$ fixed.
> * __Market:__ The market draw $g_{m}(t)$ comes from a jump hidden Markov model of `SPY`.
> * __Sector:__ The sector draw $f_{s}(t)$ comes from a jump hidden Markov model of the average single index residual of the stocks in sector $s$. The loading $\delta_{i}$ is the slope of the stock's residual on the other residuals in its sector, scaled to match the sector's co-movement in the data.
> * __Stock-specific residual:__ The component $\psi_{i}\,\tilde{\varepsilon}_{i}(t)$ models moves left after removing the baseline, market, and sector contributions, such as an earnings surprise. We center a draw from a jump hidden Markov model of this remaining residual and size it with $\psi_{i}$.
>
> On each day, every stock shares the market draw, and every stock in a sector shares the sector draw. Each stock's residual path is sampled independently of the market, the sectors, and the other stocks' residual paths. The shared market and sector draws are the only link between stocks.

Let's write out how we construct the stock-specific residual component:

> __The stock-specific residual component (fixed β):__
>
> In the data, we subtract the market fit from the observed growth rate. We then subtract the sector contribution:
>
> $$
> \begin{aligned}
> \varepsilon_{i}(t) &= g_{i}(t)-\alpha_{i}-\beta_{i}g_{m}(t)
> && \text{residual after the market fit} \\
> u_{i}(t) &= \varepsilon_{i}(t)-\delta_{i}f_{s}(t)
> && \text{residual after market and sector}.
> \end{aligned}
> $$
>
> We fit a jump hidden Markov model to the series $u_{i}(t)$. Let $u_{i}^{\mathrm{draw}}(t)$ be a draw from that model and $\bar{u}_{i}$ its long-run mean. We center the draw on zero:
>
> $$
> \tilde{\varepsilon}_{i}(t) = u_{i}^{\mathrm{draw}}(t)-\bar{u}_{i}.
> $$
>
> The simulated sector and stock residual paths are independent. Their variances add. We use the variances in the real data to set a target for their sum:
>
> $$
> \underbrace{\delta_{i}^{2}\operatorname{Var}(f_{s})}_{\text{sector variance}}
> + \underbrace{\psi_{i}^{2}\operatorname{Var}(u_{i})}_{\text{target stock residual variance}}
> = \underbrace{\operatorname{Var}(\varepsilon_{i})}_{\text{variance left after the market fit}}.
> $$
>
> Solving for the scale gives:
>
> $$
> \boxed{\psi_{i}^{2} =
> \frac{\operatorname{Var}(\varepsilon_{i})-\delta_{i}^{2}\operatorname{Var}(f_{s})}
> {\operatorname{Var}(u_{i})}}
> \qquad
> \psi_{i} \geq 0.
> $$
>
> The sector factor includes stock $i$ in its average. The remaining series $u_{i}$ can still move with that factor in the data. The scale compensates for the covariance lost when we draw the two paths independently.
>
> The variance equation sets a target. The fitted jump hidden Markov models reproduce the data variances only approximately.

### Sampling Algorithm
The algorithm below draws one synthetic path of the whole universe over $T$ days.

__Initialize__: Given the jump hidden Markov models of the market, of each sector's factor, and of each stock's residual after its sector term; each stock's single index estimates $\alpha_{i}$ and $\beta_{i}$ and sector loading $\delta_{i}$; the scale of each stock's residual component, $\psi_{i}^{2} = \left(\operatorname{Var}(\varepsilon_{i}) - \delta_{i}^{2}\operatorname{Var}(f_{s})\right)/\operatorname{Var}(\varepsilon_{i}-\delta_{i}f_{s})$, with the variances of the real data; each stock's real starting price $S_{i}(0)$; and the number of days $T$. 

Draw one market path $g_{m}(1),\dotsc,g_{m}(T)$, shared by all stocks, from the market model. For each sector $s$, draw one factor path $f_{s}(1),\dotsc,f_{s}(T)$, shared by the stocks in that sector, from its sector model. Use independent random draws to generate the different paths.

For each stock $i$ in sector $s$ __do__:
1. __Residual draw:__ Draw a path from the model of the stock's remaining residual and subtract the model's long-run mean (the mean of one long simulated path) to get $\tilde{\varepsilon}_{i}(1),\dotsc,\tilde{\varepsilon}_{i}(T)$.
2. __Compose:__ For each day $t = 1,2,\dots,T$ __do__: compose the growth rate $g_{i}(t) \gets \alpha_{i} + \beta_{i}\,g_{m}(t) + \delta_{i}\,f_{s}(t) + \psi_{i}\,\tilde{\varepsilon}_{i}(t)$, and compound the price, starting from $S_{i}(0)$: $S_{i}(t) \gets S_{i}(t-1)\exp\left(g_{i}(t)\,\Delta t\right)$.

__Output__: The market path $g_{m}(t)$, and the growth rates $g_{i}(t)$ and prices $S_{i}(t)$ of every synthetic stock, for $t = 1,\dotsc,T$.

__Notes:__
1. The sector factor averages the residuals of the stocks in sector $s$, stock $i$ included. Independent draws drop the part of the stock's leftover residual that moves with the sector factor, and $\psi_{i}$ puts that part back. In a sector with many stocks, $\psi_{i}$ is close to one.
2. With the drifting β option below, $\beta_{i}$ becomes $\beta_{i}(t)$ in step 2. We also reduce $\psi_{i}$ to preserve the variance target.

### An option: a drifting β
The coefficient $\beta_{i}$ measures stock $i$'s sensitivity to market growth. Holding $\beta_{i}$ fixed assumes that stock $i$'s sensitivity to market growth rate stays constant throughout a synthetic path. Alternatively, we can let the sensitivity change over time around its single index estimate $\hat{\beta}_{i}$. 

Each day, we update it with:

$$
\beta_{i}(t) = \hat{\beta}_{i}
+ \underbrace{(1-a)\left(\beta_{i}(t-1)-\hat{\beta}_{i}\right)}_{\text{retained deviation}}
+ \underbrace{a\,k_{i}\,z_{i}(t)}_{\text{new shock}}.
$$

Here $0<a<1$ is the exponential moving average weight. A small $a$ retains most of the previous day's deviation from $\hat{\beta}_{i}$.

The shocks $z_{i}(t)$ are independent standard normal draws, with their size controlled by $k_{i}\geq 0$. In the example, we choose $k_{i}$ using the variation in β estimates updated daily from real data. These estimates also contain sampling noise.

The market contribution is now $\beta_{i}(t)g_{m}(t)$. Let's look at an example:

## Building an Adaptive Trading System
A utility maximization tells us what to hold at one set of prices and model estimates. Once we buy those shares, prices keep moving and new observations change the estimates. Repeating the allocation requires rules for what to update and when to trade.

The system we build comes from J. D. Varner's INFORMS 2026 study, [Interpretable Allocation and Model-Predictive Rebalancing of Equity Portfolios](https://arxiv.org/abs/2611.03022). The [study repository](https://github.com/varnerlab/INFORMS-Annual-Meeting-2026-Varnerlab) contains its code and results.

The stocks and cash we hold form the __book__. Let $n_i$ be the shares held of stock $i$ in the basket $\mathcal N$, $S_i(t)$ its close price (units: USD/share), and $c(t)$ the cash balance (units: USD). The measured wealth of the book at the close of trading day $t$ is given by:

$$
W(t)=\sum_{i\in\mathcal N}n_iS_i(t)+c(t).
$$

The __controller__ is the system that decides, each trading day, what to update and whether to trade:

## Portfolio Allocation with Trading Limits
The controller chooses target holdings in two steps. A __selection rule__ picks a basket $\mathcal N$ of $K$ stocks. A __weighting rule__ then chooses how many shares of each stock to hold.

The study ranks stocks by a score that combines each stock's expected growth with the direction of the market:

> __Definition (stock score):__ Let $\alpha_i$ and $\beta_i$ be the single index intercept (units: $\mathrm{yr}^{-1}$) and market sensitivity of stock $i$. Let $\tilde g_M$ be the 252-day exponential moving average of SPY's daily growth rate (units: $\mathrm{yr}^{-1}$). The score $\gamma_i$ and the market state $\xi$ are given by:
>
> $$
> \begin{aligned}
> \gamma_i&=\tanh\left(h\,\frac{\alpha_i+\beta_i\tilde g_M}{\max(|\beta_i|,0.1)^{\xi}}\right),\\
> \xi&=-G\left(\frac{\mathrm{EMA}_{21}}{\mathrm{EMA}_{63}}-1\right),
> \end{aligned}
> $$
> where $\mathrm{EMA}_{21}$ and $\mathrm{EMA}_{63}$ are the 21-day and 63-day exponential moving averages of SPY's close price. The scale $h=1$ year makes the argument of $\tanh$ dimensionless. The gain $G>0$ keeps $\xi$ between $-1$ and $1$ on 99% of the calibration days.

__How does the score work?__ The numerator is the stock's expected growth rate under the single index model, and the score has its sign. When SPY's 21-day average falls below its 63-day average, the market is falling and $\xi>0$. For a stock with positive expected growth, dividing by $|\beta_i|^{\xi}$ then favors a small $|\beta_i|$, because such a stock falls less with the market.

In a rising market, $\xi<0$ and the tilt favors stocks with large $|\beta_i|$. The floor of 0.1 keeps the denominator away from zero, and $\tanh$ keeps every score between $-1$ and $1$.

The study's selection by score keeps the three highest-scoring stocks in each of the 11 sectors, 33 stocks in all. We hold that basket fixed and turn to the weighting rule.

The scores become the preference weights of a Cobb–Douglas utility:

> __Proposition (Cobb–Douglas allocation):__ Consider a basket with positive scores, today's close prices $S_i$ (units: USD/share), and current wealth $W$ (units: USD). Maximizing $\sum_{i\in\mathcal N}\gamma_i\ln n_i$ over share counts $n_i>0$ and cash $c\geq0$, subject to the budget $\sum_{i\in\mathcal N}n_iS_i+c=W$, gives:
>
> $$
> \boxed{w_i^\star=\frac{\gamma_i}{\sum_{j\in\mathcal N}\gamma_j},\qquad n_i^\star=\frac{W}{S_i}\,w_i^\star,\qquad c^\star=0,}
> $$
> where $w_i^\star=n_i^\star S_i/W$ is the target fraction of wealth in stock $i$. The first-order conditions equalize marginal utility per dollar, $\gamma_i/(n_iS_i)$, across stocks.

Each stock's target weight equals its score divided by the total score. We can trace every target weight back to a stock's expected growth and market sensitivity, which is why the study calls the allocation __interpretable__.

The controller also limits concentration, volatility, and trading costs:

> __Definition (constrained utility allocation):__ Let $n_i^{\mathrm{prev}}$ be the shares the book holds before the re-plan. Choose share counts $n_i>0$ and cash $c\geq0$, with target weights $w_i=n_iS_i/W$ and a utility $U(n;\gamma)$ from the table below, to solve:
>
> $$
> \boxed{\begin{aligned}
> \operatorname*{maximize}_{n_1,\ldots,n_K>0,\,c\geq0}\quad &U(n;\gamma)\\
> \text{subject to}\quad &\sum_{i\in\mathcal N}n_iS_i+c=W &&\text{(budget)}\\
> &w_i\leq\bar w\quad\text{for each }i &&\text{(concentration)}\\
> &\sqrt{w^\top\Sigma_gw}\leq\sigma_{\mathrm{SPY}} &&\text{(volatility)}\\
> &\bar c\sum_{i\in\mathcal N}|n_i-n_i^{\mathrm{prev}}|\leq B &&\text{(trading cost)}\\
> &\sum_{i\in\mathcal N}\beta_iw_i\leq\bar\beta &&\text{(optional hedge)}.
> \end{aligned}}
> $$
>
> The concentration limit $\bar w$ is the largest fraction of wealth one stock may hold. The volatility limit keeps the book's growth rate from varying more than SPY's, where $\Sigma_g$ is the covariance of the stocks' daily growth rates (units: $\mathrm{yr}^{-2}$) and $\sigma_{\mathrm{SPY}}$ is SPY's growth-rate standard deviation over the calibration period (units: $\mathrm{yr}^{-1}$).
>
> Each share bought or sold costs $\bar c$ (units: USD/share), an estimate of half the bid–ask spread. The total trading cost may not exceed $B$ (units: USD). The optional hedge limits the book's market sensitivity to $\bar\beta=\max(0,1-\xi)$, which tightens as the market falls.

A numerical solver finds the constrained optimum. Cash can remain when a limit stops further purchases.

The study compares four utilities, whose forms and budget-only allocations are:

$$
\begin{array}{lcc}
\hline
\text{Utility} & U(n;\gamma) & \text{Budget-only allocation} \\[5pt]
\hline
\text{Cobb–Douglas} & \displaystyle\sum_i\gamma_i\ln n_i & w_i\propto\gamma_i \\[9pt]
\text{CES on shares} & \displaystyle\left(\sum_i\gamma_i n_i^{\rho}\right)^{1/\rho} & n_i\propto(\gamma_i/S_i)^{\eta} \\[9pt]
\text{CES on weights} & \displaystyle\left(\sum_i\gamma_i w_i^{\rho}\right)^{1/\rho} & w_i\propto\gamma_i^{\eta} \\[9pt]
\text{Stone–Geary} & \displaystyle\sum_i\gamma_i\ln(n_i-b_i) & (n_i-b_i)S_i\propto\gamma_i \\[5pt]
\hline
\end{array}
$$

The two constant elasticity of substitution (CES) utilities use the elasticity $\eta=0.5$, with $\rho=1-1/\eta$. The Stone–Geary utility keeps a baseline of $b_i$ shares in each stock, worth 1% of wealth.

If every score is negative and the hedge is off, Cobb–Douglas and CES hold only cash. Stone–Geary keeps its baseline shares.

## Keeping the Estimates Current
The score uses the single index estimates $\alpha_i$ and $\beta_i$. A fit made once would give the controller a fixed view of each stock. We use __exponentially weighted least squares__ (EWLS) to revise the fit after every close.

> __Definition (online single index estimation):__ Let $g_i(s)$ and $g_M(s)$ be the daily growth rates of stock $i$ and SPY on trading day $s$ (units: $\mathrm{yr}^{-1}$). For a trial intercept $\alpha$ and market sensitivity $\beta$, the weighted squared-error loss is:
>
> $$
> \begin{aligned}
> \mathcal L_{i,t}(\alpha,\beta)
> &=\lambda^t\mathcal L_{i,0}(\alpha,\beta)
> +\sum_{s=1}^{t}\lambda^{t-s}\left(g_i(s)-\alpha-\beta g_M(s)\right)^2,\\
> \lambda&=2^{-1/h_{\mathrm{EW}}},\qquad h_{\mathrm{EW}}=63\;\text{trading days},
> \end{aligned}
> $$
> where $\mathcal L_{i,0}$ is the loss of the initial calibration, counted as 63 days of data. Minimizing $\mathcal L_{i,t}$ gives the updated estimates $\alpha_i(t)$ and $\beta_i(t)$. The decay $\lambda$ halves an observation's weight every 63 trading days.

A shorter half-life follows changes faster but gives each observation more influence. A longer half-life averages over more history. Each update uses only prices available at that close.

The smoothed market growth $\tilde g_M$ and the market state $\xi$ also update after every close. At a re-plan, the controller rebuilds $\Sigma_g$ from the current $\beta_i$, the market's growth-rate variance, and the last 126 days of single index residuals.

## Model-Predictive Rebalancing
The controller decides when to trade by comparing the book with a forecast:

> __Model-predictive rebalancing:__ A controller uses a model to predict the consequences of a plan over a finite horizon. It compares new measurements with that prediction and makes a new plan from the measured state when a trigger fires.

Here, the plan is to hold the target shares for the next $T=21$ trading days. The forecast of that plan stays fixed until the next re-plan, while the estimates keep updating.

### Forecast the wealth of holding the target shares
Let $t_0$ be the re-plan day, $n_i^\star$ the target shares, and $c^\star=W(t_0)-\sum_i n_i^\star S_i(t_0)$ the cash left at that day's prices. A forecast path keeps these share counts fixed. Cash grows at the risk-free rate $g_f$ (units: $\mathrm{yr}^{-1}$) of the re-plan day.

The forecast uses a multi-asset geometric Brownian motion with time step $\Delta t=1/252$ year.

For path $p$ and forecast day $\tau=1,\ldots,T$, the controller draws the growth rates and advances each price as follows:

$$
\begin{aligned}
\mathbf g^{(p)}(\tau)&\sim\mathcal N(\boldsymbol\mu_g,\Sigma_g)
&&\text{draw the stocks' growth rates together},\\
\mu_{g,i}&=\alpha_i+\beta_i\tilde g_M
&&\text{expected growth at the re-plan},\\
S_i^{(p)}(\tau)&=S_i^{(p)}(\tau-1)\exp\left(g_i^{(p)}(\tau)\Delta t\right)
&&\text{compound from }S_i^{(p)}(0)=S_i(t_0).
\end{aligned}
$$

The mean $\boldsymbol\mu_g$ and covariance $\Sigma_g$ stay at their re-plan values. Each day and each path uses an independent draw.

The wealth on path $p$ is given by:

$$
W^{(p)}(\tau)=\sum_i n_i^\star S_i^{(p)}(\tau)+c^\star\exp(g_f\tau\Delta t).
$$

We draw $N_p=1{,}000$ paths. On each forecast day $\tau$, the paths give a sample mean $\mu_W(\tau)$ and standard deviation $\sigma_W(\tau)$ of wealth (units: USD).

Under a normal approximation, the 95% __forecast band__ is given by:

$$
\boxed{\mu_W(\tau)-1.96\sigma_W(\tau)\;\leq\;W(t_0+\tau)\;\leq\;\mu_W(\tau)+1.96\sigma_W(\tau).}
$$

### Three reasons to make a new plan
After each close, the controller updates the peak wealth $W^{\mathrm{pk}}$ and counts the trading days $\tau$ since the last re-plan. It then checks three triggers in order:

1. __Drawdown:__ Re-plan if wealth has fallen more than $D_{\max}=8\%$ below its peak, $1-W(t)/W^{\mathrm{pk}}>D_{\max}$.
2. __Horizon:__ Re-plan if $\tau\geq T$, because the forecast has reached its end.
3. __Band exit:__ Re-plan if measured wealth lies outside the forecast band for day $\tau$.

After a drawdown re-plan, the peak resets to current wealth. The next drawdown trigger then responds only to a new loss.

__Why check both sides of the band?__ An unusually large gain also means the forecast no longer describes the book. The re-plan recomputes the best holdings from current estimates. The trigger itself does not say whether to buy or sell.

### The daily controller algorithm

__Initialize__: Given the daily prices and risk-free growth rates, a fixed basket $\mathcal N$, the initial model estimates, the starting wealth $W_0$, the horizon $T$, the drawdown threshold $D_{\max}$, the number of forecast paths $N_p$, and the final trading day $t_{\mathrm{end}}$.

Set $n_i\gets0$ shares for every stock $i\in\mathcal N$, cash $c\gets W_0$, the tracked peak $W^{\mathrm{pk}}\gets W_0$, and the forecast age $\tau\gets0$. Start with no forecast and no pending orders.

For each trading day $t=1,2,\dots,t_{\mathrm{end}}$ __do__:
1. __Open:__ If orders are pending, __then__ fill them at the open, selling first and scaling purchases to available cash after costs. Update the shares $n_i$ and cash $c$, then clear the pending orders.
2. __Advance:__ If $t>1$, __then__ accrue cash $c\gets c\exp\left(g_f(t)\Delta t\right)$ and advance the forecast age $\tau\gets\tau+1$.
3. __Measure:__ At the close, compute the wealth $W(t)\gets\sum_{i\in\mathcal N}n_iS_i(t)+c$ and update the tracked peak $W^{\mathrm{pk}}\gets\max\left(W^{\mathrm{pk}},W(t)\right)$.
4. __Estimate:__ Update the single index estimates, $\tilde g_M$, and $\xi$ using observations through this close. Compute the stock scores $\gamma_i$.
5. __Trigger:__ If $t=1$, __then__ the re-plan reason is the initial allocation. Otherwise, the reason is the first trigger that holds:
    - If $1-W(t)/W^{\mathrm{pk}}>D_{\max}$, __then__ the reason is drawdown.
    - If $\tau\geq T$, __then__ the reason is horizon.
    - If a forecast band exists and $W(t)$ lies outside it for day $\tau$, __then__ the reason is band exit.
6. __Re-plan:__ If no trigger holds, __then__ keep the shares and the forecast. Otherwise, make a new plan:
    - a. Solve the constrained allocation from the current account for target shares $n_i^\star$ and cash $c^\star$. If the solve fails, __then__ set $n_i^\star\gets n_i$ and $c^\star\gets c$ and place no orders. Otherwise, if $t<t_{\mathrm{end}}$, __then__ queue the orders $\Delta n_i\gets n_i^\star-n_i$ for the next open.
    - b. Simulate $N_p$ wealth paths of the target holdings over $T$ days. If the target book holds stock, __then__ store the forecast band. Otherwise, clear the band. An all-cash book has no band to exit, and the horizon trigger revisits it after $T$ days.
    - c. Set $\tau\gets0$. If the reason is drawdown, __then__ reset the tracked peak $W^{\mathrm{pk}}\gets W(t)$.
7. __Record:__ Store the day's wealth, holdings, trades, forecast, and re-plan reason.

__Output__: The daily account history and the re-plan log.

## Summary
We built a trading system that updates its estimates every day and trades only when a trigger fires. We then tested which of its choices mattered on real and synthetic markets.

> __Key Takeaways:__
>
> * __Adaptation begins with the scores:__ We updated the single index fit with exponentially weighted least squares after every close. The scores combined those estimates with the market state, while the budget and trading limits set the target holdings.
> * __Feedback determines when a target becomes a trade:__ We compared measured wealth with the forecast of holding the current plan. A drawdown, the horizon, or a band exit started a new allocation, with orders filled at the next open.
> * __Matched tests identify what each choice adds:__ On one real path, the basket mattered more than the weighting rule. In synthetic sell-offs, every book's median ended above SPY, random baskets included.
