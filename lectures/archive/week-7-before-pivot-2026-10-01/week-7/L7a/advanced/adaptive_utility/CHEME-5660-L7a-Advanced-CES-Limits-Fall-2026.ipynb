## The CES Utility and Budget Constraint
Recall the setting of the lecture: on a trading day $t$, asset $i$ has price $S_{i}(t)$ (USD per share), the investor holds $n_{i}$ shares, and the preferred set is $\mathcal{A}^{+} = \{i : \gamma_{i} > 0\}$ with preference weights $\gamma_{i}\in(0,1)$ on it; the budget available to the preferred set is $W_{\text{adj}}$ (USD). To keep the algebra light we write the CES exponent as $\rho\equiv(\eta-1)/\eta$ throughout this notebook (this $\rho$ is a local shorthand for the exponent, not the correlation of L5a). The CES utility over the share counts $\mathbf{n} = (n_{1},\ldots,n_{|\mathcal{P}|})$ is:
$$
U_{\text{CES}}(\mathbf{n}) = \left(\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{\rho}\right)^{1/\rho},\qquad\rho = \frac{\eta-1}{\eta}
$$
where $\eta > 0$ is the elasticity of substitution. The allocator maximizes $U_{\text{CES}}$ subject to the budget constraint:
$$
\sum_{i\in\mathcal{A}^{+}}S_{i}(t)\,n_{i} = W_{\text{adj}}
$$
The Lagrangian for this constrained maximization, with multiplier $\mu$ on the budget, is:
$$
\mathcal{L}(\mathbf{n},\mu) = \left(\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{\rho}\right)^{1/\rho} - \mu\left(\sum_{i\in\mathcal{A}^{+}}S_{i}(t)\,n_{i} - W_{\text{adj}}\right)
$$
___

## First-Order Conditions
Let $\Psi = \sum_{j\in\mathcal{A}^{+}}\gamma_{j}\,n_{j}^{\rho}$ so that $U = \Psi^{1/\rho}$. Differentiating $U$ with respect to $n_{i}$ by the chain rule gives:
$$
\frac{\partial U}{\partial n_{i}} = \frac{1}{\rho}\,\Psi^{1/\rho-1}\cdot\rho\,\gamma_{i}\,n_{i}^{\rho-1} = \Psi^{1/\rho-1}\,\gamma_{i}\,n_{i}^{\rho-1}
$$
Setting $\partial\mathcal{L}/\partial n_{i} = 0$ gives the first-order condition for each preferred asset $i\in\mathcal{A}^{+}$:
$$
\Psi^{1/\rho-1}\,\gamma_{i}\,n_{i}^{\rho-1} = \mu\,S_{i}(t)
$$
Taking the ratio of the conditions for assets $i$ and $j$ cancels the common factor $\Psi^{1/\rho-1}$:
$$
\frac{\gamma_{i}\,n_{i}^{\rho-1}}{\gamma_{j}\,n_{j}^{\rho-1}} = \frac{S_{i}(t)}{S_{j}(t)}
$$
Rearranging and using $\rho - 1 = -1/\eta$ gives:
$$
n_{i}^{-1/\eta}\propto\frac{S_{i}(t)}{\gamma_{i}}\quad\Longleftrightarrow\quad n_{i}\propto\left(\frac{\gamma_{i}}{S_{i}(t)}\right)^{\eta}
$$
Imposing the budget constraint $\sum_{i}S_{i}(t)\,n_{i} = W_{\text{adj}}$ and solving for the proportionality constant yields the closed-form CES share-count optimum:
$$
\boxed{\;n_{i}^{\star} = \frac{(\gamma_{i}/S_{i}(t))^{\eta}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)\,(\gamma_{j}/S_{j}(t))^{\eta}}\,W_{\text{adj}}\quad\blacksquare\;}
$$
The corresponding dollar weights $w_{i}^{\star} = S_{i}(t)\,n_{i}^{\star}/W_{\text{adj}}$ have the standard CES expenditure-share form:
$$
\boxed{\;w_{i}^{\star} = \frac{S_{i}(t)^{1-\eta}\,\gamma_{i}^{\eta}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)^{1-\eta}\,\gamma_{j}^{\eta}}\;}
$$
which is what the course package computes in `allocate_ces`. Because $\rho < 1$ for every finite $\eta > 0$, the CES utility is concave on the positive orthant, so the stationary point is the maximum.
___

## Limit 1: $\eta\to 1$ Recovers Cobb–Douglas
We first verify that the CES utility itself converges to Cobb–Douglas. For $\eta\to 1$ we have $\rho\to 0$, and the limit
$$
\lim_{\rho\to 0}\left(\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{\rho}\right)^{1/\rho}
$$
is an indeterminate form of type $1^{\infty}$ when the preference weights are normalized so that $\sum_{i}\gamma_{i} = 1$ (if they are not, the sum tends to $\sum_{i}\gamma_{i}\neq 1$ and the limit is $0$ or $\infty$; only preferences matter for the allocation, so we normalize). Taking the logarithm gives:
$$
\ln U_{\text{CES}} = \frac{1}{\rho}\ln\left(\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{\rho}\right)
$$
Both numerator and denominator vanish as $\rho\to 0$. Applying L'Hôpital's rule with respect to $\rho$ yields:
$$
\lim_{\rho\to 0}\ln U_{\text{CES}} = \left.\frac{\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{\rho}\ln n_{i}}{\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{\rho}}\right|_{\rho=0} = \sum_{i\in\mathcal{A}^{+}}\gamma_{i}\ln n_{i}
$$
Exponentiating recovers the Cobb–Douglas form:
$$
\boxed{\;\lim_{\eta\to 1}U_{\text{CES}}(\mathbf{n}) = \prod_{i\in\mathcal{A}^{+}}n_{i}^{\gamma_{i}}\quad\blacksquare\;}
$$
The optimum collapses correspondingly, and this limit needs no normalization. Setting $\eta = 1$ in the dollar-weight formula gives:
$$
w_{i}^{\star} = \frac{S_{i}(t)^{0}\,\gamma_{i}}{\sum_{j}S_{j}(t)^{0}\,\gamma_{j}} = \frac{\gamma_{i}}{\sum_{j\in\mathcal{A}^{+}}\gamma_{j}}
$$
so dollar weights track preference weights and are independent of prices, which is the lecture's Cobb–Douglas closed form. At the optimum the marginal utility per dollar, $\gamma_{i}/(S_{i}(t)\,n_{i}^{\star})$, is the same for every preferred asset.
___

## Limit 2: $\eta\to\infty$ Concentrates on the Best Asset
For $\eta\to\infty$ we have $\rho\to 1$, and the CES utility becomes linear in the share counts:
$$
\boxed{\;\lim_{\eta\to\infty}U_{\text{CES}}(\mathbf{n}) = \sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}\;}
$$
To see the optimum, write $r_{i} = \gamma_{i}/S_{i}(t)$ for the bang-for-the-buck ratio of asset $i$, let $r^{\star} = \max_{i\in\mathcal{A}^{+}}r_{i}$, and assume the maximizer $i^{\star}$ is unique. Factoring $r^{\star}$ out of the numerator and denominator of the closed form gives:
$$
n_{i}^{\star} = \frac{(r_{i}/r^{\star})^{\eta}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)\,(r_{j}/r^{\star})^{\eta}}\,W_{\text{adj}}
$$
For $i\neq i^{\star}$ the ratio $r_{i}/r^{\star} < 1$, so $(r_{i}/r^{\star})^{\eta}\to 0$ as $\eta\to\infty$, while the $i^{\star}$ term equals one. The optimum concentrates on the single best asset:
$$
\boxed{\;\lim_{\eta\to\infty}n_{i}^{\star} = \begin{cases}W_{\text{adj}}/S_{i^{\star}}(t) & i = i^{\star}\\ 0 & \text{otherwise}\end{cases}\quad\blacksquare\;}
$$
This is the linear-programming corner solution: when utility is linear in share counts, the budget-constrained maximum spends every dollar on whichever asset has the highest preference weight per dollar of price. If two assets tie for $r^{\star}$, the limit spreads the budget over the tied set rather than one asset.
___

## Limit 3: $\eta\to 0^{+}$ Equalizes Share Counts
For $\eta\to 0^{+}$ we have $\rho\to-\infty$. With $\rho < 0$, the term $n_{i}^{\rho} = 1/n_{i}^{|\rho|}$ is largest for the smallest $n_{i}$, so the sum $\sum_{j}\gamma_{j}\,n_{j}^{\rho}$ is dominated by the term with the smallest share count, and taking the $1/\rho$ power with $\rho\to-\infty$ extracts the minimum:
$$
\boxed{\;\lim_{\eta\to 0^{+}}U_{\text{CES}}(\mathbf{n}) = \min_{i\in\mathcal{A}^{+}}n_{i}\;}
$$
up to a constant factor that does not change the maximizer. This is the Leontief, or perfect-complements, form. The optimum follows from the closed form: setting $\eta = 0$ gives $(\gamma_{i}/S_{i}(t))^{0} = 1$ for every preferred asset, so:
$$
\boxed{\;\lim_{\eta\to 0^{+}}n_{i}^{\star} = \frac{W_{\text{adj}}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)}\quad\blacksquare\;}
$$
for every $i\in\mathcal{A}^{+}$: all preferred assets receive the same share count, independent of $\gamma_{i}$ or price. In dollar-weight terms $w_{i}^{\star} = S_{i}(t)/\sum_{j}S_{j}(t)$, which is price-weighted rather than equal-dollar. Equal allocation here means equal share counts, not equal dollar exposure.
___

## The Elasticity Rule
The engine ties the elasticity to the market-state signal $\xi_{t}$ of the lecture through the rule $\eta(\xi_{t}) = \eta_{\min} + (\eta_{\max}-\eta_{\min})/(1+|\xi_{t}|)$ with $0 < \eta_{\min}\leq\eta_{\max}$. Its two limits follow at once. As $\xi_{t}\to 0$ the denominator tends to one and $\eta(\xi_{t})\to\eta_{\max}$: with no signal the allocator behaves like the high-elasticity limit above, concentrating on the best bang for the buck (with the defaults $\eta_{\max} = 5$ this is strong concentration, not the corner solution). As $|\xi_{t}|\to\infty$ the fraction vanishes and $\eta(\xi_{t})\to\eta_{\min}$: a loud signal in either direction pushes the allocator toward the low-elasticity limit, spreading the budget in case the signal is noise (with $\eta_{\min} = 0.5$ this is diversification, not equal share counts). The rule is symmetric in the sign of the signal by design: what the sign does is handled by the preference weights $\gamma_{i}(t)$, and what the elasticity does is decide how aggressively to act on them.
___

## Log-Linear Utility Has the Cobb–Douglas Optimizer
The log-linear utility is $U_{\log}(\mathbf{n}) = \sum_{i\in\mathcal{A}^{+}}\gamma_{i}\ln n_{i}$, the logarithm of the Cobb–Douglas utility. Because the logarithm is strictly increasing on the positive reals, $U_{\log}(\mathbf{n}_{1}) > U_{\log}(\mathbf{n}_{2})$ if and only if $U(\mathbf{n}_{1}) > U(\mathbf{n}_{2})$, so the two utilities rank every pair of feasible portfolios the same way and have the same maximizer under the same budget: the Cobb–Douglas closed form. What changes is the __value__ of the utility, which becomes additive across assets. That is convenient when utility values are averaged across scenarios, compared between portfolios of different sizes, or used as a reward that a learning algorithm accumulates, which is how it will be used later in the course.
___

## A Caveat: Share Denomination
The CES closed form makes one property visible that the Cobb–Douglas form hides. The Cobb–Douglas dollar weights $\gamma_{i}/\sum_{j}\gamma_{j}$ do not depend on prices at all, so they do not care how a company happens to denominate its shares. The CES dollar weights $w_{i}^{\star}\propto S_{i}(t)^{1-\eta}\gamma_{i}^{\eta}$ do depend on the price level of each share whenever $\eta\neq 1$: a two-for-one stock split halves $S_{i}(t)$, changes $S_{i}(t)^{1-\eta}$, and therefore changes the dollar allocation to asset $i$, although nothing economic has happened. The reason is that the CES utility is written over share counts, and share counts are an accounting unit. This does not affect the lecture's use of CES as a concentration dial, but it means the CES weights on a given day should be read as a function of that day's price levels, and a version of the allocator written over dollar exposures $x_{i} = n_{i}S_{i}(t)$ would remove the dependence.

## Summary
In this notebook, we derived the closed-form CES optimum in share counts from the budget-constrained Lagrangian, took its three limits, read the limits of the engine's elasticity rule, showed that log-linear utility has the Cobb–Douglas optimizer, and recorded the share-denomination caveat.

> __Key Takeaways:__
>
> * __One formula, three regimes:__ The closed-form CES optimum is a single expression whose shape is set entirely by the elasticity: Cobb–Douglas at unit elasticity, concentration on the best bang for the buck as the elasticity grows, and equal share counts as it vanishes.
> * __The elasticity rule reads the signal's loudness, the preferences read its sign:__ a quiet signal sends the elasticity to its maximum and concentrates, a loud signal in either direction sends it to its minimum and diversifies, while which assets are favored is decided by the preference weights.
> * __Two things Cobb–Douglas hides:__ log-linear utility shares its optimizer but has additive values, and the CES weights for any other elasticity depend on the price level of each share because the utility is written over share counts.

The lecture's engine uses these results as a dial and a default; the L13a lecture returns to the elasticity as something a learning algorithm can tune.
