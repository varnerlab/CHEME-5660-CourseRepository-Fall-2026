# L8b Advanced: Static Replication of Piecewise-Linear Payoffs
The L8b lecture showed that adding option legs __produces__ payoff shapes: each leg bends at its own strike, so a sum of legs is a piecewise-linear function with a corner at every strike where the legs do not cancel. This notebook answers the converse question, which the lecture left open.

> __The question.__ Suppose you can name any payoff shape you want at a single expiration date. Which shapes can a portfolio of cash, the underlying, and options actually deliver? For those that can be delivered, is the portfolio that delivers them unique, and how do you find it?

The answer is a clean theorem: the payoffs reachable by a __finite static portfolio at a single expiration__ are exactly the continuous piecewise-linear ones, each has __exactly one__ representation in terms of cash, shares, and calls struck at its kinks, and the number of calls struck at $K_{i}$ is simply the jump in slope at $K_{i}$. The minimal portfolio uses exactly those strikes where the slope actually jumps.

> __Learning Objectives:__
>
> By the end of this notebook, you should be able to:
> * __State and prove the static replication theorem__ for continuous piecewise-linear payoffs, including the existence argument by matching slopes and the uniqueness argument by working outward from the leftmost interval.
> * __Decompose any strategy by inspection.__ Read off the cash, share, and call weights of a vertical spread, straddle, strangle, butterfly, covered call, and collar directly from their kinks, without algebra.
> * __Identify what is out of reach.__ Explain why a strictly curved payoff and a discontinuous payoff cannot be built from finitely many options at one expiration, and why the hypothesis $K_{1}>0$ is not a technicality.

This notebook is __optional__ and is not a prerequisite for L9a. It uses only the expiration payoff algebra from the L8b lecture: no pricing model, no probability, no discounting.

## Setup: the admissible payoffs and the building blocks
Throughout, $S_{T}\geq{0}$ is the price of the underlying at the single expiration date $T$, in USD/share. Every function below is a __payoff per share at expiration__, measured in USD/share, and we work with one contract of multiplier $q=1$ so the arithmetic stays visible. We return to $q$ and to premiums at the end.

> __Definition (admissible payoff).__ A function $f:[0,\infty)\to\mathbb{R}$ is __continuous piecewise linear with kinks__ $K_{1},\dots,K_{d}$ if
> $$
0 < K_{1} < K_{2} < \dots < K_{d} < \infty
$$
> and $f$ is continuous on $[0,\infty)$ and __affine__ (that is, of the form $\alpha+\beta S_{T}$) on each of the $d+1$ intervals
> $$
[0,K_{1}],\quad [K_{1},K_{2}],\quad\dots,\quad [K_{d-1},K_{d}],\quad [K_{d},\infty)
$$
> with the slope genuinely jumping at each $K_{i}$, and write $\mathcal{A}$ for the set of all such functions, over all finite kink sets, including the empty one.

Two features of this definition carry the whole argument, so it is worth pausing on them:

* __The kinks are strictly positive.__ We require $K_{1}>0$, so the leftmost interval $[0,K_{1}]$ has positive length. Section 5 shows that dropping this requirement destroys uniqueness, and it is not a technicality: it is the reason the theorem has a well-posed answer at all.
* __Finitely many kinks.__ The set is finite, which is what makes the resulting portfolio a finite one that you could actually trade. The empty set is allowed: an $f$ that is affine on all of $[0,\infty)$ has $d=0$, an empty sum, and decomposition $f(S_{T})=f(0)+f'(0^{+})S_{T}$, which is cash plus shares.
* __The kinks are genuine.__ We require the slope to change at each $K_{i}$, so $w_{i}\neq{0}$. Without this, a strike where the slope does not jump could be added or dropped at will with weight $0$, and the __list of strikes__ would not be unique even though the coefficients on any fixed list still are.

The building blocks are the three things an investor can hold at a single expiration:

| Instrument | Payoff per share at $T$ | Notation |
|---|---|---|
| Cash (a zero-coupon claim paying 1 USD at $T$) | $1$ | the constant function |
| The underlying asset | $S_{T}$ | the identity function |
| A long call struck at $K$ | $(S_{T}-K)^{+}$ | $h_{c}(S_{T};K)$ |

Puts are deliberately __absent__ from this list, and they are not missing by oversight. Section 4 shows every put is already a combination of the three.

> __Why calls and not puts?__ Either choice works: the theorem could be stated with puts instead. Calls are the conventional choice because $(S_{T}-K)^{+}$ vanishes on $[0,K]$, so a call struck at $K$ is invisible to the left of its strike. That makes the coefficients fall out of a left-to-right sweep, which is exactly how the proof of uniqueness proceeds.

## The static replication theorem
> __Theorem (existence and uniqueness of the static decomposition).__
>
> Let $f\in\mathcal{A}$ have kinks $0<K_{1}<\dots<K_{d}$. Then there exist unique real numbers $a$, $b$, and $w_{1},\dots,w_{d}$ such that
> $$
\boxed{
f(S_{T}) \;=\; \underbrace{a}_{\text{cash}} \;+\; \underbrace{b\,S_{T}}_{\text{shares}} \;+\; \sum_{i=1}^{d}\underbrace{w_{i}}_{\text{calls at }K_{i}}\left(S_{T}-K_{i}\right)^{+}
\qquad\text{for every } S_{T}\geq{0}}
$$
> and they are given explicitly by
> $$
a = f(0),\qquad b = f'(0^{+}),\qquad w_{i} = f'(K_{i}^{+}) - f'(K_{i}^{-})\quad i=1,\dots,d
$$
> where $f'(0^{+})$ is the slope of $f$ on the leftmost interval $[0,K_{1}]$, read as $[0,\infty)$ when $d=0$, and $f'(K_{i}^{+})$ and $f'(K_{i}^{-})$ are the slopes of $f$ immediately to the right and to the left of the kink $K_{i}$.

In words: __the cash position is the payoff at zero, the share position is the initial slope, and the number of calls struck at each kink is the amount by which the slope jumps there.__ Nothing else is required, and nothing else is possible.

Introduce notation for the slopes before proving it. Let $s_{0}$ denote the slope of $f$ on $[0,K_{1}]$, let $s_{m}$ denote its slope on $[K_{m},K_{m+1}]$ for $m=1,\dots,d-1$, and let $s_{d}$ denote its slope on $[K_{d},\infty)$. The definitions above then read $b = s_{0}$ and $w_{i}=s_{i}-s_{i-1}$.

### Proof of existence
Define $g$ to be the right-hand side, using the stated coefficients:
$$
g(S_{T}) = f(0) + s_{0}S_{T} + \sum_{i=1}^{d}\left(s_{i}-s_{i-1}\right)\left(S_{T}-K_{i}\right)^{+}
$$
We show $g=f$ everywhere on $[0,\infty)$ by checking that the two functions start at the same place and have the same slope on every interval.

__They agree at the origin.__ Every call term vanishes at $S_{T}=0$, because $K_{i}>0$ makes $(0-K_{i})^{+}=0$. Hence $g(0)=f(0)$.

__They have the same slope on every interval.__ Fix $m\in\{0,1,\dots,d\}$ and take $S_{T}$ in the interior of the $m$-th interval. A call term $(S_{T}-K_{i})^{+}$ contributes slope $1$ when $K_{i}<S_{T}$ and slope $0$ when $K_{i}>S_{T}$, so exactly the terms $i=1,\dots,m$ are active. Therefore
$$
g'(S_{T}) = s_{0} + \sum_{i=1}^{m}\left(s_{i}-s_{i-1}\right) = s_{0} + \left(s_{m}-s_{0}\right) = s_{m}
$$
where the middle step is a __telescoping sum__: consecutive terms cancel in pairs, leaving only the last slope and the first. And $s_{m}$ is precisely the slope of $f$ on that interval.

__Conclusion.__ Both $f$ and $g$ are continuous on $[0,\infty)$, both are affine on each of the $d+1$ intervals, they agree at $S_{T}=0$, and they have equal slopes on every interval. Walking left to right, they therefore agree at $K_{1}$, hence on all of $[K_{1},K_{2}]$, hence at $K_{2}$, and so on. So $f=g$. $\blacksquare$

### Proof of uniqueness
It suffices to show that the representation of the __zero function__ is trivial, since the difference of two representations of the same $f$ is a representation of zero. So suppose
$$
a + b\,S_{T} + \sum_{i=1}^{d}w_{i}\left(S_{T}-K_{i}\right)^{+} = 0\qquad\text{for every } S_{T}\geq{0}
$$
and work outward from the left.

__Step 1: the leftmost interval kills $a$ and $b$.__ On $[0,K_{1}]$ every call term is zero, so the identity reduces to $a+b\,S_{T}=0$ for all $S_{T}\in[0,K_{1}]$. Because $K_{1}>0$, this interval contains more than one point; an affine function vanishing at two distinct points is identically zero, so $a=0$ and $b=0$. When $d=0$ the step runs on $[0,\infty)$ instead, with no call terms to clear and the same conclusion, so the affine case needs no separate argument.

__Step 2: induction on the kinks.__ Suppose $w_{1}=\dots=w_{m-1}=0$ for some $m\leq{d}$. On the interval $[K_{m},K_{m+1}]$ (or $[K_{d},\infty)$ when $m=d$) the identity reduces to $w_{m}\left(S_{T}-K_{m}\right)=0$, which forces $w_{m}=0$ because that interval also has positive length.

By induction all the $w_{i}$ vanish, so the representation of zero is trivial and the decomposition of any $f\in\mathcal{A}$ is unique. $\blacksquare$

> __Where each hypothesis was used.__ Continuity is what lets us glue the intervals together in the existence proof. Finiteness of the kink set is what makes the sum a finite portfolio. And $K_{1}>0$ is what gives Step 1 an interval of positive length to work on. Section 5 shows the theorem is false without it.

## Reading the standard strategies straight off the theorem
The theorem converts a modeling question into arithmetic on slopes. To decompose a strategy, plot its payoff, note the value at zero, the initial slope, and the slope jump at each kink; the triple $(a,b,w)$ is the portfolio. No algebra is needed.

### First: every put is cash, short shares, and a call
Apply the theorem to $f(S_{T})=(K-S_{T})^{+}$, the long put payoff. Its value at zero is $K$, its slope on $[0,K]$ is $-1$, and at the kink $K$ the slope jumps from $-1$ to $0$, a jump of $+1$. Therefore
$$
\boxed{\left(K-S_{T}\right)^{+} = \underbrace{K}_{a} \;\underbrace{-\;S_{T}}_{b=-1} \;+\; \underbrace{\left(S_{T}-K\right)^{+}}_{w=1}}
$$
This identity is worth checking by hand. For $S_{T}<K$ the right side is $K-S_{T}+0 = K-S_{T}$, and for $S_{T}\geq{K}$ it is $K-S_{T}+S_{T}-K=0$. Both match the put. It also justifies leaving puts out of the building blocks: they add nothing that cash, shares, and calls do not already span.

> __This identity is put-call parity, in payoff space.__ Rearranged, it says
> $$
\left(S_{T}-K\right)^{+} - \left(K-S_{T}\right)^{+} = S_{T} - K
$$
> that is, __long a call and short a put at the same strike has the same payoff as holding the share and owing $K$ in cash__. That is a statement about payoffs alone, and it required no model, no probability, and no assumption about how prices move.
>
> Turning it into a statement about __premiums__ takes one further step: if two portfolios have identical cash flows at __every__ date and state, not merely at $T$, and both can actually be traded and shorted, then they must cost the same today, or a risk-free profit is available. The extra conditions are not decoration: dividends, early exercise of an American contract, and short-sale limits each break the argument if the payoffs match only at expiration. That step introduces discounting and the tradability assumptions, and it is where L9a picks up the story. The payoff identity above is the part that belongs to L8b.

### The catalogue
Every structure the L8b examples build, including both credit spreads, plus two that appear in L11a. All rows are for $q=1$ and are __payoffs__, not profits:

| Strategy | Construction | $a$ (cash) | $b$ (shares) | $w$ (calls) |
|---|---|---:|---:|---|
| Long call at $K$ | one call | $0$ | $0$ | $+1$ at $K$ |
| Long put at $K$ | one put | $K$ | $-1$ | $+1$ at $K$ |
| Bull call spread $K_{1}<K_{2}$ | long call $K_{1}$, short call $K_{2}$ | $0$ | $0$ | $+1$ at $K_{1}$, $-1$ at $K_{2}$ |
| Bull put credit spread $K_{2}<K_{1}$ | short put $K_{1}$, long put $K_{2}$ | $K_{2}-K_{1}$ | $0$ | $+1$ at $K_{2}$, $-1$ at $K_{1}$ |
| Bear call credit spread $K_{1}<K_{2}$ | short call $K_{1}$, long call $K_{2}$ | $0$ | $0$ | $-1$ at $K_{1}$, $+1$ at $K_{2}$ |
| Long straddle at $K$ | long call and long put at $K$ | $K$ | $-1$ | $+2$ at $K$ |
| Long strangle $K_{1}<K_{2}$ | long put $K_{1}$, long call $K_{2}$ | $K_{1}$ | $-1$ | $+1$ at $K_{1}$, $+1$ at $K_{2}$ |
| Butterfly at $K$, width $0<h<K$ | long call $K-h$, short 2 calls $K$, long call $K+h$ | $0$ | $0$ | $+1,-2,+1$ at $K-h,K,K+h$ |
| Covered call at $K$ | long share, short call $K$ | $0$ | $+1$ | $-1$ at $K$ |
| Collar $K_{1}<K_{2}$ | long share, long put $K_{1}$, short call $K_{2}$ | $K_{1}$ | $0$ | $+1$ at $K_{1}$, $-1$ at $K_{2}$ |

Two of these rows deserve comment, and two terms in the table need defining since the L8b lecture did not introduce them.

> __Butterfly:__ the three-strike structure in the table, long one call at each wing and short two at the centre. Its payoff is a triangular tent of height $h$ peaking at $K$ and zero outside $[K-h,K+h]$. __Collar:__ holding the share while buying a put below and selling a call above, which brackets the outcome. Both are taken up properly in L10b and L11a; here they are simply two more rows of arithmetic.

__The straddle has $b=-1$ and $w=+2$.__ Its decomposition is not "one call and one put" in this basis, because the put itself contributed $-S_{T}$ and a second call. Check it at $S_{T}=0$: the formula gives $K-0+2(0-K)^{+}=K$, and indeed $\lvert 0-K\rvert = K$. This is the L8b lecture's finite-downside result appearing as a coefficient: the straddle's payoff at zero is $a=K$, a number, not infinity.

__The collar has $b=0$.__ Holding a share gives $b=+1$, and the put contributes $b=-1$, so the two cancel exactly. A collar therefore has __no net share exposure__ in the decomposition; its payoff is a constant $K_{1}$ plus a bull call spread between the strikes, which is why it is clamped between $K_{1}$ and $K_{2}$ at the extremes.

## What the theorem cannot give you
The theorem is an exact characterization, so it says as much about what is impossible as about what is possible. Three consequences are worth stating precisely.

### 1. Curved payoffs are out of reach
A finite portfolio of cash, shares, and calls at finitely many strikes and one expiration produces a payoff in $\mathcal{A}$, and every member of $\mathcal{A}$ is __piecewise linear__. So a payoff that is strictly curved on some interval, such as the variance-style claim $f(S_{T})=S_{T}^{2}$, cannot be replicated exactly. ($\ln S_{T}$ fails twice over: it is strictly curved, and it is not even defined at $S_{T}=0$, so it is not a candidate on the domain $[0,\infty)$ at all. The curvature is the obstruction that matters here; the domain failure is separate.)

> __State the limitation precisely.__ The correct claim is: not reachable __by a finite static portfolio of cash, shares, and calls at finitely many strikes with a single expiration date__. It is not a claim that such payoffs are unattainable in general. Allowing a continuum of strikes turns the sum into an integral and does represent smooth payoffs exactly; allowing trading between now and $T$ opens up dynamic replication, which is the subject of L9a and L10b. Both routes leave the setting of this notebook.

### 2. Discontinuous payoffs are out of reach
Every member of $\mathcal{A}$ is continuous. A __digital__ (or binary) payoff that pays $1$ USD when $S_{T}>K$ and $0$ otherwise jumps at $K$, so it is not in $\mathcal{A}$ and no finite portfolio replicates it. Narrow bull call spreads approximate it increasingly well as the strikes converge, but "approximates" is not "replicates," and the approximating portfolios require ever larger positions.

### 3. The hypothesis $K_{1}>0$ is doing real work
Suppose we allowed a kink at $K_{1}=0$. On the domain $[0,\infty)$,
$$
\left(S_{T}-0\right)^{+} = S_{T}
$$
because $S_{T}\geq{0}$ always. The "call struck at zero" is therefore not a new instrument at all: it is the share. Its column in the basis duplicates the $b\,S_{T}$ column, so for any $w$ whatsoever
$$
a + b\,S_{T} = a + (b-w)\,S_{T} + w\left(S_{T}-0\right)^{+}
$$
gives infinitely many representations of the same payoff, and uniqueness fails. Existence survives, but the decomposition stops being well defined.

> __The economic reading.__ A call with a zero strike carries an unconditional obligation to pay $0$ and an unconditional right to receive the share, so it __is__ the share. Requiring $K_{1}>0$ is the formal way of saying the basis contains no duplicates. This is the same reason the leftmost interval in the uniqueness proof needed positive length.

## Recovering the weights when you can only evaluate the payoff
The theorem gives $w_{i}$ as a slope jump, which presumes you know the slopes. Suppose instead you can only __evaluate__ $f$ at chosen prices. The weight at a kink is then recoverable from three evaluations.

Fix a kink $K_{i}$ and choose a spacing $h>0$ __small enough that the interval $(K_{i}-h,\,K_{i}+h)$ contains no other kink__ and lies inside $(0,\infty)$. Since $f$ is affine on $[K_{i}-h,K_{i}]$ and on $[K_{i},K_{i}+h]$,
$$
f(K_{i}-h) = f(K_{i}) - h\,f'(K_{i}^{-}),\qquad f(K_{i}+h) = f(K_{i}) + h\,f'(K_{i}^{+})
$$
Adding these and rearranging gives the __normalized second difference__:
$$
\boxed{\frac{f(K_{i}+h) - 2f(K_{i}) + f(K_{i}-h)}{h} \;=\; f'(K_{i}^{+}) - f'(K_{i}^{-}) \;=\; w_{i}}
$$
The result is __exact__, not an approximation, provided the no-other-kink condition holds; and the division by $h$ is what makes it so. Note the units: $f$ is in USD/share and $h$ is in USD/share, so $w_{i}$ is dimensionless, which is right, because $w_{i}$ counts contracts.

The weights $+1,-2,+1$ are the same pattern that defines the __butterfly__ in the catalogue above, which is why the formula is often called a butterfly second difference. Be careful with that name: here the pattern is applied to __three evaluations of the payoff function $f$ at three underlying prices__, all at the single expiration date. It is a finite difference in the state variable $S_{T}$. Buying a traded butterfly is a different object: it is the payoff function $(S_{T}-(K_{i}-h))^{+}-2(S_{T}-K_{i})^{+}+(S_{T}-(K_{i}+h))^{+}$, and pricing it means combining option __premiums across strikes__. The two operations share an arithmetic pattern and nothing else.

> __Three conditions, easy to lose:__
> * __Equal spacing.__ The $1,-2,1$ pattern assumes the two wings are equidistant from the centre. For unequal wings, go back to differencing the two adjacent slopes directly; the symmetric formula does not apply.
> * __No intervening kink.__ If other kinks of $f$ lie inside $(K_{i}-h,K_{i}+h)$, the normalized second difference is no longer guaranteed to return $w_{i}$. It returns a __distance-weighted__ combination of every slope jump in the window, $\sum_{j}w_{j}\left(1-\lvert K_{j}-K_{i}\rvert/h\right)^{+}$, in which a kink nearer the edge counts for less than its full weight. A window containing no other kink therefore __guarantees__ the answer is $w_{i}$; with other kinks present the answer is $w_{i}$ only if their weighted contributions happen to cancel, which is an accident you cannot count on.
> * __This is arithmetic on payoffs.__ Everything here is a statement about the function $f$, evaluated at prices of the underlying. Performing the same second difference on option __premiums__ across strikes is a different operation with a different meaning, and it is the one that leads toward the risk-neutral density in L9a. Do not conflate the two.

## From payoff to profit, and from one share to one contract
Everything above decomposes a __payoff__. Two adjustments carry it to the quantities an account actually reports.

__Premiums shift $a$ and nothing else.__ Profit is payoff minus the net premium paid at the open, and the net premium is a constant, independent of $S_{T}$. If a strategy has payoff decomposition $(a,b,w)$ and net opening debit $D$ per share, its profit decomposition is
$$
\left(a - D,\;\; b,\;\; w\right)
$$
The share position and every call weight are unchanged. This is the precise version of the L8b lecture's statement that the net premium "shifts the diagram without changing its shape": in this basis it moves exactly one coefficient.

__The multiplier scales everything.__ Take $n$ copies of the whole per-share strategy, every leg carrying the same deliverable $q$ shares/contract. Then every coefficient scales by the same factor:
$$
\left(nq\,a,\;\; nq\,b,\;\; nq\,w\right)
$$
so the shape is fixed by $(a,b,w)$ and the size by $nq$. A standard U.S. equity contract has $q=100$. If the legs carry __different__ deliverables $q_{i}$, as a corporate-action adjustment can leave them, no single factor comes out: scale each leg by its own $n_{i}q_{i}$ first and then add the decompositions.

> __A worked check.__ The long straddle at $K=230$ from the L8b composite example has payoff decomposition $a=230$, $b=-1$, $w=+2$ at $K=230$, per share. Its net debit is $D = 21.30+21.58 = 42.88$ USD/share, so its __profit__ decomposition is $a=230-42.88=187.12$, $b=-1$, $w=+2$. Evaluating the profit at $S_{T}=0$ gives $187.12 - 0 + 0 = 187.12$ USD/share, which is exactly the finite downside the example plots and the lecture emphasizes. At $q=100$ that is $18{,}712$ USD per contract. The bound the lecture argues for verbally is, in this basis, simply the coefficient $a$.

## Summary
This notebook proved that the composite-contract rule of the L8b lecture is not merely a way to build payoffs but a complete description of what can be built at a single expiration.

> __Key Takeaways:__
>
> * __The decomposition exists and is unique.__ Every continuous piecewise-linear payoff with kinks $0<K_{1}<\dots<K_{d}$ is $a + bS_{T} + \sum_{i}w_{i}(S_{T}-K_{i})^{+}$ for exactly one choice of coefficients, namely $a=f(0)$, $b=f'(0^{+})$, and $w_{i}$ the jump in slope at $K_{i}$. Existence follows from a telescoping sum of slopes; uniqueness follows by sweeping from the leftmost interval outward.
> * __Strategies are triples, not recipes.__ Vertical spreads, straddles, strangles, butterflies, covered calls, and collars are one theorem seen from six angles. Reading the catalogue explains structural facts directly: the straddle's cash coefficient $a=K$ __is__ its finite payoff at $S_{T}=0$ (its maximum loss, once a debit $D$ is paid, is $D$ at $S_{T}=K$), and the collar's $b=0$ __is__ its lack of an unconditional share position, its remaining exposure living in the two call terms between the strikes.
> * __The boundary of the reachable set is sharp.__ Curved payoffs and discontinuous payoffs lie outside it, and the hypothesis $K_{1}>0$ is load-bearing, because a call struck at zero is the share and duplicates a basis element.

Two directions lead out of this notebook. Letting the strikes fill a continuum turns the sum into an integral and reaches smooth payoffs; letting the portfolio be rebalanced through time gives dynamic replication, which is how L9a prices a contract rather than merely describing what it pays.
