## European Call and Put Contracts
A __European__ option can be exercised only at expiration $T$. For a strike $K>0$ and terminal underlying price $S_T\geq 0$, the per-share payoffs are
$$
\boxed{h_c(S_T;K)=(S_T-K)^+,\qquad h_p(S_T;K)=(K-S_T)^+},
$$
where $x^+=\max(x,0)$. The call premium is denoted $C_0$ and the put premium $P_0$, both in USD/share at time $0$. Multiply by the contract deliverable only when converting a per-share value to account dollars.

The exercise convention is essential. BSM prices the terminal European payoff directly; it does not optimize an early-exercise decision. That additional decision is the subject of L9b.
___

## Risk-Neutral Valuation
Let $g_y$ be the continuously compounded risk-free rate in the notation of L2a, $\delta$ a continuous proportional dividend yield, and $\sigma>0$ annualized volatility. Under the BSM risk-neutral model,
$$
\frac{dS_t}{S_t}=(g_y-\delta)\,dt+\sigma\,dW_t^{\mathbb Q}.
$$
The drift $g_y-\delta$ belongs to the pricing measure $\mathbb Q$; it is not an estimate of the asset's historical or expected real-world return. For a European payoff $h(S_T)$,
$$
\boxed{V_0=e^{-g_yT}\,\mathbb E^{\mathbb Q}[h(S_T)]}.
$$
This expression is a no-arbitrage value under the model assumptions. It can be evaluated by simulation, and for vanilla calls and puts it has a closed form.

> __Model scope.__ Classical BSM assumes frictionless continuous trading, no arbitrage, constant $g_y$, $\delta$, and $\sigma$, continuous lognormal price paths, and the ability to trade the underlying and financing asset as required by the hedge. A formula can be evaluated exactly even when its assumptions approximate the market.
___

## Black-Scholes-Merton Closed Forms
Define
$$
d_1=\frac{\ln(S_0/K)+(g_y-\delta+\tfrac12\sigma^2)T}{\sigma\sqrt T},\qquad d_2=d_1-\sigma\sqrt T,
$$
and let $\Phi$ be the standard-normal cumulative distribution function. The European premiums are
$$
\boxed{C_0=S_0e^{-\delta T}\Phi(d_1)-Ke^{-g_yT}\Phi(d_2)},
$$
$$
\boxed{P_0=Ke^{-g_yT}\Phi(-d_2)-S_0e^{-\delta T}\Phi(-d_1)}.
$$
Every input convention must agree: $T$ is in years, $g_y$ is continuously compounded, $\sigma$ is annualized, and $S_0$ and $K$ use the same currency-per-share units.

### Put-Call Parity
The prices satisfy the European replication identity
$$
\boxed{C_0-P_0=S_0e^{-\delta T}-Ke^{-g_yT}}.
$$
Parity is an excellent check on two independently evaluated formulas. It is a poor numerical implementation of a tiny premium when it requires subtracting nearly equal large numbers; compute each BSM formula directly and use parity as the check.
___

## Checks Before Trusting a European Premium
A model value should pass several inexpensive checks:

1. __Nonnegativity:__ $C_0\geq0$ and $P_0\geq0$.
2. __Put-call parity:__ the replication residual should be no larger than numerical rounding permits.
3. __Strike monotonicity:__ holding other inputs fixed, call premiums are nonincreasing in $K$ and put premiums are nondecreasing in $K$.
4. __Simulation agreement:__ a risk-neutral Monte Carlo estimate should differ from the closed form by an ordinary number of estimated standard errors. A 95% interval is not guaranteed to contain the closed form on every random run.
