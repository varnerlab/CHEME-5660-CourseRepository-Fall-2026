# L9b Advanced: Where the CRR Up and Down Factors Come From
The lecture quotes the Cox-Ross-Rubinstein up and down factors $u=e^{\sigma\sqrt{\Delta t}}$ and $d=e^{-\sigma\sqrt{\Delta t}}$ without deriving them. This notebook derives them, and is equally explicit about what the derivation assumes and what it discards.

## The Calibration Problem
A one-step lattice has three unknowns: the up factor $u>0$, the down factor $d>0$, and the up probability. The diffusion it is meant to approximate supplies only two conditions, one on the mean of the one-step return and one on its variance. Two equations cannot determine three unknowns, so a third condition must be supplied by the modeler.

> __The three unknowns and where each is pinned down.__
>
> * The __variance__ condition fixes the size of the jump, and therefore $u$ and $d$ together.
> * The __mean__ condition fixes the probability. Under the risk-neutral measure this is the martingale condition the lecture uses to derive $q$.
> * The __third condition is a choice.__ Cox, Ross, and Rubinstein take $ud=1$. Jarrow and Rudd instead set the probability to $1/2$ exactly and let the factors carry the drift. Both give valid lattices that converge to the same diffusion.

This notebook derives the CRR factors from a local variance match under the $ud=1$ choice. Keeping the roles separate is what makes the derivation honest: the step below uses an auxiliary, equally weighted two-point distribution to size the jump. That auxiliary weight is a calibration device, __not__ the risk-neutral pricing probability the lecture derives.
___

## Deriving the Factors
Let $\sigma>0$ be the annualized volatility of the underlying's log return and let $\Delta t>0$ be the length of one lattice step in years. Impose the reciprocal condition:
$$
\boxed{ud = 1 \quad\Longleftrightarrow\quad d = \frac{1}{u}}
$$
and write $a \equiv \ln u$, so that $\ln d = -a$. To calibrate the jump magnitude, introduce the auxiliary symmetric two-point variable $\widetilde X$:
$$
\widetilde X \;=\;
\begin{cases}
+a & \text{with probability } 1/2,\\
-a & \text{with probability } 1/2.
\end{cases}
$$
The equal weights define only this calibration proxy. The actual lattice log return has the same two support points but uses the pricing probability $q$, derived later from the martingale condition.

The first two moments of $\widetilde X$ are exact:
$$
\begin{align*}
\mathbb{E}[\widetilde X] &= 0,\\
\mathbb{E}[\widetilde X^{2}] &= a^{2},\\
\operatorname{Var}(\widetilde X) &= a^{2}.
\end{align*}
$$
The geometric Brownian motion the lattice approximates has log-return variance $\sigma^{2}\Delta t$ over the same step. Matching the proxy variance to that diffusion variance gives:
$$
a^{2} = \sigma^{2}\Delta t \quad\Longrightarrow\quad a = \sigma\sqrt{\Delta t}
$$
where the positive root is taken because $a=\ln u$ labels the up move and $\sigma\geq 0$, $\Delta t>0$. Exponentiating, and using $d=1/u$:
$$
\boxed{
\begin{align*}
u &= e^{+\sigma\sqrt{\Delta t}}\\
d &= e^{-\sigma\sqrt{\Delta t}}
\end{align*}}
$$
which is the CRR form the lecture quotes.
___

## What the Derivation Costs
The step that deserves scrutiny is the equally weighted calibration proxy, because the lattice is not priced with those equal weights. It is priced with the risk-neutral probability $q$ the lecture derives from the martingale condition:
$$
q = \frac{e^{g_{y}\Delta t}-d}{u-d}
$$
where $g_{y}$ is the continuously compounded risk-free rate in the notation of L2a. Substituting the CRR factors and expanding for small $\Delta t$ gives:
$$
q = \frac{1}{2} + \frac{g_{y}-\sigma^{2}/2}{2\sigma}\sqrt{\Delta t} + \mathcal{O}\!\left(\Delta t^{3/2}\right)
$$
so $q$ approaches $1/2$ but is not equal to it, and the gap is of order $\sqrt{\Delta t}$. Feeding the true $q$ back through the exact variance formula gives:
$$
\operatorname{Var}_{\mathbb{Q}}(X) = 4q(1-q)a^{2} = \sigma^{2}\Delta t\left[1-\left(\frac{g_{y}-\sigma^{2}/2}{\sigma}\right)^{2}\Delta t\right] + \mathcal{O}\!\left(\Delta t^{3}\right)
$$

> __What is dropped, stated precisely.__ The CRR lattice does __not__ match the diffusion variance exactly. It matches it to leading order, with an absolute error of order $\Delta t^{2}$, that is, a relative error of order $\Delta t$. The lattice variance falls *short* of the diffusion variance by $\left(g_{y}-\sigma^{2}/2\right)^{2}\Delta t^{2}$, a term that vanishes faster than the variance itself as the step shrinks. Matching moments to this order is a necessary ingredient in, not a proof of, the convergence of the lattice price to the Black-Scholes-Merton price; the point here is only that a coarse lattice is not the same thing as the diffusion. Note also that the leading error term disappears when $g_{y}=\sigma^{2}/2$, so the discrepancy is smaller on that ray than the generic order suggests.

Nothing about the exponential is truncated: $u=e^{\sigma\sqrt{\Delta t}}$ is exact given $a$. The approximation enters when the equally weighted proxy, rather than the pricing probability $q$, is used to calibrate the jump magnitude.
___

## Where the Lattice Is Valid
The derived factors are usable only where the pricing probability is a probability. Requiring $0<q<1$ in the expression above is equivalent to the one-step no-arbitrage condition:
$$
d < e^{g_{y}\Delta t} < u
$$
and substituting the CRR factors turns it into a condition on the step size:
$$
e^{-\sigma\sqrt{\Delta t}} < e^{g_{y}\Delta t} < e^{+\sigma\sqrt{\Delta t}}
\quad\Longleftrightarrow\quad
\lvert g_{y}\rvert\sqrt{\Delta t} < \sigma
$$

> __Read the condition the right way round.__ For a fixed rate and step, __increasing__ $\sigma$ widens the band and makes the condition easier to satisfy. It is a __large time step__, or a rate extreme relative to the volatility, that breaks it. A lattice that violates the band has no equivalent risk-neutral measure for that step: $q$ falls outside $(0,1)$ and the arithmetic that follows is not a price. At the boundary itself $q$ is exactly $0$ or $1$, which is still a number in $[0,1]$ but assigns zero weight to one of the two states, so the measure is no longer equivalent to the real-world one. The remedy is a smaller $\Delta t$, that is, more steps, and never clipping $q$ back into range.

Two degenerate cases sit at the boundary. If $\sigma=0$ then $u=d=1$ and the expression for $q$ divides by zero; the CRR parameterization is singular and the deterministic risk-neutral path must be valued directly instead. For example, a European payoff is worth $e^{-g_yT}h(S_0e^{g_yT})$. If $\Delta t=0$ there is no step to take. Both are excluded by the standing assumptions $\sigma>0$ and $\Delta t>0$.

## Summary
This notebook derived the CRR up and down factors by matching the variance of a two-point lattice log return to the variance of the diffusion over one step, under the reciprocal normalization $ud=1$.

> __Key Takeaways:__
>
> * __Calibration needs a third condition:__ A one-step lattice has three unknowns while the diffusion supplies only a mean and a variance condition, so the modeler must add one more, and the reciprocal condition on the factors is the choice that defines CRR rather than a result the diffusion forces.
> * __Equal weighting is a calibration device, not the pricing measure:__ The jump size comes from an equally weighted auxiliary variable, while the lattice is priced with the risk-neutral probability, which approaches one half only in the limit and differs from it at the order of the square root of the time step.
> * __The match is first order and has a domain:__ The lattice reproduces the diffusion variance with an error one order smaller in the time step than the variance itself, and it is a valid pricing model only where the rate is small enough relative to the volatility to keep the risk-neutral probability strictly inside the unit interval.

The L9b lecture uses these factors to build the lattice and runs backward induction on it to compare American and European contract values.
