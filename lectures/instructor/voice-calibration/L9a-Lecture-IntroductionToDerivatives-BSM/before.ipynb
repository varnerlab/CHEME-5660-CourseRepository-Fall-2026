## What is a Derivative?
A derivative is a financial contract between a buyer and a seller that describes an agreement to transact an asset at a predetermined price on a specified future date. The value of a derivative is based on the performance of an underlying asset.

In the language of L1b, a derivative is an abstract asset whose cash flows depend on the underlying asset. An option buyer pays today for a future cash flow that may be zero. 

Let's watch a short video to understand the concept of derivatives better: [Derivatives Explained in 2 Minutes in Basic English | 2023](https://www.yout-ube.com/watch?v=B0jeGk5GJX0).

Common types of derivatives include:
* __Futures contracts__ are legal agreements to buy or sell an underlying asset, such as commodities, equities, or other securities, at a predetermined price on a future date. Futures are exchange-traded.
* __Forward contracts__ are similar to futures but are traded over-the-counter (OTC) and have more flexibility in contract terms. They may have non-standard terms.
* __Swaps__ are derivative contracts in which one party exchanges or swaps one asset's value or cash flows for another. Swaps are traded over-the-counter (OTC). 
* __Options contracts__ give the contract buyer the right, but not the obligation, to buy (call option) or sell (put option) an asset at a specific price. Options are exchange-traded. The largest options exchange is the [Chicago Board Options Exchange (CBOE)](https://www.cboe.com).
* __Event contracts__ pay a fixed amount, usually one dollar, if a specified event occurs and nothing otherwise. Exchanges such as [Kalshi](https://kalshi.com) list them on economic data, weather, and financial index levels.

We will focus on options contracts in this lecture.

The two types of options contracts (derivatives whose underlying asset is typically a stock or an ETF) are: __call options__ and __put options__. Let's watch a short video to understand call and put options better: [Call versus Put Options | SoFi](https://www.yout-ube.com/watch?v=bvm-zeL_u8s).

What is the business case for __long and short call options__?

> __Buyer (long)__. From the buyer's perspective, a call contract allows:
> * The buyer benefits from upside price movement of the underlying asset without purchasing it directly. Long calls have defined risk: the maximum loss for the buyer is the premium paid.
> * Long call options provide an opportunity to purchase shares at the strike price instead of the market price. 
>
> __Seller (short)__. From the seller's perspective, a call contract offers:
> * The seller receives the contract premium for selling a call contract. Short call contracts have unlimited upside risk. Call options are often sold only by investors who already own the required number of shares as a covered call (100 shares per contract). 
> * Short call options obligate the seller to sell shares at the strike price $K$ if the option is exercised.

What is the business case for __long and short put options__?

> __Buyer (long)__. From the buyer's perspective, a put contract allows:
> * The buyer benefits from downside price movement of the underlying asset without selling it directly. Long puts have defined risk: the maximum loss for the buyer is the premium paid for the contract.
> * Long put options provide an opportunity to sell shares at the strike price $K$ instead of the market price $S$.
>
> __Seller (short)__. From the seller's perspective, a put contract offers:
> * The seller receives the contract premium for selling a put contract. Short put contracts have large but bounded downside risk: if the share price falls to zero, the seller loses the strike price minus the premium per share. Put options are often sold only by investors with sufficient funds to purchase shares (100 shares per contract). This is a cash-secured put position.
> * Short put options obligate the seller to purchase shares at the strike price $K$ if the option is exercised.

Let's examine the payoff and profit for single call and put options contracts. A __call__ option gives the owner the right, but not the obligation, to __buy__ 100 shares of an underlying asset (per contract) at a specific price (strike) by a future date (expiration). A __put__ option gives the owner the right, but not the obligation, to __sell__ 100 shares at the strike price by the expiration date.

The payoff per share at expiration (or when the contract is exercised) for a __call__ and a __put__ option is:
$$
\boxed{
\begin{align*}
V_{c}(K,S(T)) & = \max\left(S(T) - K,~0\right)\\
V_{p}(K,S(T)) & = \max\left(K - S(T),~0\right)
\end{align*}}
$$
where $K$ denotes the strike price and $S(T)$ is the share price of the underlying asset at expiration (time $t=T$). The seller charges the buyer a premium for each contract, $\mathcal{P}_{c}(K,S(0))$ for a call and $\mathcal{P}_{p}(K,S(0))$ for a put.

The __buyer's profit per share__ at expiration is the payoff minus the contract premium, and the __seller's profit per share__ is the contract premium minus the payoff:
$$
\begin{align*}
P_{c}^{\text{buyer}}(K,S(T)) & = V_{c}(K,S(T)) - \mathcal{P}_{c}(K,S(0)) & P_{c}^{\text{seller}}(K,S(T)) & = \mathcal{P}_{c}(K,S(0)) - V_{c}(K,S(T))\\
P_{p}^{\text{buyer}}(K,S(T)) & = V_{p}(K,S(T)) - \mathcal{P}_{p}(K,S(0)) & P_{p}^{\text{seller}}(K,S(T)) & = \mathcal{P}_{p}(K,S(0)) - V_{p}(K,S(T))
\end{align*}
$$
Setting the buyer's profit to zero gives the __breakeven__ share price at expiration:
$$
\begin{align*}
S_{\text{BE}} & = K + \mathcal{P}_{c}(K,S(0)) && \text{call}\\
S_{\text{BE}} & = K - \mathcal{P}_{p}(K,S(0)) && \text{put}
\end{align*}
$$
Above the call breakeven, and below the put breakeven, the buyer profits and the seller loses.

## Pricing Options
What is an option's payoff worth today? Let's treat the option as an abstract asset and compute its net present value.

A European option is an abstract asset with two cash flows: the buyer pays the premium $\mathcal{P}$ today and receives the payoff $V(K,S(T))\geq 0$ at expiration. Here $V$ is $V_{c}$ for a call and $V_{p}$ for a put. This is the share purchase from L4a and L5a with the sale price $S(T)$ replaced by the payoff, so the net present value at $t=0$ is:
$$
\begin{align*}
\operatorname{NPV}_{0} = -\mathcal{P} + \mathcal{D}^{-1}_{T,0}(g_{y})\cdot V(K,S(T))
\end{align*}
$$
The profit $V-\mathcal{P}$ in the previous section adds dollars from different dates. In today's dollars, the call buyer breaks even at $S(T)=K+\mathcal{P}_{c}\cdot\mathcal{D}_{T,0}(g_{y})$ rather than $K+\mathcal{P}_{c}$.

A Treasury's cash flows are known, so in L2a its price made the NPV zero. An option's payoff is random, so we require a zero __expected__ NPV under the risk-neutral measure $\mathbb{Q}$. For a __European-style__ contract, this gives the premium:
$$
\boxed{
\begin{align*}
\mathbb{E}_{\mathbb{Q}}\bigl(\operatorname{NPV}_{0}\bigr) = 0
\quad\Longrightarrow\quad
\mathcal{P}^{\text{EU}} = \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\cdot V(K,S(T))\Bigr)
\end{align*}}
$$
Any deviation from this relationship would create a risk-free profit opportunity. If the premium were too low, traders would buy the option and sell a portfolio with the same payoff, built from shares and the risk-free asset. If the premium were too high, they would do the reverse. These arbitrage pressures keep the market price aligned with the discounted expected payoff. The [contingent claims notebook](advanced/contingent_claims/CHEME-5660-L9a-Advanced-ContingentClaims-BSM-Fall-2026.ipynb) builds this portfolio.

The exercise style changes this premium:
* __European-style__ options can only be exercised on the expiration date.
* __American-style__ options can be exercised at any time before or on the expiration date. This creates extra risk for the contract seller, so American-style options cost at least as much as otherwise identical European-style options.
* __Bermudan-style__ options, a less common hybrid, can be exercised only on specified dates up to the expiration date.

For an __American-style__ contract, the holder can exercise at any time $\tau\leq T$ chosen using only the information available at that time (a __stopping time__), which can increase the premium:
$$
\begin{align*}
\mathcal{P}^{\text{AM}} = \sup_{\tau\leq T} \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{\tau,0}(g_{y})\cdot V(K,S(\tau))\Bigr) \geq \mathcal{P}^{\text{EU}}
\end{align*}
$$
where $\mathcal{D}_{\tau,0}(g_{y})$ denotes the risk-free discount factor computed between purchase and the exercise time $\tau$ (we assume continuous compounding at the risk-free rate $g_{y}$). The American premium is at least the European one because the American holder has the flexibility to choose when to exercise.

What does $\mathbb{Q}$ mean? Under $\mathbb{Q}$, the expected share price grows at the risk-free rate $g_{y}$ (no dividends), so $\mathbb{E}_{\mathbb{Q}}\bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\,S(T)\bigr)=S(0)$ and the share also has zero expected NPV. $\mathbb{Q}$ is a pricing rule under which every traded asset has zero expected NPV. The real-world measure $\mathbb{P}$ still describes how often outcomes occur.

The buyer pays for something today that may have no value at expiration. The scaled NPV $\rho_{T}=\mathcal{D}^{-1}_{T,0}(g_{y})\,V(K,S(T))/\mathcal{P}-1$ equals $-1$ whenever the option expires worthless. For a call, that happens with real-world probability $\mathbb{P}(S(T)\leq K)$, which L5a's terminal target formula computes. A share under GBM never loses its whole value, so its scaled NPV stays above $-1$.

The same pricing condition prices an event contract, which pays $V=\mathbb{1}\{\text{event}\}$: one dollar if the event occurs and nothing otherwise. Its price $p$ is:
$$
\begin{align*}
p = \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{1}\{\text{event}\}\Bigr) = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{Q}(\text{event})
\end{align*}
$$
The price is a discounted $\mathbb{Q}$-probability, which can differ from the real-world probability $\mathbb{P}(\text{event})$.

### Black–Scholes–Merton and Put–Call Parity
European options can only be exercised at expiration, which simplifies their pricing. The price of a European option contract can be determined using the Black-Scholes-Merton (BSM) model.

> __Nobel Prize 1997:__ The 1997 Nobel Prize in Economic Sciences was awarded to Robert C. Merton and Myron S. Scholes for their work in developing the Black-Scholes-Merton model. Their contributions have had a profound impact on financial markets, risk management, and investment strategies worldwide. [See more here](https://www.nobelprize.org/prizes/economic-sciences/1997/summary/). Fischer Black was also a key contributor to the model but had passed away before the prize was awarded.

To compute the expectation, we can simulate the future share price using a [geometric Brownian motion model](https://en.wikipedia.org/wiki/Geometric_Brownian_motion). Alternatively, we can use the [Black–Scholes pricing formula](https://en.wikipedia.org/wiki/Black–Scholes_model) for a European call and put option ($t = 0$):
$$
\boxed{
\begin{align*}
\mathcal{P}_{c}(K,S(0)) & = N(d_{+})\cdot{S}(0) - N(d_{-})\cdot{K}\cdot\mathcal{D}^{-1}_{T,0}(g_{y})\\
\mathcal{P}_{p}(K,S(0)) & = N(-d_{-})\cdot{K}\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) - N(-d_{+})\cdot{S}(0)
\end{align*}}
$$
where:
$$
\begin{align*}
d_{+} & = \frac{1}{\sigma\sqrt{T}}\left[\ln(\frac{S(0)}{K}) + (g_{y}+\frac{\sigma^{2}}{2})T\right] \\
d_{-} & = d_{+} - \sigma\sqrt{T}
\end{align*}
$$
and $N(\dots)$ denotes the standard normal cumulative distribution function, $\sigma$ is the volatility of the underlying asset, $g_{y}$ is the risk-free rate, and $T$ is the time to expiration of the contract in years.

The call and put premiums are linked. For a call and a put with the same strike and expiration, the payoffs differ by $S(T)-K$ at every share price:
$$
\begin{align*}
V_{c}(K,S(T)) - V_{p}(K,S(T)) = \max\left(S(T)-K,~0\right) - \max\left(K-S(T),~0\right) = S(T) - K
\end{align*}
$$
Subtracting the put premium condition from the call premium condition gives:
$$
\begin{align*}
\mathcal{P}_{c}(K,S(0)) - \mathcal{P}_{p}(K,S(0))
& = \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\cdot\bigl[V_{c}(K,S(T)) - V_{p}(K,S(T))\bigr]\Bigr) && \text{expectation is linear}\\
& = \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\cdot\bigl[S(T) - K\bigr]\Bigr) && \text{payoff identity}\\
& = S(0) - K\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) && \mathbb{E}_{\mathbb{Q}}\bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\,S(T)\bigr) = S(0)
\end{align*}
$$
This is __put–call parity__:
$$
\boxed{
\begin{align*}
\mathcal{P}_{c}(K,S(0)) - \mathcal{P}_{p}(K,S(0)) = S(0) - K\cdot\mathcal{D}^{-1}_{T,0}(g_{y})
\end{align*}}
$$
Parity uses only the payoff identity and the share's zero expected NPV, so it holds whatever model describes the share price under $\mathbb{Q}$.
