## Matching One Lattice Step to Geometric Brownian Motion
The CRR model matches the mean and the volatility of the binomial model to those of geometric Brownian motion (GBM) for a small time step. In the binomial model, the share price moves up by a factor $u$ or down by a factor $d$. The CRR model makes a down move undo an up move, the symmetric condition $ud = 1$. Let $g$ denote the growth rate over one step of length $\Delta t$:
$$
g = \frac{1}{\Delta t}\ln\frac{S(t+\Delta t)}{S(t)} =
\begin{cases}
\dfrac{\ln(u)}{\Delta t} = +\dfrac{a}{\Delta t} & \text{with probability } p,\\[2ex]
\dfrac{\ln(d)}{\Delta t} = -\dfrac{a}{\Delta t} & \text{with probability } 1-p,
\end{cases}
$$
where $S(t)$ is the share price at time $t$ and $g$ has units of inverse years. We define $a = \ln(u)$, and the symmetric condition gives $\ln(d) = -a$. The probability $p$ of an up move is the real-world probability, as in the data-driven binomial model we discussed earlier. The mean of $g$ is:
$$
\mathbb{E}[g] = p\cdot\Bigl(+\frac{a}{\Delta t}\Bigr) + (1-p)\cdot\Bigl(-\frac{a}{\Delta t}\Bigr) = \frac{(2p-1)\,a}{\Delta t}.
$$
The second moment is:
$$
\mathbb{E}[g^2] = p\cdot\frac{a^2}{\Delta t^2} + (1-p)\cdot\frac{a^2}{\Delta t^2} = \frac{a^2}{\Delta t^2}.
$$
Using the variance formula $\mathrm{Var}(g) = \mathbb{E}[g^2] - (\mathbb{E}[g])^2$, we have:
$$
\begin{align*}
\mathrm{Var}(g) & = \mathbb{E}[g^2] - (\mathbb{E}[g])^2 && \text{variance formula}\\
& = \frac{a^2}{\Delta t^2} - \Big[\frac{(2p-1)\,a}{\Delta t}\Big]^2 && \text{substitute the two moments}\\
& = \frac{a^2}{\Delta t^2}\Big[1 - (2p-1)^2\Big] && \text{factor out } a^{2}/\Delta t^{2}\\
& = \frac{4p(1-p)\,a^2}{\Delta t^2} && 1-(2p-1)^{2} = 4p(1-p)
\end{align*}
$$
From the geometric Brownian motion model, the growth rate over $\Delta t$ has mean and variance:
$$
\mathbb{E}[g] = \mu_{g},
\qquad
\mathrm{Var}(g) = \sigma_{g}^{2} = \frac{\sigma^2}{\Delta t}
$$
where $\mu_{g}$ is the mean growth rate (units: $\mathrm{yr}^{-1}$) and $\sigma_{g}$ is the volatility of the growth rate (units: $\mathrm{yr}^{-1}$). The GBM volatility $\sigma = \sigma_{g}\sqrt{\Delta t}$ (units: $\mathrm{yr}^{-1/2}$) is the same $\sigma$ as in the Black–Scholes–Merton formulas. We set the binomial mean and variance equal to the GBM values:
$$
\begin{align*}
\frac{(2p-1)\,a}{\Delta t} & = \mu_{g} && \text{match the mean}\\
\frac{4p(1-p)\,a^2}{\Delta t^2} & = \frac{\sigma^2}{\Delta t} && \text{match the variance}
\end{align*}
$$
These two conditions fix the two unknowns, the size of a move $a$ and the probability $p$.
___

## Solving for the CRR Factors
Let's solve for $a$ first. In the variance condition, we use the second line of the variance calculation and substitute the mean condition:
$$
\begin{align*}
\frac{a^2}{\Delta t^2} - \Big[\frac{(2p-1)\,a}{\Delta t}\Big]^2 & = \frac{\sigma^2}{\Delta t} && \text{match the variance}\\
\frac{a^2}{\Delta t^2} - \mu_{g}^{2} & = \frac{\sigma^2}{\Delta t} && \text{substitute the mean condition}\\
a^2 & = \sigma^2\Delta t + \mu_{g}^{2}\Delta t^{2} && \text{multiply by } \Delta t^{2} \text{ and solve for } a^{2}
\end{align*}
$$
The second term is the first term times $(\mu_{g}/\sigma)^{2}\Delta t$, which goes to zero as the time step shrinks. Cox, Ross, and Rubinstein keep only the first term:
$$
a^2 = \sigma^2 \Delta t
\quad\Longrightarrow\quad
a = \sigma\sqrt{\Delta t}.
$$
Recall that we defined $u = e^{a}$ and $d = e^{-a}$ from the symmetric condition $ud = 1$. Therefore:
$$
\boxed{
u = e^{+\sigma\sqrt{\Delta t}},
\qquad
d = e^{-\sigma\sqrt{\Delta t}}.
}
$$
That is exactly the Cox–Ross–Rubinstein form! The mean condition then gives the probability of an up move:
$$
\begin{align*}
p & = \frac{1}{2}\Bigl(1 + \frac{\mu_{g}\Delta t}{a}\Bigr) && \text{solve the mean condition for } p\\
& = \frac{1}{2}\Bigl(1 + \frac{\mu_{g}}{\sigma}\sqrt{\Delta t}\Bigr) && \text{substitute } a = \sigma\sqrt{\Delta t}
\end{align*}
$$
The volatility sets the size of a move, and the mean growth rate tilts the probability away from one half. As the time step shrinks, $p$ approaches $\tfrac{1}{2}$, so up and down moves become equally likely.

With this $p$, the lattice matches the GBM mean growth rate exactly. Its variance, $a^{2}/\Delta t^{2} - \mu_{g}^{2} = \sigma^{2}/\Delta t - \mu_{g}^{2}$, falls short of $\sigma^{2}/\Delta t$ by $\mu_{g}^{2}$, which is small next to $\sigma^{2}/\Delta t$ for a small time step.
___

## Pricing With the Risk-Neutral Probability
The lecture keeps the CRR factors but prices with the risk-neutral probability $q$ in place of $p$:
$$
q = \frac{\mathcal{D}_{1,0}(g_{y}) - d}{u - d}
$$
where $\mathcal{D}_{1,0}(g_{y}) = e^{g_{y}\Delta t}$ is the risk-free accumulation factor for one step and $g_{y}$ is the continuously compounded risk-free rate (units: $\mathrm{yr}^{-1}$). We chose $p$ to match the GBM mean growth rate, but ruling out arbitrage fixes $q$. Does the lattice still match GBM? Under the risk-neutral pricing rule $\mathbb{Q}$, which replaces the drift $\mu = \mu_{g} + \sigma^{2}/2$ with $g_{y}$, the GBM growth rate over $\Delta t$ has mean $g_{y} - \sigma^{2}/2$ and variance $\sigma^{2}/\Delta t$.

Let's expand $q$ for a small time step. With $a = \sigma\sqrt{\Delta t}$, we expand each exponential through order $\Delta t$, where $\mathcal{O}(\Delta t^{k})$ collects the terms that shrink at least as fast as $\Delta t^{k}$:
$$
\begin{align*}
q & = \frac{e^{g_{y}\Delta t} - e^{-a}}{e^{a} - e^{-a}} && \text{CRR factors, } a = \sigma\sqrt{\Delta t}\\
& = \frac{\bigl(1 + g_{y}\Delta t\bigr) - \bigl(1 - a + \tfrac{1}{2}a^{2}\bigr) + \mathcal{O}(\Delta t^{3/2})}{2a + \mathcal{O}(\Delta t^{3/2})} && \text{expand each exponential through order } \Delta t\\
& = \frac{a + \bigl(g_{y} - \tfrac{\sigma^{2}}{2}\bigr)\Delta t + \mathcal{O}(\Delta t^{3/2})}{2a + \mathcal{O}(\Delta t^{3/2})} && \text{collect terms, } a^{2} = \sigma^{2}\Delta t\\
& = \frac{\tfrac{1}{2} + \frac{g_{y} - \sigma^{2}/2}{2\sigma}\sqrt{\Delta t} + \mathcal{O}(\Delta t)}{1 + \mathcal{O}(\Delta t)} && \text{divide the top and bottom by } 2a\\
& = \frac{1}{2} + \frac{g_{y} - \sigma^{2}/2}{2\sigma}\sqrt{\Delta t} + \mathcal{O}(\Delta t) && 1/\bigl(1 + \mathcal{O}(\Delta t)\bigr) = 1 + \mathcal{O}(\Delta t)
\end{align*}
$$
Apart from the remainder, $q$ is the formula for $p$ with $g_{y} - \sigma^{2}/2$ in place of $\mu_{g}$. Now we repeat the moment calculation with $q$ in place of $p$:
$$
\begin{align*}
\mathbb{E}_{\mathbb{Q}}[g] & = \frac{(2q-1)\,a}{\Delta t} && \text{mean of } g \text{ with } q \text{ in place of } p\\
& = \Bigl[\frac{g_{y} - \sigma^{2}/2}{\sigma}\sqrt{\Delta t} + \mathcal{O}(\Delta t)\Bigr]\frac{\sigma\sqrt{\Delta t}}{\Delta t} && \text{substitute the expansion of } q \text{ and } a = \sigma\sqrt{\Delta t}\\
& = g_{y} - \frac{\sigma^{2}}{2} + \mathcal{O}(\Delta t^{1/2}) && \text{multiply}\\
\mathrm{Var}_{\mathbb{Q}}(g) & = \frac{a^{2}}{\Delta t^{2}} - \Big[\frac{(2q-1)\,a}{\Delta t}\Big]^{2} && \text{variance of } g \text{ with } q \text{ in place of } p\\
& = \frac{\sigma^{2}}{\Delta t} - \Bigl(g_{y} - \frac{\sigma^{2}}{2}\Bigr)^{2} + \mathcal{O}(\Delta t^{1/2}) && \text{substitute } a = \sigma\sqrt{\Delta t} \text{ and the mean}
\end{align*}
$$
As the time step shrinks, the lattice's mean growth rate approaches $g_{y} - \sigma^{2}/2$, the GBM value under $\mathbb{Q}$. Its variance falls short of $\sigma^{2}/\Delta t$ by about $(g_{y} - \sigma^{2}/2)^{2}$, as it fell short by $\mu_{g}^{2}$ under $p$.

__When does $q$ lie between zero and one?__ From its definition, $0 < q < 1$ holds when $d < \mathcal{D}_{1,0}(g_{y}) < u$. For the CRR factors, this condition becomes:
$$
\begin{align*}
& e^{-\sigma\sqrt{\Delta t}} < e^{g_{y}\Delta t} < e^{\sigma\sqrt{\Delta t}} && \text{CRR factors}\\
& {-\frac{\sigma}{\sqrt{\Delta t}}} < g_{y} < \frac{\sigma}{\sqrt{\Delta t}} && \text{take logs, divide by } \Delta t\\
& \lvert g_{y}\rvert\sqrt{\Delta t} < \sigma && \text{multiply by } \sqrt{\Delta t}
\end{align*}
$$
The risk-free growth rate must lie between the down and up growth rates of the lattice, so the volatility must exceed a floor that shrinks with the time step. The [CRR pricing example](../../CHEME-5660-L9b-Example-AmericanContractsPremium-Fall-2026.ipynb) checks this volatility floor, `σ_min`, for every contract. With the example's short time steps, the floor sits far below the volatilities in the options chain.

## Summary
We derived the CRR up and down factors by matching the mean and variance of the lattice's one-step growth rate to geometric Brownian motion. We then showed that, for a small time step, the lattice still matches geometric Brownian motion when we price with the risk-neutral probability.

> __Key Takeaways:__
>
> * __Matching moments gives the factors:__ We wrote the lattice's one-step growth rate as a two-point random variable whose mean and variance depend on the size of a move and the probability of an up move. Matching both to geometric Brownian motion, with a down move that undoes an up move, gave the factors and the probability for a small time step.
> * __The volatility sets the size of a move:__ We found that, for a small time step, the size of a move is the volatility times the square root of the time step. The mean growth rate tilts the probability of an up move away from one half, and the tilt vanishes as the step shrinks.
> * __The factors carry over to the risk-neutral probability:__ We expanded the risk-neutral probability and found that, for a small time step, the lattice still matches the mean and variance of the growth rate under the pricing rule. The probability lies between zero and one only when the volatility exceeds a floor that shrinks with the time step.

The up and down factors depend only on the volatility and the time step, so one lattice serves both the real world and the pricing rule. Only the probability of an up move changes.
