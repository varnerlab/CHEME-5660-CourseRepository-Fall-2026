## Reading the Notation

<a id="reading-the-notation"></a>

In the tables, **1** means dimensionless and **yr** means year. Matrix dimensions appear in the definition column. Read the subscript and lecture context along with the letter.

| Symbol | Meaning and context | Units |
|---|---|---|
| $\ln x$, $\log x$ | Natural logarithm; price ratios inside a logarithm are dimensionless. | 1 |
| $\mathbb{E}[X]$ | Expected value of a random quantity $X$. | Same as $X$ |
| $\operatorname{Var}(X)$ | Variance; the square of the standard deviation. | Units of $X$ squared |
| $\operatorname{Cov}(X,Y)$ | Covariance of two random quantities. | Units of $X$ times units of $Y$ |
| $\mathbb{P}$, $\mathbb{Q}$ | Real-world and risk-neutral probability laws (L3b). | Probabilities: 1 |
| $\hat{x}$ | Usually an estimate from observations. The augmented design matrix $\hat{\mathbf{X}}$ is a separate convention (L4b, L6b). | Same as $x$ |
| $g^{\prime}$ | Sample mean growth rate; the prime marks a mean in these lectures. | $\mathrm{yr}^{-1}$ |
| $\mathbf{x}$, $\mathbf{A}$ | Bold symbols denote vectors or matrices; subscripts identify entries. | Context dependent |
| $\mathbf{A}^{\top}$, $\mathbf{A}^{-1}$ | Transpose and inverse, respectively. An inverse requires a square, nonsingular matrix. | Context dependent |
| $\mathbf{1}$, $\mathbf{I}_M$ | Vector of ones of the required length; $M\times M$ identity matrix. | 1 |
| $\lVert\mathbf{x}\rVert_2^2$ | Sum of squared entries of $\mathbf{x}$. | Entry units squared |

**Read an expression as a calculation.** For example, $g_j=\ln(S_j/S_{j-1})/\Delta t$ starts with a price ratio, takes its natural logarithm, and divides by the elapsed time. The price units cancel in the ratio; dividing by years gives a growth rate in inverse years. The L3a section works through this calculation.

**Separate inputs from outputs.** In a forecast we supply an initial price, parameter estimates, and a holding period. The model then supplies a distribution of future prices. An expectation averages over that distribution; a probability adds up the portion satisfying a stated event, such as exceeding a sale-price target.

**Distinguish an estimate from the quantity it describes.** The sample mean $g^{\prime}$ estimates the population mean growth rate $\mu_g$. A hat usually marks a fitted quantity, such as $\hat\mu_g$.

L1a is the orientation slide deck. Its option-preview symbol $\delta$ denotes option-price sensitivity to the underlying price; its detailed development comes later in the course.

## L1b: Time Value of Money and Net Present Value

<a id="l1b-time-value-of-money-and-net-present-value"></a>

We move cash flows to a common date before comparing them. The accumulation factor moves money forward in time, and its inverse discounts a future payment to time zero.

| Symbol | Meaning and context | Units |
|---|---|---|
| $P$, $F_j$ | Present value at time zero and future value after $j$ compounding periods. | USD |
| $y$ | Nominal annual yield, quoted as a decimal, with its compounding convention. | $\mathrm{yr}^{-1}$ |
| $n$ | Compounding frequency: number of periods per year. | $\mathrm{yr}^{-1}$ |
| $j$, $\ell$, $N$ | Period indices and final cash-flow index; $j/n$ is elapsed time in years. | Counts |
| $T$ | Elapsed time or investment horizon; $nT$ is the number of periods. | yr |
| $i_\ell$ | Effective rate during compounding period $\ell$. | 1 |
| $\mathcal{D}_{j,0}$ | Accumulation factor from date 0 to period $j$; for a constant yield, $(1+y/n)^j$. | 1 |
| $\mathcal{D}_{j,0}^{-1}$ | Discount factor from period $j$ to date 0. | 1 |
| $g(t)$, $g$ | Time-varying or constant continuously compounded growth rate. | $\mathrm{yr}^{-1}$ |
| $g_y$ | Continuous growth rate equivalent to the quoted yield: $n\log(1+y/n)$. | $\mathrm{yr}^{-1}$ |
| $\mathcal{A}$ | Abstract asset represented by its dated net cash flows. | Cash-flow collection |
| $c_{j,k}$, $\mathbf{c}_j$ | Nonnegative amount of cash flow $k$ at date $j$; vector of the $m_j$ amounts. | USD |
| $\nu_{j,k}$, $\boldsymbol{\nu}_j$ | Cash-flow direction: $+1$ for money received, $-1$ for money paid. | 1 |
| $m_j$ | Number of cash-flow components at date $j$. | Count |
| $\bar c_j$, $\bar{\mathbf c}$ | Signed net cash flow at date $j$; vector of net cash flows at dates $0,\ldots,N$. | USD |
| $\mathcal{D}_{\star,0}^{-1}$ | Vector of the $N+1$ discount factors for those cash-flow dates. | 1 |
| $\operatorname{NPV}_0$ | Net present value: sum of discounted signed cash flows. | Time-0 USD |


**How do the cash-flow symbols become a value?** At date $j$, multiply each cash amount $c_{j,k}$ by its direction $\nu_{j,k}$ and add the results. Then discount that date's net cash flow to time zero:

$$
\bar c_j=\sum_{k=1}^{m_j}c_{j,k}\nu_{j,k},\qquad
\operatorname{NPV}_0=\sum_{j=0}^{N}\mathcal D_{j,0}^{-1}\bar c_j.
$$

The first sum combines payments made on the same date. The second combines their present values across dates. For example, paying 100 USD today and receiving 110 USD in one year at a 5% annual benchmark gives $\operatorname{NPV}_0=-100+110/1.05\approx4.76$ USD. The positive result means that the payment exceeds the benchmark requirement by 4.76 USD in today's money. The discount factor is dimensionless, so each term in the NPV retains units of USD.

## L2a: Treasury Bills, Notes, and Bonds

<a id="l2a-treasury-bills-notes-and-bonds"></a>

The yield convention determines the accumulation and discount factors. A bill's bank-discount quote uses a 360-day basis; the investment horizon $T$ uses the year-fraction convention selected for valuation.

| Symbol | Meaning and context | Units |
|---|---|---|
| $V_B$ | Purchase price or modeled value of the Treasury security. | USD |
| $V_P$ | Par (face) value paid at maturity. | USD |
| $d$ | Annual bank-discount quote for a Treasury bill, entered as a decimal. | Annual quote, 360-day basis |
| $D$ | Actual number of days remaining to bill maturity. | Days |
| $T$ | Time to maturity under the selected valuation day-count convention. | yr |
| $y$, $n$, $g_y$ | Nominal annual yield, compounding frequency, and equivalent continuous rate from L1b. | $\mathrm{yr}^{-1}$ |
| $g_B$ | Realized continuously compounded growth for a bill held to maturity. | $\mathrm{yr}^{-1}$ |
| $r_B$ | Bill's holding-period log return: $r_B=Tg_B$. | 1 |
| $c$ | Nominal annual coupon rate as a decimal. | $\mathrm{yr}^{-1}$ |
| $C$ | Cash coupon per payment date: $C=(c/n)V_P$. | USD |
| $N$ | Number of remaining coupon payments; $N=nT$ on the lecture's coupon-date grid. | Count |
| $\mathcal{D}_{N,0}^{-1}$ | Discount factor for the par repayment date. | 1 |


**From a bill quote to investment growth.** The bank-discount quote expresses the discount as a fraction of par. We first calculate the purchase price, then use the purchase-to-par ratio to calculate growth over the holding period:

$$
V_B=V_P\left(1-d\frac{D}{360}\right),\qquad
g_B=\frac{1}{T}\ln\left(\frac{V_P}{V_B}\right).
$$

For $V_P=100$ USD, $d=0.04$, and $D=90$ days, the bill costs 99 USD. If we select $T=90/365$ years for valuation, $g_B\approx0.04076\,\mathrm{yr}^{-1}$. The quote uses par and a 360-day basis; the growth calculation uses the amount invested and the selected holding-time convention.

**What changes for a coupon bond?** We receive coupons along the way as well as par at maturity. On a coupon date, with a constant discount yield, its value is the sum of those discounted payments:

$$
V_B=\underbrace{V_P\mathcal D_{N,0}^{-1}}_{\text{present value of par}}
+\underbrace{C\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}}_{\text{present value of coupons}}.
$$

Each discount factor applies to the date of its payment. This is the L1b cash-flow calculation with the bond's payment schedule supplied.

## L2b: Yield Sensitivity, Duration, Convexity, and the Curve

<a id="l2b-yield-sensitivity-duration-convexity-and-the-curve"></a>

Duration and convexity summarize a price change when the quoted yield changes. The spot curve supplies a separate discount factor for each payment date.

| Symbol | Meaning and context | Units |
|---|---|---|
| $V_B(y,T)$ | Security price as a function of yield and time to maturity. | USD |
| $X_j$ | Promised cash flow at payment $j$, including par at maturity. | USD |
| $t_j=j/n$ | Time of payment $j$. | yr |
| $V_{B,y}$ | Yield-sensitivity magnitude: $-\partial V_B/\partial y$. | $\mathrm{USD}\cdot\mathrm{yr}$ |
| $V_{B,T}$ | Bill maturity-sensitivity quantity: $-\partial V_B/\partial T$. | $\mathrm{USD}/\mathrm{yr}$ |
| $V_{B,c}$ | Coupon-rate sensitivity: $\partial V_B/\partial c$. | $\mathrm{USD}\cdot\mathrm{yr}$ |
| $D_{\mathrm{mod}}$ | Modified duration: $-(1/V_B)(dV_B/dy)$. | yr |
| $D_{\mathrm{mac}}$ | Macaulay duration: payment times averaged using their shares of present value. | yr |
| $K(y)$ | Convexity: $(1/V_B)(d^2V_B/dy^2)$. | $\mathrm{yr}^{2}$ |
| $\Delta y$, $\Delta V_B$ | Finite changes in quoted yield and price. Enter yield changes as decimals; 1 basis point is 0.0001. | $\mathrm{yr}^{-1}$; USD |
| $\bar y$ | Yield to maturity: one constant yield reproducing the market price from promised payments. | $\mathrm{yr}^{-1}$ |
| $D(0,t_j)$ | Spot-curve discount factor for a payment at $t_j$. This is the date-indexed counterpart of $\mathcal D_{j,0}^{-1}$. | 1 |
| $c_{\mathrm{par}}$ | Coupon rate that prices a hypothetical bond at par under the spot curve. | $\mathrm{yr}^{-1}$ |


**How much will a yield change move the price?** Modified duration gives the first-order price sensitivity; convexity corrects for curvature. With the lecture's nominal-yield convention:

$$
D_{\mathrm{mod}}=\frac{D_{\mathrm{mac}}}{1+y/n},\qquad
\frac{\Delta V_B}{V_B}\approx-D_{\mathrm{mod}}\Delta y+\frac12K(\Delta y)^2.
$$

For $D_{\mathrm{mod}}=5$ yr, $K=30\,\mathrm{yr}^2$, and a yield increase $\Delta y=0.01\,\mathrm{yr}^{-1}$, the estimated price change is $-0.05+0.0015=-0.0485$, or a 4.85% decrease. This local approximation is most useful for small yield changes.

**How do we value payments with a yield curve?** The curve supplies a discount factor $D(0,t_j)$ for each payment date. Adding the discounted payments gives the price; setting that price equal to par and solving for the coupon rate gives:

$$
V_B=\sum_{j=1}^{N}X_jD(0,t_j),\qquad
c_{\mathrm{par}}=n\frac{1-D(0,t_N)}{\sum_{j=1}^{N}D(0,t_j)}.
$$

The par coupon makes the present value of the coupons equal the difference between par and the present value of the final par repayment.

## L3a: Observed Growth Rates and Stylized Facts

<a id="l3a-observed-growth-rates-and-stylized-facts"></a>

**What persists when direction is hard to predict?** Volatility clustering describes persistence in the **size** of movements. A turbulent period can contain both gains and losses. We measure direction with signed growth rates and movement size with their absolute values.

First convert consecutive prices into growth rates. For asset $i$, the observation interval $\Delta t_j$ divides the log price ratio:

$$
g_{j,j-1}^{(i)}=\frac{1}{\Delta t_j}\ln\left(\frac{S_j^{(i)}}{S_{j-1}^{(i)}}\right).
$$

A change from 100 to 101 USD/share over $1/252$ year gives a log return $\ln(1.01)\approx0.00995$ and a growth rate $252\ln(1.01)\approx2.508\,\mathrm{yr}^{-1}$. The rate is large because we divide a small one-day change by a small fraction of a year. If that same log-price change occurred on each of 252 trading days, the accumulated log change would be 2.508. Here, we use the per-year scale to record one observed day; a forecast requires a model for the days that follow.

![The L3a lecture figure: daily SPY and AMD growth rates, a shared high-volatility period, and an isolated AMD price jump.](figs/Fig-L3a-AMD-Volatility-Clustering.svg)

**What do we see?** This is the figure from L3a's *Volatility Clustering* section. The shaded March-2020 period contains many large movements of both signs in SPY and AMD. The isolated April-2016 AMD spike is one large jump. Clustering concerns a sequence of large movements, so we distinguish that turbulent period from the isolated spike.

The lecture measures persistence using **autocorrelation**, the linear association between observations separated by a specified number of time steps. Applying it to signed growth rates asks whether positive or negative deviations tend to recur. Applying it to growth magnitudes asks whether large or small movements tend to recur. Positive magnitude autocorrelation across several lags is evidence of persistence in movement size.

**Why does this matter for a model?** A fixed volatility uses the same uncertainty scale after calm and turbulent days. Clustering suggests that recent movement sizes contain information about upcoming risk. When we reach the lattice and GBM models, we can ask whether their assumptions preserve or discard that information.

**Check.** If signed growth has little autocorrelation but magnitude has persistent positive autocorrelation, can we infer the sign of tomorrow's change? **Answer:** this evidence supports persistence in movement size. It supplies no directional prediction or fixed date when a turbulent episode must end.

**Reading the two autocorrelations.** For signed growth, $\hat R(k)$ averages products of deviations from the sample mean, and $\rho(k)$ divides by the zero-lag variance. For magnitude autocorrelation $\hat\rho_i(k)$, apply the same operation to $|g_t^{(i)}|$ after subtracting its own mean. The normalization cancels the growth units, allowing us to compare persistence on the same scale.

**Growth variability versus diffusion volatility.** Multiplying growth rates by $\Delta t$ gives interval log returns, so $\sigma_r(\Delta t)=\Delta t\,\sigma_g$. Under GBM, the log-return standard deviation is also $\sigma\sqrt{\Delta t}$. Equating these gives $\sigma=\sqrt{\Delta t}\,\sigma_g$. As an illustrative daily-scale conversion, $\sigma_g\approx3.1749\,\mathrm{yr}^{-1}$ corresponds to $\sigma\approx0.20\,\mathrm{yr}^{-1/2}$. Scaling observed quantities at one interval is a change of units; extrapolating variability across horizons adds model assumptions.

## L3b: Binomial Price Lattices

<a id="l3b-binomial-price-lattices"></a>


**Why does one tree use two probabilities?** Suppose a share costs 100 USD today and will be worth either 120 or 90 USD after one step. The up and down factors are $u=1.20$ and $d=0.90$. Keep these prices fixed while we ask two different questions.

**Forecasting: how often might each outcome occur?** Let the estimated real-world up probability be $p=0.65$. Weighting the two prices by their forecast probabilities gives:

$$
\mathbb E_{\mathbb P}[S_1]=0.65(120)+0.35(90)=109.50\ \mathrm{USD/share}.
$$

This average describes the forecast distribution. Its quality depends on whether the estimated probabilities describe later outcomes adequately.

**Pricing: what should a specified payoff cost today?** Suppose a claim pays $H_u=20$ USD after an up move and $H_d=0$ after a down move. Let the risk-free accumulation factor over the same step be $R_f=1.02$. With frictionless trading, short positions allowed, and borrowing or lending at this rate, no-arbitrage requires the risk-neutral weight $q$ to satisfy:

$$
qu+(1-q)d=R_f
\quad\Longrightarrow\quad
q=\frac{R_f-d}{u-d}=\frac{1.02-0.90}{1.20-0.90}=0.40.
$$

The condition $d<R_f<u$ gives $0<q<1$. We use these weights to average the claim's payoffs and then discount one step:

$$
V_0=\frac{qH_u+(1-q)H_d}{R_f}
=\frac{0.40(20)+0.60(0)}{1.02}\approx7.84\ \mathrm{USD}.
$$

**Why is that the price?** We can build the claim's payoff from the share and a loan. The claim's payoff differs by $20-0=20$ USD between the two states, while the share price differs by $120-90=30$ USD/share. Holding $20/30=2/3$ of a share therefore reproduces the payoff difference. In the down state those shares are worth 60 USD and the claim pays zero, so we need a loan repayment of 60 USD. Borrowing $60/1.02$ USD today supplies that repayment after one step. After one step the shares are worth 80 or 60 USD, and the loan repayment is 60 USD. The net payoff is therefore 20 or zero, exactly the claim's payoff. The initial cost is $100(2/3)-60/1.02\approx7.84$ USD. A different claim price would let us trade the claim against this replicating position for an arbitrage profit under the stated assumptions.

Thus $p$ describes forecast frequencies, while $q$ values a payoff through a trading argument. Risk-neutral pricing does not assert that the stock is risk-free or that 40% of actual moves will be upward.

**Check.** Change only $p$ from 0.65 to 0.50. What changes? **Answer:** the forecast expected price becomes 105 USD/share. The replicated payoff still costs about 7.84 USD because the two possible share prices and the borrowing rate are unchanged.

| $\tau$, $\rho_g(\tau)$ | Lag in lattice steps and growth autocorrelation at that lag. | Count; 1 |


**Extending the forecast tree.** With fixed multipliers and independent moves, a path containing $k$ up moves in $t$ steps ends at $S_0u^kd^{t-k}$. There are $\binom tk$ orders of those moves, so the node's price and total probability are:

$$
S_{t,k}=S_0u^kd^{t-k},\qquad
\mathbb P(k\text{ up moves})=\binom tk p^k(1-p)^{t-k}.
$$

For two steps and $p=0.65$, the middle node is $100(1.20)(0.90)=108$ USD/share. It can be reached by up-down or down-up, so its probability is $2(0.65)(0.35)=0.455$. This explains why the binomial coefficient appears. The three node probabilities sum to one.

## L4a: Quotes, Trade NPV, and Lattice Targets

<a id="l4a-quotes-trade-npv-and-lattice-targets"></a>

We evaluate a long position at a scheduled sale time. Scaled NPV measures the discounted gain relative to the initial investment.

| Symbol | Meaning and context | Units |
|---|---|---|
| $a_t$, $b_t$ | Best ask and best bid at time $t$. | USD/share |
| $s_t$, $m_t$ | Quoted spread $a_t-b_t$ and midpoint $(a_t+b_t)/2$. | USD/share |
| $n_0$ | Number of shares bought at entry and held to sale. | Shares |
| $S_0$, $S_T$ | Entry and terminal model share prices. | USD/share |
| $N$, $T=N\Delta t$ | Number of lattice steps and elapsed holding time. | Count; yr |
| $g_y$ | Continuous benchmark growth rate used to discount sale proceeds. | $\mathrm{yr}^{-1}$ |
| $\rho_T$ | Scaled trade NPV: $\operatorname{NPV}_0/(n_0S_0)$. | 1 |
| $\rho_\star$ | Investor's target scaled NPV, with $\rho_\star>-1$. | 1 |
| $\rho(k)$ | Scaled NPV at the terminal node with $k$ up moves. | 1 |
| $K_N$ | Random number of up moves in $N$ steps; binomial with parameters $N,p$. | Count |
| $\tau(\rho_\star)$ | Real-valued up-count threshold obtained by solving $\rho(k)>\rho_\star$. | Step-count threshold |
| $\kappa$, $k_{\min}$ | $\kappa=\lfloor\tau\rfloor+1$ is the first strictly successful integer; $k_{\min}$ clips it to $[0,N+1]$. | Counts |
| $m$, $f_j$, $p_j$ | Number of branches, positive price factor on branch $j$, and its probability in an $m$-ary lattice. | Count; 1; 1 |
| $x_j$, $\mathbf x$, $\mathbf X_t$ | Number of branch-$j$ moves, count vector, and random count vector after $t$ steps; $\sum_jx_j=t$. | Counts |
| $S_t(\mathbf x)$ | Price at count state $\mathbf x$: $S_0\prod_{j=1}^{m}f_j^{x_j}$. | USD/share |
| $L_t$ | Number of count states: $\binom{t+m-1}{m-1}$. Distinct count states can have the same price. | Count |


**Which lattice outcomes beat our target?** Buying $n_0$ shares costs $n_0S_0$ today; their sale brings $n_0S_T$ at time $T$. Ignoring dividends and trading costs, discount and divide by the purchase cost:

$$
\rho_T=\frac{-n_0S_0+n_0S_Te^{-g_yT}}{n_0S_0}
=\frac{S_T}{S_0}e^{-g_yT}-1.
$$

At a terminal lattice node with $k$ up moves, substitute $S_T/S_0=u^kd^{N-k}$. Taking logs of $\rho(k)>\rho_\star$ and collecting the terms in $k$ gives, for $u>d>0$:

$$
k>\tau(\rho_\star)
=\frac{\ln(1+\rho_\star)+g_yT-N\ln d}{\ln(u/d)}.
$$

The up count is an integer, so the first strictly successful count is $\lfloor\tau\rfloor+1$. Add the binomial probabilities for that count and all higher counts. If $\tau=2$ exactly, success starts at three up moves; two only meets the target. Clipping the first count to $[0,N+1]$ represents the cases where every node or no node succeeds. This turns a trading requirement into a probability over the model's terminal outcomes.

## L4b: Single-Asset Geometric Brownian Motion

<a id="l4b-single-asset-geometric-brownian-motion"></a>


**Which average do drift and mean growth describe?** Geometric Brownian motion uses drift $\mu$ for expected price growth and volatility $\sigma$ for the size of random changes. With constant parameters, the price equation is:

$$
\frac{dS(t)}{S(t)}=\mu\,dt+\sigma\,dW(t).
$$

Here $W(t)$ is a Wiener process; its increment over $\Delta t$ has variance $\Delta t$. The random price increment has zero conditional mean, so averaging the price equation gives $d\mathbb E[S(t)]/dt=\mu\mathbb E[S(t)]$. Solving this mean equation gives $\mathbb E[S_T]=S_0e^{\mu T}$. We will compare that expression with the average obtained from the log-price solution to find how drift relates to mean growth.

For the price at a fixed time $T$, the solution in terms of mean growth $\mu_g$ is:

$$
\ln(S_T/S_0)=\mu_gT+\sigma\sqrt T\,Z,
\qquad Z\sim\mathcal N(0,1).
$$

Averaging this equation makes the random term vanish because $\mathbb E[Z]=0$. Thus $\mu_g$ is expected log growth per unit time. To average the price itself, we exponentiate first. The normal identity $\mathbb E[e^{aZ}]=e^{a^2/2}$, valid for any real number $a$, gives:

$$
\mathbb E[S_T]
=S_0e^{\mu_gT}\mathbb E[e^{\sigma\sqrt T Z}]
=S_0e^{(\mu_g+\sigma^2/2)T}.
$$

Comparing this result with $\mathbb E[S_T]=S_0e^{\mu T}$ gives $\mu=\mu_g+\sigma^2/2$, or $\mu_g=\mu-\sigma^2/2$. This is the **half-variance correction**. Averaging prices and averaging their logarithms are different operations; the normal exponential average supplies the precise difference for GBM. The L4b derivation obtains the same correction from Itô's lemma.

**A numerical comparison.** Let $S_0=100$ USD/share, $\mu=0.08\,\mathrm{yr}^{-1}$, $\sigma=0.20\,\mathrm{yr}^{-1/2}$, and $T=1$ year. Then $\mu_g=0.06\,\mathrm{yr}^{-1}$, giving:

$$
\begin{aligned}
\mathbb E[S_T]&=100e^{0.08}\approx108.33\ \mathrm{USD/share},\\
\operatorname{median}(S_T)&=100e^{0.06}\approx106.18\ \mathrm{USD/share}.
\end{aligned}
$$

The median splits the possible prices into equal lower and upper halves. It follows from the median $Z=0$ and the increasing exponential function. The right tail pulls the expected price above the median; neither summary identifies a particular future path.

**Which number belongs in a simulation?** The equivalent updates are:

$$
S_{t+\Delta t}=S_t e^{\mu_g\Delta t+\sigma\sqrt{\Delta t}\,Z}
=S_t e^{(\mu-\sigma^2/2)\Delta t+\sigma\sqrt{\Delta t}\,Z}.
$$

The same draw $Z$ gives the same price in both forms. If we estimate $\mu_g$ from observed growth rates, the correction is already included; subtracting it again would lower the simulated mean growth incorrectly.

**Check.** Hold $\mu=0.08\,\mathrm{yr}^{-1}$ fixed and increase $\sigma$ to $0.40\,\mathrm{yr}^{-1/2}$. **Answer:** $\mu_g=0$, the expected one-year price remains about 108.33 USD/share, and the median becomes 100 USD/share. Higher volatility separates the two averages further.

## L4b: What Can a Price Model Help Us Decide?

<a id="l4b-what-can-a-price-model-help-us-decide"></a>

**What can we decide if we cannot predict the exact price?** Suppose we buy a share for $S_0=100$ USD/share and plan to sell after $T=1$ year. We want the proceeds to exceed an investment growing at the benchmark rate $g_y=0.02\,\mathrm{yr}^{-1}$. We assume a scheduled sale, no dividends or trading costs, and a constant benchmark rate. Discounting the sale price $S_T$ gives the scaled net present value:

$$
\rho_T=\frac{-S_0+S_Te^{-g_yT}}{S_0}.
$$

The target $\rho_T>0$ means that the discounted proceeds exceed the purchase cost. Solving this inequality tells us which future prices count as successes:

$$
\rho_T>0\quad\Longleftrightarrow\quad
S_T>S_0e^{g_yT}=100e^{0.02}\approx102.02\ \mathrm{USD/share}.
$$

**Where does the probability come from?** Under geometric Brownian motion (GBM), the log price ratio at the sale date is normally distributed. Let $\mu_g=0.06\,\mathrm{yr}^{-1}$ be mean growth and $\sigma=0.20\,\mathrm{yr}^{-1/2}$ be volatility. The model writes:

$$
\ln(S_T/S_0)=\mu_gT+\sigma\sqrt T\,Z,
\qquad Z\sim\mathcal N(0,1).
$$

Here $Z$ is a standard normal random variable. Substitute the model into the target condition, then subtract the mean and divide by the positive scale $\sigma\sqrt T$:

$$
\begin{aligned}
\mu_gT+\sigma\sqrt T\,Z&>g_yT,\\
Z&>\frac{(g_y-\mu_g)T}{\sigma\sqrt T}
=\frac{0.02-0.06}{0.20}=-0.20.
\end{aligned}
$$

The normal cumulative distribution function $\Phi(z)$ gives the probability to the left of $z$. Our target uses the area to the right, so:

$$
\mathbb P(\rho_T>0)=1-\Phi(-0.20)\approx0.5793.
$$

We obtain a **57.9% model probability of beating the benchmark**, and a 42.1% probability of finishing at or below it. This supplies a risk comparison for the specified trade. It concerns the sale date; exiting when a price first reaches a boundary is a different trading rule.

**How much should we trust the answer?** If mean growth were $0.02\,\mathrm{yr}^{-1}$ instead, the threshold for $Z$ would be zero and the success probability would be 50%. Estimated inputs therefore matter. The model also assumes constant parameters and independent, normally distributed log-price increments. Clustering or heavier tails in the data can change the risk beyond what a new mean estimate captures.

To check the forecast, fit on earlier observations and save probabilities for later, fixed-horizon trades before their outcomes are known. Over many comparable cases, compare the forecast probabilities with the frequency of the same event. For a prediction interval at one date, we check whether the price at that date lies inside it. To check a band intended to contain an entire path, every observed price must remain inside its bounds; one crossing is enough to fail that check. The L5a out-of-sample example develops that distinction.

**Check.** Keep the parameters above but require a 5% scaled NPV. What changes? **Answer:** the price threshold becomes $100(1.05)e^{0.02}\approx107.12$, and the probability falls to about 48.2%. We raise the target before asking the model for its probability.

## L4b: Regression and Parameter Uncertainty

<a id="l4b-regression-and-parameter-uncertainty"></a>

The lecture also fits a straight line to log price against time. These symbols describe that regression; L6b reuses the regression structure with different response and predictor variables.

| Symbol | Meaning and context | Units |
|---|---|---|
| $\mathbf y$ | Vector of the $N+1$ observed log prices. A fixed price unit is implicit when taking log price. | 1 |
| $\hat{\mathbf X}$ | Augmented design matrix: $N+1$ rows, columns of ones and times $t_j$. | Columns: 1; yr |
| $\boldsymbol\theta$ | Parameter vector $(\ln S_0,\mu_g)^{\top}$. | Entries: 1; $\mathrm{yr}^{-1}$ |
| $\hat{\boldsymbol\theta}$ | Least-squares estimate of the intercept and slope. | Same as parameters |
| $\epsilon_j$ | Cumulative stochastic deviation of log price from the line at time $t_j$. | 1 |
| $\mathbf r$ | Fitted residual vector: $\mathbf y-\hat{\mathbf X}\hat{\boldsymbol\theta}$. | 1 |
| $n=N+1$, $p=2$ | Number of regression observations and number of fitted parameters. | Counts |
| $\nu=n-p$ | Residual degrees of freedom. | Count |
| $\hat\sigma_\epsilon^2$ | Residual variance estimate under the lecture's simplifying regression-error model. | 1 |
| $\operatorname{SE}(\hat\theta_j)$ | Standard error of fitted coefficient $j$. | Same as coefficient |
| $\alpha$ | Significance level; the stated confidence level is $1-\alpha$. | 1 |
| $t_{1-\alpha/2,\nu}$ | Student-$t$ quantile used in a two-sided coefficient interval. | 1 |


**What are we fitting, and what does the interval describe?** For observation $j$, the lecture fits log price as an intercept plus mean growth times elapsed time. Stacking all observations gives:

$$
\ln S_j=\ln S_0+\mu_gt_j+\epsilon_j
\quad\Longleftrightarrow\quad
\mathbf y=\hat{\mathbf X}\boldsymbol\theta+\boldsymbol\epsilon.
$$

Each row of the design matrix is $[1,t_j]$. Least squares chooses the intercept and slope to minimize the squared residuals. For a fitted coefficient $\hat\theta_j$, the standard error estimates its uncertainty; a smaller value indicates greater precision. The lecture's two-sided confidence interval is:

$$
\hat\theta_j\ \pm\ t_{1-\alpha/2,\nu}\operatorname{SE}(\hat\theta_j).
$$

This is an interval for a model coefficient. A prediction interval for a future price must also account for future random price changes. The classical coefficient interval assumes independent, constant-variance, normally distributed errors. GBM log-price errors accumulate through time and share earlier shocks, so this simplifying regression-error assumption limits the interval's interpretation. Fitting a line alone does not validate its uncertainty calculation.

## L5a: Trade Targets and Updating GBM Estimates

<a id="l5a-trade-targets-and-updating-gbm-estimates"></a>

We retain the GBM and trade-NPV notation from L4a-L4b. The new symbols describe a terminal price target and exponential moving estimates of growth-rate moments.

| Symbol | Meaning and context | Units |
|---|---|---|
| $K$ | Terminal price threshold corresponding to target scaled NPV: $S_0(1+\rho_\star)e^{g_yT}$. | USD/share |
| $\rho_T$, $\rho_\star$ | Realized scaled trade NPV and the investor's target. | 1 |
| $s$, $k$ | Entry observation index and subsequent update index. | Counts |
| $g_k$ | Latest growth observation: $\ln(S_k/S_{k-1})/\Delta t$. | $\mathrm{yr}^{-1}$ |
| $\lambda$ | EMA decay factor, with $0<\lambda<1$. | 1 |
| $H_{\mathrm{half}}$ | Number of observations over which an old weight halves; $\lambda=2^{-1/H_{\mathrm{half}}}$. | Observation count |
| $\delta_k$ | Innovation relative to the previous mean: $g_k-m_{k-1}$. | $\mathrm{yr}^{-1}$ |
| $m_k$ | Exponentially weighted mean growth rate after observation $k$. | $\mathrm{yr}^{-1}$ |
| $v_k$ | Exponentially weighted growth-rate variance after observation $k$. | $\mathrm{yr}^{-2}$ |
| $\hat\mu_{g,k}$ | Updated GBM mean log growth: $m_k$. | $\mathrm{yr}^{-1}$ |
| $\hat\sigma_k$ | Updated GBM volatility: $\sqrt{v_k\Delta t}$. | $\mathrm{yr}^{-1/2}$ |
| $\hat\mu_k$ | Updated GBM drift: $\hat\mu_{g,k}+\hat\sigma_k^2/2$. | $\mathrm{yr}^{-1}$ |
| $\hat\mu_{g,0}$, $\hat\sigma_0$ | Initial training estimates used to initialize $m_s$ and $v_s$. | $\mathrm{yr}^{-1}$; $\mathrm{yr}^{-1/2}$ |


**What changes when another price arrives?** The new growth observation $g_k$ differs from the previous mean by the innovation $\delta_k=g_k-m_{k-1}$. We retain a fraction $\lambda$ of the old information and give the new observation weight $1-\lambda$:

$$
m_k=\lambda m_{k-1}+(1-\lambda)g_k
=m_{k-1}+(1-\lambda)\delta_k.
$$

For a 21-observation half-life, $\lambda=2^{-1/21}\approx0.9675$, so the new observation receives about 3.25% of the mean's weight. Half-life describes estimation memory; $T$ describes how far ahead we forecast.

**Why is the variance update different?** Variance is measured about the updated mean. Measuring the old observations about this new center adds the squared shift of the mean to the old variance. Combining their contribution with the new observation gives:

$$
v_k=\lambda\left[v_{k-1}+(m_{k-1}-m_k)^2\right]
+(1-\lambda)(g_k-m_k)^2.
$$

The mean update gives $m_k-m_{k-1}=(1-\lambda)\delta_k$, and the new observation's deviation is $g_k-m_k=\lambda\delta_k$. Substituting these two deviations and collecting terms yields:

$$
\begin{aligned}
v_k&=\lambda v_{k-1}+\lambda(1-\lambda)^2\delta_k^2
+(1-\lambda)\lambda^2\delta_k^2\\
&=\lambda\left[v_{k-1}+(1-\lambda)\delta_k^2\right].
\end{aligned}
$$

Initialize $m_s=\hat\mu_{g,0}$ and $v_s=\hat\sigma_0^2/\Delta t$ from the training data. The updated growth variance has units $\mathrm{yr}^{-2}$; multiplying by $\Delta t$ and taking a square root gives GBM volatility $\hat\sigma_k=\sqrt{v_k\Delta t}$. Each new forecast uses the updated estimates. In the course example, those estimates remain fixed over that forecast's horizon, so updating them does not by itself model future volatility clusters within the horizon.

## L5b: Multiple-Asset GBM and Covariance

<a id="l5b-multiple-asset-gbm-and-covariance"></a>


**How can independent inputs produce correlated asset movements?** Give the assets a shared random input. Suppose their volatility parameters are 0.20 and $0.30\,\mathrm{yr}^{-1/2}$, with correlation 0.50. The required covariance rate and a loading matrix that produces it are:

$$
\mathbf C=\begin{bmatrix}0.04&0.03\\0.03&0.09\end{bmatrix}\mathrm{yr}^{-1},
\qquad
\mathbf A=\begin{bmatrix}0.20&0\\0.15&\sqrt{0.0675}\end{bmatrix}\mathrm{yr}^{-1/2}.
$$

The first diagonal loading is $\sqrt{0.04}=0.20$. The shared loading is $0.03/0.20=0.15$; the remaining loading is $\sqrt{0.09-0.15^2}=\sqrt{0.0675}\approx0.259808$. These choices make $\mathbf A\mathbf A^\top=\mathbf C$.

Let $Z_1$ and $Z_2$ be independent standard normal variables. Write $\mathbf X=\mathbf A\mathbf Z$ for the two noise contributions before time scaling. Expanding the matrix product gives:

$$
X_1=0.20Z_1,\qquad X_2=0.15Z_1+\sqrt{0.0675}\,Z_2.
$$

Both assets use $Z_1$, which produces their covariance. Independence removes the cross term containing $Z_1Z_2$, and each input has variance one, so:

$$
\begin{aligned}
\operatorname{Var}(X_1)&=0.20^2=0.04,\\
\operatorname{Var}(X_2)&=0.15^2+0.0675=0.09,\\
\operatorname{Cov}(X_1,X_2)&=0.20(0.15)=0.03.
\end{aligned}
$$

These values have units $\mathrm{yr}^{-1}$. Dividing the covariance by $0.20(0.30)$ recovers correlation 0.50. One draw illustrates the multiplication: if $\mathbf Z=(1,-1)^\top$, then $\mathbf X\approx(0.20,-0.109808)^\top\,\mathrm{yr}^{-1/2}$. Positive correlation allows individual draws with opposite signs; it describes the joint distribution across draws.

**How do these numbers enter a price step?** For asset $i$, add mean log growth and the time-scaled noise in the exponent:

$$
S_i(t+\Delta t)=S_i(t)\exp\!\left[\mu_{g,i}\Delta t+\sqrt{\Delta t}\,X_i\right],
\qquad \mu_{g,i}=\mu_i-C_{ii}/2.
$$

For $\Delta t=1/252$ yr, starting prices of 100 USD/share, and both mean growth rates $0.06\,\mathrm{yr}^{-1}$, the draw above gives next prices about 101.29 and 99.33 USD/share. The same vector is shared across assets within the step; we draw a fresh vector at the next step.

**Why use a loading matrix?** For independent standard normal inputs, $\operatorname{Cov}(\mathbf A\mathbf Z)=\mathbf A\mathbf A^\top$. Multiplying by $\mathbf C$ itself would produce $\mathbf C\mathbf C^\top$, with the wrong covariance and units.

**Check.** Replace the loading 0.15 by $-0.15$. **Answer:** both variances stay the same because they use squared loadings. Covariance becomes $-0.03\,\mathrm{yr}^{-1}$ and correlation becomes $-0.50$.

\qquad
\mathbf\Sigma_g=\operatorname{Cov}(\mathbf g_k)=\frac{\mathbf C}{\Delta t}.
$$

To estimate that growth covariance, center each data column on its own mean. The $(i,j)$ entry of $\tilde{\mathbf G}^{\top}\tilde{\mathbf G}$ adds the products of aligned deviations for assets $i$ and $j$. Divide by $N-1$, then multiply by the observation interval to estimate the covariance rate:

$$
\hat{\mathbf\Sigma}_g=\frac{\tilde{\mathbf G}^{\top}\tilde{\mathbf G}}{N-1},
\qquad \hat{\mathbf C}=\Delta t\,\hat{\mathbf\Sigma}_g.
$$

Thus growth covariance has units $\mathrm{yr}^{-2}$, covariance rate has units $\mathrm{yr}^{-1}$, and the log-price shock covariance over a step is $\Delta t\,\mathbf C$, which is dimensionless. ## L5b: Portfolio Weights and Dirichlet Sampling

<a id="l5b-portfolio-weights-and-dirichlet-sampling"></a>

Initial weights divide the budget among assets. With buy-and-hold trading, share counts stay fixed while the fractions of wealth change as prices move.

| Symbol | Meaning and context | Units |
|---|---|---|
| $W_0$, $W_t$ | Initial budget and portfolio wealth at time $t$. | USD |
| $w_i$, $\mathbf w$ | Initial fraction in asset $i$ and length-$M$ allocation vector. | 1 |
| $n_i$ | Shares purchased: $w_iW_0/S_i(0)$. | Shares |
| $w_i(t)$ | Current wealth fraction: $n_iS_i(t)/W_t$. | 1 |
| $\Delta^{M-1}$ | Simplex of long-only weights with $w_i\geq0$ and $\sum_iw_i=1$. | Set of allocations |
| $\boldsymbol\alpha$, $\alpha_i$ | Positive Dirichlet concentration parameters used to sample allocations. | 1 |
| $\alpha_0$ | Total concentration: $\sum_i\alpha_i$. | 1 |
| $\alpha$ (symmetric case) | Common concentration when every $\alpha_i=\alpha$. | 1 |
| $\mathbf g$ | Length-$M$ vector of asset growth rates over one interval. | $\mathrm{yr}^{-1}$ |
| $g_p$ | Linear one-period portfolio growth proxy: $\mathbf w^{\top}\mathbf g$. | $\mathrm{yr}^{-1}$ |
| $g_p^{\prime}$ | Sample mean of the proxy: $\mathbf w^{\top}\mathbf g^{\prime}$. | $\mathrm{yr}^{-1}$ |
| $\sigma_{g,p}^2$ | Proxy's sample variance: $\mathbf w^{\top}\hat{\mathbf\Sigma}_g\mathbf w$. | $\mathrm{yr}^{-2}$ |


**How do weights become wealth?** Allocate the fraction $w_i$ of initial wealth $W_0$ to asset $i$, buying $n_i=w_iW_0/S_i(0)$ shares. Holding those shares fixed gives:

$$
W_t=\sum_i n_iS_i(t)=W_0\sum_iw_i\frac{S_i(t)}{S_i(0)}.
$$

If two equal allocations experience price factors 1.20 and 0.90, the wealth factor is $0.5(1.20)+0.5(0.90)=1.05$. The current wealth fractions change as prices move, even though the share counts stay fixed.

**Why do portfolio calculations use a weighted growth rate?** Over one short interval, substitute the asset price factors $e^{g_i\Delta t}$ into the wealth ratio:

$$
\frac{1}{\Delta t}\ln(W_{\Delta t}/W_0)
=\frac{1}{\Delta t}\ln\left(\sum_iw_ie^{g_i\Delta t}\right)
\approx\sum_iw_ig_i=g_p.
$$

The approximation uses $e^x\approx1+x$, $\ln(1+x)\approx x$, and weights summing to one. It is useful for small interval changes; the wealth formula supplies the exact buy-and-hold calculation. Dirichlet sampling generates candidate long-only weight vectors. Equal concentration parameters of one give uniform sampling over the simplex, while larger equal concentrations favor weights near equal allocation. Comparing sampled portfolios explores possible allocations; optimization in L6a determines the minimum for the specified constraints.

## L6a: Estimated Growth and Portfolio Risk

<a id="l6a-estimated-growth-and-portfolio-risk"></a>

**How do individual growth rates become portfolio reward and risk?** We use fixed weights to form the linear growth proxy $g_p=\mathbf w^{\top}\mathbf g$. Its expected value measures reward, and its variance measures risk. The L6a Dirichlet example compares sampled allocations this way; the data example chooses weights by optimization.

| Symbol | Meaning and context | Units |
|---|---|---|
| $M$, $N$ | Number of assets and number of aligned growth observations. | Counts |
| $\mathbf G$ | $N\times M$ data matrix: one observation per row, one asset per column. | $\mathrm{yr}^{-1}$ |
| $\boldsymbol\mu_g$ | Length-$M$ vector of population mean growth rates. | $\mathrm{yr}^{-1}$ |
| $\mathbf g^{\prime}$, $\hat{\mathbf g}$ | Sample-mean vector; stored as `ĝ` in the examples. | $\mathrm{yr}^{-1}$ |
| $\mathbf\Sigma_g$, $\hat{\mathbf\Sigma}_g$ | Population and sample growth covariance matrices. | $\mathrm{yr}^{-2}$ |
| $g_p$, $\sigma_{g,p}$ | Weighted growth proxy and its standard deviation. | $\mathrm{yr}^{-1}$ |
| $\rho_{12}$ | Correlation of two asset growth rates. | 1 |
| $\mathbf C$, $\mathbf\Sigma_r$ | Covariance rate $\Delta t\,\mathbf\Sigma_g$; interval log-return covariance $(\Delta t)^2\mathbf\Sigma_g$. | $\mathrm{yr}^{-1}$; 1 |

**Estimate the inputs.** Each entry of $\mathbf G$ comes from an aligned price pair: $g_{i,k}=\ln[S_i(t_k)/S_i(t_{k-1})]/\Delta t$. Average each column and then average the products of its centered observations with those of another column:

$$
g_i^{\prime}=\frac{1}{N}\sum_{k=1}^{N}g_{i,k},\qquad
\hat\Sigma_{g,ij}=\frac{1}{N-1}\sum_{k=1}^{N}(g_{i,k}-g_i^{\prime})(g_{j,k}-g_j^{\prime}).
$$

The diagonal records individual variances. The off-diagonal records whether assets tend to be above or below their own means together.

**Work through two assets.** Suppose the estimated means are $(0.06,0.12)\,\mathrm{yr}^{-1}$, their growth-rate standard deviations are $(2,4)\,\mathrm{yr}^{-1}$, and their correlation is $0.25$. For weights $(0.75,0.25)$, the covariance is $0.25(2)(4)=2\,\mathrm{yr}^{-2}$. Thus:

$$
\begin{aligned}
\hat{\mathbf\Sigma}_g&=\begin{bmatrix}4&2\\2&16\end{bmatrix}\,\mathrm{yr}^{-2},\\
\widehat{\mathbb E}[g_p]&=0.75(0.06)+0.25(0.12)=0.075\,\mathrm{yr}^{-1},\\
\widehat{\operatorname{Var}}(g_p)&=0.75^2(4)+0.25^2(16)+2(0.75)(0.25)(2)
=4\,\mathrm{yr}^{-2}.
\end{aligned}
$$

The portfolio standard deviation is $2\,\mathrm{yr}^{-1}$. For $\Delta t=1/252$ yr, its corresponding diffusion-volatility scale is $\sqrt{\Delta t}\,\sigma_{g,p}=0.126\,\mathrm{yr}^{-1/2}$. The examples use this conversion when comparing growth-rate variability with GBM volatility.

**Check.** Keep the weights and individual variances fixed, but set the correlation to zero. What changes? **Answer:** mean growth stays $0.075\,\mathrm{yr}^{-1}$, while variance falls to $3.25\,\mathrm{yr}^{-2}$ and standard deviation to $1.803\,\mathrm{yr}^{-1}$. The smaller covariance reduces portfolio risk.

## L6a: Data-Driven Minimum-Variance Portfolios

<a id="l6a-data-driven-minimum-variance-portfolios"></a>

**Which fully invested allocation has the smallest variance?** The global minimum-variance (GMV) problem supplies the growth covariance and chooses the weights. A positive-definite covariance gives a unique minimum. We first allow signed weights, so the budget is the only constraint:

$$
\underset{\mathbf w}{\operatorname{minimize}}\quad
\frac12\mathbf w^{\top}\mathbf\Sigma_g\mathbf w
\qquad\text{subject to}\quad\mathbf1^{\top}\mathbf w=1.
$$

| Symbol | Meaning and context | Units |
|---|---|---|
| $\mathbf w_{\mathrm{GMV}}$ | Fully invested global minimum-variance weights. | 1 |
| $\sigma_{g,\mathrm{GMV}}^2$ | Smallest growth variance under the stated weight constraints. | $\mathrm{yr}^{-2}$ |
| $\mathcal L$, $\lambda$ | Lagrangian and budget-constraint multiplier for the half-variance objective. | $\mathrm{yr}^{-2}$ |
| $\mathbf x$ | Solution of $\mathbf\Sigma_g\mathbf x=\mathbf1$ before normalization. | $\mathrm{yr}^{2}$ |

**Why does the inverse covariance appear?** We attach a multiplier to the budget constraint and set the gradient to zero:

$$
\begin{aligned}
\mathcal L(\mathbf w,\lambda)
&=\tfrac12\mathbf w^{\top}\mathbf\Sigma_g\mathbf w
-\lambda(\mathbf1^{\top}\mathbf w-1),\\
\mathbf\Sigma_g\mathbf w&=\lambda\mathbf1
\quad\Longrightarrow\quad\mathbf w=\lambda\mathbf\Sigma_g^{-1}\mathbf1.
\end{aligned}
$$

The factor $1/2$ simplifies differentiation. Requiring the weights to sum to one determines $\lambda$ and gives:

$$
\mathbf w_{\mathrm{GMV}}
=\frac{\mathbf\Sigma_g^{-1}\mathbf1}{\mathbf1^{\top}\mathbf\Sigma_g^{-1}\mathbf1},
\qquad
\sigma_{g,\mathrm{GMV}}^2
=\frac{1}{\mathbf1^{\top}\mathbf\Sigma_g^{-1}\mathbf1}.
$$

The numerator accounts for variances and covariances; the denominator normalizes the allocation. The data example solves $\hat{\mathbf\Sigma}_g\mathbf x=\mathbf1$ and divides $\mathbf x$ by the sum of its entries. Mean-growth estimates enter only when we add a growth requirement.

**A two-asset calculation.** For uncorrelated assets with growth variances $1$ and $4\,\mathrm{yr}^{-2}$, the unnormalized vector is $(1,1/4)\,\mathrm{yr}^{2}$. Normalizing gives weights $(0.8,0.2)$ and variance $1/(1+1/4)=0.8\,\mathrm{yr}^{-2}$. The lower-variance asset receives more of the budget.

**What does a negative weight mean?** With 1,000 USD and weights $(1.2,-0.2)$, we borrow and sell 200 USD of the second asset and buy 1,200 USD of the first. The short position is a liability: a price increase raises the cost of buying back the borrowed shares. The lecture's calculation omits borrowing fees, collateral, and dividends owed.

**Check.** If an unrestricted GMV solution contains a negative weight, what happens when we require $w_i\geq0$? **Answer:** we solve again over the smaller feasible set. The minimum variance increases for a positive-definite covariance. If the unrestricted weights were already nonnegative, they would remain optimal.

## L6a: Exact Growth Targets and Growth Floors

<a id="l6a-exact-growth-targets-and-growth-floors"></a>

**How much additional risk does a growth requirement impose?** The lecture first requires the weighted sample means to equal a target, $(\mathbf g^{\prime})^{\top}\mathbf w=g_\star$, while allowing short positions. The minimum variance at each target traces the minimum-variance frontier.

| Symbol | Meaning and context | Units |
|---|---|---|
| $g_\star$ | Exact target in the signed-weight problem; lower bound in the long-only problem. | $\mathrm{yr}^{-1}$ |
| $\mathbf w_\star(g_\star)$ | Minimum-variance weights for an exact growth target. | 1 |
| $a$, $b$, $c$, $d$ | Scalar frontier coefficients defined below using the lecture's sample means. | $\mathrm{yr}^{2}$; yr; 1; $\mathrm{yr}^{2}$ |
| $\sigma_g^2(g_\star)$ | Minimum growth variance at the exact target. | $\mathrm{yr}^{-2}$ |
| $R$ (solver input) | Name used for the growth floor $g_\star$ in the portfolio examples. | $\mathrm{yr}^{-1}$ |

For a positive-definite $\mathbf\Sigma_g$ and unequal asset sample means, calculate:

$$
\begin{aligned}
a&=\mathbf1^{\top}\mathbf\Sigma_g^{-1}\mathbf1,&
b&=\mathbf1^{\top}\mathbf\Sigma_g^{-1}\mathbf g^{\prime},\\
c&=(\mathbf g^{\prime})^{\top}\mathbf\Sigma_g^{-1}\mathbf g^{\prime},&d&=ac-b^2>0.
\end{aligned}
$$

The budget and growth constraints determine the coefficients of two vectors. The resulting weights and variance are:

$$
\begin{aligned}
\mathbf w_\star(g_\star)
&=\frac{c-bg_\star}{d}\mathbf\Sigma_g^{-1}\mathbf1
+\frac{ag_\star-b}{d}\mathbf\Sigma_g^{-1}\mathbf g^{\prime},\\
\sigma_g^2(g_\star)
&=\frac{ag_\star^2-2bg_\star+c}{d}
=\frac1a+\frac ad\left(g_\star-\frac ba\right)^2.
\end{aligned}
$$

The last expression separates GMV variance, $1/a$, from the extra variance imposed by the target. The GMV mean is $b/a$. The upper branch, $g_\star\geq b/a$, is efficient: a point on the lower branch has a higher-growth counterpart at the same risk.

**What changes with a floor and long-only weights?** We solve the following problem numerically using estimated inputs:

$$
\begin{aligned}
\underset{\mathbf w}{\operatorname{minimize}}\quad&\mathbf w^{\top}\hat{\mathbf\Sigma}_g\mathbf w\\
\text{subject to}\quad &(\mathbf g^{\prime})^{\top}\mathbf w\geq g_\star,\qquad
\mathbf1^{\top}\mathbf w=1,\qquad 0\leq w_i\leq1.
\end{aligned}
$$

Return to uncorrelated variances $(1,4)\,\mathrm{yr}^{-2}$ and suppose the means are $(0.04,0.14)\,\mathrm{yr}^{-1}$. The GMV weights $(0.8,0.2)$ give mean growth $0.06\,\mathrm{yr}^{-1}$. A floor of $0.05$ leaves this allocation unchanged. An exact target of $0.05$ instead requires weights $(0.9,0.1)$ and variance $0.85\,\mathrm{yr}^{-2}$. Equality forces us below the efficient branch; the floor accepts the better GMV allocation.

**Check.** Raise the floor to $0.09\,\mathrm{yr}^{-1}$. **Answer:** it binds at weights $(0.5,0.5)$, giving variance $1.25\,\mathrm{yr}^{-2}$. A long-only floor above $0.14\,\mathrm{yr}^{-1}$ is infeasible. Setting the floor to the smallest asset mean, as in the examples' GMV solve, allows every fully invested long-only allocation.

## L6a: From Portfolio Weights to Wealth and NPV

<a id="l6a-from-portfolio-weights-to-wealth-and-npv"></a>

**How do we evaluate the investment after choosing its weights?** We convert the initial dollar allocations to shares, keep the share counts fixed, and value them at later prices. The data and multiple-asset GBM examples use the same wealth calculation on observed and simulated prices.

| Symbol | Meaning and context | Units |
|---|---|---|
| $W_0$, $W_t$ | Initial investment and net portfolio wealth at time $t$. | USD |
| $w_i$, $n_i$ | Initial wealth fraction and fixed share count for asset $i$. | 1; shares |
| $S_i(0)$, $S_i(t)$ | Initial and later price per share. | USD/share |
| $T$, $\Delta t$ | Holding period and observation interval. | yr |
| $g_y$ | Constant benchmark growth rate used to discount wealth. | $\mathrm{yr}^{-1}$ |
| $\rho_T$, $\rho_\star$ | Scaled terminal NPV and its target. | 1 |

**Work from dollars to shares.** Suppose $W_0=1{,}000$ USD, weights are $(0.6,0.4)$, and initial share prices are $(100,50)$ USD. We buy six shares of the first asset and eight of the second. In symbols:

$$
n_i=\frac{w_iW_0}{S_i(0)},\qquad
W_t=\sum_i n_iS_i(t)=W_0\sum_i w_i\frac{S_i(t)}{S_i(0)}.
$$

If the prices after one year are $(110,45)$ USD, wealth is $6(110)+8(45)=1{,}020$ USD. The first asset now represents $660/1{,}020=64.7\%$ of wealth. Its initial fraction was 60%; the share count stayed six as its wealth fraction changed.

**Did we exceed the benchmark?** Discount terminal wealth before comparing it with the initial investment:

$$
\operatorname{NPV}(g_y,T)=-W_0+W_Te^{-g_yT},\qquad
\rho_T=\frac{W_T}{W_0}e^{-g_yT}-1.
$$

For $g_y=0.03\,\mathrm{yr}^{-1}$ and $T=1$ yr, the scaled NPV is $1.02e^{-0.03}-1\approx-0.01015$. Wealth rose by 20 USD, but its discounted value fell short of the initial investment by about 10.15 USD. To exceed a scaled target $\rho_\star>-1$, terminal wealth must satisfy:

$$
W_T>W_0(1+\rho_\star)e^{g_yT}.
$$

For a zero target, the required wealth is $1{,}030.45$ USD. This translates the NPV question into a terminal-wealth event that we can count in simulated paths.

**How do we use the testing data?** The data example estimates the inputs from 2014 to 2024 and evaluates both GMV allocations, equal weights, and SPY on 2025 prices. Realized mean growth is $\ln(W_T/W_0)/T$ when wealth is positive. Realized risk is the sample standard deviation of daily wealth growth rates. These quantities describe the investment actually held, whose weights can drift through time.

**Check.** If we double $W_0$ and keep the weights and price ratios fixed, what changes? **Answer:** share counts, wealth, and dollar NPV double. Scaled NPV and the probability of exceeding a scaled target stay the same. The calculation allows fractional shares and omits trading costs and dividends.

## L6a: Simulated Wealth and Target Probabilities

<a id="l6a-simulated-wealth-and-target-probabilities"></a>

**What does a band around simulated wealth describe?** The multiple-asset GBM example holds the fitted parameters and initial allocation fixed, simulates correlated prices, and computes wealth on every path. At a given date, the paths give a distribution of possible wealth outcomes under that model.

| Symbol | Meaning and context | Units |
|---|---|---|
| $n$, $k$ | Number of simulated paths and path index on this page. | Counts |
| $W_t^{(k)}$, $\bar W_t$ | Wealth on path $k$ and mean wealth across paths at date $t$. | USD |
| $s_W(t)$ | Sample standard deviation of wealth across paths. | USD |
| $s_W(t)/\sqrt n$ | Monte Carlo standard error of the estimated mean wealth. | USD |
| $\hat p$, $\operatorname{SE}(\hat p)$ | Estimated target probability and its Monte Carlo standard error. | 1 |

**Separate outcome spread from precision of the mean.** Average the wealth values across paths, then measure their spread:

$$
\bar W_t=\frac1n\sum_{k=1}^{n}W_t^{(k)},\qquad
s_W(t)=\sqrt{\frac{1}{n-1}\sum_{k=1}^{n}(W_t^{(k)}-\bar W_t)^2}.
$$

Suppose 2,000 independent paths give a mean of 1,100 USD and a standard deviation of 150 USD at a selected date. The example's approximate 95% pointwise prediction band is:

$$
\bar W_t\pm1.96s_W(t)=1{,}100\pm294\ \mathrm{USD}.
$$

It describes the spread of wealth outcomes on that date. Its normal approximation can be imperfect because wealth is skewed. By contrast, the Monte Carlo standard error of the mean is $150/\sqrt{2{,}000}=3.35$ USD, giving an approximate 95% interval of $1{,}100\pm6.57$ USD for the model's expected wealth. Pointwise coverage refers to a single date; whole-path coverage is a different event.

**Estimate a target probability.** Discount each path's terminal wealth to obtain $\rho_T^{(k)}$. Count the paths strictly above $\rho_\star$ and divide by the number of paths:

$$
\hat p=\frac{\#\{k:\rho_T^{(k)}>\rho_\star\}}{n},\qquad
\operatorname{SE}(\hat p)=\sqrt{\frac{\hat p(1-\hat p)}{n}}.
$$

If 1,360 of 2,000 paths exceed the target, $\hat p=0.68$ and its standard error is $0.0104$, about one percentage point. An approximate 95% Monte Carlo interval is $0.68\pm0.0204$. This quantifies the effect of using finitely many random paths while keeping the model inputs fixed.

The simulation first checks that generated growth rates reproduce the fitted means and covariance. The later comparison with observed 2025 wealth asks a separate question: how well those fitted inputs describe a new period.

**Check.** Increase the number of independent paths from 2,000 to 8,000. What should shrink? **Answer:** Monte Carlo standard errors are approximately halved. The estimated spread of individual wealth outcomes remains near 150 USD because the model and allocation have not changed. More paths improve numerical precision; uncertainty in the fitted inputs requires a separate analysis.

## L6b: Single Index Model and Estimation

<a id="l6b-single-index-model-and-estimation"></a>

The single index model (SIM) relates an asset's growth rate to market growth. The subscript $M$ means market; the number of portfolio assets is $\lvert\mathcal P\rvert$. The release uses SPY as the market proxy and fits growth rates from aligned trading days.

| Symbol| Meaning and context | Units |
|---|---|---|
| $g_{i,t}$, $g_{M,t}$ | Asset and market growth rates on aligned observation $t$. | $\mathrm{yr}^{-1}$ |
| $N$, $t$ | Number of aligned observations and observation index. | Counts |
| $\alpha_i$ | Intercept of the asset's linear projection on market growth. | $\mathrm{yr}^{-1}$ |
| $\beta_i$ | Market exposure: $\operatorname{Cov}(g_i,g_M)/\operatorname{Var}(g_M)$. | 1 |
| $\varepsilon_{i,t}$ | Asset-specific residual growth after removing the intercept and market contribution. | $\mathrm{yr}^{-1}$ |
| $\rho_{iM}$ | Correlation between asset and market growth. | 1 |
| $\sigma_{g,i}$, $\sigma_{g,M}$ | Asset and market growth-rate standard deviations. | $\mathrm{yr}^{-1}$ |
| $\sigma_{g,\varepsilon,i}^2$ | Population residual growth variance for asset $i$. | $\mathrm{yr}^{-2}$ |
| $\mathbf y$, $\hat{\mathbf y}$ | Length-$N$ vectors of observed and fitted asset growth rates. | $\mathrm{yr}^{-1}$ |
| $\hat{\mathbf X}$ | $N\times2$ augmented design matrix, with columns of ones and market growth rates. | Columns: 1; $\mathrm{yr}^{-1}$ |
| $\boldsymbol\theta_i$, $\hat{\boldsymbol\theta}_i$ | Parameter vector $(\alpha_i,\beta_i)^{\top}$ and its least-squares estimate. | Entries: $\mathrm{yr}^{-1}$; 1 |
| $\mathbf r_i$ | Fitted residual series for asset $i$: $\mathbf y-\hat{\mathbf y}$. | $\mathrm{yr}^{-1}$ |
| $\operatorname{SE}(\hat\alpha_i)$, $\operatorname{SE}(\hat\beta_i)$ | Standard errors of the fitted coefficients. | $\mathrm{yr}^{-1}$; 1 |
| $R^2$ | Fraction of sample growth variation explained by the fitted regression with an intercept. | 1 |
| $g_i^{\prime}$ | Sample mean of the observed growth rates for asset $i$. | $\mathrm{yr}^{-1}$ |
| $g_M^{\prime}$, $s_{g,M}^2$ | Training-period sample mean and sample variance of market growth. | $\mathrm{yr}^{-1}$; $\mathrm{yr}^{-2}$ |
| $s_{g,\varepsilon,i}^2$ | Residual variance estimate: $\lVert\mathbf r_i\rVert_2^2/(N-2)$. | $\mathrm{yr}^{-2}$ |
| $m$, $f_{j,t}$ | Number of additional factors and value of factor $j$ on day $t$ in the opening factor-model discussion. | Count; factor-specific |
| $\gamma_{ij}$ | Sensitivity to factor $j$; $\gamma_{ij}f_{j,t}$ contributes a growth rate. | $\mathrm{yr}^{-1}$ per factor unit |
| $\mathrm{SMB}_t$, $\mathrm{HML}_t$; $s_i$, $h_i$ | Size and value factors (small minus big; high minus low book-to-market), and their sensitivities in the three-factor model. | Growth factors: $\mathrm{yr}^{-1}$; sensitivities: 1 |


**Reading the design matrix.** Each row of $\hat{\mathbf X}$ pairs the number one with that day's market growth. Multiplying by $(\alpha_i,\beta_i)^{\top}$ produces the intercept plus market contribution for each day. The fitted residual vector is what remains after subtracting those fitted values from observations.

The estimation example solves the same least-squares problem by normal equations, a QR-based solve, and singular value decomposition (SVD). For the thin SVD $\hat{\mathbf X}=\mathbf U\mathbf\Sigma\mathbf V^{\top}$, $\mathbf U$ is $N\times2$, and $\mathbf\Sigma$ and $\mathbf V$ are $2\times2$. Here $\mathbf\Sigma$ contains singular values; $\mathbf\Sigma_g$ denotes a growth covariance. Nonzero singular values give full column rank and a unique fit.


### L6b: Reading Alpha, Beta, Residuals, and Fit

**What does an asset's beta tell us?** The single index model separates the fitted market response from the asset-specific residual:

$$
g_{i,t}=\alpha_i+\beta_i g_{M,t}+\varepsilon_{i,t},\qquad
\beta_i=\frac{\operatorname{Cov}(g_i,g_M)}{\operatorname{Var}(g_M)}
=\rho_{iM}\frac{\sigma_{g,i}}{\sigma_{g,M}}.
$$

Beta includes both correlation and the ratio of growth-rate scales. With correlation 0.60 and asset variability twice the market's, beta is 1.20. A $0.10\,\mathrm{yr}^{-1}$ increase in market growth then raises fitted asset growth by $0.12\,\mathrm{yr}^{-1}$.

Suppose $\alpha_i=0.02\,\mathrm{yr}^{-1}$ and the observed market growth on a day is $0.10\,\mathrm{yr}^{-1}$. The fitted asset growth is $0.02+1.2(0.10)=0.14\,\mathrm{yr}^{-1}$. If the observed asset growth is $-0.06\,\mathrm{yr}^{-1}$, its fitted residual is $-0.20\,\mathrm{yr}^{-1}$. The residual records the part of that observation left unexplained by the line.

**How do we fit the line?** Choose alpha and beta to minimize the squared residuals. For a full-column-rank design matrix, the lecture gives:

$$
\hat{\boldsymbol\theta}_i
=\arg\min_{\boldsymbol\theta}\lVert\mathbf y-\hat{\mathbf X}\boldsymbol\theta\rVert_2^2
=(\hat{\mathbf X}^{\top}\hat{\mathbf X})^{-1}\hat{\mathbf X}^{\top}\mathbf y.
$$

The equivalent sample formulas explain the fitted coefficients:

$$
\hat\beta_i=\frac{\widehat{\operatorname{Cov}}(g_i,g_M)}{s_{g,M}^2},\qquad
\hat\alpha_i=g_i^{\prime}-\hat\beta_i g_M^{\prime}.
$$

The intercept ensures that the fitted line passes through the sample means. The SVD example computes the same estimate as $\mathbf V\mathbf\Sigma^{-1}\mathbf U^{\top}\mathbf y$. Agreement among the three solves checks the numerical calculation.

**How much variation did the fit explain?** Compare squared residuals with squared deviations from the asset's sample mean:

$$
R^2=1-\frac{\lVert\mathbf r_i\rVert_2^2}{\sum_{t=1}^{N}(g_{i,t}-g_i^{\prime})^2}.
$$

If residuals retain 40% of the original sum of squares, $R^2=0.60$. In this one-predictor regression with an intercept, $R^2$ equals squared sample correlation. It measures in-sample fit; later observations are needed to assess forecast usefulness.

**Check.** Suppose the sample correlation is 0.60 and the asset's sample standard deviation is twice the market's. What are $\hat\beta_i$ and $R^2$? **Answer:** $\hat\beta_i=1.20$ and $R^2=0.36$. Beta measures sensitivity to market growth, while $R^2$ measures the fraction of variation explained. The client interview uses both when screening firms.

## L6b: Standard Errors and Bootstrap Uncertainty

<a id="l6b-standard-errors-and-bootstrap-uncertainty"></a>

**How much might alpha and beta change with another dataset?** Conditional on the observed market growth rates, the classical calculation assumes zero-mean errors with constant variance and no correlation across days. We estimate the error variance after fitting two coefficients:

$$
s_{g,\varepsilon,i}^2=\frac{\lVert\mathbf r_i\rVert_2^2}{N-2},\qquad
\widehat{\operatorname{Cov}}(\hat{\boldsymbol\theta}_i)
=s_{g,\varepsilon,i}^2(\hat{\mathbf X}^{\top}\hat{\mathbf X})^{-1}.
$$

Square roots of the two diagonal entries give $\operatorname{SE}(\hat\alpha_i)$ and $\operatorname{SE}(\hat\beta_i)$. Alpha's standard error has units of inverse years; beta's is dimensionless.

| Symbol | Meaning and context | Units |
|---|---|---|
| $N-2$ | Residual degrees of freedom after fitting an intercept and slope. | Count |
| $t_{0.975,N-2}$ | 97.5th percentile of a Student-$t$ law with $N-2$ degrees of freedom. | 1 |
| $z_{0.975}$ | Standard-normal critical value, approximately 1.96. | 1 |
| $K$, $k$ | Number of bootstrap datasets and dataset index. | Counts |
| $\boldsymbol\varepsilon^{(k)}$, $\mathbf y^{(k)}$ | Synthetic errors and synthetic asset-growth observations. | $\mathrm{yr}^{-1}$ |
| $\hat{\boldsymbol\theta}^{(k)}$ | Alpha and beta refitted on synthetic dataset $k$. | $\mathrm{yr}^{-1}$; 1 |

**Read an interval.** With thousands of observations, the 95% interval is approximately the estimate plus or minus 1.96 standard errors. Suppose $\hat\alpha_i=0.02\,\mathrm{yr}^{-1}$ with standard error $0.03\,\mathrm{yr}^{-1}$, and $\hat\beta_i=1.20$ with standard error $0.05$. Then:

$$
\begin{aligned}
\alpha_i:&\quad 0.02\pm1.96(0.03)=[-0.0388,0.0788]\,\mathrm{yr}^{-1},\\
\beta_i:&\quad 1.20\pm1.96(0.05)=[1.102,1.298].
\end{aligned}
$$

The alpha interval includes zero, while the beta interval lies above one. Under the regression assumptions, these data estimate market exposure more precisely relative to its magnitude. The estimation example uses the Student-$t$ critical value; the large-sample Gaussian value is used in the bootstrap comparison. Gaussian errors give the exact classical $t$ interval.

**What does the Gaussian bootstrap hold fixed?** The uncertainty example keeps $\hat{\mathbf X}$ fixed, draws independent synthetic residuals from a fitted normal distribution, and adds them to fitted growth:

$$
\mathbf y^{(k)}=\hat{\mathbf X}\hat{\boldsymbol\theta}_i+\boldsymbol\varepsilon^{(k)}.
$$

We refit the model to each of the $K=1{,}000$ synthetic datasets. The standard deviation of the refitted alphas or betas estimates its standard error. Because these datasets follow the fitted Gaussian model, agreement with the classical calculation checks that calculation. The optional residual-diagnostics example separately examines heavy tails and dependence across days.

**Check.** Does increasing $K$ from 1,000 to 4,000 make the original beta estimate twice as precise? **Answer:** it improves the numerical precision of the bootstrap comparison. The original observations and their information about beta remain the same.

## L6b: SIM Portfolio Inputs and Risk Decomposition

<a id="l6b-sim-portfolio-inputs-and-risk-decomposition"></a>

The SIM covariance separates a common market contribution from asset-specific residual variances. Its diagonal residual matrix imposes zero cross-asset residual covariance. This is an additional assumption beyond fitting each asset by least squares.

| Symbol| Meaning and context | Units |
|---|---|---|
| $\boldsymbol\alpha$, $\boldsymbol\beta$ | Intercept and beta vectors, each of length $\lvert\mathcal P\rvert$. | $\mathrm{yr}^{-1}$; 1 |
| $\boldsymbol\varepsilon$ | Vector of asset residual growth rates for one observation. | $\mathrm{yr}^{-1}$ |
| $\mu_{g,M}$, $\sigma_{g,M}^2$ | Population mean and variance of market growth. | $\mathrm{yr}^{-1}$; $\mathrm{yr}^{-2}$ |
| $\mathbf D_g$ | Diagonal matrix with entries $\sigma_{g,\varepsilon,i}^2$. | $\mathrm{yr}^{-2}$ |
| $\mathbf\Omega_\varepsilon$, $\hat{\mathbf\Omega}_\varepsilon$ | Population and sample residual covariance matrices, allowing nonzero off-diagonal entries. | $\mathrm{yr}^{-2}$ |
| $\hat{\mathbf D}_g$ | Diagonal matrix of fitted residual variances $s_{g,\varepsilon,i}^2$. | $\mathrm{yr}^{-2}$ |
| $\hat{\mathbf g}_{\mathrm{SIM}}$ | Fitted mean vector: $\hat{\boldsymbol\alpha}+\hat{\boldsymbol\beta}g_M^{\prime}$. | $\mathrm{yr}^{-1}$ |
| $\hat{\mathbf\Sigma}_{g,\mathrm{SIM}}$ | Fitted SIM growth covariance, using the same training observations as the coefficient fits. | $\mathrm{yr}^{-2}$ |
| $\beta_p$ | Portfolio market exposure: $\mathbf w^{\top}\boldsymbol\beta$. | 1 |
| $\varepsilon_p$ | Portfolio residual: $\mathbf w^{\top}\boldsymbol\varepsilon$. | $\mathrm{yr}^{-1}$ |
| $\bar\sigma^2$ (residual bound) | Common upper bound on residual variances in the diversification argument. | $\mathrm{yr}^{-2}$ |

**How does the market model supply portfolio inputs?** Assume residuals have zero mean, are uncorrelated with market growth, and are uncorrelated across assets. Taking means and covariances gives:

$$
\boldsymbol\mu_g=\boldsymbol\alpha+\boldsymbol\beta\mu_{g,M},\qquad
\mathbf\Sigma_g=\sigma_{g,M}^2\boldsymbol\beta\boldsymbol\beta^{\top}+\mathbf D_g.
$$

Substitute estimates from the same training observations to construct:

$$
\hat{\mathbf g}_{\mathrm{SIM}}=\hat{\boldsymbol\alpha}+\hat{\boldsymbol\beta}g_M^{\prime},\qquad
\hat{\mathbf\Sigma}_{g,\mathrm{SIM}}=s_{g,M}^2\hat{\boldsymbol\beta}\hat{\boldsymbol\beta}^{\top}+\hat{\mathbf D}_g.
$$

The diagonal contains both market and residual variance. Off-diagonal entries contain only the shared market contribution. Positive residual variances make the fitted covariance positive definite. These inputs replace the empirical ones in L6a's allocation problem.


### L6b: What the SIM Keeps and Omits

**Which input changes when we use the SIM?** With an intercept and the same training days, $\hat\alpha_i=g_i^{\prime}-\hat\beta_i g_M^{\prime}$, so $\hat{\mathbf g}_{\mathrm{SIM}}=\mathbf g^{\prime}$. The reward inputs agree. The covariance changes because the model removes residual co-movement.

For different assets, the sample covariance splits exactly into:

$$
\hat\Sigma_{g,ij}
=\underbrace{\hat\beta_i\hat\beta_j s_{g,M}^2}_{\text{SIM keeps}}
+\underbrace{\widehat{\operatorname{Cov}}(\mathbf r_i,\mathbf r_j)}_{\text{SIM omits}},\qquad i\ne j.
$$

On the diagonal, the SIM divides residual sums of squares by $N-2$, whereas the sample residual covariance uses $N-1$. Thus, the SIM residual contribution is larger by the factor $(N-1)/(N-2)$, nearly one for thousands of observations.

**Build a two-asset covariance.** Suppose the fitted betas are $(1,1.5)$, the market growth variance is $4\,\mathrm{yr}^{-2}$, and fitted residual variances are $(1,4)\,\mathrm{yr}^{-2}$. Then:

$$
\hat{\mathbf\Sigma}_{g,\mathrm{SIM}}
=4\begin{bmatrix}1&1.5\\1.5&2.25\end{bmatrix}
+\begin{bmatrix}1&0\\0&4\end{bmatrix}
=\begin{bmatrix}5&6\\6&13\end{bmatrix}\,\mathrm{yr}^{-2}.
$$

For fixed weights, portfolio beta is $\beta_p=\mathbf w^{\top}\boldsymbol\beta$. Multiplying the covariance by the weights on both sides gives the general risk decomposition:

$$
\operatorname{Var}(g_p)=\beta_p^2\sigma_{g,M}^2+\sum_i w_i^2\sigma_{g,\varepsilon,i}^2.
$$

At equal weights, fitted portfolio beta is $0.5(1)+0.5(1.5)=1.25$. Substituting the fitted inputs gives:

$$
\widehat{\operatorname{Var}}_{\mathrm{SIM}}(g_p)
=\underbrace{1.25^2(4)}_{6.25\text{ market}}
+\underbrace{0.5^2(1)+0.5^2(4)}_{1.25\text{ residual}}
=7.5\,\mathrm{yr}^{-2}.
$$

If residual covariance were $1\,\mathrm{yr}^{-2}$, keeping these diagonal entries fixed would add the cross term $2(0.5)(0.5)(1)=0.5\,\mathrm{yr}^{-2}$. The variance would be $8\,\mathrm{yr}^{-2}$. Positive residual co-movement increases the risk of these positive-weight holdings.

**Which risk can diversification reduce?** With $\lvert\mathcal P\rvert$ equally weighted assets and uncorrelated residuals of common variance $v$, residual risk is $v/\lvert\mathcal P\rvert$. More generally, a common upper bound $\bar\sigma^2$ gives a bound $\bar\sigma^2/\lvert\mathcal P\rvert$. Market risk remains $\beta_p^2\sigma_{g,M}^2$, determined by the portfolio beta.

**How do the portfolio examples compare allocations?** The risky-assets example scores both weight vectors under the sample covariance at common growth floors, then tests both on 2025 prices. The data-optimal weights have at most the sample-covariance risk of the SIM weights at the same floor. Using the same covariance separates the effect of choosing different holdings from the effect of measuring risk differently.

**Check.** In the two-asset calculation, replace residual covariance $1$ by $-1\,\mathrm{yr}^{-2}$. **Answer:** with the same diagonal entries, variance becomes $7\,\mathrm{yr}^{-2}$. The SIM now overstates this portfolio's risk by omitting a negative covariance.

## L6b: Risk-Free Allocation and the Capital Allocation Line

<a id="l6b-risk-free-allocation-and-the-capital-allocation-line"></a>

We combine a risky portfolio with an asset whose growth rate is known over the holding period. The risk-free weight controls lending and borrowing.

| Symbol| Meaning and context | Units |
|---|---|---|
| $g_f$ | Known growth rate of the risk-free asset over the matched holding period. | $\mathrm{yr}^{-1}$ |
| $w_f$ | Risk-free wealth fraction; negative values represent borrowing. | 1 |
| $p$, $c$ (subscripts) | Risky portfolio and complete portfolio that combines risky and risk-free positions. | Portfolio labels |
| $g_p$, $g_c$ | Growth-rate proxies for the risky and complete portfolios. | $\mathrm{yr}^{-1}$ |
| $\sigma_{g,p}$, $\sigma_{g,c}$ | Corresponding growth-rate standard deviations. | $\mathrm{yr}^{-1}$ |
| $\mathrm{SR}_p$ | CAL slope: $(\mathbb E[g_p]-g_f)/\sigma_{g,p}$ for the observation interval. | 1 |
| $\tau=1\,\mathrm{yr}$ | Reference year used when annualizing the Sharpe ratio. | yr |
| $\mathrm{SR}_p^{\mathrm{annual}}$ | Annualized ratio: $\sqrt{\tau/\Delta t}\,\mathrm{SR}_p$ under independent-increment scaling. | 1 |
| $\mathbf e$ | Excess-growth vector: $\boldsymbol\mu_g-g_f\mathbf1$. | $\mathrm{yr}^{-1}$ |
| $\mathcal T$, $\mathbf w_{\mathcal T}$ | Tangent risky portfolio and its weights, normalized to sum to one. | Label; 1 |
| $\mathbf w$ (complete allocation) | Risky weights as fractions of total initial wealth; $\mathbf1^{\top}\mathbf w=1-w_f$. | 1 |
| $g_\star$ | Required mean growth of the complete portfolio. | $\mathrm{yr}^{-1}$ |

**What happens when we lend or borrow?** Put fraction $w_f$ in the risk-free asset and fraction $1-w_f$ in a fixed risky portfolio. Its known growth contributes to the mean and adds zero variance:

$$
\mathbb E[g_c]=g_f+(1-w_f)(\mathbb E[g_p]-g_f),\qquad
\sigma_{g,c}=|1-w_f|\sigma_{g,p}.
$$

For daily observations ($\Delta t=1/252$ yr), suppose $g_f=0.02\,\mathrm{yr}^{-1}$, $\mathbb E[g_p]=0.08\,\mathrm{yr}^{-1}$, and $\sigma_{g,p}=3.0\,\mathrm{yr}^{-1}$. A 50% risk-free allocation gives mean growth $0.05$ and standard deviation $1.5\,\mathrm{yr}^{-1}$. Borrowing 50% of initial wealth gives $w_f=-0.5$: mean growth becomes $0.11$ and standard deviation $4.5\,\mathrm{yr}^{-1}$. Borrowing increases both risky exposure and uncertainty.

For $w_f\leq1$, eliminating the risky fraction gives the capital allocation line:

$$
\mathbb E[g_c]=g_f+\mathrm{SR}_p\,\sigma_{g,c}.
$$

Its slope is $(0.08-0.02)/3=0.02$. The annualized ratio is $\sqrt{252}(0.02)=0.3175$ under independent-increment scaling. Expressing daily growth in inverse years does not by itself annualize this ratio.

**Check.** For a target mean growth of $0.065\,\mathrm{yr}^{-1}$, what risk-free fraction is needed? **Answer:** $1-w_f=(0.065-0.02)/(0.08-0.02)=0.75$, so $w_f=0.25$ and standard deviation is $2.25\,\mathrm{yr}^{-1}$.

## L6b: Tangent Weights and the Complete Investment

<a id="l6b-tangent-weights-and-the-complete-investment"></a>

**Which risky portfolio defines the steepest capital allocation line?** The tangent portfolio maximizes excess mean growth per unit standard deviation. With a positive-definite covariance, signed risky weights allowed, and GMV mean growth exceeding $g_f$, the lecture obtains:

$$
\mathbf w_{\mathcal T}=\frac{\mathbf\Sigma_g^{-1}\mathbf e}{\mathbf1^{\top}\mathbf\Sigma_g^{-1}\mathbf e},
\qquad \mathbf e=\boldsymbol\mu_g-g_f\mathbf1.
$$

The inverse covariance adjusts the excess-growth vector for risk; the denominator makes the risky fund's weights sum to one. Long-only restrictions require a numerical solve. In the release, SIM-2 uses risky weights measured as fractions of total initial wealth:

$$
\begin{aligned}
\underset{\mathbf w}{\operatorname{minimize}}\quad
&\mathbf w^{\top}\hat{\mathbf\Sigma}_{g,\mathrm{SIM}}\mathbf w\\
\text{subject to}\quad
&(\hat{\mathbf g}_{\mathrm{SIM}}-g_f\mathbf1)^{\top}\mathbf w\geq g_\star-g_f,\\
&0\leq w_i\leq1,\qquad w_f=1-\mathbf1^{\top}\mathbf w.
\end{aligned}
$$

The risky weights can sum above one because borrowing is allowed. The bounds $w_i\leq1$ limit each risky position to the initial wealth, so they can eventually restrict borrowing along a fixed risky direction.

**Recover the risky fund.** For targets $g_\star>g_f$, until an upper bound binds, normalize a nonzero solution by its risky fraction. Every such target selects the same long-only tangent mix:

$$
\mathbf w_{\mathcal T}=\frac{\mathbf w}{\mathbf1^{\top}\mathbf w},\qquad
\mathbf w(g_\star)=(1-w_f)\mathbf w_{\mathcal T},\qquad
1-w_f=\frac{g_\star-g_f}{\hat{\mathbf g}_{\mathrm{SIM}}^{\top}\mathbf w_{\mathcal T}-g_f}.
$$

Suppose the risky solution is $(0.30,0.45)$. Its sum is $0.75$, so $w_f=0.25$ and the normalized tangent weights are $(0.4,0.6)$. Borrowing 50% instead would give complete-portfolio risky weights $(0.6,0.9)$, with the same normalized mix. At larger exposures, check the upper bounds before using this scaling.

**Follow the complete investment.** The risky/risk-free example fixes the initial share counts and lets the risk-free position accumulate. With the tangent weights normalized to one, wealth is:

$$
W_t=W_0\left[w_fe^{g_ft}+(1-w_f)\sum_i w_{\mathcal T,i}\frac{S_i(t)}{S_i(0)}\right].
$$

For $w_f<0$, the first term is the growing debt. The example assumes the same borrowing and lending rate. Buy-and-hold lets the risk-free fraction and firm fractions drift, so realized growth and risk need not equal the fixed-weight CAL predictions. The comparison uses 2025 prices reserved from estimation and scores both tangent funds under the same sample covariance.

**Check.** Let $W_0=1{,}000$ USD, $w_f=-0.5$, $g_f=0.02\,\mathrm{yr}^{-1}$, and suppose the risky fund gains 10% over one year. **Answer:** net wealth is $1{,}000[-0.5e^{0.02}+1.5(1.10)]=1{,}139.90$ USD. The 1,650 USD risky holding is offset by a 510.10 USD repayment obligation.
