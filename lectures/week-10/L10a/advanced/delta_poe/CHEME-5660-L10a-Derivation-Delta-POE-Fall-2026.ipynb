# Derivation: Delta as a Probability of Exercise
The [lecture](../../CHEME-5660-L10a-Lecture-American-Derivatives-Greeks-Fall-2026.ipynb) states that a call's Delta on the CRR tree is a probability of exercise. Where does this come from, and how far is Delta from the risk-neutral probability of exercise?

## The Call Premium as a Sum over Terminal Nodes
Consider a call with strike price $K$ and time to expiration $T$ (years) on a share with no dividends. Its American and European premiums are then equal, so we can use the European backward induction.

The CRR tree has $h$ steps of length $\Delta t=T/h$. The share price at level $t$ after $k$ up moves is $S_{t,k}=S_{0}u^{k}d^{t-k}$, with up factor $u=e^{\sigma\sqrt{\Delta t}}$, down factor $d=e^{-\sigma\sqrt{\Delta t}}$, and volatility $\sigma$.

The risk-neutral up probability is:
$$
q=\frac{e^{g_{y}\Delta t}-d}{u-d}
$$
where $g_{y}$ is the annual risk-free rate. At expiration, node $k$ pays $\hat{V}_{h,k}=\max\left(S_{h,k}-K,~0\right)$.

Each step back weights the two next values by $q$ and $1-q$ and discounts one step:
$$
\hat{V}_{t,k}=e^{-g_{y}\Delta t}\left(q\,\hat{V}_{t+1,k+1}+(1-q)\,\hat{V}_{t+1,k}\right)
$$
Let's unroll it from the root for two steps:
$$
\begin{align*}
\hat{V}_{0,0} & = e^{-g_{y}\Delta t}\left(q\,\hat{V}_{1,1}+(1-q)\,\hat{V}_{1,0}\right) && \text{one step back from level 1}\\
& = e^{-g_{y}\Delta t}\,q\,e^{-g_{y}\Delta t}\left(q\,\hat{V}_{2,2}+(1-q)\,\hat{V}_{2,1}\right) && \text{substitute } \hat{V}_{1,1}\\
& \quad+e^{-g_{y}\Delta t}\,(1-q)\,e^{-g_{y}\Delta t}\left(q\,\hat{V}_{2,1}+(1-q)\,\hat{V}_{2,0}\right) && \text{substitute } \hat{V}_{1,0}\\
& = e^{-2g_{y}\Delta t}\left(q^{2}\,\hat{V}_{2,2}+2q(1-q)\,\hat{V}_{2,1}+(1-q)^{2}\,\hat{V}_{2,0}\right) && \text{collect, two paths reach node } (2,1)
\end{align*}
$$
Each path picks up a factor $q$ per up move and $1-q$ per down move, and $\binom{t}{k}$ paths reach node $(t,k)$. After $h$ steps, the discount is $e^{-g_{y}h\Delta t}=e^{-g_{y}T}=\mathcal{D}^{-1}_{T,0}(g_{y})$.

So the premium is:
$$
\boxed{
\begin{align*}
\mathcal{P}_{c} = \mathcal{D}^{-1}_{T,0}(g_{y})\sum_{k=0}^{h}\binom{h}{k}q^{k}(1-q)^{h-k}\max\left(S_{0}u^{k}d^{h-k}-K,~0\right)
\end{align*}}
$$
The weight $\binom{h}{k}q^{k}(1-q)^{h-k}$ is the risk-neutral probability of reaching terminal node $k$. Let $\text{ITM}=\left\{k : S_{0}u^{k}d^{h-k}>K\right\}$ be the set of in-the-money terminal nodes.

Summing their weights gives the risk-neutral probability of exercise:
$$
\mathbb{Q}\left(S_{T}>K\right)=\sum_{k\,\in\,\text{ITM}}\binom{h}{k}q^{k}(1-q)^{h-k}
$$
___

## Differentiating the Premium
Delta is the slope of the premium with respect to the share price today, $\Delta_{c}=\partial\mathcal{P}_{c}/\partial S_{0}$. The tree's $u$, $d$, and $q$ do not depend on $S_{0}$, so only the payoffs move.

The slope of one payoff is:
$$
\frac{\partial}{\partial S_{0}}\max\left(S_{0}u^{k}d^{h-k}-K,~0\right)=\begin{cases}
u^{k}d^{h-k} & \text{if } S_{0}u^{k}d^{h-k}>K \text{ (in the money)}\\
0 & \text{if } S_{0}u^{k}d^{h-k}<K \text{ (out of the money)}
\end{cases}
$$
To turn the sum into a probability, we absorb the discount factor into the up and down probabilities.

The risk-neutral condition follows from the definition of $q$:
$$
\begin{align*}
qu+(1-q)d & = d+q\,(u-d) && \text{collect the terms in } q\\
& = e^{g_{y}\Delta t} && \text{substitute } q=\left(e^{g_{y}\Delta t}-d\right)/(u-d)
\end{align*}
$$
Define $q^{*}=q\,u\,e^{-g_{y}\Delta t}$. Then the risk-neutral condition gives its complement:
$$
\begin{align*}
q\,u\,e^{-g_{y}\Delta t}+(1-q)\,d\,e^{-g_{y}\Delta t} & = 1 && \text{divide by } e^{g_{y}\Delta t}\\
(1-q)\,d\,e^{-g_{y}\Delta t} & = 1-q^{*} && \text{substitute } q^{*} \text{ and solve}
\end{align*}
$$
Both terms in the first line are positive and add to one, so $q^{*}$ is also a probability.

Then Delta is:
$$
\begin{align*}
\Delta_{c} & = \frac{\partial\mathcal{P}_{c}}{\partial S_{0}} && \text{definition of Delta}\\
& = \mathcal{D}^{-1}_{T,0}(g_{y})\sum_{k\,\in\,\text{ITM}}\binom{h}{k}q^{k}(1-q)^{h-k}\,u^{k}d^{h-k} && \text{only in-the-money payoffs have a slope}\\
& = \sum_{k\,\in\,\text{ITM}}\binom{h}{k}\left(q\,u\,e^{-g_{y}\Delta t}\right)^{k}\left((1-q)\,d\,e^{-g_{y}\Delta t}\right)^{h-k} && \mathcal{D}^{-1}_{T,0}(g_{y})=e^{-g_{y}h\Delta t}=\left(e^{-g_{y}\Delta t}\right)^{k}\left(e^{-g_{y}\Delta t}\right)^{h-k}\\
& = \sum_{k\,\in\,\text{ITM}}\binom{h}{k}\left(q^{*}\right)^{k}\left(1-q^{*}\right)^{h-k} && \text{substitute } q^{*} \text{ and } 1-q^{*}
\end{align*}
$$
Next to the probability of exercise, this gives the lecture's result:
$$
\boxed{
\begin{align*}
\Delta_{c} & = \sum_{k\,\in\,\text{ITM}}\binom{h}{k}\left(q^{*}\right)^{k}\left(1-q^{*}\right)^{h-k} && \text{Delta: up probability } q^{*}\\
\mathbb{Q}\left(S_{T}>K\right) & = \sum_{k\,\in\,\text{ITM}}\binom{h}{k}\,q^{k}\left(1-q\right)^{h-k} && \text{risk-neutral POE: up probability } q
\end{align*}}
$$
So Delta is a probability of exercise, computed with $q^{*}$ in place of $q$.

__Why is Delta larger?__ For $q<1$, we need $e^{g_{y}\Delta t}-d<u-d$, that is, $u>e^{g_{y}\Delta t}$. So $u\,e^{-g_{y}\Delta t}>1$, and $q^{*}=q\,u\,e^{-g_{y}\Delta t}>q$. The in-the-money nodes are the ones with the most up moves, so a larger up probability puts more weight on them, and a call's Delta is larger than its risk-neutral probability of exercise.
___

## How Far Apart Are Delta and the POE?
Both sums run over the same nodes, so the gap depends on how much larger $q^{*}$ is than $q$.

For a small time step:
$$
\begin{align*}
q^{*}-q & = q\,u\,e^{-g_{y}\Delta t}-q && \text{definition of } q^{*}\\
& = q\left(e^{\sigma\sqrt{\Delta t}-g_{y}\Delta t}-1\right) && \text{substitute } u=e^{\sigma\sqrt{\Delta t}}\\
& = q\left(\sigma\sqrt{\Delta t}+\mathcal{O}(\Delta t)\right) && e^{x}=1+x+\mathcal{O}(x^{2})\\
& = \frac{1}{2}\sigma\sqrt{\Delta t}+\mathcal{O}(\Delta t) && q=\tfrac{1}{2}+\mathcal{O}(\sqrt{\Delta t})
\end{align*}
$$
where $\mathcal{O}(\cdot)$ collects terms that shrink at least as fast as its argument. The last line uses the result from the [CRR factors notebook](../../../../week-9/L9b/advanced/crr_factors/CHEME-5660-L9b-Advanced-CRR-Factors-Fall-2026.ipynb) that $q$ approaches one half as the time step shrinks.

Each step adds only about $\frac{1}{2}\sigma\sqrt{\Delta t}$ to the up probability, but a tree has $h$ steps. The number of up moves $N$ is binomial, with mean $hq$ and standard deviation $\sqrt{hq(1-q)}$ under $q$, and switching to $q^{*}$ moves its mean by $h(q^{*}-q)$.

In standard deviations, the move is:
$$
\begin{align*}
\frac{h\left(q^{*}-q\right)}{\sqrt{hq(1-q)}} & \approx \frac{h\cdot\frac{1}{2}\sigma\sqrt{\Delta t}}{\frac{1}{2}\sqrt{h}} && \text{small time step: } q^{*}-q\approx\tfrac{1}{2}\sigma\sqrt{\Delta t}\text{ and } q(1-q)\approx\tfrac{1}{4}\\
& = \sigma\sqrt{h\,\Delta t} && \text{cancel } \tfrac{1}{2}\text{, and } h/\sqrt{h}=\sqrt{h}\\
& = \sigma\sqrt{T} && h\,\Delta t=T
\end{align*}
$$
The move does not depend on the number of tree steps. When $\sigma\sqrt{T}$ is small, the two distributions of up moves nearly overlap, and Delta is close to the probability of exercise. Delta is a good estimate for a 30-day contract with $\sigma=0.30$, where $\sigma\sqrt{T}\approx0.09$, and a poor one for a one-year contract with $\sigma=0.50$, where $\sigma\sqrt{T}=0.50$.
___

## Summary
We wrote a call's premium on the CRR tree as a sum over its terminal nodes and differentiated it with respect to the share price. Delta turned out to be a probability of exercise with a slightly larger up probability, and we measured the gap between the two.

> __Key Takeaways:__
>
> * __The premium is a sum over terminal nodes:__ We unrolled the European backward induction, which prices a call on a share with no dividends. Each terminal payoff is weighted by the risk-neutral probability of reaching its node and discounted to today.
> * __Delta is a probability of exercise:__ We differentiated the premium, keeping only the in-the-money payoffs, and split the discount factor over the steps. Delta became a probability of finishing in the money, with a slightly larger up probability than the risk-neutral one.
> * __The gap grows with the volatility and the time to expiration:__ Over the whole tree, the larger up probability shifts the distribution of up moves by about the volatility times the square root of the time to expiration, in standard deviations. So Delta is close to the probability of exercise for short-dated, low-volatility contracts.
