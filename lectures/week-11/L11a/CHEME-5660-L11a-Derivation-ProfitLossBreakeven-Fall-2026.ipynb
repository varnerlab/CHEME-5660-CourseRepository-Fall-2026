## The Recipe
The profit at expiration of a composite contract $\mathcal{C}$ (USD/share) is its payoff minus its composite premium grown to expiration:
$$
\begin{align*}
\mathcal{P}_{\mathcal{C}} &= \sum_{i\in\mathcal{C}}\theta_{i}\,n_{i}\,\mathcal{P}_{i} && \text{composite premium}\\
\hat{V}_{\mathcal{C}}(S(T)) &= \sum_{i\in\mathcal{C}}\theta_{i}\,n_{i}\,V_{i}(K_{i},S(T)) && \text{composite payoff}\\
\hat{P}_{\mathcal{C}}(S(T)) &= \hat{V}_{\mathcal{C}}(S(T))-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}} && \text{profit at expiration}
\end{align*}
$$
where $S(T)$ is the share price at expiration, leg $i$ has strike price $K_{i}$, premium $\mathcal{P}_{i}>0$, copy number $n_{i}$, and orientation $\theta_{i}=1$ (long) or $\theta_{i}=-1$ (short), $V_{i}$ is its payoff, $(S(T)-K_{i})^{+}$ for a call or $(K_{i}-S(T))^{+}$ for a put, and $\mathcal{D}_{T,0}(g_{y})=e^{g_{y}T}$ is the accumulation factor at the risk-free rate $g_{y}$.

Every premium $\mathcal{P}_{i}$ is positive, and the orientation sets its direction. Because $\theta_{i}=1$ for a long leg and $\theta_{i}=-1$ for a short leg, the composite premium splits into the premium we pay and the premium we collect:
$$
\mathcal{P}_{\mathcal{C}}=\underbrace{\sum_{\text{long legs}}n_{i}\,\mathcal{P}_{i}}_{\text{premium we pay}}-\underbrace{\sum_{\text{short legs}}n_{i}\,\mathcal{P}_{i}}_{\text{premium we collect}}
$$
If we pay more than we collect, $\mathcal{P}_{\mathcal{C}}>0$, and the contract opens for a __debit__. If we collect more than we pay, $\mathcal{P}_{\mathcal{C}}<0$, and it opens for a __credit__ of $-\mathcal{P}_{\mathcal{C}}$ USD/share. The profit then has two minus signs to track, the one in front of $\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}}$ and the one in front of the short legs' sum inside $\mathcal{P}_{\mathcal{C}}$. For a credit, $\mathcal{P}_{\mathcal{C}}<0$, so subtracting it adds the credit grown to expiration, $-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}}>0$.

From this expression, we find the maximum profit, maximum loss, and breakevens of any composite contract in three steps:
1. Compute the composite premium $\mathcal{P}_{\mathcal{C}}$ one leg at a time, with a minus sign for each short leg. If $\mathcal{P}_{\mathcal{C}}<0$, the contract opens for a credit of $-\mathcal{P}_{\mathcal{C}}$ (a positive term in the profit at expiration).
2. Sort the strikes and write the profit on each interval between them. On each interval, each leg's payoff is either linear, $S(T)-K_{i}$ for a call or $K_{i}-S(T)$ for a put in the money, or zero. So the profit is a straight line on each interval.
3. A straight line takes its largest and smallest values at the ends of its interval, so we evaluate the profit at the strikes, at $S(T)=0$, and as $S(T)$ grows without limit. The largest value is the maximum profit, and the negative of the smallest value is the maximum loss. If the profit keeps falling as $S(T)$ grows, the maximum loss is unlimited. The breakevens $S^{\ddagger}$ are the share prices where the profit crosses zero, $\hat{P}_{\mathcal{C}}(S^{\ddagger})=0$. We solve for them on each sloped interval and keep a solution only if it lies in that interval.

The lecture tables list the maximum loss as a positive amount (USD/share), the negative of the smallest profit. Let's work through the three steps for a put credit spread, then for a short strangle.

## Put Credit Spread
A put credit spread sells a put at strike $K_{2}$ and buys a put at a lower strike $K_{1}<K_{2}$, one contract of each, with premiums $\mathcal{P}_{2}$ and $\mathcal{P}_{1}$:

| Leg $i$ | Contract | Strike | Orientation $\theta_{i}$ | Copy number $n_{i}$ | Premium |
|:--|:--|:--|:--|:--|:--|
| 1 | put | $K_{1}$ | $1$ (long) | $1$ | $\mathcal{P}_{1}$ |
| 2 | put | $K_{2}$ | $-1$ (short) | $1$ | $\mathcal{P}_{2}$ |

### The Composite Premium
We sum over the two legs:
$$
\begin{align*}
\mathcal{P}_{\mathcal{C}} &= \theta_{1}\,n_{1}\,\mathcal{P}_{1}+\theta_{2}\,n_{2}\,\mathcal{P}_{2} && \text{sum over the two legs}\\
&= (1)(1)\,\mathcal{P}_{1}+(-1)(1)\,\mathcal{P}_{2} && \text{long put at } K_{1}\text{, short put at } K_{2}\\
&= -(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{factor out the minus sign}
\end{align*}
$$
A put with a higher strike is worth more, so $\mathcal{P}_{2}>\mathcal{P}_{1}$ and $\mathcal{P}_{\mathcal{C}}<0$. The spread opens for a credit of $\mathcal{P}_{2}-\mathcal{P}_{1}$. For example, if $\mathcal{P}_{1}=1.00$ and $\mathcal{P}_{2}=2.50$ USD/share, then $\mathcal{P}_{\mathcal{C}}=1.00-2.50=-1.50$ USD/share, a credit of 1.50 USD/share.

The composite payoff weights each put payoff by $\theta_{i}n_{i}$, and the profit subtracts the composite premium grown to expiration:
$$
\begin{align*}
\hat{V}_{\mathcal{C}}(S(T)) &= (1)(1)\,(K_{1}-S(T))^{+}+(-1)(1)\,(K_{2}-S(T))^{+} && \text{composite payoff}\\
\hat{P}_{\mathcal{C}}(S(T)) &= \hat{V}_{\mathcal{C}}(S(T))-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}} && \text{profit at expiration}\\
&= (K_{1}-S(T))^{+}-(K_{2}-S(T))^{+}-\mathcal{D}_{T,0}(g_{y})\cdot\bigl(-(\mathcal{P}_{2}-\mathcal{P}_{1})\bigr) && \text{substitute } \hat{V}_{\mathcal{C}} \text{ and } \mathcal{P}_{\mathcal{C}}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-S(T))^{+}+(K_{1}-S(T))^{+} && \text{the two minus signs cancel}
\end{align*}
$$
The last line is the lecture's profit at expiration for the put credit spread. Its first term is the credit grown to expiration.

### The Profit on Each Interval
A put is in the money when $S(T)<K_{i}$, with payoff $K_{i}-S(T)$, and pays zero otherwise. The two strikes split the share prices at expiration into three intervals.

Below $K_{1}$, both puts are in the money:
$$
\begin{align*}
\hat{P}_{\mathcal{C}}(S(T)) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-S(T))+(K_{1}-S(T)) && \text{both payoffs are linear}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-K_{1}) && S(T) \text{ cancels}
\end{align*}
$$
Between the strikes, only the short put is in the money, and at or above $K_{2}$, neither put is.

Collecting the three intervals:
$$
\hat{P}_{\mathcal{C}}(S(T))=\begin{cases}
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-K_{1}) & \text{if } 0\leq S(T)<K_{1} \text{ (both puts in the money)}\\
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-S(T)) & \text{if } K_{1}\leq S(T)<K_{2} \text{ (short put in the money)}\\
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) & \text{if } S(T)\geq K_{2} \text{ (neither put in the money)}
\end{cases}
$$
The profit is flat below $K_{1}$, rises one for one with the share price between the strikes, and is flat again above $K_{2}$.

### Maximum Profit, Maximum Loss, and Breakeven
The middle line rises from its value at $K_{1}$ to its value at $K_{2}$, so the largest profit is on the upper flat piece and the smallest on the lower one:
$$
\begin{align*}
\text{maximum profit} &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{for } S(T)\geq K_{2}\\
\text{maximum loss} &= -\Bigl(\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-K_{1})\Bigr) && \text{negative of the smallest profit, for } S(T)\leq K_{1}\\
&= (K_{2}-K_{1})-\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{distribute the minus sign}
\end{align*}
$$
The profit can cross zero only on the middle interval, where it slopes:
$$
\begin{align*}
0 &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-S^{\ddagger}) && \text{set the middle line to zero at } S(T)=S^{\ddagger}\\
S^{\ddagger} &= K_{2}-\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{solve for } S^{\ddagger}
\end{align*}
$$
The breakeven sits below the short strike by the credit grown to expiration. It lies in the middle interval when the credit grown to expiration is less than the strike width $K_{2}-K_{1}$, which is also when the maximum loss is positive. These are the put credit spread's entries in the lecture table, and the spread profits when $S(T)>S^{\ddagger}$.

## Short Strangle
A short strangle sells a put at strike $K_{1}$ and a call at a higher strike $K_{2}>K_{1}$, one contract of each, with premiums $\mathcal{P}_{1}$ and $\mathcal{P}_{2}$:

| Leg $i$ | Contract | Strike | Orientation $\theta_{i}$ | Copy number $n_{i}$ | Premium |
|:--|:--|:--|:--|:--|:--|
| 1 | put | $K_{1}$ | $-1$ (short) | $1$ | $\mathcal{P}_{1}$ |
| 2 | call | $K_{2}$ | $-1$ (short) | $1$ | $\mathcal{P}_{2}$ |

### The Composite Premium
Both legs are short, so both premiums enter with a minus sign:
$$
\begin{align*}
\mathcal{P}_{\mathcal{C}} &= \theta_{1}\,n_{1}\,\mathcal{P}_{1}+\theta_{2}\,n_{2}\,\mathcal{P}_{2} && \text{sum over the two legs}\\
&= (-1)(1)\,\mathcal{P}_{1}+(-1)(1)\,\mathcal{P}_{2} && \text{short put at } K_{1}\text{, short call at } K_{2}\\
&= -(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{factor out the minus sign}
\end{align*}
$$
So $\mathcal{P}_{\mathcal{C}}<0$ for any premiums, and the strangle opens for a credit of $\mathcal{P}_{1}+\mathcal{P}_{2}$. For example, if $\mathcal{P}_{1}=1.20$ and $\mathcal{P}_{2}=1.40$ USD/share, then $\mathcal{P}_{\mathcal{C}}=-1.20-1.40=-2.60$ USD/share, a credit of 2.60 USD/share.

The composite payoff and the profit at expiration are:
$$
\begin{align*}
\hat{V}_{\mathcal{C}}(S(T)) &= (-1)(1)\,(K_{1}-S(T))^{+}+(-1)(1)\,(S(T)-K_{2})^{+} && \text{composite payoff}\\
\hat{P}_{\mathcal{C}}(S(T)) &= \hat{V}_{\mathcal{C}}(S(T))-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}} && \text{profit at expiration}\\
&= -(K_{1}-S(T))^{+}-(S(T)-K_{2})^{+}-\mathcal{D}_{T,0}(g_{y})\cdot\bigl(-(\mathcal{P}_{1}+\mathcal{P}_{2})\bigr) && \text{substitute } \hat{V}_{\mathcal{C}} \text{ and } \mathcal{P}_{\mathcal{C}}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(K_{1}-S(T))^{+}-(S(T)-K_{2})^{+} && \text{the two minus signs cancel}
\end{align*}
$$
The last line is the lecture's profit at expiration for the short strangle.

### The Profit on Each Interval
The put is in the money when $S(T)<K_{1}$, and the call when $S(T)>K_{2}$. At most one leg is in the money at any share price, so each interval keeps at most one payoff:
$$
\hat{P}_{\mathcal{C}}(S(T))=\begin{cases}
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(K_{1}-S(T)) & \text{if } 0\leq S(T)<K_{1} \text{ (put in the money)}\\
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) & \text{if } K_{1}\leq S(T)\leq K_{2} \text{ (neither leg in the money)}\\
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(S(T)-K_{2}) & \text{if } S(T)>K_{2} \text{ (call in the money)}
\end{cases}
$$
The profit rises one for one with the share price up to $K_{1}$, is flat between the strikes, and falls one for one above $K_{2}$.

### Maximum Profit, Maximum Losses, and Breakevens
The largest profit is on the flat piece between the strikes. Below $K_{1}$, the profit is smallest at $S(T)=0$. Above $K_{2}$, it keeps falling as the share price grows, so the loss on that side has no limit:
$$
\begin{align*}
\text{maximum profit} &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{for } K_{1}\leq S(T)\leq K_{2}\\
\text{maximum loss below } K_{1} &= -\Bigl(\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-K_{1}\Bigr) && \text{negative of the profit at } S(T)=0\\
&= K_{1}-\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{distribute the minus sign}\\
\text{maximum loss above } K_{2} &= \text{unlimited} && \hat{P}_{\mathcal{C}}(S(T))\to-\infty \text{ as } S(T)\to\infty
\end{align*}
$$
The profit slopes on both outer intervals, so we set each outer line to zero:
$$
\begin{align*}
0 &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(K_{1}-S^{\ddagger}_{-}) && \text{lower line at } S(T)=S^{\ddagger}_{-}\\
S^{\ddagger}_{-} &= K_{1}-\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{solve for } S^{\ddagger}_{-}\\
0 &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(S^{\ddagger}_{+}-K_{2}) && \text{upper line at } S(T)=S^{\ddagger}_{+}\\
S^{\ddagger}_{+} &= K_{2}+\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{solve for } S^{\ddagger}_{+}
\end{align*}
$$
Each breakeven sits outside its strike by the credit grown to expiration. The upper one always lies above $K_{2}$. The lower one lies in its interval, $S^{\ddagger}_{-}>0$, when the credit grown to expiration is less than $K_{1}$, which is also when the maximum loss below $K_{1}$ is positive. These are the short strangle's entries in the lecture table, and the strangle profits when $S^{\ddagger}_{-}<S(T)<S^{\ddagger}_{+}$. Setting $K_{1}=K_{2}=K$ shrinks the flat middle piece to the single point $S(T)=K$ and gives the short straddle's entries, with the put premium $\mathcal{P}_{p}$ and the call premium $\mathcal{P}_{c}$ in place of $\mathcal{P}_{1}$ and $\mathcal{P}_{2}$.

## Summary
We started from the profit at expiration of a composite contract and derived the maximum profit, maximum loss, and breakevens of a put credit spread and a short strangle.

> __Key Takeaways:__
>
> * __The composite premium carries the sign of each leg:__ We added up the legs' premiums weighted by orientation and copy number, so each short leg entered with a minus sign. A negative composite premium means the contract opens for a credit, and the profit at expiration then starts from the credit grown to expiration.
> * __The profit at expiration is a straight line between strikes:__ We split the share prices at expiration at the strikes and kept only the legs in the money on each interval. Each payoff is then zero or linear in the share price, so the profit is a straight line on each interval.
> * __The table entries come from the ends of each interval:__ We evaluated the profit on its flat pieces, at a share price of zero, and as the share price grows, and set the sloped pieces to zero for the breakevens, keeping each solution that lies in its interval. This gave the lecture's table entries for both contracts, and the same three steps work for the other composite contracts.
