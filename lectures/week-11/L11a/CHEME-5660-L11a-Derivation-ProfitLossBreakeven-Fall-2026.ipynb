## The General Case
The profit at expiration of a composite contract $\mathcal{C}$ (USD/share) is its payoff minus its composite premium grown to expiration:
$$
\begin{align*}
\mathcal{P}_{\mathcal{C}} &= \sum_{i\in\mathcal{C}}\theta_{i}\,n_{i}\,\mathcal{P}_{i} && \text{composite premium}\\
\hat{V}_{\mathcal{C}}(S(T)) &= \sum_{i\in\mathcal{C}}\theta_{i}\,n_{i}\,V_{i}(K_{i},S(T)) && \text{composite payoff}\\
\hat{P}_{\mathcal{C}}(S(T)) &= \hat{V}_{\mathcal{C}}(S(T))-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}} && \text{profit at expiration}
\end{align*}
$$
where $T$ is the time to expiration (years) and $S(T)$ is the share price at expiration. Leg $i$ has strike price $K_{i}$, premium $\mathcal{P}_{i}>0$, copy number $n_{i}$, orientation $\theta_{i}=1$ (long) or $\theta_{i}=-1$ (short), and payoff $V_{i}(K_{i},S(T))$. The accumulation factor at the risk-free rate $g_{y}$ is $\mathcal{D}_{T,0}(g_{y})=e^{g_{y}T}$.

### The Sign of the Composite Premium
Each premium $\mathcal{P}_{i}$ is positive, and the orientation $\theta_{i}$ sets the sign of its term. Splitting the sum into long and short legs:
$$
\begin{align*}
\mathcal{P}_{\mathcal{C}} &= \sum_{i\in\mathcal{C}}\theta_{i}\,n_{i}\,\mathcal{P}_{i} && \text{composite premium}\\
&= \sum_{\text{long legs}}(1)\,n_{i}\,\mathcal{P}_{i}+\sum_{\text{short legs}}(-1)\,n_{i}\,\mathcal{P}_{i} && \theta_{i}=1 \text{ for long legs, } \theta_{i}=-1 \text{ for short legs}\\
&= \underbrace{\sum_{\text{long legs}}n_{i}\,\mathcal{P}_{i}}_{\text{premium we pay}}-\underbrace{\sum_{\text{short legs}}n_{i}\,\mathcal{P}_{i}}_{\text{premium we collect}} && \text{factor }-1\text{ out of the short-leg sum}
\end{align*}
$$
If we pay more than we collect, $\mathcal{P}_{\mathcal{C}}>0$, and the contract opens for a __debit__. If we collect more than we pay, $\mathcal{P}_{\mathcal{C}}<0$, and it opens for a __credit__ of $-\mathcal{P}_{\mathcal{C}}>0$ USD/share. For a credit, the premium term in the profit at expiration is positive:
$$
\begin{align*}
\hat{P}_{\mathcal{C}}(S(T)) &= \hat{V}_{\mathcal{C}}(S(T))-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}} && \text{profit at expiration}\\
&= \hat{V}_{\mathcal{C}}(S(T))+\underbrace{\mathcal{D}_{T,0}(g_{y})\cdot(-\mathcal{P}_{\mathcal{C}})}_{\text{credit grown to expiration}} && \text{move the minus sign onto }\mathcal{P}_{\mathcal{C}}
\end{align*}
$$

### The Profit Between the Strikes
Each leg's payoff, with $x^{+}=\max(x,0)$, is zero on one side of its strike and a straight line in $S(T)$ on the other:
$$
\begin{align*}
(S(T)-K_{i})^{+} &= \begin{cases}S(T)-K_{i} & \text{if } S(T)>K_{i} \text{ (in the money)}\\ 0 & \text{if } S(T)\leq K_{i}\end{cases} && \text{call}\\
(K_{i}-S(T))^{+} &= \begin{cases}K_{i}-S(T) & \text{if } S(T)<K_{i} \text{ (in the money)}\\ 0 & \text{if } S(T)\geq K_{i}\end{cases} && \text{put}
\end{align*}
$$
Between two neighboring strikes, each leg stays in the money or stays out of the money. So on each interval, the profit is a constant plus a sum of straight lines, which is a straight line. Its largest and smallest values are at the strikes, at $S(T)=0$, or as $S(T)$ grows.

### The Three Steps
This gives three steps for any composite contract:
1. Compute the composite premium $\mathcal{P}_{\mathcal{C}}$, with a minus sign for each short leg. If $\mathcal{P}_{\mathcal{C}}<0$, the contract opens for a credit of $-\mathcal{P}_{\mathcal{C}}$ (a positive term in the profit at expiration).
2. Sort the strikes and write the profit on each interval between them. On each interval, replace each payoff by its in-the-money line or by zero.
3. Evaluate the profit at $S(T)=0$, at the strikes, and as $S(T)$ grows. The largest value is the maximum profit, and the negative of the smallest value is the maximum loss. If the profit keeps falling as $S(T)$ grows, the maximum loss is unlimited. Set the profit to zero on each sloped interval to find the breakevens $S^{\ddagger}$, keeping each solution that lies in its interval.

The lecture tables list the maximum loss as a positive amount (USD/share). Let's apply the three steps to a put credit spread, then to a short strangle.

## Put Credit Spread
A put credit spread sells a put at strike $K_{2}$ and buys a put at a lower strike $K_{1}<K_{2}$, one contract of each with the same expiration, with premiums $\mathcal{P}_{2}$ and $\mathcal{P}_{1}$:

| Leg $i$ | Contract | Strike | Orientation $\theta_{i}$ | Copy number $n_{i}$ | Premium |
|:--|:--|:--|:--|:--|:--|
| 1 | put | $K_{1}$ | $1$ (long) | $1$ | $\mathcal{P}_{1}$ |
| 2 | put | $K_{2}$ | $-1$ (short) | $1$ | $\mathcal{P}_{2}$ |

### The Composite Premium
We substitute the two legs into the composite premium:
$$
\begin{align*}
\mathcal{P}_{\mathcal{C}} &= \theta_{1}\,n_{1}\,\mathcal{P}_{1}+\theta_{2}\,n_{2}\,\mathcal{P}_{2} && \text{sum over the two legs}\\
&= (1)(1)\,\mathcal{P}_{1}+(-1)(1)\,\mathcal{P}_{2} && \theta_{1}=1,\ \theta_{2}=-1,\ n_{1}=n_{2}=1\\
&= \mathcal{P}_{1}-\mathcal{P}_{2} && \text{premium we pay minus premium we collect}\\
&= -(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{minus the credit } \mathcal{P}_{2}-\mathcal{P}_{1}>0
\end{align*}
$$
A put with a higher strike is worth more, so $\mathcal{P}_{2}>\mathcal{P}_{1}$ and $\mathcal{P}_{\mathcal{C}}<0$. The spread opens for a credit of $\mathcal{P}_{2}-\mathcal{P}_{1}$. For example, if $\mathcal{P}_{1}=1.00$ and $\mathcal{P}_{2}=2.50$ USD/share, then $\mathcal{P}_{\mathcal{C}}=1.00-2.50=-1.50$ USD/share, a credit of 1.50 USD/share.

### The Profit at Expiration
We substitute the legs into the composite payoff, then the payoff and the composite premium into the profit at expiration:
$$
\begin{align*}
\hat{V}_{\mathcal{C}}(S(T)) &= \theta_{1}\,n_{1}\,(K_{1}-S(T))^{+}+\theta_{2}\,n_{2}\,(K_{2}-S(T))^{+} && \text{composite payoff of two puts}\\
&= (K_{1}-S(T))^{+}-(K_{2}-S(T))^{+} && \theta_{1}=1,\ \theta_{2}=-1,\ n_{1}=n_{2}=1\\
\hat{P}_{\mathcal{C}}(S(T)) &= \hat{V}_{\mathcal{C}}(S(T))-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}} && \text{general profit at expiration}\\
&= (K_{1}-S(T))^{+}-(K_{2}-S(T))^{+}-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}} && \text{substitute } \hat{V}_{\mathcal{C}}\\
&= (K_{1}-S(T))^{+}-(K_{2}-S(T))^{+}-\mathcal{D}_{T,0}(g_{y})\cdot\bigl(-(\mathcal{P}_{2}-\mathcal{P}_{1})\bigr) && \text{substitute } \mathcal{P}_{\mathcal{C}}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-S(T))^{+}+(K_{1}-S(T))^{+} && \text{the two minus signs cancel; write the credit first}
\end{align*}
$$
The last line is the lecture's profit at expiration for the put credit spread. Its first term is the credit grown to expiration.

### The Profit on Each Interval
The two strikes split the share prices at expiration into three intervals. A put is in the money when $S(T)<K_{i}$, and then its payoff is $K_{i}-S(T)$.

Below $K_{1}$, both puts are in the money:
$$
\begin{align*}
\hat{P}_{\mathcal{C}}(S(T)) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-S(T))+(K_{1}-S(T)) && \text{short put pays } K_{2}-S(T)\text{, long put pays } K_{1}-S(T)\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-K_{2}+S(T)+K_{1}-S(T) && \text{remove the parentheses}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-K_{1}) && S(T) \text{ cancels}
\end{align*}
$$
Between the strikes, only the short put is in the money:
$$
\begin{align*}
\hat{P}_{\mathcal{C}}(S(T)) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-S(T))+0 && \text{short put pays } K_{2}-S(T)\text{, long put pays zero}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-K_{2}+S(T) && \text{remove the parentheses}
\end{align*}
$$
At or above $K_{2}$, neither put is in the money:
$$
\begin{align*}
\hat{P}_{\mathcal{C}}(S(T)) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-0+0 && \text{both puts pay zero}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{remove the zero terms}
\end{align*}
$$
Together, the profit is flat below $K_{1}$, rises one for one with $S(T)$ between the strikes, and is flat above $K_{2}$:
$$
\hat{P}_{\mathcal{C}}(S(T))=\begin{cases}
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-K_{1}) & \text{if } 0\leq S(T)<K_{1}\\
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-S(T)) & \text{if } K_{1}\leq S(T)<K_{2}\\
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) & \text{if } S(T)\geq K_{2}
\end{cases}
$$

### Maximum Profit and Maximum Loss
We evaluate the profit at the ends of the intervals:
$$
\begin{align*}
\hat{P}_{\mathcal{C}}(0) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-K_{1}) && \text{lower interval, flat}\\
\hat{P}_{\mathcal{C}}(K_{1}) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-K_{1}) && \text{middle line at } S(T)=K_{1}\\
\hat{P}_{\mathcal{C}}(K_{2}) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{upper interval, flat}
\end{align*}
$$
The profit stays at $\hat{P}_{\mathcal{C}}(K_{2})$ as $S(T)$ grows. The largest of these values is the maximum profit, and the negative of the smallest is the maximum loss:
$$
\begin{align*}
\text{maximum profit} &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{for } S(T)\geq K_{2}\\
\text{maximum loss} &= -\Bigl(\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-K_{1})\Bigr) && \text{negative of } \hat{P}_{\mathcal{C}}(0)\text{, for } S(T)\leq K_{1}\\
&= (K_{2}-K_{1})-\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{distribute the minus sign}
\end{align*}
$$

### Breakeven
The profit slopes only on the middle interval, so we set the middle line to zero at $S(T)=S^{\ddagger}$:
$$
\begin{align*}
0 &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})-(K_{2}-S^{\ddagger}) && \text{middle line at } S(T)=S^{\ddagger}\\
K_{2}-S^{\ddagger} &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{add } K_{2}-S^{\ddagger} \text{ to both sides}\\
S^{\ddagger} &= K_{2}-\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1}) && \text{solve for } S^{\ddagger}
\end{align*}
$$
We keep this solution only if it lies in the middle interval:
$$
\begin{align*}
K_{1} &< S^{\ddagger}<K_{2} && \text{strictly inside the middle interval}\\
K_{1} &< K_{2}-\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})<K_{2} && \text{substitute } S^{\ddagger}\\
K_{1}-K_{2} &< -\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})<0 && \text{subtract } K_{2}\\
0 &< \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{2}-\mathcal{P}_{1})<K_{2}-K_{1} && \text{multiply by }-1\text{ and flip the inequalities}
\end{align*}
$$
The breakeven sits below the short strike by the credit grown to expiration. The left inequality holds because the spread opens for a credit. The right inequality says the credit grown to expiration is less than the strike width, which is also when the maximum loss is positive. These are the put credit spread's entries in the lecture table. The spread makes a profit when $S(T)>S^{\ddagger}$.

## Short Strangle
A short strangle sells a put at strike $K_{1}$ and a call at a higher strike $K_{2}>K_{1}$, one contract of each with the same expiration, with premiums $\mathcal{P}_{1}$ and $\mathcal{P}_{2}$:

| Leg $i$ | Contract | Strike | Orientation $\theta_{i}$ | Copy number $n_{i}$ | Premium |
|:--|:--|:--|:--|:--|:--|
| 1 | put | $K_{1}$ | $-1$ (short) | $1$ | $\mathcal{P}_{1}$ |
| 2 | call | $K_{2}$ | $-1$ (short) | $1$ | $\mathcal{P}_{2}$ |

### The Composite Premium
Both legs are short, so both premiums enter with a minus sign:
$$
\begin{align*}
\mathcal{P}_{\mathcal{C}} &= \theta_{1}\,n_{1}\,\mathcal{P}_{1}+\theta_{2}\,n_{2}\,\mathcal{P}_{2} && \text{sum over the two legs}\\
&= (-1)(1)\,\mathcal{P}_{1}+(-1)(1)\,\mathcal{P}_{2} && \theta_{1}=\theta_{2}=-1,\ n_{1}=n_{2}=1\\
&= 0-(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{no long legs: we pay nothing, we collect } \mathcal{P}_{1}+\mathcal{P}_{2}\\
&= -(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{minus the credit } \mathcal{P}_{1}+\mathcal{P}_{2}>0
\end{align*}
$$
So $\mathcal{P}_{\mathcal{C}}<0$ for any premiums, and the strangle opens for a credit of $\mathcal{P}_{1}+\mathcal{P}_{2}$. For example, if $\mathcal{P}_{1}=1.20$ and $\mathcal{P}_{2}=1.40$ USD/share, then $\mathcal{P}_{\mathcal{C}}=-1.20-1.40=-2.60$ USD/share, a credit of 2.60 USD/share.

### The Profit at Expiration
We substitute the legs into the composite payoff, then the payoff and the composite premium into the profit at expiration:
$$
\begin{align*}
\hat{V}_{\mathcal{C}}(S(T)) &= \theta_{1}\,n_{1}\,(K_{1}-S(T))^{+}+\theta_{2}\,n_{2}\,(S(T)-K_{2})^{+} && \text{composite payoff of a put and a call}\\
&= -(K_{1}-S(T))^{+}-(S(T)-K_{2})^{+} && \theta_{1}=\theta_{2}=-1,\ n_{1}=n_{2}=1\\
\hat{P}_{\mathcal{C}}(S(T)) &= \hat{V}_{\mathcal{C}}(S(T))-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}} && \text{general profit at expiration}\\
&= -(K_{1}-S(T))^{+}-(S(T)-K_{2})^{+}-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{\mathcal{C}} && \text{substitute } \hat{V}_{\mathcal{C}}\\
&= -(K_{1}-S(T))^{+}-(S(T)-K_{2})^{+}-\mathcal{D}_{T,0}(g_{y})\cdot\bigl(-(\mathcal{P}_{1}+\mathcal{P}_{2})\bigr) && \text{substitute } \mathcal{P}_{\mathcal{C}}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(K_{1}-S(T))^{+}-(S(T)-K_{2})^{+} && \text{the two minus signs cancel; write the credit first}
\end{align*}
$$
The last line is the lecture's profit at expiration for the short strangle.

### The Profit on Each Interval
The put is in the money when $S(T)<K_{1}$, with payoff $K_{1}-S(T)$. The call is in the money when $S(T)>K_{2}$, with payoff $S(T)-K_{2}$.

Below $K_{1}$, only the put is in the money:
$$
\begin{align*}
\hat{P}_{\mathcal{C}}(S(T)) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(K_{1}-S(T))-0 && \text{put pays } K_{1}-S(T)\text{, call pays zero}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-K_{1}+S(T) && \text{remove the parentheses}
\end{align*}
$$
Between the strikes, neither leg is in the money:
$$
\begin{align*}
\hat{P}_{\mathcal{C}}(S(T)) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-0-0 && \text{both legs pay zero}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{remove the zero terms}
\end{align*}
$$
Above $K_{2}$, only the call is in the money:
$$
\begin{align*}
\hat{P}_{\mathcal{C}}(S(T)) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-0-(S(T)-K_{2}) && \text{put pays zero, call pays } S(T)-K_{2}\\
&= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})+K_{2}-S(T) && \text{remove the parentheses}
\end{align*}
$$
Together, the profit rises one for one with $S(T)$ below $K_{1}$, is flat between the strikes, and falls one for one above $K_{2}$:
$$
\hat{P}_{\mathcal{C}}(S(T))=\begin{cases}
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(K_{1}-S(T)) & \text{if } 0\leq S(T)<K_{1}\\
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) & \text{if } K_{1}\leq S(T)\leq K_{2}\\
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(S(T)-K_{2}) & \text{if } S(T)>K_{2}
\end{cases}
$$

### Maximum Profit and Maximum Losses
We evaluate the profit at the ends of the intervals:
$$
\begin{align*}
\hat{P}_{\mathcal{C}}(0) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-K_{1} && \text{lower line at } S(T)=0\\
\hat{P}_{\mathcal{C}}(K_{1}) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{middle interval, flat}\\
\hat{P}_{\mathcal{C}}(K_{2}) &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{middle interval, flat}
\end{align*}
$$
Above $K_{2}$, the profit falls without limit as $S(T)$ grows. The largest value is the maximum profit. The maximum loss below $K_{1}$ is the negative of $\hat{P}_{\mathcal{C}}(0)$, and above $K_{2}$ it is unlimited:
$$
\begin{align*}
\text{maximum profit} &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{for } K_{1}\leq S(T)\leq K_{2}\\
\text{maximum loss below } K_{1} &= -\Bigl(\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-K_{1}\Bigr) && \text{negative of } \hat{P}_{\mathcal{C}}(0)\\
&= K_{1}-\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{distribute the minus sign}\\
\text{maximum loss above } K_{2} &= \text{unlimited} && \hat{P}_{\mathcal{C}}(S(T))\to-\infty \text{ as } S(T)\to\infty
\end{align*}
$$

### Breakevens
The profit slopes on both outer intervals, so we set each outer line to zero:
$$
\begin{align*}
0 &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(K_{1}-S^{\ddagger}_{-}) && \text{lower line at } S(T)=S^{\ddagger}_{-}\\
K_{1}-S^{\ddagger}_{-} &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{add } K_{1}-S^{\ddagger}_{-} \text{ to both sides}\\
S^{\ddagger}_{-} &= K_{1}-\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{solve for } S^{\ddagger}_{-}\\
0 &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})-(S^{\ddagger}_{+}-K_{2}) && \text{upper line at } S(T)=S^{\ddagger}_{+}\\
S^{\ddagger}_{+}-K_{2} &= \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{add } S^{\ddagger}_{+}-K_{2} \text{ to both sides}\\
S^{\ddagger}_{+} &= K_{2}+\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) && \text{solve for } S^{\ddagger}_{+}
\end{align*}
$$
We keep each solution only if it lies in its interval:
$$
\begin{align*}
0 &< S^{\ddagger}_{-}<K_{1} && \text{strictly inside the lower interval}\\
0 &< K_{1}-\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})<K_{1} && \text{substitute } S^{\ddagger}_{-}\\
-K_{1} &< -\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})<0 && \text{subtract } K_{1}\\
0 &< \mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2})<K_{1} && \text{multiply by }-1\text{ and flip the inequalities}\\
S^{\ddagger}_{+} &> K_{2} && \text{inside the upper interval}\\
K_{2}+\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) &> K_{2} && \text{substitute } S^{\ddagger}_{+}\\
\mathcal{D}_{T,0}(g_{y})\cdot(\mathcal{P}_{1}+\mathcal{P}_{2}) &> 0 && \text{subtract } K_{2}
\end{align*}
$$
Each breakeven sits outside its strike by the credit grown to expiration. The upper one always lies in its interval because the credit is positive. The lower one does when the credit grown to expiration is less than $K_{1}$, which is also when the maximum loss below $K_{1}$ is positive. These are the short strangle's entries in the lecture table. The strangle makes a profit when $S^{\ddagger}_{-}<S(T)<S^{\ddagger}_{+}$.

For the short straddle, set $K_{1}=K_{2}=K$, $\mathcal{P}_{1}=\mathcal{P}_{p}$ (put premium), and $\mathcal{P}_{2}=\mathcal{P}_{c}$ (call premium). The flat middle interval shrinks to the single point $S(T)=K$, and the same steps give the straddle's entries in the lecture table.

## Summary
We started from the profit at expiration of a composite contract and derived the maximum profit, maximum loss, and breakevens of a put credit spread and a short strangle.

> __Key Takeaways:__
>
> * __The composite premium carries the sign of each leg:__ We added up the legs' premiums weighted by orientation and copy number, so each short leg entered with a minus sign. A negative composite premium means the contract opens for a credit, and the credit grown to expiration is a positive term in the profit at expiration.
> * __The profit at expiration is a straight line between strikes:__ We split the share prices at expiration at the strikes and kept only the legs in the money on each interval. Each payoff is then zero or linear in the share price, so the profit is a straight line on each interval.
> * __The table entries come from the ends of each interval:__ We evaluated the profit at a share price of zero, at the strikes, and as the share price grows. We set each sloped piece to zero for the breakevens and kept each solution that lies in its interval. This gave the lecture's table entries for both contracts, and the same three steps work for the other composite contracts.
