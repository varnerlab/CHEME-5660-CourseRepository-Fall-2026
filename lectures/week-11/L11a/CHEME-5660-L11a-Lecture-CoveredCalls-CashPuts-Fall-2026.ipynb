# L11a: Covered Calls and Cash Secured Puts
In this lecture, we analyze two popular strategies for generating income from options: covered calls and cash secured puts. Both sell an option and collect its premium. What can each trade earn and lose, and how does its value change with the share price?

> __Learning Objectives:__
> 
> By the end of this lecture, you should be able to:
> * __Analyze a covered call:__ Derive the profit at expiration of shares plus a short call, identify the two regimes based on whether the call is exercised or expires worthless, and compute the breakeven, maximum profit, and maximum loss.
> * __Analyze a cash secured put:__ Derive the profit at expiration of a short put backed by enough cash to buy the shares, identify the two regimes based on whether the put is exercised or expires worthless, and compute the breakeven, maximum profit, and maximum loss.
> * __Calculate the net Delta of a position:__ Compute the net Delta of a portfolio of shares and options by adding the shares to the weighted Deltas of the option legs, and find the number of shares that makes a position delta-neutral.

Some results in this lecture are core, and some are here to build understanding:

> __What you need to know:__
>
> __Core concepts:__
> * A covered call owns $m$ shares for each call it sells. Its profit at expiration is $\min(S(T),K)-\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)-\mathcal{P}_{c}\right)$ per share: the premium lowers the breakeven, and the call caps the profit at the strike.
> * A cash secured put sells a put and sets aside the cash to buy the shares if it is exercised. Its profit at expiration is the short put's, $\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}-(K-S(T))^{+}$ per share.
> * The net Delta is the number of shares plus the options Delta, $\Delta_{\text{net}}=N_{S}+\Delta_{\text{opts}}$, and holding $N_{S}=-\Delta_{\text{opts}}$ shares makes a position delta-neutral. Covered calls and cash secured puts both have a positive net Delta.

## Covered Calls
A __covered call__ owns shares and sells a call on them, collecting the premium. Investors use it when they expect the share price to stay flat or rise slightly. Let's watch a short video from SMB Capital: [Simple & Low Risk Investing Strategy](https://www.yout-ube.com/watch?v=Tid1rfVIc9c).

> __Naked calls:__ Selling a call without owning the shares is a __naked call__, which carries unlimited risk if the share price rises. The shares in a covered call cover that obligation.

You don't need to do this yourself. Covered call ETFs, such as the [Global X NASDAQ 100 Covered Call ETF (QYLD)](https://www.globalxetfs.com/funds/qyld/) and the [REX FANG & Innovation Equity Premium Income ETF (FEPI)](https://www.rexshares.com/fang/), run this strategy for you.

### Profit at Expiration
We buy shares at $S(0)$ and sell one call per $m$ shares, with strike $K$, premium $\mathcal{P}_{c}$, and expiration $T$. Per share, we pay $S(0)-\mathcal{P}_{c}$ today. At expiration, we own a share worth $S(T)$ and owe the call's payoff $(S(T)-K)^{+}$. As for a composite contract, the profit at expiration $\hat{P}(S(T))$ (USD/share) is the value at expiration minus today's cost grown with the accumulation factor $\mathcal{D}_{T,0}(g_{y})=e^{g_{y}T}$:
$$
\begin{align*}
\hat{P}(S(T)) &= S(T) - \left(S(T) - K\right)^{+} - \mathcal{D}_{T,0}(g_{y})\cdot\left(S(0) - \mathcal{P}_{c}\right)\\
&= \underbrace{\left(S(T) - \mathcal{D}_{T,0}(g_{y})\cdot S(0)\right)}_{\text{shares}} + \underbrace{\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c} - \left(S(T) - K\right)^{+}}_{\text{short call}}
\end{align*}
$$
We assume no dividends and no early exercise. The call's payoff is zero when $S(T)\leq K$, so:
$$
\boxed{
\hat{P}(S(T)) = \begin{cases}
\overbrace{K - \mathcal{D}_{T,0}(g_{y})\cdot S(0) + \mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c}}^{\text{realized gain}}, & \text{if } S(T) > K\;\text{(call exercised)}\\
\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c} + \underbrace{\left(S(T) - \mathcal{D}_{T,0}(g_{y})\cdot S(0)\right)}_{\text{unrealized gain}}, & \text{if } S(T) \leq K\;\text{(call expires worthless)}
\end{cases}}
$$
For $n$ contracts, multiply by $mn$. The two regimes are:

* __Call exercised, $S(T)>K$:__ We must sell the shares at $K$, so the whole profit is __realized__. It does not grow with $S(T)$: the call caps it.
* __Call expires worthless, $S(T)\leq K$:__ We keep the premium, a __realized gain__, and the shares. The gain (or loss) on the shares is __unrealized__ because we still own them.

> __What??__
>
> We measured the share gain from $S(0)$, the price we paid today. Suppose instead we bought the shares earlier, at a price $S^{\dagger}$. If the call is exercised, the realized gain on the shares over what we originally paid is $K-S^{\dagger}$. Given my background, I tend to think of $S^{\dagger}$ as the __cost basis__ of the stock position.

The maximum profit, breakeven $S^{\ddagger}$, and maximum loss are:

* __Maximum profit:__ $K-\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)-\mathcal{P}_{c}\right)$, when $S(T)\geq K$.
* __Breakeven:__ $S^{\ddagger}=\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)-\mathcal{P}_{c}\right)$.
* __Maximum loss:__ $\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)-\mathcal{P}_{c}\right)$, when $S(T)=0$.

Compared with owning the shares alone, the premium lowers the breakeven, and the call gives up the gains above the strike.

## Cash Secured Puts
A __cash secured put__ sells a put and sets aside enough cash to buy the shares if the put is exercised. Investors use it when they are willing to buy the shares at a lower price and want the premium in the meantime.

### Profit at Expiration
We sell one put with strike $K$, premium $\mathcal{P}_{p}$, and expiration $T$, and set aside $K$ in cash per share. The cash earns the risk-free rate whether or not we sell the put, so we leave its interest out of the profit. We again assume no early exercise. The premium grows to $\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}$ by expiration, and if the put is exercised, we pay $K$ for a share worth $S(T)$:
$$
\boxed{
\hat{P}(S(T)) = \begin{cases}
\overbrace{\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}}^{\text{realized gain}}, & \text{if } S(T) \geq K\;\text{(put expires worthless)}\\
\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p} + \underbrace{\left(S(T) - K\right)}_{\text{unrealized loss}}, & \text{if } S(T) < K\;\text{(put exercised)}
\end{cases}}
$$
This is the profit at expiration of a short put, $\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}-(K-S(T))^{+}$; the cash guarantees we can pay for the shares. The two regimes are:

* __Put expires worthless, $S(T)\geq K$:__ We keep the premium, a __realized gain__, and buy no shares.
* __Put exercised, $S(T)<K$:__ We must buy the shares at $K$, more than their market value $S(T)$. The loss on the shares, $K-S(T)$, is __unrealized__ because we still own them. The position still makes money if $S(T)$ is above the breakeven.

The maximum profit, breakeven $S^{\ddagger}$, and maximum loss are:

* __Maximum profit:__ $\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}$, when $S(T)\geq K$.
* __Breakeven:__ $S^{\ddagger}=K-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}$.
* __Maximum loss:__ $K-\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}$, when $S(T)=0$.

If the put is exercised, we own the shares at an effective price equal to the breakeven.

## Delta of Covered Calls and Cash Secured Puts
Previously, we used Delta ($\Delta$) as a sensitivity, an estimate of the probability of exercise, and the number of shares that hedges one contract. A portfolio's Delta is the sum of its parts' Deltas, so the same idea applies to a whole position. __Hedging__ offsets potential losses by taking an opposite position in a related __counter asset__, often the underlying shares. A position whose overall Delta is zero is insensitive to small changes in the share price.

### The Net Delta of a Portfolio
Consider a portfolio of cash $N_{C}$ (USD), $N_{S}$ shares, and the legs of a composite contract $\mathcal{C}$ on the same shares. For leg $i$, $\theta_{i}$ is the orientation ($1$ long, $-1$ short), $n_{i}$ is the number of contracts, $m$ is the contract multiplier (shares per contract), and $\mathcal{P}_{i}(S,t)$ is the premium (USD/share) when the share price is $S$ at time $t$. The value of the portfolio is:
$$
\Pi(S,t)=N_{C}+N_{S}\,S+m\sum_{i\in\mathcal{C}}\theta_{i}n_{i}\mathcal{P}_{i}(S,t)
$$
The net Delta is its sensitivity to the share price:
$$
\begin{align*}
\Delta_{\text{net}}(S,t)&=\frac{\partial \Pi}{\partial S} = \underbrace{\frac{\partial N_C}{\partial S}}_{=0} + \underbrace{\frac{\partial (N_S S)}{\partial S}}_{=N_S} + m\sum_{i\in\mathcal{C}} \theta_i n_i \underbrace{\frac{\partial \mathcal{P}_i}{\partial S}}_{=\Delta_i(S,t)}\\
&= N_S + \underbrace{m\sum_{i\in\mathcal{C}} \theta_i n_i \Delta_i(S,t)}_{= \Delta_{\text{opts}}(S,t)}
\end{align*}
$$
The shares contribute $N_{S}$ and the options contribute $\Delta_{\text{opts}}=m\,\Delta_{\mathcal{C}}$, where $\Delta_{\mathcal{C}}(S,t)=\sum_{i\in\mathcal{C}}\theta_{i}n_{i}\Delta_{i}(S,t)$ is the __composite Delta__, the Delta of the composite contract per share.

> __Delta-neutral position:__ A position is delta-neutral if $\Delta_{\text{net}}(S,t)=0$, which we get by holding $N_{S}=-\Delta_{\text{opts}}(S,t)$ shares. For example, short $10$ calls at $\Delta=0.6$ with $m=100$ have $\Delta_{\text{opts}}=10\times(-1)\times 0.6\times 100=-600$, so holding $N_{S}=600$ shares makes $\Delta_{\text{net}}=0$.

If leg $i$ is exercised or assigned at time $\tau\le T$, set $\Delta_{i}=0$ for $t\geq\tau$ and update the shares: $N_S \leftarrow N_S + \theta_i m n_i$ for a call and $N_S \leftarrow N_S - \theta_i m n_i$ for a put.

### Covered Call and Cash Secured Put Delta
A covered call holds $N_{S}=mn$ shares and is short $n$ calls with Delta $\Delta_{c}$. A cash secured put holds no shares and is short $n$ puts with Delta $\Delta_{p}$; the cash does not contribute. Their net Deltas are:
$$
\Delta_{\text{net}}(S,t) = mn\left(1-\Delta_{c}(S,t)\right) \quad\text{(covered call)}\qquad
\Delta_{\text{net}}(S,t) = -mn\,\Delta_{p}(S,t) \quad\text{(cash secured put)}
$$
Since $0\leq\Delta_{c}\leq 1$ and $-1\leq\Delta_{p}\leq 0$, both net Deltas are positive: like the shares, both positions gain when the share price rises.

> __When are they delta-neutral?__
>
> * __Covered call:__ when $\Delta_{c}\approx 1$, i.e., the call is __deep in the money__ ($S(t)\gg K$) and behaves almost like the shares. The long shares ($+mn$) and the short call ($-mn$) then nearly offset.
> * __Cash secured put:__ when $\Delta_{p}\approx 0$, i.e., the put is __deep out of the money__ ($S(t)\gg K$). The put's value then changes very little with the share price, and the probability of exercise is negligible.

### Delta of the Common Composite Contracts
The same calculation applies to any composite contract. Using the common strike rule, short legs at Deltas of $\pm 0.25$, protecting long legs at $\pm 0.10$, and an at-the-money straddle at $\pm 0.5$, the composite Delta at entry and the shares that hedge one contract ($m=100$) are:

| Composite contract | Composite Delta, $\Delta_{\mathcal{C}}$ | Shares to hedge, $N_{S}$ |
|:--|:--|:--|
| Short put | $-(-0.25)=0.25$ | $-25$ |
| Short call | $-0.25$ | $+25$ |
| Put credit spread | $-0.10-(-0.25)=0.15$ | $-15$ |
| Call credit spread | $-0.25+0.10=-0.15$ | $+15$ |
| Short straddle | $-(-0.5)-0.5\approx0$ | $\approx0$ |
| Short strangle | $-(-0.25)-0.25=0$ | $0$ |
| Protected strangle | $-(-0.25)-0.25+0.10=0.10$ | $-10$ |
| Iron condor | $-0.10-(-0.25)-0.25+0.10=0$ | $0$ |

The sign of $\Delta_{\mathcal{C}}$ matches each contract's outlook. A Delta of zero is not zero risk: as the share price moves, Gamma pulls the Delta of a short strangle or iron condor away from zero, so keeping a position delta-neutral means updating $N_{S}$ as the share price and the time to expiration change.

## Summary
This lecture analyzes covered calls and cash secured puts, two strategies that generate income by selling options, and computes the net Delta of a portfolio of shares and options.

> __Key Takeaways:__
> 
> * __Covered calls combine stock ownership with short call options:__ We derived the profit at expiration of shares plus a short call. If the call is exercised, we sell the shares at the strike price and realize the gain or loss, so the call caps the profit. If it expires worthless, we keep the premium and the shares, and the premium lowers the breakeven.
> * __Cash secured puts generate income without initial stock ownership:__ We derived the profit at expiration of a short put backed by enough cash to buy the shares. If the put expires worthless, we keep the premium. If it is exercised, we buy the shares at the strike price, and we lose money only when the share price falls below the strike minus the premium grown to expiration.
> * __Portfolio Delta combines stock and options contributions:__ We derived the net Delta of a portfolio as the number of shares plus the weighted Deltas of the option legs, and found the number of shares that makes a position delta-neutral. Both covered calls and cash secured puts gain when the share price rises.

Understanding covered calls and cash secured puts lets us build income-generating positions from options combined with shares or cash.
