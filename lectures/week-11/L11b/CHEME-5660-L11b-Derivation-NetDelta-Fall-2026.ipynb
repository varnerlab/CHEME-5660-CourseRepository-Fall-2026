## Delta and Hedging
Previously, we used Delta ($\Delta$) as an estimate of the probability of exercise. Let's explore the other secret life of Delta: its use in hedging.

__Hedging__ offsets potential losses by taking an opposite position in a related __counter asset__, often the underlying shares. A position whose overall Delta is zero is insensitive to small changes in the share price.

### The Net Delta of a Portfolio
Consider a portfolio of cash $N_{C}$ (USD), $N_{S}$ shares, and the legs of a composite contract $\mathcal{C}$ on the same shares. For leg $i$, $\theta_{i}$ is the orientation ($1$ long, $-1$ short), $n_{i}$ is the number of contracts, $m$ is the contract multiplier (shares per contract), and $\mathcal{P}_{i}(S,t)$ is the premium (USD/share) when the share price is $S$ at time $t$. The wealth of the portfolio at time $t$ is:
$$
W_{t}=N_{C}+N_{S}\,S+m\sum_{i\in\mathcal{C}}\theta_{i}n_{i}\mathcal{P}_{i}(S,t)
$$
The net Delta is its sensitivity to the share price:
$$
\begin{align*}
\Delta_{\text{net}}(S,t)&=\frac{\partial W_{t}}{\partial S} = \underbrace{\frac{\partial N_C}{\partial S}}_{=0} + \underbrace{\frac{\partial (N_S S)}{\partial S}}_{=N_S} + m\sum_{i\in\mathcal{C}} \theta_i n_i \underbrace{\frac{\partial \mathcal{P}_i}{\partial S}}_{=\Delta_i(S,t)}\\
&= N_S + \underbrace{m\sum_{i\in\mathcal{C}} \theta_i n_i \Delta_i(S,t)}_{= \Delta_{\text{opts}}(S,t)}
\end{align*}
$$
The shares contribute $N_{S}$ and the options contribute $\Delta_{\text{opts}}=m\,\Delta_{\mathcal{C}}$, where $\Delta_{\mathcal{C}}(S,t)=\sum_{i\in\mathcal{C}}\theta_{i}n_{i}\Delta_{i}(S,t)$ is the __composite Delta__, the Delta of the composite contract per share.

> __Delta-neutral position:__ A position is delta-neutral if $\Delta_{\text{net}}(S,t)=0$, which we get by holding $N_{S}=-\Delta_{\text{opts}}(S,t)$ shares. For example, short $10$ calls at $\Delta=0.6$ with $m=100$ have $\Delta_{\text{opts}}=100\times(-1)\times 10\times 0.6=-600$, so holding $N_{S}=600$ shares makes $\Delta_{\text{net}}=0$.

If leg $i$ is exercised or assigned at time $\tau\le T$, set $\Delta_{i}=0$ for $t\geq\tau$ and update the shares: $N_S \leftarrow N_S + \theta_i m n_i$ for a call and $N_S \leftarrow N_S - \theta_i m n_i$ for a put.

## Covered Call and Collar Delta
Let's apply the net Delta to the two positions from the lecture. A covered call holds $N_{S}=mn$ shares and is short $n$ calls, each with Delta $\Delta_{c}$. A collar holds the same shares and short calls, with the calls at the strike $K_{2}$, and adds $n$ long puts at the strike $K_{1}$, each with Delta $\Delta_{p}$. The cash does not contribute. Their net Deltas are:
$$
\Delta_{\text{net}}(S,t) = mn\left(1-\Delta_{c}(S,t)\right) \quad\text{(covered call)}\qquad
\Delta_{\text{net}}(S,t) = mn\left(1+\Delta_{p}(S,t)-\Delta_{c}(S,t)\right) \quad\text{(collar)}
$$
Since $0\leq\Delta_{c}\leq 1$, the covered call's net Delta is never negative: like the shares, it does not lose value when the share price rises. Since $-1\leq\Delta_{p}\leq 0$, the collar's long put and short call can only lower its net Delta from the $mn$ of the shares alone.

> __When are they delta-neutral?__
>
> * __Covered call:__ when $\Delta_{c}\approx 1$, i.e., the call is __deep in the money__ ($S(t)\gg K$) and behaves almost like the shares. The long shares ($+mn$) and the short call ($-mn$) then nearly offset.
> * __Collar:__ beyond either strike. Far below the floor ($S(t)\ll K_{1}$), the put is deep in the money, $\Delta_{p}\approx -1$ and $\Delta_{c}\approx 0$. Far above the cap ($S(t)\gg K_{2}$), the call is deep in the money, $\Delta_{c}\approx 1$ and $\Delta_{p}\approx 0$. Either way, the collar's value barely changes with the share price.

## Delta of the Common Composite Contracts
The same calculation applies to any composite contract from [L11a](../L11a/CHEME-5660-L11a-Lecture-CompositeContracts-Fall-2026.ipynb). Suppose we choose the strikes by Delta, a common rule of thumb: the short legs at Deltas of $\pm 0.25$, the protecting long legs at $\pm 0.10$, and the straddle at the money, at about $\pm 0.5$. The composite Delta at entry and the shares that hedge one contract ($m=100$) are:

| Composite contract | Composite Delta, $\Delta_{\mathcal{C}}$ | Shares to hedge, $N_{S}$ |
|:--|:--|:--|
| Short put | $-(-0.25)=0.25$ | $-25$ |
| Short call | $-0.25$ | $+25$ |
| Put credit spread | $-0.10-(-0.25)=0.15$ | $-15$ |
| Call credit spread | $-0.25+0.10=-0.15$ | $+15$ |
| Short straddle | $-(-0.5)-0.5\approx0$ | $\approx0$ |
| Short strangle | $-(-0.25)-0.25=0$ | $0$ |
| Protected strangle | $-(-0.25)-0.25+0.10=0.10$ | $-10$ |
| Iron condor | $-0.10-(-0.25)-0.25+0.10=0$ | $0$ |

The sign of $\Delta_{\mathcal{C}}$ matches each contract's outlook. A Delta of zero is not zero risk: as the share price moves, Gamma pulls the Delta of a short strangle or iron condor away from zero, so keeping a position delta-neutral means updating $N_{S}$ as the share price and the time to expiration change.

## Summary
We derived the net Delta of a portfolio of shares and options and applied it to covered calls, collars, and the common composite contracts.

> __Key Takeaways:__
>
> * __Portfolio Delta combines stock and options contributions:__ We differentiated the wealth of a portfolio of cash, shares, and options with respect to the share price. The cash drops out, the shares contribute $N_{S}$, and the options contribute the weighted Deltas of their legs, $\Delta_{\text{opts}}$.
> * __Holding shares can cancel the options Delta:__ We found that holding $N_{S}=-\Delta_{\text{opts}}$ shares makes a position delta-neutral, and that an exercised or assigned leg moves its shares into $N_{S}$. A covered call's net Delta is never negative, and a collar's put and call can only lower its net Delta from that of the shares.
> * __Zero Delta is not zero risk:__ We computed the composite Delta of the common composite contracts at entry. Several start near zero, but Gamma moves the Delta as the share price changes, so a delta-neutral position must update $N_{S}$ as the share price and the time to expiration change.
