# Example: Analysis of Covered Call Positions on AMD
In this example, we analyze a covered call position on Advanced Micro Devices, Inc. (AMD) using our `AMD` options chain. A covered call is a strategy in which equity is purchased and call contracts are sold on a share-for-share basis against the equity.

> __Learning Objectives:__
>
> By the end of this example, you should be able to:
> * __Compute the profit of a covered call:__ Combine long shares with a short call into one profit at expiration, and compute the breakeven, maximum profit, and maximum loss.
> * __Estimate the probabilities of the outcomes:__ Simulate the share price at expiration with geometric Brownian motion, and compute the probabilities of a loss, of the maximum profit, and of an opportunity loss, under our belief $\mathbb{P}$ and under the risk-neutral probabilities $\mathbb{Q}$.
> * __Evaluate the opportunity loss:__ Find the share price above which holding the shares alone earns more than the covered call, and compare the probability of passing it with the probability of the maximum profit.

## Task 2: Analysis of the Profit and Loss of the Covered Call at Expiration
In this task, let's compute the breakeven point on this position and the profit and loss diagram for the covered call position at expiration. The covered call consists of long shares and a short-call contract. The profit expression (per share for a single covered call instance) is given by:
$$
\begin{align*}
\hat{P}(S(T)) &= S(T) - \left(S(T) - K\right)^{+} - \mathcal{D}_{T,0}(g_{y})\cdot\left(S(0) - \mathcal{P}_{c}\right)\\
&= \underbrace{\left(S(T) - \mathcal{D}_{T,0}(g_{y})\cdot S(0)\right)}_{\text{shares}} + \underbrace{\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c} - \left(S(T) - K\right)^{+}}_{\text{short call}}
\end{align*}
$$
where $(S(T) - K)^{+} \equiv \max(S(T) - K,0)$ is the call payoff. The deal flow for different share prices at expiration (assuming no cash position, no extra shares owned, and no extrinsic value at expiration) can be written (on a per share basis) as:
$$
\boxed{
\begin{align*}
\hat{P}(S(T)) &= \begin{cases}
\overbrace{K - \mathcal{D}_{T,0}(g_{y})\cdot\left(S(0) - \mathcal{P}_{c}\right)}^{\text{realized gain}}, & \text{if } S(T) > K\;\text{(call exercised)}\\
\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c} + \underbrace{\left(S(T) - \mathcal{D}_{T,0}(g_{y})\cdot S(0)\right)}_{\text{unrealized gain (no share sale)}}, & \text{if } S(T) \leq K\;\text{(call expires worthless)}
\end{cases}
\end{align*}}
$$
If we already owned the shares at a cost basis $S^{\dagger}$, the realized gain on the shares when the call is exercised is $K-S^{\dagger}$. The max profit, max loss, and breakeven point are given by (maximum losses are listed as positive amounts):

```julia
max_profit, max_loss, B = let

    # setup
    K = call_contract_model.K; # short strike
    𝒫 = call_contract_model.premium;

    # scenarios -
    max_profit = K - 𝒟*(Sₒ - 𝒫); # max profit case: the call is exercised
    max_loss = 𝒟*(Sₒ - 𝒫); # max loss: share price goes to zero, a positive number
    B = 𝒟*(Sₒ - 𝒫); # breakeven: the net cost grown to expiration

    # return
    max_profit, max_loss, B
end;
```

Let's make a pretty table:

```julia
let

    # show the maximum profit, the breakeven, and the maximum loss -
    df = DataFrame(quantity = ["maximum profit, S(T) ≥ K", "breakeven S‡", "maximum loss, S(T) = 0"],
        value = [max_profit, B, max_loss], units = ["USD/share", "USD/share", "USD/share"]);
    pretty_table(df; column_labels = ["quantity", "value", "units"], alignment = [:l, :r, :l],
        formatters = [fmt__printf("%.2f", [2])], fit_table_in_display_horizontally = false,
        table_format = TextTableFormat(borders = text_table_borders__compact));
end
```

## Task 3: Probability of Outcomes
In this task, let's compute the probability of observing exit scenarios for our covered call position on `AMD`. We'll consider three scenarios:

> __Scenarios:__
>
> * __Scenario 1: Loss__. In this scenario, the share price at expiration $S(T)$ is below the breakeven $S^{\ddagger}$ (stored as `B` in the code). The probability of this scenario occurring is given by $\mathbb{P}(S(T)<S^{\ddagger}) = F_{S}\left(S^{\ddagger}\right)$.
> * __Scenario 2: Maximum profit__. In this scenario, the share price at expiration $S(T)$ is at or above the strike price $K$, so the call caps the profit. The probability of this scenario occurring is given by $\mathbb{P}(S(T)\geq K) = \mathbb{P}(S(T)>K) = 1 - F_{S}\left(K\right)$.
> * __Scenario 3: Upside opportunity loss__. In this scenario, the share price at expiration $S(T)$ is above the alternative opportunity point $O^{+}=K+\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c}$ (stored as `OP` in the code), the share price above which holding the shares alone earns more than the covered call. The probability of this scenario occurring is given by $\mathbb{P}(S(T)>O^{+}) = 1 - F_{S}\left(O^{+}\right)$.

To start, let's build a model of the future underlying share price at expiration $S(T)$. Build an instance of the [MyGeometricBrownianMotionEquityModel type](https://varnerlab.org/CHEME-5660-CourseRepository-Fall-2026/dev/equity/#VLQuantitativeFinancePackage.MyGeometricBrownianMotionEquityModel) which holds values for the drift `μ` and volatility `σ` parameters using a [custom build(...) method](https://varnerlab.org/CHEME-5660-CourseRepository-Fall-2026/dev/equity/#VLQuantitativeFinancePackage.build-Tuple{Type{MyGeometricBrownianMotionEquityModel},%20NamedTuple}). 

We'll save the model in the `model::MyGeometricBrownianMotionEquityModel` variable, and the implied volatility value in the `IV::Float64` variable.

```julia
model, IV = let

    # Step 1: get the implied volatility 
    IV = call_leg_data[:IV];

    # Step 2: build the GBM model (risk-neutral pricing)
    model = build(MyGeometricBrownianMotionEquityModel, (
        μ = risk_free_rate, # under Q drift parameter is the risk-free rate gᵧ (1/yr)
        σ = IV # under Q volatility parameter is the implied volatility from the options chain
    ));

    model, IV
end;
```

Now that we have a populated [MyGeometricBrownianMotionEquityModel instance](https://varnerlab.org/CHEME-5660-CourseRepository-Fall-2026/dev/equity/#VLQuantitativeFinancePackage.MyGeometricBrownianMotionEquityModel) holding the `AMD` data, let's generate `number_of_samples` possible future share price values using the [sample(...) function exported by the VLQuantitativeFinancePackage.jl package](https://varnerlab.org/CHEME-5660-CourseRepository-Fall-2026/dev/equity/#VLQuantitativeFinancePackage.sample-Tuple{MyMultipleAssetGeometricBrownianMotionEquityModel,%20NamedTuple}).
* The [sample method](https://varnerlab.org/CHEME-5660-CourseRepository-Fall-2026/dev/equity/#VLQuantitativeFinancePackage.sample-Tuple{MyMultipleAssetGeometricBrownianMotionEquityModel,%20NamedTuple}) takes a `model::MyGeometricBrownianMotionEquityModel` instance as the first argument, along with the time values that we wish to sample: `T₁::Float64` is the start time, `T₂::Float64` is the stop time (in units of `years`), `Δt::Float64` is the time step (in units of `years`), and `Sₒ::Float64` denotes the initial share price (in units of `USD/share`). Finally, we pass in the number of samples we want to generate in the `number_of_paths` argument.
* The [sample method](https://varnerlab.org/CHEME-5660-CourseRepository-Fall-2026/dev/equity/#VLQuantitativeFinancePackage.sample-Tuple{MyMultipleAssetGeometricBrownianMotionEquityModel,%20NamedTuple}) generates `number_of_paths` sample paths from $T_{1}\rightarrow{T}_{2}$ in steps of $\Delta{t}$. However, we only want the share prices corresponding to the duration of the contract, i.e., $T_{2} = \text{DTE}$. Thus, we pass the full `samples` array [using the Julia pipe operator](https://docs.julialang.org/en/v1/manual/functions/#Function-composition-and-piping) to a slicing operation where we select columns `2:end` of the last row (col 1 holds the time). We store the `number_of_paths` samples of the share price at $T_{2}$ in the `endpoint::Array{Float64,1}` variable.

So what do we get?

```julia
endpoint = let
    number_of_samples = 25000; # TODO: adjust me!
    endpoint = VLQuantitativeFinancePackage.sample(model, (
        T₁ = 0.0,
        T₂ = (DTE)*(Δt),
        Δt = Δt,
        Sₒ = Sₒ
    ), number_of_paths = number_of_samples) |> samples -> samples[end,2:end]; # get last row from col 2 -> number_of_paths

    endpoint;
end;
```

We've previously shown that sampling a GBM model emits a LogNormal share price distribution. Fit a [LogNormal distribution exported by the Distributions.jl package](https://juliastats.org/Distributions.jl/stable/univariate/#Distributions.LogNormal) using the [fit_mle(...) function exported by Distributions.jl](https://juliastats.org/Distributions.jl/stable/fit/#Distributions.fit_mle-Tuple{Any,%20Any}), and save this in the `d_gbm::LogNormal` variable.

```julia
d_gbm = fit_mle(LogNormal, endpoint);
```

### Risk-neutral probabilities $\mathbb{Q}$
Next, estimate [the cumulative distribution function (CDF)](https://en.wikipedia.org/wiki/Cumulative_distribution_function) for the risk-neutral share price distribution. 
We can query the cumulative distribution function $F_{S}(\star)$ for the risk-neutral distribution `d_gbm` using [the `cdf(...)` function exported by the Distributions.jl package](https://github.com/JuliaStats/Distributions.jl?tab=readme-ov-file).

Now that we have the cumulative distribution function for the risk-neutral share price distribution, let's compute the probability of observing each of the three scenarios outlined above.

> __Scenario 1: Loss__. In this scenario, the share price at expiration $S(T)$ is below the breakeven $S^{\ddagger}$. The probability of this scenario occurring is given by $\mathbb{P}(S(T)<S^{\ddagger}) = F_{S}\left(S^{\ddagger}\right)$.

Let's save the probability of this scenario occurring in the `p_downside_loss_gbm::Float64` variable.

```julia
p_downside_loss_gbm = let

    # compute P(S(T) < B) for the GBM model
    p_downside_loss_gbm = cdf(d_gbm, B);

    # return data -
    p_downside_loss_gbm
end;
println("Scenario 1: Probability of loss under ℚ = $(p_downside_loss_gbm)")
```

Next, let's compute the probability of maximum profit.

> __Scenario 2: Maximum profit__. In this scenario, the share price at expiration $S(T)$ is at or above the short call strike price $K$. The probability of this scenario occurring is given by $\mathbb{P}(S(T)\geq K) = \mathbb{P}(S(T)>K) = 1 - F_{S}\left(K\right)$.

We'll save the probability of this scenario occurring in the `p_max_profit_gbm::Float64` variable.

```julia
p_max_profit_gbm = let

    # initialize
    K = call_contract_model.K; # strike price (USD/share)

    # compute the probability P(S(T) ≥ K) for the GBM
    p_max_profit_gbm = 1 - cdf(d_gbm, K);

    # return data -
    p_max_profit_gbm
end;
println("Scenario 2: Probability of max profit under ℚ = $(p_max_profit_gbm)")
```

Would we be better off holding the shares instead of selling the covered call? Let's compute the probability of opportunity loss.

> __Scenario 3: Upside opportunity loss__. In this scenario, the share price at expiration $S(T)$ is above the alternative opportunity point $O^{+}=K+\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c}$. The probability of this scenario occurring is given by $\mathbb{P}(S(T)>O^{+}) = 1 - F_{S}\left(O^{+}\right)$.

We'll save the probability of this scenario occurring in the `p_upside_opp_loss::Float64` variable.

```julia
p_upside_opp_loss = let

    # initialize
    K = call_contract_model.K; # strike price (USD/share)
    OP = K + 𝒟*call_contract_model.premium; # O⁺: above it, the shares alone earn more (USD/share)

    # compute the probability P(S(T) > O⁺) for the GBM
    p_upside_opp_loss = 1 - cdf(d_gbm, OP)

    # return data -
    p_upside_opp_loss
end;
println("Scenario 3: Probability of upside opportunity loss under ℚ = $(p_upside_opp_loss)")
```

Let's summarize our probability results in a table.

```julia
let

    # initialize -
    df = DataFrame();

    # populate the data frame -
    row_df = (
        ticker = ticker,
        Sₒ = Sₒ,
        DTE = DTE,
        strike_price_K = call_contract_model.K,
        premium = call_contract_model.premium,
        max_profit = max_profit,
        breakeven_B = B,
        IV = IV,
        p_downside_loss_gbm = p_downside_loss_gbm,
        p_max_profit_gbm = p_max_profit_gbm,
        p_upside_opp_loss = p_upside_opp_loss
    );
    push!(df, row_df); # store the row

    # make a table -
    pretty_table(
         df;
         fit_table_in_display_horizontally = false,
         
         table_format = TextTableFormat(borders = text_table_borders__compact)
    );
end
```

### Our belief $\mathbb{P}$
So far, we used the risk-neutral probabilities $\mathbb{Q}$, the probabilities inside the market price. Our belief $\mathbb{P}$ uses the SAGBM parameters instead: geometric Brownian motion with drift $\hat{\mu}=\hat{\mu}_{g}+\hat{\sigma}^{2}/2$ and volatility $\hat{\sigma}$, where $\hat{\mu}_{g}$ (the `drift` column) and $\hat{\sigma}$ (the `volatility` column) are the AMD estimates we loaded earlier. 

Let's compute the probabilities of the three scenarios for two distributions of the share price at expiration:
* __Our belief $\mathbb{P}$:__ log-normal, with drift $\hat{\mu}$ and volatility $\hat{\sigma}$.
* __Risk-neutral $\mathbb{Q}$:__ log-normal, with drift $g_{y}$ and the call's implied volatility, the distribution `d_gbm` above.

So what do we get?

```julia
share_price_distributions = let

    # initialize: the two share price distributions at expiration -
    T = DTE*Δt; # time to expiration (years)
    K = call_contract_model.K; # strike price (USD/share)
    OP = K + 𝒟*call_contract_model.premium; # O⁺: above it, the shares alone earn more (USD/share)
    amd = only(eachrow(filter(:ticker => ==("AMD"), parameters_df))); # AMD's SAGBM estimates
    μ̂_g = amd.drift; # mean growth rate (1/yr)
    σ̂ = amd.volatility; # historical volatility (1/sqrt(yr))
    μ̂ = μ̂_g + σ̂^2/2; # GBM drift of our belief ℙ (1/yr)
    cases = [ # (label, distribution of S(T)) for each row of the table
        ("our belief ℙ", LogNormal(log(Sₒ) + (μ̂ - σ̂^2/2)*T, σ̂*sqrt(T))),
        ("risk-neutral ℚ", d_gbm)
    ];

    # the probability of each scenario: loss below B, maximum profit above K, opportunity loss above O⁺ -
    df = DataFrame(case = String[], p_loss = Float64[], p_max_profit = Float64[], p_opp_loss = Float64[]); # one row per distribution
    for (label, d) ∈ cases
        push!(df, (label, cdf(d, B), 1 - cdf(d, K), 1 - cdf(d, OP)));
    end

    # show the results -
    pretty_table(df; column_labels = ["distribution of S(T)", "P(loss)", "P(max profit)", "P(opportunity loss)"],
        alignment = [:l, :r, :r, :r], formatters = [fmt__printf("%.4f", [2, 3, 4])],
        table_format = TextTableFormat(borders = text_table_borders__compact),
        fit_table_in_display_horizontally = false);
    println("our belief ℙ: μ̂_g = $(round(μ̂_g, digits = 4)), σ̂ = $(round(σ̂, digits = 4)), μ̂ = $(round(μ̂, digits = 4)) (1/yr)");
    println("risk-neutral ℚ: gᵧ = $(round(risk_free_rate, digits = 4)) (1/yr), IV = $(IV)");

    # checks: the ℚ row matches the probabilities above, and O⁺ > K makes an opportunity loss less likely than the maximum profit -
    @assert isapprox(df.p_loss[2], p_downside_loss_gbm; atol = 1e-12) "the ℚ row does not match d_gbm"
    @assert all(df.p_opp_loss .< df.p_max_profit) "P(opportunity loss) should be smaller than P(max profit)"

    cases; # return the (label, distribution) pairs
end;
```

> __What do we see?__
>
> * __Loss:__ Our belief $\mathbb{P}$ gives a lower probability of a loss compared to $\mathbb{Q}$. The actual drift $\hat{\mu}$ is much larger than $g_{y}$, and its volatility $\hat{\sigma}$ is lower than the implied volatility, so a share price below the breakeven is less likely under $\mathbb{P}$.
> * __Maximum profit and opportunity loss:__ The larger drift also makes a share price above the strike price more likely. Under $\mathbb{P}$, the covered call is more likely to earn its maximum profit, and the share price is more likely to finish above $O^{+}$, where the shares alone would have earned more.

`Unhide` the code block below to see how we plotted the cumulative distribution function $F_{S}(x)=\mathbb{P}(S(T)\leq x)$ of each distribution in the table (solid lines), and one minus it, $1-F_{S}(x)=\mathbb{P}(S(T)>x)$ (dashed lines, same color).

> __What do we see?__ The navy curves ($\mathbb{P}$) sit to the right of the blue curves ($\mathbb{Q}$). At the breakeven $S^{\ddagger}$ (green line), the solid navy curve is lower than the solid blue curve, so a loss is less likely under $\mathbb{P}$. At $K$ (purple line) and $O^{+}$ (gray line), the dashed navy curve is higher than the dashed blue curve, so the maximum profit and an opportunity loss are more likely under $\mathbb{P}$.

How do the two distributions compare?

```julia
let

    # initialize: the strike, the opportunity point, and the share prices at expiration to plot -
    K = call_contract_model.K; # strike price (USD/share)
    OP = K + 𝒟*call_contract_model.premium; # O⁺ (USD/share)
    x = range(minimum(quantile(d, 0.005) for (_, d) ∈ share_price_distributions),
        stop = maximum(quantile(d, 0.995) for (_, d) ∈ share_price_distributions), length = 500) |> collect; # share price at expiration (USD/share)
    colors = [:navy, :deepskyblue1]; # our belief ℙ, risk-neutral ℚ (the L11a colors)

    # plot: F_S(x) = P(S(T) ≤ x) solid, and 1 - F_S(x) = P(S(T) > x) dashed, in the same color for each row of the table -
    p = plot(bg="gray95", background_color_outside="white", framestyle = :box, fg_legend = :transparent,
        xminorticks = true, yminorticks = true, minorgrid = true, legend = :right, legendfontsize = 8);
    for ((label, d), c) ∈ zip(share_price_distributions, colors)
        plot!(p, x, cdf.(d, x), lw = 3, c = c, label = "$(label): P(S(T) ≤ x)");
        plot!(p, x, 1 .- cdf.(d, x), lw = 3, c = c, ls = :dash, label = "$(label): P(S(T) > x)");
    end

    # read each table entry across to the probability axis: P(loss) on the solid curve at S‡,
    # P(max profit) and P(opportunity loss) on the dashed curve at K and O⁺ -
    for ((_, d), c) ∈ zip(share_price_distributions, colors)
        for (v, h) ∈ ((B, cdf(d, B)), (K, 1 - cdf(d, K)), (OP, 1 - cdf(d, OP))) # (share price, height) (USD/share, -)
            plot!(p, [x[1], v], [h, h], lw = 1, c = c, ls = :dot, label = "");
            scatter!(p, [v], [h], ms = 5, c = c, msc = :black, msw = 1, label = ""); # where the line meets the curve, outlined
        end
    end
    xlims!(p, (x[1], x[end])); # start the plot at x[1], so the horizontal lines reach the axis

    # mark the breakeven, the strike, and the opportunity point -
    vline!(p, [B], lw = 2, c = :green, ls = :dash, label = "S‡ = $(round(B, digits = 2)) USD/share");
    vline!(p, [K], lw = 1, c = :purple, ls = :dash, label = "K = $(K) USD/share");
    vline!(p, [OP], lw = 1, c = :gray40, ls = :dash, label = "O⁺ = $(round(OP, digits = 2)) USD/share");

    # labels -
    xlabel!(p, "AMD share price at expiration, x (USD/share)", fontsize=18);
    ylabel!(p, "Probability", fontsize=18);
    p # return
end
```

## Summary
In this example, we bought `AMD` shares and sold a call on them, computed the covered call's breakeven, maximum profit, and maximum loss, and estimated the probabilities of a loss, of the maximum profit, and of an opportunity loss under the risk-neutral probabilities and under our belief.

> __Key Takeaways:__
>
> * __The premium lowers the breakeven and the call caps the profit:__ We combined long shares and a short call into one profit at expiration. The premium lowers the breakeven, and the call caps the profit at the strike price. The maximum loss equals the breakeven, reached if the share price falls to zero.
> * __Our belief and the risk-neutral probabilities give different answers:__ We simulated the share price at expiration and read the probability of each outcome from its cumulative distribution. Under our belief, built from AMD's estimated drift and volatility, a loss was less likely and the maximum profit more likely than under the risk-neutral probabilities.
> * __Selling the call gives up the gains above the opportunity point:__ We found the share price above which the shares alone earn more than the covered call. It sits above the strike price, so an opportunity loss is less likely than the maximum profit.

Understanding covered call positions enables traders to generate income from equity holdings while accepting limited upside potential.
