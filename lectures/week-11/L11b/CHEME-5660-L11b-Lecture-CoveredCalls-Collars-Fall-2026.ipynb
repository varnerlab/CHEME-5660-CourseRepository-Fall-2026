# L11b: Covered Calls and Protective Collars
In this lecture, we analyze two strategies that combine shares with options: covered calls, which sell a call on the shares to collect its premium, and protective collars, which add a put to protect the shares. What can each trade earn and lose, and how does its value change with the share price?

> __Learning Objectives:__
> 
> By the end of this lecture, you should be able to:
> * __Analyze a covered call:__ Derive the profit at expiration of shares plus a short call, identify the two regimes based on whether the call is exercised or expires worthless, and compute the breakeven, maximum profit, and maximum loss.
> * __Analyze a protective collar:__ Derive the profit at expiration of shares plus a long put and a short call, identify the three regimes based on which option, if any, is exercised, and compute the breakeven, maximum profit, and maximum loss.
> * __Calculate the net Delta of a position:__ Compute the net Delta of a portfolio of shares and options by adding the shares to the weighted Deltas of the option legs, and find the number of shares that makes a position delta-neutral.

Some results in this lecture are core, and some are here to build understanding:

> __What you need to know:__
>
> __Core concepts:__
> * A covered call owns $m$ shares for each call it sells. Its profit at expiration is $\min(S(T),K)-\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)-\mathcal{P}_{c}\right)$ per share: the premium lowers the breakeven, and the call caps the profit at the strike.
> * A protective collar adds a long put at $K_{1}$ to a covered call with its call at $K_{2}>K_{1}$. Its profit at expiration is $S(T)+(K_{1}-S(T))^{+}-(S(T)-K_{2})^{+}-\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)+\mathcal{P}_{1}-\mathcal{P}_{2}\right)$ per share: the put sets a floor, and the call a cap.
> * The net Delta is the number of shares plus the options Delta, $\Delta_{\text{net}}=N_{S}+\Delta_{\text{opts}}$, and holding $N_{S}=-\Delta_{\text{opts}}$ shares makes a position delta-neutral. A covered call's net Delta is never negative, and a collar's put and call can only lower its net Delta from that of the shares.
>

## Covered Calls
A __covered call__ owns shares and sells a call on them, collecting the premium. Investors use it when they expect the share price to stay flat or rise slightly. Let's watch a short video from SMB Capital, a proprietary trading firm in New York: [Simple & Low Risk Investing Strategy](https://www.youtube.com/embed/Tid1rfVIc9c?rel=0). SMB's YouTube channel has [many more covered call videos](https://www.youtube.com/@smbcapital/search?query=covered%20call).

> __Naked calls:__ Selling a call without owning the shares is a __naked call__, which carries unlimited risk if the share price rises. The shares in a covered call cover that obligation.

You don't need to do this yourself. Covered call ETFs, such as the [Global X NASDAQ 100 Covered Call ETF (QYLD)](https://www.globalxetfs.com/funds/qyld/) and the [REX FANG & Innovation Equity Premium Income ETF (FEPI)](https://www.rexshares.com/fang/), run this strategy for you.

### Profit at Expiration
We buy shares at $S(0)$ and sell one call per $m$ shares, with strike $K$, premium $\mathcal{P}_{c}$, and expiration $T$. Per share, we pay $S(0)-\mathcal{P}_{c}$ today. At expiration, we own shares worth $S(T)$ and owe the call's payoff $(S(T)-K)^{+}$. As for a composite contract, the profit at expiration $\hat{P}(S(T))$ (USD/share) is the value at expiration minus today's cost grown with the accumulation factor $\mathcal{D}_{T,0}(g_{y})=e^{g_{y}T}$:
$$
\begin{align*}
\hat{P}(S(T)) &= S(T) - \left(S(T) - K\right)^{+} - \mathcal{D}_{T,0}(g_{y})\cdot\left(S(0) - \mathcal{P}_{c}\right)\\
&= \underbrace{\left(S(T) - \mathcal{D}_{T,0}(g_{y})\cdot S(0)\right)}_{\text{shares}} + \underbrace{\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c} - \left(S(T) - K\right)^{+}}_{\text{short call}}
\end{align*}
$$
We assume no dividends and no early exercise. The call's payoff is zero when $S(T)\leq K$, so:
$$
\boxed{
\hat{P}(S(T)) = \begin{cases}
\overbrace{K - \mathcal{D}_{T,0}(g_{y})\cdot S(0) + \mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c}}^{\text{realized gain}}, & \text{if } S(T) > K\;\text{(call exercised)}\\
\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c} + \underbrace{\left(S(T) - \mathcal{D}_{T,0}(g_{y})\cdot S(0)\right)}_{\text{unrealized gain}}, & \text{if } S(T) \leq K\;\text{(call expires worthless)}
\end{cases}}
$$
For $n$ contracts, multiply by $mn$. The two regimes are:

* __Call exercised, $S(T)>K$:__ We must sell the shares at $K$, so the whole profit is __realized__. It does not grow with $S(T)$: the call caps it.
* __Call expires worthless, $S(T)\leq K$:__ We keep the premium, a __realized gain__, and the shares. The gain (or loss) on the shares is __unrealized__ because we still own them.

> __What??__
>
> We measured the share gain from $S(0)$, the price we paid today. Suppose instead we bought the shares earlier, at a price $S^{\dagger}$. If the call is exercised, the realized gain on the shares over what we originally paid is $K-S^{\dagger}$. Given my background, I tend to think of $S^{\dagger}$ as the __cost basis__ of the stock position.

The maximum profit, breakeven $S^{\ddagger}$, and maximum loss are given below. Maximum losses are listed as positive amounts (USD/share):

* __Maximum profit:__ $K-\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)-\mathcal{P}_{c}\right)$, when $S(T)\geq K$.
* __Breakeven:__ $S^{\ddagger}=\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)-\mathcal{P}_{c}\right)$.
* __Maximum loss:__ $\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)-\mathcal{P}_{c}\right)$, when $S(T)=0$.

Compared with owning the shares alone, the premium lowers the breakeven, and the call gives up the gains above the strike.

## Protective Collars
A __protective collar__ adds a long put to a covered call: we own the shares, buy a put with a strike $K_{1}$ below the share price, and sell a call with a strike $K_{2}>K_{1}$ above it, all with the same expiration. The put sets a floor under the shares, the call caps them, and the call premium pays for some or all of the put. Investors use a collar to protect shares they want to keep holding.

### Profit at Expiration
We buy shares at $S(0)$, and for every $m$ shares we buy one put at $K_{1}$ with premium $\mathcal{P}_{1}$ and sell one call at $K_{2}$ with premium $\mathcal{P}_{2}$. Per share, we pay $S(0)+\mathcal{P}_{1}-\mathcal{P}_{2}$ today. At expiration, we own shares worth $S(T)$, the put pays $(K_{1}-S(T))^{+}$, and we owe the call's payoff $(S(T)-K_{2})^{+}$. As for the covered call, the profit at expiration $\hat{P}(S(T))$ (USD/share) is:
$$
\hat{P}(S(T)) = S(T) + \left(K_{1} - S(T)\right)^{+} - \left(S(T) - K_{2}\right)^{+} - \mathcal{D}_{T,0}(g_{y})\cdot\left(S(0) + \mathcal{P}_{1} - \mathcal{P}_{2}\right)
$$
We again assume no dividends and no early exercise. At most one of the two options pays at expiration, so:
$$
\boxed{
\hat{P}(S(T)) = \begin{cases}
K_{1} - \mathcal{D}_{T,0}(g_{y})\cdot\left(S(0) + \mathcal{P}_{1} - \mathcal{P}_{2}\right), & \text{if } S(T) < K_{1}\;\text{(put exercised)}\\
S(T) - \mathcal{D}_{T,0}(g_{y})\cdot\left(S(0) + \mathcal{P}_{1} - \mathcal{P}_{2}\right), & \text{if } K_{1} \leq S(T) \leq K_{2}\;\text{(both expire worthless)}\\
K_{2} - \mathcal{D}_{T,0}(g_{y})\cdot\left(S(0) + \mathcal{P}_{1} - \mathcal{P}_{2}\right), & \text{if } S(T) > K_{2}\;\text{(call exercised)}
\end{cases}}
$$
For $n$ contracts, multiply by $mn$. The three regimes are:

* __Put exercised, $S(T)<K_{1}$:__ We sell the shares at $K_{1}$, so the loss is __realized__ and stops growing: the put sets a floor.
* __Both options expire worthless, $K_{1}\leq S(T)\leq K_{2}$:__ We keep the shares, and the gain or loss on them is __unrealized__. The profit moves one for one with the share price.
* __Call exercised, $S(T)>K_{2}$:__ We must sell the shares at $K_{2}$, so the profit is __realized__ and capped, as for the covered call.

The maximum profit, breakeven $S^{\ddagger}$, and maximum loss, a positive amount (USD/share), are:

* __Maximum profit:__ $K_{2}-\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)+\mathcal{P}_{1}-\mathcal{P}_{2}\right)$, when $S(T)\geq K_{2}$.
* __Breakeven:__ $S^{\ddagger}=\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)+\mathcal{P}_{1}-\mathcal{P}_{2}\right)$, when it lies between $K_{1}$ and $K_{2}$.
* __Maximum loss:__ $\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)+\mathcal{P}_{1}-\mathcal{P}_{2}\right)-K_{1}$, when $S(T)\leq K_{1}$. If $\mathcal{D}_{T,0}(g_{y})\cdot\left(S(0)+\mathcal{P}_{1}-\mathcal{P}_{2}\right)\leq K_{1}$, the collar cannot lose, and there is no maximum loss.

When $\mathcal{P}_{1}=\mathcal{P}_{2}$, the call premium pays for the whole put: a __zero-cost collar__. Compared with the covered call, the put limits the loss, and it costs some of the call premium.

## Delta of Covered Calls and Collars
Previously, we used Delta ($\Delta$) as an estimate of the probability of exercise. Let's explore the other secret life of Delta: its use in hedging.

__Hedging__ offsets potential losses by taking an opposite position in a related __counter asset__, often the underlying shares. A position whose overall Delta is zero is insensitive to small changes in the share price.

### The Net Delta of a Portfolio
Consider a portfolio of cash $N_{C}$ (USD), $N_{S}$ shares, and the legs of a composite contract $\mathcal{C}$ on the same shares. For leg $i$, $\theta_{i}$ is the orientation ($1$ long, $-1$ short), $n_{i}$ is the number of contracts, $m$ is the contract multiplier (shares per contract), and $\mathcal{P}_{i}(S,t)$ is the premium (USD/share) when the share price is $S$ at time $t$. The wealth of the portfolio at time $t$ is:
$$
W_{t}=N_{C}+N_{S}\,S+m\sum_{i\in\mathcal{C}}\theta_{i}n_{i}\mathcal{P}_{i}(S,t)
$$
The net Delta is its sensitivity to the share price:
$$
\begin{align*}
\Delta_{\text{net}}(S,t)&=\frac{\partial W_{t}}{\partial S} = \underbrace{\frac{\partial N_C}{\partial S}}_{=0} + \underbrace{\frac{\partial (N_S S)}{\partial S}}_{=N_S} + m\sum_{i\in\mathcal{C}} \theta_i n_i \underbrace{\frac{\partial \mathcal{P}_i}{\partial S}}_{=\Delta_i(S,t)}\\
&= N_S + \underbrace{m\sum_{i\in\mathcal{C}} \theta_i n_i \Delta_i(S,t)}_{= \Delta_{\text{opts}}(S,t)}
\end{align*}
$$
The shares contribute $N_{S}$ and the options contribute $\Delta_{\text{opts}}=m\,\Delta_{\mathcal{C}}$, where $\Delta_{\mathcal{C}}(S,t)=\sum_{i\in\mathcal{C}}\theta_{i}n_{i}\Delta_{i}(S,t)$ is the __composite Delta__, the Delta of the composite contract per share.

> __Delta-neutral position:__ A position is delta-neutral if $\Delta_{\text{net}}(S,t)=0$, which we get by holding $N_{S}=-\Delta_{\text{opts}}(S,t)$ shares. For example, short $10$ calls at $\Delta=0.6$ with $m=100$ have $\Delta_{\text{opts}}=100\times(-1)\times 10\times 0.6=-600$, so holding $N_{S}=600$ shares makes $\Delta_{\text{net}}=0$.

If leg $i$ is exercised or assigned at time $\tau\le T$, set $\Delta_{i}=0$ for $t\geq\tau$ and update the shares: $N_S \leftarrow N_S + \theta_i m n_i$ for a call and $N_S \leftarrow N_S - \theta_i m n_i$ for a put.

### Covered Call and Collar Delta
A covered call holds $N_{S}=mn$ shares and is short $n$ calls with Delta $\Delta_{c}$. A collar adds $n$ long puts with Delta $\Delta_{p}$, so its call is the one at $K_{2}$ and its put the one at $K_{1}$. The cash does not contribute. Their net Deltas are:
$$
\Delta_{\text{net}}(S,t) = mn\left(1-\Delta_{c}(S,t)\right) \quad\text{(covered call)}\qquad
\Delta_{\text{net}}(S,t) = mn\left(1+\Delta_{p}(S,t)-\Delta_{c}(S,t)\right) \quad\text{(collar)}
$$
Since $0\leq\Delta_{c}\leq 1$, the covered call's net Delta is never negative: like the shares, it does not lose value when the share price rises. Since $-1\leq\Delta_{p}\leq 0$, the collar's long put and short call can only lower its net Delta from the $mn$ of the shares alone.

> __When are they delta-neutral?__
>
> * __Covered call:__ when $\Delta_{c}\approx 1$, i.e., the call is __deep in the money__ ($S(t)\gg K$) and behaves almost like the shares. The long shares ($+mn$) and the short call ($-mn$) then nearly offset.
> * __Collar:__ beyond either strike. Far below the floor ($S(t)\ll K_{1}$), the put is deep in the money, $\Delta_{p}\approx -1$ and $\Delta_{c}\approx 0$. Far above the cap ($S(t)\gg K_{2}$), the call is deep in the money, $\Delta_{c}\approx 1$ and $\Delta_{p}\approx 0$. Either way, the collar's value barely changes with the share price.

### Delta of the Common Composite Contracts
The same calculation applies to any composite contract. Suppose we choose the strikes by Delta, a common rule of thumb: the short legs at Deltas of $\pm 0.25$, the protecting long legs at $\pm 0.10$, and the straddle at the money, at about $\pm 0.5$. The composite Delta at entry and the shares that hedge one contract ($m=100$) are:

| Composite contract | Composite Delta, $\Delta_{\mathcal{C}}$ | Shares to hedge, $N_{S}$ |
|:--|:--|:--|
| Short put | $-(-0.25)=0.25$ | $-25$ |
| Short call | $-0.25$ | $+25$ |
| Put credit spread | $-0.10-(-0.25)=0.15$ | $-15$ |
| Call credit spread | $-0.25+0.10=-0.15$ | $+15$ |
| Short straddle | $-(-0.5)-0.5\approx0$ | $\approx0$ |
| Short strangle | $-(-0.25)-0.25=0$ | $0$ |
| Protected strangle | $-(-0.25)-0.25+0.10=0.10$ | $-10$ |
| Iron condor | $-0.10-(-0.25)-0.25+0.10=0$ | $0$ |

The sign of $\Delta_{\mathcal{C}}$ matches each contract's outlook. A Delta of zero is not zero risk: as the share price moves, Gamma pulls the Delta of a short strangle or iron condor away from zero, so keeping a position delta-neutral means updating $N_{S}$ as the share price and the time to expiration change.

## Summary
This lecture analyzes covered calls and protective collars, two strategies that combine shares with options, and computes the net Delta of a portfolio of shares and options.

> __Key Takeaways:__
> 
> * __Covered calls combine stock ownership with short call options:__ We derived the profit at expiration of shares plus a short call. If the call is exercised, we sell the shares at the strike price and realize the gain or loss, so the call caps the profit. If it expires worthless, we keep the premium and the shares, and the premium lowers the breakeven.
> * __Protective collars put a floor and a cap on the shares:__ We derived the profit at expiration of shares plus a long put and a short call. Below the put's strike, the put sets a floor under the loss; above the call's strike, the call caps the profit; in between, the position moves with the shares. The call premium pays for some or all of the put.
> * __Portfolio Delta combines stock and options contributions:__ We derived the net Delta of a portfolio as the number of shares plus the weighted Deltas of the option legs, and found the number of shares that makes a position delta-neutral. A covered call's net Delta is never negative, and a collar's put and call both reduce how much it gains or loses.

Understanding covered calls and protective collars lets us build positions that trade some of the upside of the shares for income or for protection.
