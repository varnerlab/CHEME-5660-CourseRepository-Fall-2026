## Notation: growth versus product price change
For a horizon $\Delta t$ measured in years, our course convention is

$$r_{t,\Delta t}=\log\left(\frac{S_{t+\Delta t}}{S_t}\right)=(\Delta t)g_{t,\Delta t}.$$

Option and ETF outcome diagrams generally use a cumulative __simple price change__. Denote it by

$$R_{t,\Delta t}=\frac{S_{t+\Delta t}-S_t}{S_t}=\exp(r_{t,\Delta t})-1=\exp((\Delta t)g_{t,\Delta t})-1.$$

A 90-day collar outcome or a two-year ETF cap is cumulative over its stated horizon, not an annual growth rate.
___

## The standard-listed protective collar
At $t=0$, combine one share, one long put with strike $K_P$, and one short call with strike $K_C$, where $K_P<S_0<K_C$. Ignoring dividends, financing, exercise frictions, and the contract multiplier, terminal value is

$$V_T=S_T+(K_P-S_T)^+-(S_T-K_C)^+.$$

Therefore,

$$V_T=\begin{cases}K_P,&S_T<K_P,\\S_T,&K_P\leq S_T\leq K_C,\\K_C,&S_T>K_C.\end{cases}$$

An ordinary standard equity or ETF option contract generally represents $m=100$ shares. A matched one-contract collar therefore uses 100 shares, one put, and one call with the same expiration. Corporate actions can alter the deliverable, so the contract specification must still be checked.
___

## A practical retail order ticket
For an existing 100-share position, buy to open one put and sell to open one call on the same underlying and expiration. For a new position, some brokers support a stock-plus-options combination order; otherwise legging the position introduces execution and interim-exposure risk.

Use executable prices when planning the trade:

* value the purchased put at its ask;
* value the sold call at its bid;
* submit a multi-leg net limit order when supported; and
* treat any midpoint improvement as execution benefit, not an assumption.

Verify the ticker, deliverable, call/put, strike, expiration, multiplier, option approval level, buying power, and ex-dividend date. Standard equity/ETF options are generally American-style, so early assignment is possible.
___

## Payoff is not profit
Let $P_0^{ask}$ be the put ask and $C_0^{bid}$ the call bid per share. The conservative initial cost of one matched standard collar is

$$\Pi_0=m\left(S_0+P_0^{ask}-C_0^{bid}\right),\qquad m=100.$$

Ignoring financing and dividends,

$$\text{maximum loss}=\Pi_0-mK_P,\qquad \text{maximum profit}=mK_C-\Pi_0,$$

and the interior breakeven price is $\Pi_0/m=S_0+P_0^{ask}-C_0^{bid}$. A so-called zero-cost collar only has $P_0\approx C_0$; it does not remove market risk, execution cost, assignment risk, taxes, or foregone upside.
___

## The listed chain constrains the design
Standard options are selected from available expirations and strikes. Raising $K_P$ improves the floor but generally makes the put more expensive. Lowering $K_C$ generally collects more call premium but caps upside sooner. A reproducible selection rule might maximize $K_C$ subject to a minimum $K_P$, maximum opening debit, minimum volume/open interest, and maximum bid--ask spread.

A conventional 10%-down/10%-up collar does __not__ provide a 100% downside buffer: it participates in the first 10% decline and is protected only below the put strike. A put near $S_0$ provides stronger protection but normally requires a larger debit or a lower call cap.
___

## Expiration and assignment

* Below $K_P$, the put can sell the matched shares at the floor strike.
* Between $K_P$ and $K_C$, both options normally expire without exercise and the shares remain.
* Above $K_C$, call assignment sells the shares at the cap strike.
* Before expiration, the short call can be assigned early, particularly near an ex-dividend date when its remaining extrinsic value is small.

The operator should decide before entry how to handle dividends, early assignment, expiration, closing, rolling, and any mismatch between the share count and option deliverables.

## TJUL is the comparison—not the required instrument
TJUL remains useful because it packages a defined outcome in an exchange-traded fund. The fund uses customized FLEX options on `SPY`; its disclosed construction uses purchased call exposure, a purchased put, and a sold call that creates the cap. Students do not need FLEX access: they can buy or analyze TJUL shares as a product, or build the self-managed collar using ordinary listed options.

For a full-buffer classroom abstraction, let $R_{SPY}$ be the complete-period cumulative simple `SPY` price change and $c$ the starting cap. Before fees and expenses,

$$R_{TJUL}^{gross}=\begin{cases}0,&R_{SPY}<0,\\R_{SPY},&0\leq R_{SPY}\leq c,\\c,&R_{SPY}>c.\end{cases}$$

This is an outcome map, not a daily guarantee or a valuation model for a mid-period purchaser.

## Standard collar versus TJUL

| Feature | Self-managed standard collar | TJUL |
|:--|:--|:--|
| Instrument accessed | Listed shares, put, and call | ETF shares |
| Options inside position | Standard listed options chosen by investor | FLEX options managed inside fund |
| Capital granularity | Usually 100-share blocks | Whole or fractional ETF shares, broker dependent |
| Floor and cap | Determined by available strikes, premiums, and expiration | Dated starting buffer and cap |
| Management | Investor handles execution, assignment, dividends, and rolls | Fund manages option package; investor manages entry/exit timing |
| Costs | Bid--ask spreads, commissions/fees, financing, taxes | Expense ratio, ETF spread, premium/discount, taxes |

TJUL references `SPY` __price__ return. A direct `SPY` holder may receive distributions, so compare complete wealth using consistent price-return or total-return conventions.

## Summary

> __Key Takeaways__
>
> * The core collar is retail-replicable with 100 shares, one ordinary listed put, and one ordinary listed call; FLEX options are not required.
> * Use the put ask and call bid for a conservative opening estimate, and distinguish payoff from profit.
> * Use $r=(\Delta t)g$ for course log growth and $R=\exp((\Delta t)g)-1$ for cumulative simple price change.
> * Standard strike availability determines the actual floor, cap, and debit; a typical OTM collar is not a full downside buffer.
> * TJUL is a comparison product whose fund manager uses FLEX options internally; investors can access the ETF without trading those options directly.
