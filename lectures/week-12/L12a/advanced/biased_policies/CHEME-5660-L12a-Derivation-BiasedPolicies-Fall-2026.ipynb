## Biased Policies
We use the minority game of the lecture with its price extension. Each of the $N$ agents holds $J$ policies. Policy $\pi_{ij}$ of agent $i$ gives the action $\pi_{ij}(\mu)$, buy ($+1$) or sell ($-1$), for each history $\mu$ of the last $m$ minority actions, and there are $P=2^{m}$ histories.

The attendance $A_{t}$ is the number of buyers minus the number of sellers in round $t$, and $\langle\cdot\rangle$ averages over the rounds. After round $t$, the score of every policy changes by $-\pi_{ij}(\mu_{t})A_{t}/P$.

The growth rate of the price is:

$$
g_{t+1}=\frac{A_{t}}{\lambda\,\Delta t}+\frac{\sigma_{S}}{\sqrt{\Delta t}}\,Z_{t+1}
$$

Here $\lambda$ is the liquidity, one round lasts $\Delta t$ years, $Z_{t+1}\sim\mathcal{N}(0,1)$ is the independent price noise, and $\sigma_{S}$ is its annualized volatility.

Suppose instead that each policy entry is buy with probability $b\geq 1/2$, so that the agents lean toward buying.

If the agents ignored their scores and chose their policies at random, the average attendance per agent would equal the expected value of a policy entry, the __bias__:

$$
\mathbb{E}\left[\pi_{ij}(\mu)\right]=b-(1-b)=2b-1
$$

How much of this bias can agents who follow their scores remove?
___

## The Threshold and the Locked Market
We consider the greedy game, in which every agent uses its best-scoring policy.

> __Theorem (bias in the greedy game):__ In the greedy game with $N$ agents and $J$ policies per agent, let each policy entry be buy with probability $b\geq 1/2$. For large $N$, the average attendance per agent, $\langle A\rangle/N$, can be close to zero only if $b$ is at or below the threshold:
>
> $$
> b\leq b_{c}=2^{-1/J}
> $$
>
> For $b>b_{c}$ and with probability close to one for large $N$, the attendance locks at a constant value, and the growth rate has a constant average:
>
> $$
> \frac{A}{N}\approx 2b^{J}-1\qquad\qquad\langle g\rangle\approx\frac{N}{\lambda\,\Delta t}\left(2b^{J}-1\right)
> $$
>
> with only the noise moving the growth rate around its average.

__Proof:__ Fix a history $\mu$. Each of an agent's $J$ policies is a table with one row for each history. The rows for $\mu$ in these $J$ tables either all agree or they do not, which sorts the agents into three groups:
* __Frozen buyers:__ the row for $\mu$ says buy in all $J$ tables, with probability $b^{J}$. The agent buys at $\mu$, whatever its scores.
* __Frozen sellers:__ the row for $\mu$ says sell in all $J$ tables, with probability $(1-b)^{J}$. The agent sells at $\mu$, whatever its scores.
* __Switchers:__ the rows for $\mu$ disagree, with probability $1-b^{J}-(1-b)^{J}$. The agent can buy or sell, depending on its scores.

An agent's group depends on the history, because only the rows for $\mu$ have to agree. For example, an agent with $J=2$ policies could be a frozen buyer at one history and a switcher at another.

For large $N$, the frozen agents add a fixed net buying of about $b^{J}-(1-b)^{J}$ per agent. Even if the switchers could choose freely, they could add at most $1-b^{J}-(1-b)^{J}$ per agent when all of them buy, and subtract at most the same amount when all of them sell.

Let's compare two buy probabilities for agents with $J=2$ policies, where the threshold is $b_{c}=1/\sqrt{2}\approx 0.71$:

| Quantity at a history | Expression | $b=0.6$ (below $b_{c}$) | $b=0.8$ (above $b_{c}$) |
|:---|:---:|:---:|:---:|
| Share of frozen buyers | $b^{J}$ | $0.36$ | $0.64$ |
| Share of frozen sellers | $(1-b)^{J}$ | $0.16$ | $0.04$ |
| Share of switchers | $1-b^{J}-(1-b)^{J}$ | $0.48$ | $0.32$ |
| Net buying of the frozen agents, per agent | $b^{J}-(1-b)^{J}$ | $0.20$ | $0.60$ |
| Attendance per agent if every switcher sells | $2b^{J}-1$ | $-0.28$ | $0.28$ |

The attendance per agent can be close to zero only if there are enough switchers to offset the frozen agents:

$$
1-b^{J}-(1-b)^{J}\;\geq\;b^{J}-(1-b)^{J}\quad\Longleftrightarrow\quad b^{J}\leq\frac{1}{2}\quad\Longleftrightarrow\quad b\leq 2^{-1/J}
$$

For $b>b_{c}$, the frozen buyers alone are a majority at every history, with probability close to one for large $N$. The minority action is then $-1$ in every round, even before any agent adapts. After $m$ rounds, the history is $(-1,\dots,-1)$, and it stays there.

At this history, every policy that says sell gains score each round and every policy that says buy loses score. The gap between any sell policy and any buy policy of an agent grows by $2A_{t}/P$ every round. As this score gap grows, each switcher eventually uses a sell policy.

With $X$ frozen buyers, the attendance then locks at $A=2X-N$. Because $X/N\approx b^{J}$ for large $N$, the $(1-b)^{J}$ terms cancel:

$$
\frac{A}{N}\approx\underbrace{\left[b^{J}-(1-b)^{J}\right]}_{\text{frozen agents}}-\underbrace{\left[1-b^{J}-(1-b)^{J}\right]}_{\text{switchers, all selling}}=2b^{J}-1
$$

Because the attendance no longer changes, the growth rate $g_{t+1}=\frac{A}{\lambda\,\Delta t}+\frac{\sigma_{S}}{\sqrt{\Delta t}}Z_{t+1}$ has the average $\frac{A}{\lambda\,\Delta t}\approx\frac{N}{\lambda\,\Delta t}\left(2b^{J}-1\right)$ and the variance $\sigma_{S}^{2}/\Delta t$ over time. $\blacksquare$
___

## What the Result Means
Three remarks on the result:
* __Adaptation removes part of the bias:__ Agents who chose at random would carry the full bias $2b-1$. Above $b_{c}$, agents who follow their scores bring it down to about $2b^{J}-1$. With $J=2$ and $b=0.8$, that is $0.28$ instead of $0.6$.

  Above $b_{c}$, a temperature $T>0$ only delays the lock. As the score gaps keep growing, the probability of using a buy policy at the locked history falls toward zero. Only at $T=\infty$, where the agents ignore their scores, does the full bias $2b-1$ remain.
* __Zero is possible, not guaranteed:__ Below $b_{c}$, the switchers could offset the frozen agents. However, each agent keeps one score per policy, which ties its choice at one history to its choices at all the others. Because the agent cannot lean toward selling at each history separately, some bias can survive.
* __The mean square is no longer the variance:__ With a bias, the average growth rate is not zero, and $\langle g^{2}\rangle=\operatorname{Var}(g)+\langle g\rangle^{2}$. In the locked market, the attendance adds nothing to the variance of the growth rate over time: the agents produce a steady trend, and all of that variance comes from the noise.

The [thermal minority game example](../../CHEME-5660-L12a-Example-ThermalMinorityGame-Fall-2026.ipynb) has a `buy_probability` constant, the $b$ of this notebook. Setting `buy_probability = 0.8` puts the game above the threshold, where the attendance plot becomes a flat line.

## Summary
We let the policy entries of the minority game lean toward buying, derived how much of that bias agents who follow their scores can remove, and found what happens to the market when they cannot remove all of it.

> __Key Takeaways:__
>
> * __Agents who ignore their scores carry the full bias:__ We computed the expected action of a policy entry when buy is more likely than sell. Agents who chose their policies at random would pass this whole bias into the attendance and give the price a drift.
> * __Only the agents who can switch push back:__ We sorted the agents at each history into those whose policies all agree and those who can switch. The switchers can offset the frozen agents only when buying is not too likely, which gave a threshold that rises with the number of policies per agent.
> * __Above the threshold, the market locks:__ We showed that the minority is then the sellers in every round. As a result, the history stops changing, and every switcher ends up selling.
>
>   The attendance locks at a constant value, the price follows a steady trend, and all the volatility of the growth rate comes from the noise.
