## The Maximum-Entropy Problem
An agent chooses among $J$ policies with scores $U_{1},\dots,U_{J}$, and it uses policy $\pi_{j}$ with probability $p_{j}$. The __entropy__ of the choice measures how spread out its probabilities are:

$$
\mathcal{H}(p)=-\sum_{j=1}^{J}p_{j}\ln p_{j}
$$

The entropy is zero when the agent picks one policy for certain, and largest, $\ln J$, when every policy has probability $1/J$.

If we require the choice to reach some expected score but assume nothing else about it, the choice that commits to the least is the one with the largest entropy. We fix a target expected score $\bar{U}$ strictly between the average and the largest score, and look for the probabilities that solve:

$$
\max_{p}\;\mathcal{H}(p)\quad\text{subject to}\quad\sum_{j=1}^{J}p_{j}=1,\qquad\sum_{j=1}^{J}p_{j}U_{j}=\bar{U}
$$
___

## The Boltzmann Rule
> __Theorem (maximum-entropy choice):__ Among all probabilities $p_{1},\dots,p_{J}$ that sum to one and give $\sum_{j}p_{j}U_{j}=\bar{U}$, the entropy $\mathcal{H}(p)$ is largest for the Boltzmann rule:
>
> $$
> p_{j}=\frac{1}{\mathcal{Z}}\exp\left(\frac{U_{j}}{T}\right)\qquad\text{where}\qquad\mathcal{Z}=\sum_{k=1}^{J}\exp\left(\frac{U_{k}}{T}\right)
> $$
>
> The partition function $\mathcal{Z}$ makes the probabilities sum to one. The temperature $T>0$, in the units of the scores, is the value at which these probabilities give the target expected score $\bar{U}$.

__Proof:__ We use __Lagrange multipliers__, one number attached to each constraint, which turn the constrained maximization into finding where the derivative of a single function, the Lagrangian $\mathcal{L}$, is zero. With $\phi_{0}$ for the normalization and $\phi_{1}$ for the expected score:

$$
\mathcal{L}=-\sum_{j}p_{j}\ln p_{j}-\phi_{0}\Bigl(\sum_{j}p_{j}-1\Bigr)+\phi_{1}\Bigl(\sum_{j}p_{j}U_{j}-\bar{U}\Bigr)
\qquad
\frac{\partial\mathcal{L}}{\partial p_{j}}=-\ln p_{j}-1-\phi_{0}+\phi_{1}U_{j}=0
$$

So $p_{j}=e^{-1-\phi_{0}}e^{\phi_{1}U_{j}}$. Because the factor $e^{-1-\phi_{0}}$ is the same for every policy, the normalization sets it to $1/\mathcal{Z}$. We define $T=1/\phi_{1}$.

Because the second derivatives, $-1/p_{j}$, are negative and the mixed derivatives are zero, this point is the maximum. $\blacksquare$
___

## What the Temperature Means
Two remarks on the result:
* __Lowering $T$ raises the expected score:__ Substituting the Boltzmann rule into the constraint gives the expected score as a function of the temperature, $\bar{U}=\frac{1}{\mathcal{Z}}\sum_{j}U_{j}\exp(U_{j}/T)$. Its derivative is negative unless all scores are equal:

  $$
  \begin{align*}
  \frac{d\mathcal{Z}}{dT} & = -\frac{1}{T^{2}}\sum_{j}U_{j}\exp\left(\frac{U_{j}}{T}\right) && \text{differentiate } \mathcal{Z} \text{ term by term}\\
  & = -\frac{\mathcal{Z}\,\bar{U}}{T^{2}} && \text{use the expression for } \bar{U}\\
  \frac{d\bar{U}}{dT} & = -\frac{1}{\mathcal{Z}\,T^{2}}\sum_{j}U_{j}^{2}\exp\left(\frac{U_{j}}{T}\right)-\frac{\bar{U}}{\mathcal{Z}}\frac{d\mathcal{Z}}{dT} && \text{product rule on } \bar{U}\\
  & = -\frac{1}{T^{2}}\sum_{j}p_{j}U_{j}^{2}+\frac{\bar{U}^{2}}{T^{2}} && \text{substitute } p_{j} \text{ and } d\mathcal{Z}/dT\\
  & = -\frac{1}{T^{2}}\operatorname{Var}_{p}(U) && \text{where } \operatorname{Var}_{p}(U)=\sum_{j}p_{j}U_{j}^{2}-\bar{U}^{2}
  \end{align*}
  $$

  As $T\to\infty$, the expected score is the average score. As $T\to 0$, it approaches the largest score. Because the expected score falls continuously between these limits, exactly one $T>0$ gives each target $\bar{U}$ strictly between them.
* __Temperature in thermodynamics:__ Raising $\bar{U}$ forces the choice onto the best policies and lowers the maximum entropy at the rate $-d\mathcal{H}/d\bar{U}=\phi_{1}=1/T$:

  $$
  \begin{align*}
  \mathcal{H} & = -\sum_{j}p_{j}\left(\frac{U_{j}}{T}-\ln\mathcal{Z}\right) && \text{substitute the Boltzmann rule for } \ln p_{j}\\
  & = -\frac{\bar{U}}{T}+\ln\mathcal{Z} && \text{use the two constraints}\\
  \frac{d\mathcal{H}}{dT} & = -\frac{1}{T}\frac{d\bar{U}}{dT}+\frac{\bar{U}}{T^{2}}+\frac{1}{\mathcal{Z}}\frac{d\mathcal{Z}}{dT} && \text{differentiate with respect to } T\\
  & = -\frac{1}{T}\frac{d\bar{U}}{dT} && \text{substitute } d\mathcal{Z}/dT\\
  \frac{d\mathcal{H}}{d\bar{U}} & = \frac{d\mathcal{H}/dT}{d\bar{U}/dT}=-\frac{1}{T} && \text{chain rule}
  \end{align*}
  $$

  Thermodynamics defines temperature the same way, with the energy in place of minus the score.
___

## Summary
We wrote the choice of a policy as a constrained maximization of entropy, solved it with Lagrange multipliers, and interpreted the temperature that the solution introduces.

> __Key Takeaways:__
>
> * __Maximum entropy is the least committed choice:__ We wrote the entropy of a choice and required the probabilities to sum to one and to reach a given expected score. Among all such choices, the one with the largest entropy assumes nothing beyond the expected score.
> * __Lagrange multipliers give the Boltzmann rule:__ We attached one multiplier to each constraint and set the derivative of the Lagrangian to zero. The solution is the Boltzmann rule, with the temperature equal to the inverse of the multiplier on the expected score.
> * __The temperature trades score for entropy:__ We showed that lowering the temperature raises the expected score, from the average toward the largest score. The maximum entropy falls as the expected score rises, at a rate equal to the inverse temperature, as in thermodynamics.
