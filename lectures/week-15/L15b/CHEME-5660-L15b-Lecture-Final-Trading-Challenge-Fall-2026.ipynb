## Notation and Operating State
The production loop carries forward a small set of model and control objects:

> * $\widehat\alpha_{i,t}$, $\widehat\beta_{i,t}$, and $\widehat\sigma_{\varepsilon,i,t}$ are the current online SIM estimates.
> * $\gamma_{i,t}$ is the current preference weight and $w_{i,t}^{\star}$ is the target portfolio weight.
> * $\Delta w_{i,t}=w_{i,t}^{\star}-w_{i,t}^{\mathrm{current}}$ is the proposed allocation change.
> * $d_t$ is portfolio drawdown, $\tau_t$ is proposed one-way turnover, and $s_{i,t}\in[0,1]$ is captured news severity.
> * $\mathcal G$ is the declared gate configuration and $\mathcal Q_t$ is the human-review queue.

Severity measures the magnitude of an information event, not its favorable or unfavorable direction.
___


## From a Backtest to an Operating Loop
A backtest consumes a completed path. An operating loop receives one observation at a time and must preserve state between evaluations. At each scheduled evaluation:

1. ingest the latest captured market observation;
2. update online model estimates;
3. compute the market-state signal and preference weights;
4. produce a target allocation and proposed trades;
5. evaluate position, turnover, drawdown, and event-severity gates;
6. approve, block, or route each proposal to human review;
7. append the observation, model state, proposal, gate results, and authorized action to the decision log.

The course implementation uses a captured tape and paper-trading interface. Changing the observation cadence requires retuning time-step and half-life conversions, not changing the logic of the EWLS recursion.
___


## Captured News and Event Severity
The course does not require a live news or language-model service. Instead, a captured event table contains timestamped ticker-level sentiment, severity, and provenance fields. This makes the exercise reproducible and keeps the scoring model outside the critical execution path.

At evaluation time, the engine loads the most recent event available at or before the decision timestamp. The signed sentiment may affect a preference score; the absolute severity determines whether the proposal requires review. Missing events default to a neutral signal and zero severity, and that fallback is recorded.

> __Model-risk rule:__ A generated or externally scored signal is an input—not ground truth. Preserve its source, timestamp, model/version identifier when available, and the transformation used by the portfolio rule.
___


## Human Review and the Decision Ticket
The review queue contains only proposals that crossed a declared escalation boundary. Each record must show the proposed trade, the relevant portfolio snapshot, every gate result, and the reason for escalation.

The reviewer may accept the proposal, reject it, or modify the authorized quantity. The resulting ticket preserves both the engine proposal and the reviewer action; it never overwrites the original proposal. This separation makes the decision auditable and supports later attribution of model versus human interventions.

## Final Trading Challenge
Each team receives the same initial capital, investable universe, evaluation tape, and benchmark series. Submit the final allocation policy together with its evidence package.

> __Required evidence package:__ State the valuation and return conventions, identify the data used for calibration and testing, report reward and risk metrics, document transaction-cost and turnover assumptions, list every active risk control, and preserve the decision log produced during the run.

The evaluation should consider terminal wealth, drawdown, tail loss, turnover, benchmark-relative performance, rule violations, and the quality of the model defense. A strategy that earns less but respects its declared limits can be preferable to a fragile strategy that wins one realized path.

Conclude by answering three questions: What did the system learn? When did it refuse or escalate an action? What evidence would be required before using the design outside the classroom?


## Summary
This lecture converted the validated Autotrader into a reproducible, human-supervised operating process.

> __Key Takeaways:__
>
> * __Production means state, controls, and records:__ A model output becomes an operational decision only after gate evaluation and authorization.
> * __Captured inputs make the exercise reproducible:__ The course workflow does not require live brokerage, news, or language-model credentials.
> * __Human intervention must remain visible:__ Proposed and authorized actions are separate objects in the decision log.
> * __The final defense evaluates the process:__ Strategy performance, assumptions, validation evidence, limits, and failure modes matter alongside terminal wealth.

The course closes with a system that can explain what it observed, what it proposed, what was authorized, and why.
