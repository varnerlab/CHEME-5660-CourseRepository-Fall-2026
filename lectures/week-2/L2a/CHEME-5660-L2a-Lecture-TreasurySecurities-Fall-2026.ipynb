## Concept Review
Last time we introduced the ideas of the time value of money, abstract assets and net present value. Let's do a quick review before we get started today.

### Time Value of Money, Yields, and Accumulation Factors
> __Time value of money.__ Dated dollars have different units. A dollar at time $0$ and a dollar at a future date cannot be added until a stated conversion rule expresses them at a common valuation date.

Suppose we have $P$ USD today. Let $y$ denote a quoted nominal annual yield used for valuation, and let the positive integer $n$ denote the number of compounding intervals per year. At compounding-period index $j$, the equivalent future amount $F_j$ is:
$$
\begin{align*}
F_j = \underbrace{\left(1+y/n\right)^{j}}_{\mathcal{D}_{j,0}(y;n)}\;P.
\end{align*}
$$
The forward accumulation factor $\mathcal{D}_{j,0}(y;n)$ converts time-0 dollars to period-$j$ dollars. Its inverse $\mathcal{D}_{j,0}^{-1}(y;n)=(1+y/n)^{-j}$ converts period-$j$ dollars to time-0 dollars. The continuously compounded equivalent of the nominal yield is given by:
$$
 g_y:=n\log(1+y/n),
$$
so over a horizon of $T$ years, $\mathcal{D}_{T,0}(y;n)=(1+y/n)^{nT}=e^{g_yT}$. We reserve $g$ for a continuously compounded growth rate and $r=(\Delta t)g$ for its associated dimensionless log return; a market-quoted yield $y$ must first be converted before it is compared with either quantity.

### Abstract Assets and Net Present Value
> __Abstract assets__ describe financial value as dated signed cash flows. Because dated dollars have different units, each cash flow must be converted to a common valuation date before the values are added.

Let $N$ be the final cash-flow index and let $\bar c_j$ be the signed net cash flow at index $j$. Under a constant nominal yield $y$, the time-0 net present value is
$$
\boxed{
\begin{align*}
\operatorname{NPV}_{0}(y) = \sum_{j=0}^{N} \mathcal{D}_{j,0}^{-1}(y;n)\;\bar{c}_{j}.
\end{align*}}
$$
Every term in the sum is measured in time-0 dollars.

## Treasury Bills
A U.S. Treasury bill (T-bill) is a short-term, _zero-coupon debt security_ issued by the [U.S. Department of the Treasury](https://www.treasurydirect.gov) with maturities ranging from a few days to one year. T-bills are sold below face value and do not pay periodic interest. The difference $V_P-V_B$ between face value and purchase price is a nominal dollar gain, not a return.

> **Where can I buy T-bills?** You can buy T-bills directly through [TreasuryDirect](https://www.treasurydirect.gov), or through a bank or broker. They may be held to maturity or traded in the secondary market.

### Valuing a T-Bill
A T-bill has two cash-flow events from the investor's perspective: the purchase price $-V_B$ at time $0$ and the face-value receipt $+V_P$ at maturity time $T>0$ years. Let $y$ be a nominal annual yield used for valuation, let $n$ be its compounding frequency, and define $g_y=n\log(1+y/n)$. The zero-NPV price is

    <center>
        <img src="figs/Fig-L2a-TBill-CashFlows.svg" width="505" alt="Timeline for a zero-coupon Treasury bill showing its purchase price today and par-value payment at maturity"/>
    </center>
</div>

$$
\begin{align*}
\underbrace{\operatorname{NPV}_{0}(y)}_{=\,0}&=-V_B+\mathcal D_{T,0}^{-1}(y;n)V_P\quad\text{solving for price $V_B$}\\
V_B&=\mathcal D_{T,0}^{-1}(y;n)V_P\\
&=\left(1+\frac{y}{n}\right)^{-nT}V_P=e^{-g_yT}V_P.\quad\blacksquare
\end{align*}
$$
The exponent $nT$ may be noninteger for a zero-coupon claim; it is not a discrete coupon index. The bill's continuously compounded holding-period growth rate and corresponding log return are
$$
 g_B=\frac{1}{T}\log\left(\frac{V_P}{V_B}\right),\qquad r_B=Tg_B.
$$
Under this flat equivalent-yield model, $g_B=g_y$.

> **Quick Example**: A 100 USD bill matures in six months. Under a nominal annual yield $y=5\%$ compounded semiannually ($n=2$), $V_B=100(1.025)^{-1}=97.56$ USD. The dollar gain is 2.44 USD; the continuously compounded annual growth rate is $g_B=2\log(100/97.56)$.

__Quotation convention matters.__ Treasury bills are commonly quoted using bank-discount and investment-rate conventions based on actual days to maturity. Those quoted rates are not automatically the nominal yield $y$ or continuously compounded growth rate $g_y$ used above. Reconcile day-count and compounding conventions before comparing rates.

## Treasury Notes and Bonds
United States Treasury notes and bonds pay semiannual coupons and repay face value at maturity. Notes and bonds share the same cash-flow structure but have different standard maturity ranges.

### Valuing Notes and Bonds
Let $T>0$ be maturity in years, let the positive integer $n$ be the coupon frequency per year, and let $N=nT$ be the integer number of remaining coupon dates. From the investor's perspective, the dated net cash flows are $\bar c_0=-V_B$, $\bar c_j=C$ for $j=1,\ldots,N-1$, and $\bar c_N=C+V_P$. The annual coupon rate is $c$, so each coupon is $C=(c/n)V_P$.

        <img src="figs/Fig-L2a-NoteBond-CashFlows.svg" width="680" alt="Timeline for a coupon Treasury security showing its purchase price, semiannual coupon payments, and par-value payment at maturity"/>
    </center>
</div>

Let $y$ be the constant nominal annual yield used for valuation. The zero-NPV price is
$$
\begin{align*}
\underbrace{\operatorname{NPV}_{0}(y)}_{=\,0}&=\underbrace{-V_B}_{\text{cost}}+\underbrace{\mathcal D_{N,0}^{-1}(y;n)V_P}_{\text{par payment}}+\underbrace{C\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}(y;n)}_{\text{coupon payments}}\quad\Longrightarrow\text{solving for price $V_B$}\\
V_B&=\mathcal D_{N,0}^{-1}(y;n)V_P+C\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}(y;n),\quad\blacksquare
\end{align*}
$$
where $\mathcal D_{j,0}^{-1}(y;n)=(1+y/n)^{-j}$. The nominal yield $y$ is a quotation and valuation convention. Its continuously compounded equivalent is $g_y=n\log(1+y/n)$, but a coupon security's realized holding-period growth also depends on coupon reinvestment and any sale price before maturity.

## Are Treasury securities risk-free?
Not completely. U.S. Treasury securities are often called *risk-free* because the likelihood of missing a promised nominal-dollar payment is low. However, this does not guarantee a profit over every holding period.

Two risks are especially important here:

* __Default risk__: The issuer may fail to make a promised payment. This risk is low for U.S. Treasury securities, but it is not equally low for every sovereign issuer.
* __Interest-rate risk__: If market yields change, the security's price changes. An investor who sells before maturity may realize a gain or loss.

The following examples illustrate default risk and its broader market effects.

### Case Study: Default Risk and Long-Term Capital Management (LTCM)
[Long-Term Capital Management (LTCM)](https://en.wikipedia.org/wiki/Long-Term_Capital_Management) was a highly leveraged hedge fund founded in 1994 by Nobel Prize-winning economists and experienced traders that relied on complex quantitative models. At its peak, LTCM held over $1 trillion in notional positions with leverage ratios approaching 25:1.

* __What happened__? LTCM's highly leveraged fixed-income strategy was initially successful, with annualized returns (after fees) of around 21% in its first year, 43% in its second year, and 41% in its third year. However, in 1998, it lost approximately 4.6 billion USD in four months due to exposure to the [1997 Asian financial crisis](https://en.wikipedia.org/wiki/1997_Asian_financial_crisis) and [1998 Russian financial crisis](https://en.wikipedia.org/wiki/1998_Russian_financial_crisis).
* __What was the aftermath?__ LTCM's losses were so significant (and connected to many Wall Street institutions) that they threatened the global financial system's stability (financial contagion). The [Federal Reserve Bank of New York](https://www.newyorkfed.org) organized a bailout by a consortium of banks and financial institutions to prevent a systemic crisis. The bailout was completed in October 1998, with the consortium providing $3.6 billion in capital to LTCM.

__Credit Risk Models and LTCM__: LTCM's collapse highlighted the importance of credit risk modeling in financial markets. The fund's strategies involved complex arbitrage positions that assumed certain relationships between bond prices would hold, but they failed to adequately model the credit risk of sovereign debt during financial crises. 

When Russia defaulted, it created a _flight to quality_ that broke many of LTCM's assumptions about price relationships between different government bonds. Modern credit risk models attempt to quantify the probability of default and the expected loss given default for various debt instruments, including sovereign bonds. 
> __Corporate Models__: Well-known models include the __Jarrow-Turnbull model__ (which models default as a Poisson process), the __Merton structural model__ (which treats default as occurring when a firm's asset value falls below its debt obligations), and the __CreditMetrics model__ developed by J.P. Morgan. These models use statistical techniques, historical data, and market indicators to assess credit risk - topics that could be explored in our advanced course materials.

## Summary
This lecture introduced the cash-flow structures, valuation, and risks of U.S. Treasury securities.

> __Key Takeaways:__
>
> * __Cash flows determine value:__ A Treasury bill has one payment at maturity, while notes and bonds pay coupons and return face value at maturity. In each case, the price follows from converting all cash flows to a common valuation date and setting NPV equal to zero.
> * __Rate conventions matter:__ Yield quotations, compounding frequency, and day-count conventions must be reconciled before rates or prices are compared.
> * __Risk-free does not mean price-risk free:__ U.S. Treasury securities have low nominal default risk, but their market prices can change when yields change. Selling before maturity may therefore produce a gain or loss.

Treasury securities provide a useful setting for connecting dated cash flows, discounting, and market risk.
