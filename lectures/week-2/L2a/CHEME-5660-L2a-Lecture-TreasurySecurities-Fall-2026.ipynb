## Concept Review
Last time we introduced the time value of money, abstract assets and the net present value. Let's do a quick review before we get started today.

### Time Value of Money, Yields, and Accumulation Factors
> __Time value of money.__ Dated dollars have different units. A dollar at time $0$ and a dollar at a future date cannot be combined or compared until they are converted to a common valuation date. The time value of money is the principle that a dollar today is worth more than a dollar tomorrow.

Suppose we have $P$ USD today. Let $y$ denote an annual yield (assumed to be constant) used for valuation, and let the positive integer $n$ denote the number of compounding intervals per year. At compounding-period index $j$, the equivalent future amount $F_j$ given $P$ dollars today and a nominal yield $y$ is given by:
$$
\begin{align*}
F_j = \underbrace{\left(1+y/n\right)^{j}}_{\mathcal{D}_{j,0}(y;n)}\;P.
\end{align*}
$$
The forward accumulation factor $\mathcal{D}_{j,0}(y;n)$ converts time-0 dollars to period-$j$ dollars. Its inverse $\mathcal{D}_{j,0}^{-1}(y;n)=(1+y/n)^{-j}$ converts period-$j$ dollars to time-0 dollars. The continuously compounded equivalent growth rate, given the nominal yield $y$, is given by:
$$
 g_y:=n\log(1+y/n),
$$
Here and throughout, $\log$ denotes the natural logarithm. Thus, over a horizon of $T$ years, $\mathcal{D}_{T,0}(y;n)=(1+y/n)^{nT}=e^{g_yT}$.

### Abstract Assets and Net Present Value
> __Abstract assets__ describe financial value of something, e.g., a car, a house or an idea, as set of dated (current and future) signed cash flows. Because dated dollars have different units, each cash flow must be converted to a common valuation date before the values can be combined. 

Let $N\geq 0$ be the final cash-flow index and let $\bar c_j$ be the signed net cash flow at index $j$. Under a constant nominal yield $y$, the time-0 net present value is given by:
$$
\boxed{
\begin{align*}
\operatorname{NPV}_{0}(y) = \sum_{j=0}^{N} \mathcal{D}_{j,0}^{-1}(y;n)\;\bar{c}_{j}.
\end{align*}}
$$
Every term in the sum is measured in time-0 dollars. The sign of NPV compares the proposed cash flows with the declared discounting benchmark. It is a valuation statement, not a guarantee of accounting profit, realized return, or forecast accuracy.

* __Positive__: After converting every cash flow to today's dollars, the amount received is greater than the amount paid. Under the chosen discount rate, the investment is worth more than its price. Thus, the asset is an attractive investment under the stated assumptions.
* __Negative__: After converting every cash flow to today's dollars, the amount received is less than the amount paid. Under the chosen discount rate, the investment is worth less than its price. Thus, the asset is not an attractive investment under the stated assumptions.
* __Zero__: After converting every cash flow to today's dollars, the amounts received and paid are equal. Under the chosen discount rate, the investment is worth exactly its price. This is the pricing condition we use for Treasury bills, notes, and bonds.

## Treasury Bills

A [U.S. Treasury bill (T-bill)](https://www.treasurydirect.gov/marketable-securities/treasury-bills/) is a short-term, _zero-coupon debt security_ issued by the U.S. Department of the Treasury. Regular bills have original maturities from 4 to 52 weeks. Treasury also issues [Cash Management Bills (CMBs)](https://www.treasurydirect.gov/marketable-securities/cash-management-bills/) with variable maturities that can range from a few days up to one year.

### Regular Bills versus Cash Management Bills
Both T-bills and CMBs are zero-coupon Treasury bills that mature within one year. The distinction between them is primarily how Treasury schedules and offers them, not their valuation mechanics.

| Feature | Regular Treasury bills | Cash Management Bills |
|:---|:---|:---|
| **Term** | Standardized 4-, 6-, 8-, 13-, 17-, 26-, or 52-week terms | Variable term selected for a particular financing need |
| **Auction schedule** | Recurring schedule | Auctioned only when needed |
| **Auction access** | TreasuryDirect or a bank, broker, or dealer | Bank, broker, or dealer only |

See [TreasuryDirect: Cash Management Bills](https://www.treasurydirect.gov/marketable-securities/cash-management-bills/) for current details.

### Par Amount, Purchase Price, and Discount
T-bills (or CMBs) have two cash-flow events from the investor's perspective: the purchase price $-V_B$ at time $0$ and the face-value receipt $+V_P$ at maturity time $T>0$ years. 

The __par amount__, also called face value, is the amount Treasury promises to pay at maturity. At Treasury auctions, bills are sold at a discount ($V_B<V_P$) or at par ($V_B=V_P$); a secondary-market price can theoretically exceed par.

For a bill purchased below par and held to maturity, $V_P-V_B$ is the dollar interest, equivalently the dollar holding-period gain.

| Quantity | Meaning | Role in a T-bill |
|:---|:---|:---|
| Par amount $V_P$ | Face value paid by Treasury at maturity | Fixed maturity payment |
| Purchase price $V_B$ | Cash paid to acquire the bill | Set at auction; changes in secondary-market trading |
| Dollar discount $V_P-V_B$ | Difference between the maturity payment and purchase price | Interest earned when a below-par bill is held to maturity |
| Bank discount rate $d$ | Annualized auction quotation relative to $V_P$ on a 360-day year | Converts the requested par amount into an auction price |
| Valuation yield $y$ | Annualized return under a stated compounding convention | Discounts $V_P$ to obtain $V_B$ in the valuation model below |

In a competitive bill auction, investors bid a bank discount rate $d$; a noncompetitive bidder requests a par amount and accepts the rate determined at auction. All successful bidders receive the single price associated with the highest accepted discount rate. If the bill has $D$ days to maturity, Treasury computes the auction price from
$$
V_B=V_P\left(1-d\frac{D}{360}\right).
$$
Thus, an award of $10{,}000~\mathrm{USD}$ in par amount is a claim to receive $10{,}000~\mathrm{USD}$ at maturity; it does not mean the investor pays $10{,}000~\mathrm{USD}$ today. For example, a price of $99.72$ per $100$ of par means that $10{,}000~\mathrm{USD}$ of par costs $9{,}972~\mathrm{USD}$ and produces $28~\mathrm{USD}$ of dollar interest if held to maturity. See [TreasuryDirect's bill-pricing explanation](https://www.treasurydirect.gov/marketable-securities/understanding-pricing/) and [TreasuryDirect's auction FAQ](https://www.treasurydirect.gov/help-center/auction-faqs/).

> __Key distinction:__ The par amount determines the maturity payment; the purchase price determines the amount invested. The quoted rate or yield describes the return connecting those two cash flows under a stated convention.

### Valuing a T-Bill
Let $y$ be a nominal annual yield used for valuation, let $n$ be its compounding frequency, and define the growth rate $g_y=n\log(1+y/n)$. The zero-NPV price is given by:

$$
\begin{align*}
\underbrace{\operatorname{NPV}_{0}(y)}_{=\,0}&=-V_B+\mathcal D_{T,0}^{-1}(y;n)V_P\quad\Longrightarrow\text{solving for price $V_B$}\\
V_B&=\mathcal D_{T,0}^{-1}(y;n)V_P\\
&=\left(1+\frac{y}{n}\right)^{-nT}V_P=e^{-g_yT}V_P.\quad\blacksquare
\end{align*}
$$
The exponent $nT$ may be noninteger for a zero-coupon T-bill or CMB. The bill's __price-implied annualized log yield__ and holding-period log return are given by:
$$
 g_B=\frac{1}{T}\log\left(\frac{V_P}{V_B}\right),\qquad r_B=Tg_B.
$$
When our model generates the price $V_B$, then the growth rates are equal: $g_B=g_y$; however, when $V_B$ is an observed secondary-market price, $g_B$ is implied directly by that price and $g_B$ may be different from $g_y$.

> **Quick Example**: 
>
> A 100 USD bill matures in six months. Under a nominal annual yield $y=5\%$ compounded semiannually ($n=2$), $V_B=100(1.025)^{-1}=97.56098\approx 97.56$ USD. The dollar interest, or dollar holding-period gain, is approximately 2.44 USD. 
> 
> The annualized continuously compounded growth rate is $g_B=2\log(1.025)\approx 0.049385$ ($4.9385\%$), while the six-month holding-period log return is $r_B=\log(1.025)\approx 0.024693$ ($2.4693\%$).

__Treasury quotation conventions matter.__ T-bills are quoted using a bank discount rate (par value and a 360-day year) or an investment/coupon-equivalent yield (purchase price and a 365- or 366-day year). Neither is automatically the nominal yield $y$ or continuous rate $g_y$ used above, so match conventions before comparing rates. 

See [Treasury's rate definitions](https://home.treasury.gov/resource-center/data-chart-center/interest-rates/TextView?type=daily) and [official auction formulas](https://www.treasurydirect.gov/files/laws-and-regulations/auction-regulations-uoc/auct-reg-gsr-356122303.pdf).


__What does zero NPV mean at auction?__ The auction price of a T-bill makes the present value of the promised payments equal to the amount paid when those payments are discounted at the yield set by the market. If investors perceived greater risk, they would demand a higher yield. 

Zero NPV does not mean zero risk or zero return; it means the price already reflects the return required by the market.

## Treasury Notes and Bonds

[U.S. Treasury notes](https://www.treasurydirect.gov/marketable-securities/treasury-notes/) are medium-term, fixed-rate coupon securities issued for 2, 3, 5, 7, or 10 years. [U.S. Treasury bonds](https://www.treasurydirect.gov/marketable-securities/treasury-bonds/) are long-term, fixed-rate coupon securities issued for 20 or 30 years. Both pay interest every six months and repay face value at maturity.

At original issue, a $T$-year note or bond with semiannual coupons ($n=2$) will have $N=nT$ coupons. In a reopening or secondary-market trade, $N$ instead denotes the number of remaining coupons.

### Coupon Rate versus Auction Yield
The annual coupon rate $c$ determines the security's semiannual coupon payment, $C=(c/n)V_P$. At maturity, Treasury pays the final coupon and repays the par amount $V_P$. The auction yield $y$ determines the price investors pay for these coupon and par payments. Treasury auction results label $c$ as the `Interest Rate`; here we call it the coupon rate.

| Quantity | Meaning | After original issue |
|:---|:---|:---|
| Par amount $V_P$ | Principal repaid at maturity | Fixed |
| Coupon rate $c$ | Annual rate that determines the contractual coupon payments | Fixed |
| Market yield $y$ | Return required by investors for the remaining cash flows | Changes with market conditions |
| Price $V_B$ | Present value of the remaining payments discounted at $y$ | Changes as $y$ changes |

At an original-issue auction, investors bid yields rather than coupon rates. So who sets the coupon rate? The Treasury sets the coupon rate after the auction, based on the highest accepted yield:

* __Auction determines the highest accepted yield__: The auction determines the highest accepted yield, and all successful bidders receive the security at the price corresponding to that yield. The Treasury then sets the coupon rate in increments of $1/8$ of one percentage point, or $0.125$ percentage points, choosing the rate that produces a price closest to, but not above, par when evaluated at the auction yield. 

If the auction yield is below $0.125\%$, the Treasury applies a minimum coupon rate of $0.125\%$, and the resulting price may be above par. 

__Example: a new 10-year note.__ The August 12, 2026 auction for CUSIP `91282CRF0` produced a high yield of $4.683\%$. Of the neighboring permitted coupon rates, $4.625\%$ produced a price closest to, but not above, par value, so the Treasury set the coupon rate at $4.625\%$. The reported auction price was $99.5407~\mathrm{USD}$ per $100~\mathrm{USD}$ of par.

See [TreasuryDirect's pricing explanation](https://www.treasurydirect.gov/marketable-securities/understanding-pricing/) and [the auction rule in 31 CFR 356.20](https://www.ecfr.gov/current/title-31/subtitle-B/chapter-II/subchapter-A/part-356/section-356.20).


### Valuing Notes and Bonds
Assume we have $N$ remaining coupon periods left on a note or bond. For Treasury notes and bonds, $n=2$ is both the number of coupon payments and yield-compounding periods per year, and $T=N/n$ is the remaining time to maturity in years. If $c$ is the annual coupon rate and $V_P$ is the face (par) value of the note or bond, then each coupon payment is $C=(c/n)V_P$. 

Thus, the investor has $N+1$ cash flow events: $\bar c_0=-V_B$ (purchase price), $\bar c_j=C$ for $j=1,\ldots,N-1$ (coupon payments), and $\bar c_N=C+V_P$ (par payment and final coupon).

Let $y$ be the nominal annual yield to maturity (YTM), compounded semiannually. The zero-NPV price for a coupon note or bond is given by:
$$
\begin{align*}
\underbrace{\operatorname{NPV}_{0}(y)}_{=\,0}&=\underbrace{-V_B}_{\text{cost}}+\underbrace{\mathcal D_{N,0}^{-1}(y;n)V_P}_{\text{par payment}}+\underbrace{C\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}(y;n)}_{\text{coupon payments}}\quad\Longrightarrow\text{solving for price $V_B$}\\
V_B&=\mathcal D_{N,0}^{-1}(y;n)V_P+C\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}(y;n),\quad\blacksquare
\end{align*}
$$
where $\mathcal D_{j,0}^{-1}(y;n)=(1+y/n)^{-j}$. This yield to maturity (YTM) convention discounts every cash flow at the same rate; a term-structure (time-varying) valuation would instead use maturity-specific spot rates.

## Are Treasury securities risk-free?
U.S. Treasury securities are commonly treated as low-default-risk nominal-dollar benchmarks. This does not mean that all government debt, or all debt issued in the United States, is risk-free. The relevant question is: __who promised the payment, and under what terms?__

__Default risk__ is the possibility that a borrower fails to make a promised payment in full and on time. Default may take the form of missed or suspended payments, or a distressed restructuring in which creditors receive less than originally promised.  

### Case Studies: Sovereign Default Risk

U.S. Treasury debt does not provide a comparable modern example of an outright sovereign default or distressed restructuring, but there are relatively recent examples from other sovereign issuers:

* __Russia, 1998__: On August 17, 1998, Russia announced a de facto default on ruble-denominated government bills and bonds and subsequently restructured the affected debt. Creditors suffered losses, and the event contributed to a broader international financial and liquidity crisis. The event occurred in August 1998; the linked [International Monetary Fund retrospective](https://www.imf.org/en/news/articles/2015/09/14/01/49/pr9935), published on July 28, 1999, describes it in the program summary.
* __Greece, 2012__: Greece restructured government debt held by private investors. Approximately 97% of the privately held bonds covered by the exchange participated, and investors accepted a 53.5% reduction in face value. See the [European Stability Mechanism's explanation of the 2012 private-sector debt restructuring](https://www.esm.europa.eu/content/what-was-private-sector-debt-restructuring-march-2012).

These cases demonstrate that government securities are not inherently free of default risk. Sovereign default can involve payment suspension or a restructuring in which the timing or amount of the promised cash flows changes. Thus, both the probability of default and the amount recovered following default affect the value of a security.

### Case Study: Mortgage Defaults and the 2008 U.S. Financial Crisis
The United States has not experienced a comparable modern default on its sovereign debt, but it has experienced a major default-driven financial crisis. The collapse of the housing bubble began the chain of events that led to the 2008 crisis. As housing prices fell, rising mortgage delinquencies and defaults produced losses on mortgage-backed securities. Because these securities were widely held and financed with substantial leverage and short-term borrowing, the losses weakened financial institutions and increased uncertainty about counterparties' ability to meet their own obligations.

Investors responded by moving capital toward U.S. Treasury securities—a _flight to quality_. The contrast is important: mortgage-backed securities and Treasury securities were both dollar-denominated debt instruments operating within the U.S. financial system, but the identity of the borrower and the structure of the promised payments produced very different default risks.

The Financial Crisis Inquiry Commission concluded that the crisis was avoidable and emphasized the role of human decisions and ignored warnings:

> *To paraphrase Shakespeare, the fault lies not in the stars, but in us.*  
> —[Financial Crisis Inquiry Commission Report](https://www.govinfo.gov/content/pkg/GPO-FCIC/pdf/GPO-FCIC.pdf), Conclusions, p. xvii

> __Watch:__ [The 2008 Financial Crisis](https://www.youtube.com/watch?v=GPOv72Awo68), a short overview suitable for class. For a deeper investigation, read the [Financial Crisis Inquiry Commission Report](https://www.govinfo.gov/content/pkg/GPO-FCIC/pdf/GPO-FCIC.pdf).

### Takeaway
Russia and Greece illustrate sovereign default; the 2008 crisis shows mortgage defaults spreading through securities and financial institutions. U.S. Treasuries are a relative benchmark because their promised nominal payments are considered unlikely to be missed—not because the United States is immune to default crises.

## Summary
This lecture introduced the cash-flow structures, valuation, and risks of U.S. Treasury securities.

> __Key Takeaways:__
>
> * __Cash flows determine value.__ A Treasury bill's par amount is its maturity payment, not necessarily its purchase price. Notes and bonds make coupon payments and return par amount at maturity. In each case, the price follows from converting all cash flows to a common valuation date.
> * __Coupon rates and yields do different jobs.__ The coupon rate fixes a note's or bond's promised payments, while the market yield discounts those payments and determines price. Treasury bill rates and bond yields also use stated day-count and compounding rules, and a note or bond purchased between coupon dates may include accrued interest.
> * __Risk-free is a relative term.__ Treasury securities have low nominal default risk and often attract investors during financial stress. Their market prices can still fall when yields rise.

Treasury securities provide a useful setting for connecting dated cash flows, discounting, and market risk.
