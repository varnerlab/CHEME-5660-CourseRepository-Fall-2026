# L2b: Yield, Duration, Convexity, and the Yield Curve

Fixed-income securities promise cash flows at specified dates, but their market values depend on the yields used to discount those cash flows. When yields change, prices change. The size of the response depends on the timing of the payments and the shape of the price-yield relationship.

This lecture begins with the interest-rate losses at Silicon Valley Bank. We then derive price sensitivity for a zero-coupon Treasury bill, extend the argument to coupon-bearing notes and bonds, and introduce duration and convexity. Malkiel's bond-pricing theorems organize these results, while the yield curve shows why one security's yield is not enough to value every payment date.

## Treasury Securities, Default Risk, and Interest-Rate Risk

U.S. Treasury securities are commonly treated as low-default-risk nominal-dollar benchmarks because they are obligations of the U.S. government. That description is conditional. A promised nominal payment can still expose its holder to several forms of risk.

* __Default risk__ is the possibility that the issuer fails to make a promised payment. This risk is low for U.S. Treasury securities, but sovereign securities from other issuers must be evaluated under their own legal, monetary, and credit arrangements.
* __Interest-rate risk__ arises because a change in required yields changes the secondary-market value of fixed promised cash flows. A holder who sells before maturity can realize a gain or loss, and an institution that marks positions to market can be affected even without a sale.

Treasury securities also retain inflation, liquidity, reinvestment, and operational risk. The rest of this lecture focuses on interest-rate risk: how much the market value changes when the relevant yield changes. The failure of Silicon Valley Bank illustrates why this distinction matters.

## Zero-Coupon Treasury Bill

A Treasury bill is the simplest security for this calculation because it makes one payment at maturity and has no coupon stream. If $V_P$ is the face value, $y$ is the nominal annual yield, $n$ is the number of compounding periods per year, and $T$ is the time to maturity in years, then

$$
V_B(y,T)=V_P\mathcal D_{T,0}^{-1}(y;n)
=V_P\left(1+\frac{y}{n}\right)^{-nT}.
$$

The discount factor converts the single payment at maturity into a time-0 price. Under the standing assumption $y>0$, the factor is less than one, so the bill trades below face value.

The yield enters through the per-period factor $1+y/n$. For a fixed exponent $k$,

$$
\frac{d}{dy}\left(1+\frac{y}{n}\right)^{-k}
=-\frac{k}{n}\left(1+\frac{y}{n}\right)^{-k-1}.
$$

This chain-rule calculation supplies both the sign and the scale of the bill's yield sensitivity.

### Price Sensitivity of a Treasury Bill

The bill price is a smooth function of $y$ and $T$. Its total differential is

$$
\begin{aligned}
dV_B
&=\frac{\partial V_B}{\partial y}\,dy
+\frac{\partial V_B}{\partial T}\,dT\\
&=-\underbrace{V_P\left(1+y/n\right)^{-nT}
\left(\frac{T}{1+y/n}\right)}_{V_{B,y}}\,dy\\
&\quad-\underbrace{V_P\left(1+y/n\right)^{-nT}
\left[n\log\left(1+y/n\right)\right]}_{V_{B,T}}\,dT\\
&=-V_{B,y}\,dy-V_{B,T}\,dT.
\end{aligned}
$$

Here $V_{B,y}$ and $V_{B,T}$ are sensitivity magnitudes; the partial derivatives are their negatives. The coupon term is absent because the contract fixes the bill's coupon rate at zero, so $dc=0$. It would be incorrect to claim that a coupon-bond pricing formula has $\partial V_B/\partial c=0$ when evaluated at $c=0$. If the coupon were allowed to change, the price would respond.

The yield magnitude $V_{B,y}$ is positive whenever $1+y/n>0$ and $T>0$. The maturity magnitude $V_{B,T}$ is positive when $y>0$ because $\log(1+y/n)>0$. This distinction matters when the yield is zero or negative.

### Interpreting the Bill Sensitivities

A higher yield discounts the face value more heavily. Holding maturity fixed, the bill price therefore decreases as the yield increases.

> __Example.__ Suppose the yield on a 52-week bill rises from 2% to 3% after the auction. The bill's secondary-market price falls because the same face value is now discounted at a higher rate.

The maturity term requires a distinction between two experiments. Comparing a 6-month bill with a 12-month bill changes the contractual maturity; for a bill, this is a smooth comparison because there is still only one payment. Holding one bill while calendar time passes is different: remaining maturity falls, so $dT=-dt$.

> __Example.__ Six months after purchasing a 52-week bill, its remaining maturity has fallen from one year to one-half year. If the yield is unchanged and positive, the price is higher because fewer periods remain to discount.

At $y=0$, every discount factor equals one and the bill is worth face value at every maturity. For $-n<y<0$, a longer bill is worth more and trades above face value. The claim that price decreases with maturity therefore requires the positive-yield assumption.

## Coupon-Bearing Treasury Notes and Bonds

A coupon-bearing Treasury note or bond pays a sequence of coupons and returns principal at maturity. With $N=nT$ remaining payment dates, its price is

$$
V_B
=V_P\mathcal D_{N,0}^{-1}(y;n)
+C\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}(y;n),
$$

where $C=(c/n)V_P$ is the coupon paid each period and

$$
\mathcal D_{j,0}^{-1}(y;n)
=\left(1+\frac{y}{n}\right)^{-j}.
$$

For a fixed payment schedule, the yield and coupon sensitivities are

$$
\frac{\partial V_B}{\partial y}
=-\underbrace{\frac{1}{n(1+y/n)}
\left[NV_P\mathcal D_{N,0}^{-1}
+C\sum_{j=1}^{N}j\mathcal D_{j,0}^{-1}\right]}_{V_{B,y}}<0,
$$

$$
\frac{\partial V_B}{\partial c}
=\underbrace{\frac{V_P}{n}\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}}_{V_{B,c}}>0.
$$

Every promised cash flow is discounted more heavily when $y$ increases, while an increase in $c$ adds coupon cash flow. These two derivatives therefore have definite and opposite signs. The underbraced factors $V_{B,y}$ and $V_{B,c}$ are positive sensitivity magnitudes, so for a fixed payment schedule the total differential is

$$
dV_B=-V_{B,y}\,dy+V_{B,c}\,dc.
$$

A maturity term is absent because the payment schedule is held fixed.

### Interpreting the Coupon-Bond Sensitivities

__Yield.__ A higher yield lowers the present value of both the coupon stream and the principal payment. A 10-year note paying a 3% coupon will therefore fall in price if comparable securities begin to yield 4%.

__Coupon rate.__ A higher coupon raises the price, but the price is not proportional to the coupon rate. Doubling $c$ doubles the coupon stream and leaves the principal payment unchanged. Thus, a 4% coupon bond is worth more than an otherwise identical 2% coupon bond, but not twice as much.

__Maturity.__ For a coupon security, $N$ is a discrete schedule index. Extending maturity adds coupon payments and postpones principal, so differentiating a finite sum whose upper limit is $nT$ is not a valid maturity comparison without a continuous extension. Complete schedules must be compared instead.

At a fixed yield and measured on coupon dates, a discount bond with $c<y$ moves up toward face value as maturity approaches, while a premium bond with $c>y$ moves down toward face value. A par bond with $c=y$ remains at face value. Maturity alone therefore does not determine the direction of a coupon bond's price change.

The table summarizes how the price of each security responds to each change:

| Change | Treasury bill | Treasury note or bond |
|:---|:---|:---|
| Yield rises ($dy>0$) | Price falls: $\partial V_B/\partial y=-V_{B,y}<0$ | Price falls: $\partial V_B/\partial y=-V_{B,y}<0$ |
| Maturity extends ($dT>0$) | Price falls when $y>0$: $\partial V_B/\partial T=-V_{B,T}<0$ | No smooth derivative in $T$; comparing complete schedules at a fixed yield, a longer schedule lowers a discount bond's price ($c<y$), raises a premium bond's price ($c>y$), and leaves a par bond ($c=y$) at face value |
| Coupon rate rises ($dc>0$) | Not applicable: the contract fixes the coupon at zero, so $dc=0$ | Price rises: $\partial V_B/\partial c=V_{B,c}>0$, less than proportionally |

Malkiel's maturity result concerns sensitivity to a yield change, not the direction in which price moves as time passes. Duration and convexity provide the appropriate language for that comparison.

## Duration and Convexity

The bill and coupon-bond formulas can be written in one form. Let $X_j\geq 0$ be the promised cash flow at period $j$, with at least one positive payment. For a fixed schedule of $N$ payments,

$$
V_B(y)=\sum_{j=1}^{N}X_j
\left(1+\frac{y}{n}\right)^{-j}.
$$

A coupon note has $X_j=C$ at each coupon date and $X_N=C+V_P$ at maturity. A bill has $X_j=0$ for $j<N$ and $X_N=V_P$. Only the cash-flow vector changes.

This representation applies directly to an __option-free__ security whose promised cash flows do not change when yields change. Conventional fixed-rate, noncallable Treasury bills, notes, and bonds satisfy this assumption. Callable bonds, mortgage-backed securities, and inflation-linked securities require additional care because a rate or state change can alter the cash flows themselves.

Writing the cash flows in a common form allows the shape of the price-yield curve to be derived once for the entire class.

### Price-Yield Geometry

Holding the cash flows and schedule fixed, differentiating the price twice gives

$$
\frac{dV_B}{dy}
=-\frac{1}{n}\sum_{j=1}^{N}jX_j
\left(1+\frac{y}{n}\right)^{-j-1}<0,
$$

$$
\frac{d^2V_B}{dy^2}
=\frac{1}{n^2}\sum_{j=1}^{N}j(j+1)X_j
\left(1+\frac{y}{n}\right)^{-j-2}>0.
$$

The strict signs follow when $1+y/n>0$ and at least one cash flow is positive. No term in either sum can reverse the stated sign. The price-yield curve is therefore decreasing and strictly convex.

The negative first derivative says that price and yield move in opposite directions. The positive second derivative says that the slope becomes less negative as yield increases. This curvature is the source of the asymmetric price response to equal upward and downward yield changes.

### Duration and Convexity as Risk Measures

Dividing the derivatives by the current price removes the scale of the position. Modified duration and convexity are defined by

$$
D_{\mathrm{mod}}(y)
:=-\frac{1}{V_B(y)}\frac{dV_B}{dy},
\qquad
K(y)
:=\frac{1}{V_B(y)}\frac{d^2V_B}{dy^2}.
$$

Modified duration measures the local percentage slope of price with respect to yield, while convexity measures the local curvature. When $y$ is quoted as a decimal annual yield, $D_{\mathrm{mod}}$ has units of years and $K$ has units of years squared.

Modified duration is related to Macaulay duration, the present-value-weighted average time to payment, by

$$
D_{\mathrm{mod}}
=\frac{D_{\mathrm{mac}}}{1+y/n}.
$$

For a zero-coupon bill, every dollar arrives at maturity. Hence $D_{\mathrm{mac}}=T$ and $D_{\mathrm{mod}}=T/(1+y/n)$, which reproduces the normalized bill sensitivity derived earlier.

### Estimating a Price Change

A second-order Taylor expansion converts duration and convexity into a local repricing formula. For a yield change $\Delta y$,

$$
\frac{\Delta V_B}{V_B}
\approx-D_{\mathrm{mod}}\Delta y
+\frac{1}{2}K(\Delta y)^2.
$$

The duration term is the straight-line estimate. The convexity term corrects for curvature and is positive for either sign of $\Delta y$ when $K>0$.

> __Example.__ Suppose a note has $D_{\mathrm{mod}}=7.2$ years and $K=65$ years$^2$. For a 50-basis-point increase in its yield, $\Delta y=0.0050$. The duration term is $-7.2(0.0050)=-0.0360$, and the convexity term is $\tfrac{1}{2}(65)(0.0050)^2=+0.0008$. The estimated price change is $-3.52\%$, compared with $-3.60\%$ from duration alone.

One basis point is $10^{-4}$ in decimal yield units. A 50-basis-point change in the federal funds target does not imply a 50-basis-point change in every Treasury yield; different maturities can move by different amounts.

Duration and convexity are local measures, so the approximation becomes less reliable as $|\Delta y|$ grows. Large shocks should be evaluated by repricing the complete cash-flow schedule.

## Malkiel's Bond-Pricing Theorems

Malkiel's theorems summarize the price-sensitivity results developed above. They assume an option-free security with fixed positive cash flows, and each comparison holds the remaining contract terms fixed. The maturity theorems compare bonds that begin at par, so $c=y>0$.

The par restriction can be seen from the closed-form coupon-bond price:

$$
\frac{V_B}{V_P}
=\frac{c}{y}
+\left(1-\frac{c}{y}\right)
\left(1+\frac{y}{n}\right)^{-N}.
$$

When $c=y$, the bond is worth $V_P$ for every maturity. Differentiate with respect to $y$ while holding $c$ fixed, and only then evaluate at $c=y$. The modified duration of a par bond is

$$
D_{\mathrm{mod}}(N)
=\frac{1-\left(1+y/n\right)^{-N}}{y}.
$$

The order of these steps matters. Setting $c=y$ before differentiating would move along the par locus, where the price is constant, rather than measuring the response of one bond to a yield change.

### Yield and Maturity Results

* __Theorem 1: Bond prices move inversely to bond yields.__ This is the sign $dV_B/dy<0$. A higher yield discounts every positive promised payment more heavily. Equivalently, modified duration is positive.

* __Theorem 2: Longer maturity increases yield sensitivity.__ For a par bond, $D_{\mathrm{mod}}(N)=[1-(1+y/n)^{-N}]/y$ is strictly increasing in $N$. Because every bond in the comparison begins at $V_P$, the percentage and dollar sensitivity rankings agree.

* __Theorem 3: The sensitivity gain diminishes as maturity lengthens.__ Adding one period increases par-bond modified duration by

$$
D_{\mathrm{mod}}(N+1)-D_{\mathrm{mod}}(N)
=\frac{\left(1+y/n\right)^{-N-1}}{n}.
$$

Successive increments shrink by the factor $(1+y/n)^{-1}<1$. Duration continues to increase with maturity, but each additional period contributes less than the previous one. In the limit, modified duration approaches $1/y$, the value for a perpetuity.

Together, the first three theorems describe the direction of a price change and the way maturity changes its scale.

### Convexity and Coupon Results

* __Theorem 4: Equal yield changes produce asymmetric price changes.__ A yield decrease raises price by more than an equal yield increase lowers it, measured from the same starting yield. Positive convexity, $d^2V_B/dy^2>0$, produces the asymmetry. In the second-order approximation, the duration term changes sign with $\Delta y$, while the convexity term $\tfrac{1}{2}K(\Delta y)^2$ remains positive.

* __Theorem 5: Higher-coupon bonds are less sensitive to yield changes.__ This is a statement about percentage sensitivity and requires at least two remaining payment dates. A larger coupon moves more present value to earlier dates, reducing Macaulay and modified duration. Dollar sensitivity moves in the other direction in a fixed-schedule comparison because a higher-coupon bond has more cash flow and a higher price. When $N=1$, every payment occurs on the same date and duration does not depend on the coupon.

These theorems separate two effects that are easy to conflate. Convexity determines the asymmetry between equal upward and downward yield changes, while the timing of the cash flows determines how coupon rate affects percentage sensitivity.

### Role of the Par Restriction

Outside the par comparison, percentage and dollar sensitivity can have different maturity rankings. A zero-coupon bill provides a useful counterexample. Its modified duration is

$$
D_{\mathrm{mod}}=\frac{T}{1+y/n},
$$

which increases with $T$, while its dollar sensitivity is

$$
-\frac{dV_B}{dy}
=V_PT\left(1+\frac{y}{n}\right)^{-nT-1}.
$$

The dollar expression combines a growing duration term with a shrinking discounted price. For $V_P=100$, $n=1$, and $y=5\%$, it peaks near $T=20.5$ years and then declines. A longer bill can therefore have greater percentage sensitivity but smaller dollar sensitivity.

The par restriction holds the starting price fixed across maturities. That is why the percentage and dollar rankings agree in Malkiel's maturity theorems.

## From One Yield to the Yield Curve

Yield to maturity is the single internal rate that reproduces one security's price from its promised cash flows. It is useful for quoting and comparison, but it does not describe the full term structure.

A __spot rate__ applies to one payment date, so a spot curve supplies the maturity-specific discount factors used to value individual cash flows. A __par yield__ is the coupon rate that makes a hypothetical security trade at face value under that spot curve.

For payments $X_j$ at times $t_j$, curve-based pricing is

$$
V_B=\sum_{j=1}^{N}X_jD(0,t_j),
$$

where $D(0,t_j)$ is the time-0 discount factor for maturity $t_j$. Curve level, slope, and curvature describe how discounting varies by maturity; they do not by themselves identify a unique forecast of future short rates.

## Summary

The lecture began with a single discounted payment and ended with a complete yield curve. The same principle connects each step: the value of a fixed-income security depends on the timing of its cash flows and the rates used to discount them.

> __Key Takeaways:__
>
> * __Price and yield move in opposite directions.__ For fixed positive cash flows, price decreases as yield increases; modified duration measures the local slope, and convexity measures the curvature of the price-yield relationship.
> * __Sensitivity depends on maturity, coupon, and yield level.__ Malkiel's theorems describe how yield, maturity, and coupon rate change price sensitivity under specific assumptions.
> * __One yield does not value every payment date.__ Yield to maturity summarizes one security with a single rate, while a spot curve values each payment at its own maturity.

These tools extend beyond Treasury securities: duration and convexity estimate how the value of any fixed stream of promised payments responds to a small change in its yield.
