## Review of Treasury Securities, Default Risk, and Interest-Rate Risk

U.S. Treasury securities are commonly treated as low-default-risk nominal-dollar benchmarks because they are obligations of the U.S. government. That description is conditional. A promised nominal payment can still expose its holder to several forms of risk.

* __Default risk__ is the possibility that the issuer fails to make a promised payment. This risk is low for U.S. Treasury securities, but sovereign securities from other issuers must be evaluated under their own legal, monetary, and credit arrangements.
* __Interest-rate risk__ arises because a change in required yields changes the secondary-market value of fixed promised cash flows. A holder who sells before maturity can realize a gain or loss, and an institution that marks positions to market can be affected even without a sale. This is the risk that sank Silicon Valley Bank.

Treasury securities also retain inflation, liquidity, reinvestment, and operational risk. The rest of this lecture focuses on interest-rate risk: how much the market value changes when the relevant yield changes.

## Motivation

Suppose you manage a \$500 million portfolio of Treasury securities. The Federal Reserve meets today, and a decision on interest rates is expected at 2:00 PM. If yields move, the value of the portfolio moves with them, and your clients will want to know by how much. Answering that question before the announcement requires a formula, not a guess.

We saw this risk with Silicon Valley Bank: the bank held long-maturity government securities, yields rose, and the market value of those holdings fell. When depositors withdrew funds, the bank had to sell at a loss and failed. Our previous discussion was qualitative. Today, we'll develop quantitative expressions for how much the price of a fixed-income security changes when market or other factors change. 

### Malkiel's Bond-Pricing Theorems

Malkiel organized the behavior of option-free bond prices into five theorems. Each statement below concerns a security with fixed positive promised cash flows, each comparison holds the remaining contract terms fixed, and the maturity comparisons are made between bonds priced at par, meaning the coupon rate equals the yield:

* __Theorem 1:__ Bond prices move inversely to bond yields: when the yield rises, the price falls.
* __Theorem 2:__ Bonds with longer maturities are more sensitive to a yield change than bonds with shorter maturities.
* __Theorem 3:__ The sensitivity gain from added maturity diminishes as maturity lengthens.
* __Theorem 4:__ Equal-sized yield changes produce asymmetric price changes: a yield decrease raises the price by more than an equal yield increase lowers it.
* __Theorem 5:__ Bonds with higher coupon rates are less sensitive, in percentage terms, to a yield change than bonds with lower coupon rates.

Today, we'll derive the price sensitivities of a Treasury bill, extend the derivation to coupon-bearing notes and bonds, define duration and convexity as standardized risk measures, and then revisit each theorem with these tools. The lecture closes by relaxing the single-yield assumption and introducing the yield curve.

## Zero-Coupon Treasury Bill

A Treasury bill is the simplest security for measuring price sensitivity because it makes only one payment at maturity and has no coupon stream. If $V_P$ is the face value, $y$ is the nominal annual yield, $n$ is the number of compounding periods per year, and $T$ is the time to maturity in years, then the bill's price is given by:

$$
V_B(y,T)=V_P\mathcal D_{nT,0}^{-1}(y;n)
=V_P\left(1+\frac{y}{n}\right)^{-nT}.
$$

The discount factor converts the single payment at maturity into a time-0 price. We assume $y>0$ throughout this lecture unless stated otherwise; the discount factor is then less than one, so the bill trades below face value.

The price depends on two variables that can change: the yield $y$ and the time to maturity $T$. We want to know how much the price moves when either one moves. Both variables live inside the power $\left(1+y/n\right)^{-nT}$, but in different places: the yield sits in the base, and the maturity sits in the exponent. Each therefore needs its own derivative.

Start with the yield. For a fixed exponent $k$, the chain rule gives:

$$
\underbrace{\frac{d}{dy}\left(1+\frac{y}{n}\right)^{-k}}_{\text{yield sensitivity}}=-\frac{k}{n}\left(1+\frac{y}{n}\right)^{-k-1}.
$$

Setting $k=nT$ and multiplying by the face value $V_P$ turns this into the yield derivative of the bill price. We give the rule for a general exponent $k$ because we will reuse it for coupon securities, where every payment date contributes its own power. The next section computes both derivatives and assembles them into a single expression for the price change.

### Price Sensitivity of a Treasury Bill

Let's consider the yield first. Setting $k=nT$ in the chain-rule result and multiplying by $V_P$ gives the yield derivative:

$$
\begin{aligned}
\frac{\partial V_B}{\partial y}
&=V_P\;\underbrace{\frac{d}{dy}\left(1+\frac{y}{n}\right)^{-nT}}_{\text{yield sensitivity}}\\
&=-V_P\,\frac{nT}{n}\left(1+\frac{y}{n}\right)^{-nT-1}\\
&=-V_P\,T\left(1+\frac{y}{n}\right)^{-nT-1}\\
&=-V_P\left(1+\frac{y}{n}\right)^{-nT}\left(\frac{T}{1+y/n}\right).\quad\blacksquare
\end{aligned}
$$

The second line applies the chain rule with $k=nT$, the third simplifies $nT/n=T$, and the last line splits off one factor of $(1+y/n)^{-1}$ so that the price $V_P\left(1+y/n\right)^{-nT}$ stands on its own. Written this way, the derivative is the price times $T/(1+y/n)$, a factor that returns later as the modified duration of a zero-coupon bill.

The maturity needs a different step because $T$ sits in the exponent. Rewrite the power as an exponential, $\left(1+y/n\right)^{-nT}=e^{-nT\log\left(1+y/n\right)}$, and differentiate with respect to $T$:

$$
\begin{aligned}
\frac{\partial V_B}{\partial T}
&=V_P\frac{d}{dT}\,e^{-nT\log\left(1+y/n\right)}\\
&=-V_P\left[n\log\left(1+\frac{y}{n}\right)\right]e^{-nT\log\left(1+y/n\right)}\\
&=-V_P\left(1+\frac{y}{n}\right)^{-nT}\left[n\log\left(1+\frac{y}{n}\right)\right].\quad\blacksquare
\end{aligned}
$$

The exponent is linear in $T$ with slope $-n\log(1+y/n)$, so differentiating brings that slope down as a factor; the last line converts the exponential back to power form.

A small yield change $dy$ and a small maturity change $dT$ each move the price through its own derivative. The total differential adds the two first-order effects, and is given by:

$$
\begin{aligned}
dV_B
&=\frac{\partial V_B}{\partial y}\,dy
+\frac{\partial V_B}{\partial T}\,dT\\
&=-\underbrace{V_P\left(1+y/n\right)^{-nT}
\left(\frac{T}{1+y/n}\right)}_{V_{B,y}}\,dy\\
&\quad-\underbrace{V_P\left(1+y/n\right)^{-nT}
\left[n\log\left(1+y/n\right)\right]}_{V_{B,T}}\,dT\\
&=-V_{B,y}\,dy-V_{B,T}\,dT.
\end{aligned}
$$

The underbraced factors $V_{B,y}$ and $V_{B,T}$ are sensitivity magnitudes: positive numbers that measure the size of each effect. The minus signs in front of them carry the direction, so the partial derivatives are $\partial V_B/\partial y=-V_{B,y}$ and $\partial V_B/\partial T=-V_{B,T}$. 

Under the standing assumption $y>0$, both magnitudes are positive: $V_{B,y}$ because $1+y/n>0$ and $T>0$, and $V_{B,T}$ because $\log(1+y/n)>0$. The signs of the two price responses are therefore fixed, and the next section interprets them.

### Interpreting the Bill Sensitivities

A higher yield discounts the face value more heavily, so the bill price falls as the yield rises. This is Theorem 1 in its simplest setting: the bill makes a single promised payment, and nothing offsets the heavier discounting.

The maturity term requires more care because a change in $T$ can mean two different things: comparing bills with different contractual maturities, or holding one bill while calendar time passes, in which case the remaining maturity falls and $dT=-dt$. For a bill, both readings use the same derivative because there is only one payment.

> __Example.__ Suppose you buy a 52-week bill at auction when its yield is 2%. Both sensitivities show up in this one position. If the yield rises to 3% the next day, the secondary-market price falls because the same face value is now discounted at a higher rate. If instead the yield holds at 2% for six months, the remaining maturity falls from one year to one-half year, and the price rises above the purchase price because fewer periods remain to discount.

The maturity conclusion depends on the positive-yield assumption: at $y=0$ every discount factor equals one and the bill is worth face value at every maturity, while for $-n<y<0$ a longer bill is worth more. Price decreases with maturity only when $y>0$.

## Coupon-Bearing Treasury Notes and Bonds

A coupon-bearing Treasury note or bond pays a sequence of coupons and returns principal at maturity. With $N=nT$ remaining payment dates, its price $V_B$ is given by:

$$
V_B
=V_P\mathcal D_{N,0}^{-1}(y;n)
+C\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}(y;n),
$$

where $C=(c/n)V_P$ is the coupon paid each period and the discount factor is given by:

$$
\mathcal D_{j,0}^{-1}(y;n)
=\left(1+\frac{y}{n}\right)^{-j}.
$$

Before differentiating, notice that maturity plays a new role here. For the bill, $T$ sat in an exponent, and the price changed smoothly as $T$ changed. For a coupon security, maturity sets the number of payments $N=nT$, which is a count: extending the maturity adds whole coupon payments to the sum. 

A count has no derivative. We therefore treat the three inputs in two groups: the yield $y$ and the coupon rate $c$ change smoothly and get derivatives, while the maturity changes in whole periods and gets a difference. For a fixed schedule of $N$ payments, the total differential is given by:

$$
dV_B=\frac{\partial V_B}{\partial y}\,dy+\frac{\partial V_B}{\partial c}\,dc.
$$

### Yield derivative.
Apply the chain-rule result from the bill section to each discount factor, term by term:

$$
\begin{aligned}
\frac{\partial V_B}{\partial y}
&=V_P\frac{d}{dy}\left(1+\frac{y}{n}\right)^{-N}
+C\sum_{j=1}^{N}\frac{d}{dy}\left(1+\frac{y}{n}\right)^{-j}\\
&=-\frac{N}{n}V_P\left(1+\frac{y}{n}\right)^{-N-1}
-\frac{C}{n}\sum_{j=1}^{N}j\left(1+\frac{y}{n}\right)^{-j-1}\\
&=-\underbrace{\frac{1}{n(1+y/n)}
\left[NV_P\mathcal D_{N,0}^{-1}
+C\sum_{j=1}^{N}j\,\mathcal D_{j,0}^{-1}\right]}_{V_{B,y}\,>\,0}.
\end{aligned}
$$

The point of the last line is the sign. Every factor under the brace is positive: the face value, the coupon payment, the discount factors, the period indices $j$, and the prefactor $1/[n(1+y/n)]$. The bracket therefore defines a positive magnitude $V_{B,y}$, and the minus sign in front of it is explicit, so $\partial V_B/\partial y=-V_{B,y}<0$ no matter the coupon rate or the maturity. 

A higher yield discounts every promised cash flow more heavily, and the present value of a later payment falls by a larger percentage. Thus, the price of a coupon security decreases as the yield increases, just like a bill. This establishes Theorem 1 for coupon-bearing securities: the inverse price-yield relationship holds for every coupon rate and maturity.

### Coupon derivative. 
Now, let's consider the effect of a change in the coupon rate $c$. Only the per-period coupon payment $C=(c/n)V_P$ depends on $c$, and it enters the price linearly:

$$
\frac{\partial V_B}{\partial c}
=\frac{\partial C}{\partial c}\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}
=\underbrace{\frac{V_P}{n}\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}}_{V_{B,c}\,>\,0}.
$$

The same sign logic applies: every term in the sum is a positive discount factor, so $V_{B,c}>0$ and $\partial V_B/\partial c=V_{B,c}>0$. A higher coupon rate adds cash flow at every coupon date, so the price rises.

Because $V_{B,c}$ does not depend on $c$, the coupon effect is the same size at every coupon rate, while the discounted principal contributes a term that does not change with $c$. Doubling the coupon rate therefore doubles the coupon stream but not the price: a 4% coupon bond is worth more than an otherwise identical 2% coupon bond, but not twice as much.

Putting the two derivatives together completes the smooth part of the story. For a fixed payment schedule, the first-order price change and its positive sensitivity magnitudes are given by:

$$
\boxed{
\begin{aligned}
dV_B&=-V_{B,y}\,dy+V_{B,c}\,dc,\quad\text{where:}\\[6pt]
V_{B,y}&=\frac{1}{n(1+y/n)}
\left[NV_P\,\mathcal D_{N,0}^{-1}
+C\sum_{j=1}^{N}j\,\mathcal D_{j,0}^{-1}\right],\\
V_{B,c}&=\frac{V_P}{n}\sum_{j=1}^{N}\mathcal D_{j,0}^{-1}.\quad\blacksquare
\end{aligned}
}
$$

The signs are built in: the price falls when the yield rises, and rises when the coupon rate rises. One input remains, and it is the discrete one.

### Maturity difference.
To measure the maturity effect, let's extend the schedule by one period and subtract the original price. Only two things change: the principal payment moves from period $N$ to period $N+1$, and one new coupon appears at period $N+1$. 

The difference between the new price and the original price is given by:

$$
\begin{aligned}
V_B(N+1)-V_B(N)
&=V_P\left(\mathcal D_{N+1,0}^{-1}-\mathcal D_{N,0}^{-1}\right)
+C\,\mathcal D_{N+1,0}^{-1}\\
&=\mathcal D_{N+1,0}^{-1}\left[C-\frac{y}{n}V_P\right]\\
&=\frac{V_P}{n}\left(c-y\right)\left(1+\frac{y}{n}\right)^{-(N+1)}.
\end{aligned}
$$

The sign of the price change is the sign of $c-y$. Recall the vocabulary from L2a: a __premium bond__ has $c>y$ and prices above face value, a __discount bond__ has $c<y$ and prices below face value, and a __par bond__ has $c=y$ and prices at face value. Extending the maturity by one period therefore raises the price of a premium bond, lowers the price of a discount bond, and leaves a par bond unchanged. 

The same difference also determines what happens as calendar time passes. Holding the yield fixed, each period that elapses removes one payment from the remaining schedule, which is the maturity difference applied with $\Delta N=-1$, so the sign of the price change flips: a discount bond's price ($c<y$) rises toward face value as maturity approaches, a premium bond's price ($c>y$) falls toward face value, and a par bond ($c=y$) stays at face value.

### Interpreting the Coupon-Bond Sensitivities

The table summarizes how the price of each security responds to each change:

| Change | Treasury bill | Treasury note or bond |
|:---|:---|:---|
| Yield rises ($dy>0$) | Price falls: $\partial V_B/\partial y=-V_{B,y}<0$ | Price falls: $\partial V_B/\partial y=-V_{B,y}<0$ |
| Maturity extends | Price falls when $y>0$ ($dT>0$): $\partial V_B/\partial T=-V_{B,T}<0$ | One added period ($\Delta N=+1$): sign of $c-y$, so price falls if $c<y$, rises if $c>y$, unchanged if $c=y$ |
| Coupon rate rises ($dc>0$) | Not applicable: the contract fixes the coupon at zero | Price rises: $\partial V_B/\partial c=V_{B,c}>0$, less than proportionally |

Note what the maturity row does and does not say. The row holds the yield fixed and compares prices across schedule lengths: adding a period to the schedule moves the price in the direction given by the sign of $c-y$. Theorems 2 and 3 instead let the yield move and compare price responses: when yields change, does a 20-year note gain or lose more than an otherwise identical 10-year note? 

Answering that question requires a standardized measure of yield sensitivity, which is the job of duration and convexity, which we develop next.

## Duration and Convexity

The bill and coupon-bond formulas can be written in one form. Let $X_j\geq 0$ be the promised cash flow at period $j$, with at least one positive payment ($j=1,\ldots,N$). For a fixed schedule of $N$ payments, the price is given by:

$$
V_B(y)=\sum_{j=1}^{N}X_j
\left(1+\frac{y}{n}\right)^{-j}.
$$

A coupon note or bond has $X_j=C$ at each coupon date and $X_N=C+V_P$ at maturity. A bill has $X_j=0$ for $j<N$ and $X_N=V_P$. Only the cash-flow vector changes.

> __More to this story?__ 
> 
> This cash-flow form applies directly to an __option-free__ security whose promised cash flows do not change when yields change. Conventional fixed-rate, noncallable Treasury bills, notes, and bonds satisfy this assumption. 
> * [Callable bonds](https://en.wikipedia.org/wiki/Callable_bond) and [mortgage-backed securities](https://en.wikipedia.org/wiki/Mortgage-backed_security) are not option-free: the issuer or the borrower holds an option whose exercise changes the cash flows. 
> * Inflation-linked securities, such as [Treasury Inflation-Protected Securities (TIPS)](https://www.treasurydirect.gov/marketable-securities/tips/), fall outside this form for a different reason: their payments are indexed to inflation, so the nominal cash flows are not fixed.

Writing the cash flows in a common form allows the shape of the price-yield curve to be derived once for all of these securities.

### Price-Yield Geometry

Holding the cash flows and schedule fixed, the first derivative of the price with respect to yield is given by:

$$
\frac{dV_B}{dy}
=-\frac{1}{n}\sum_{j=1}^{N}jX_j
\left(1+\frac{y}{n}\right)^{-j-1}<0.
$$

Differentiating a second time gives:

$$
\frac{d^2V_B}{dy^2}
=\frac{1}{n^2}\sum_{j=1}^{N}j(j+1)X_j
\left(1+\frac{y}{n}\right)^{-j-2}>0.
$$

The signs of both derivatives follow from term-by-term positivity: when $1+y/n>0$ and at least one cash flow is positive, no term in either sum can change it's sign, so the price-yield curve is decreasing and strictly convex. 

The negative first derivative is Theorem 1 in its general form, the positive second derivative supplies the curvature behind the asymmetric price response promised by Theorem 4, and the next two sections turn both derivatives into standardized risk measures.

### Duration and Convexity as Risk Measures

The two derivatives of the price-yield curve measure dollar changes, so their size depends on the size of the position: a \$10 million holding responds ten times as much as a \$1 million holding of the same security. Dividing each derivative by the current price removes that scale and produces risk measures that can be compared across securities. Thus, modified duration and convexity are defined as:

$$
D_{\mathrm{mod}}(y)
:=-\frac{1}{V_B(y)}\frac{dV_B}{dy},
\qquad
K(y)
:=\frac{1}{V_B(y)}\frac{d^2V_B}{dy^2}.
$$

Modified duration measures the local percentage slope of price with respect to yield, while convexity measures the local curvature. Both numbers are positive for an option-free security, the first because the minus sign in its definition offsets the negative slope, and the second because the price-yield curve is convex. When $y$ is quoted as a decimal annual yield, $D_{\mathrm{mod}}$ has units of years and $K$ has units of years squared.

Modified duration is built from a derivative, and the derivative alone does not explain why one cash-flow schedule is more sensitive than another. A second duration measure supplies the explanation by summarizing the timing of the cash flows directly. Macaulay duration is the present-value-weighted average time to payment, defined by:

$$
D_{\mathrm{mac}}
:=\frac{1}{V_B}\sum_{j=1}^{N}\frac{j}{n}\,X_j\left(1+\frac{y}{n}\right)^{-j}.
$$

Each payment date $j/n$, measured in years, is weighted by the fraction of the price contributed by that payment's present value. The weights are nonnegative and sum to one, so $D_{\mathrm{mac}}$ is a weighted average of the payment times: shifting present value toward earlier dates lowers it, and shifting present value toward later dates raises it. Differentiating the price term by term shows that the two duration measures are related by:

$$
D_{\mathrm{mod}}
=\frac{D_{\mathrm{mac}}}{1+y/n}.
$$

This relation ties sensitivity to timing: whatever moves $D_{\mathrm{mac}}$ moves $D_{\mathrm{mod}}$ in the same direction, and Theorem 5 turns on that observation. For a zero-coupon bill, every dollar arrives at maturity, so $D_{\mathrm{mac}}=T$ and $D_{\mathrm{mod}}=T/(1+y/n)$, which reproduces the normalized bill sensitivity derived earlier.

### Estimating a Price Change

Duration and convexity summarize the slope and curvature of the price-yield curve at the current yield, and together they answer the question posed in the motivation: how much does the price change when the yield moves? A second-order Taylor expansion in the yield change $\Delta y$ turns the two measures into a local repricing formula:

$$
\frac{\Delta V_B}{V_B}
\approx-D_{\mathrm{mod}}\Delta y
+\frac{1}{2}K(\Delta y)^2.
$$

The duration term is the straight-line estimate, and the convexity term corrects it for curvature. The correction is positive for either sign of $\Delta y$ when $K>0$, which is the mechanism behind Theorem 4: the duration term changes sign with $\Delta y$ while the convexity term does not, so a yield decrease raises the price by more than an equal yield increase lowers it.

> __Example.__ Suppose a note has $D_{\mathrm{mod}}=7.2$ years and $K=65$ years$^2$. For a 50-basis-point increase in its yield, $\Delta y=0.0050$. The duration term is $-7.2(0.0050)=-0.0360$, and the convexity term is $\tfrac{1}{2}(65)(0.0050)^2=+0.0008$. The estimated price change is $-3.52\%$, compared with $-3.60\%$ from duration alone.

The example is quoted in basis points, the standard unit for yield changes: one basis point is $10^{-4}$ in decimal yield units. Note that a 50-basis-point change in the federal funds target does not imply a 50-basis-point change in every Treasury yield; different maturities can move by different amounts.

Duration is useful for two reasons beyond the single-security estimate:

* __A common scale__: Duration puts securities with different coupons, maturities, and prices on one scale, so the interest-rate risk of a 2-year note and a 10-year note can be compared directly.
* __A portfolio measure__: The duration of a portfolio is the market-value-weighted average of the durations of the securities it holds, so the \$500 million portfolio from the motivation can be repriced in one step rather than security by security.

The same properties support risk management: offsetting positions can bring a portfolio's net duration near zero, while a bank that funds long-duration securities with short-duration deposits, as Silicon Valley Bank did, carries a duration mismatch.

Finally, duration and convexity are local measures, so the estimate becomes less reliable as $|\Delta y|$ grows. A large shock should be evaluated by repricing the complete cash-flow schedule.

## Revisiting Malkiel's Theorems

The machinery is in place, so let's return to the five theorems stated at the start of the lecture and attach each one to a derived result. Throughout, the security is option-free with fixed positive cash flows, each comparison holds the remaining contract terms fixed, and the maturity comparisons in Theorems 2 and 3 are made between bonds priced at par, so $c=y>0$.

* __Theorem 1: Bond prices move inversely to bond yields.__ This is the sign $dV_B/dy<0$, established three times: for the bill, for the coupon bond term by term, and for the general cash-flow form. Equivalently, modified duration is positive.

* __Theorem 2: Longer maturity increases yield sensitivity.__ Extending a par bond's schedule moves the principal payment, the largest single cash flow, to a later date. The weighted-average payment time rises, so Macaulay and modified duration rise with maturity. Because every bond in the comparison starts at face value, the percentage and dollar sensitivity rankings agree.

* __Theorem 3: The sensitivity gain diminishes as maturity lengthens.__ Each added period arrives more heavily discounted than the one before, so it carries less present-value weight and adds less to duration. Duration continues to increase with maturity, but each additional period contributes less than the previous one; in the limit, the modified duration of a par bond approaches $1/y$, the value for a perpetuity.

* __Theorem 4: Equal yield changes produce asymmetric price changes.__ Positive convexity produces the asymmetry. In the second-order repricing formula, the duration term changes sign with $\Delta y$ while the convexity term $\tfrac{1}{2}K(\Delta y)^2$ remains positive, so a yield decrease raises the price by more than an equal yield increase lowers it.

* __Theorem 5: Higher-coupon bonds are less sensitive to yield changes.__ This is a statement about percentage sensitivity and requires at least two remaining payment dates. A larger coupon places more present value at earlier dates, which lowers Macaulay duration and, through $D_{\mathrm{mod}}=D_{\mathrm{mac}}/(1+y/n)$, modified duration. Dollar sensitivity can rank in the other direction because a higher-coupon bond has a higher price. When $N=1$, every payment occurs on the same date and duration does not depend on the coupon.

The par restriction in Theorems 2 and 3 does real work: it holds the starting price at face value across maturities, which is what makes the percentage and dollar rankings agree. Outside the par comparison, the two rankings can separate. A zero-coupon bill's modified duration $T/(1+y/n)$ grows without bound as $T$ increases, but its dollar sensitivity eventually declines because the discounted price shrinks faster than the duration grows. The optional advanced note listed at the end of the lecture derives the par-bond duration formula that makes Theorems 2 and 3 quantitative and works through this counterexample.

## From One Yield to the Yield Curve

Every calculation so far discounted all of a security's cash flows at one yield $y$, so a change $dy$ moved the discount rate at every payment date by the same amount. Actual yields need not move this way: rates at different maturities can change by different amounts, and a single number cannot describe discounting date by date. Markets therefore quote several rate measures, and they answer different questions. This section defines the three we use.

__Yield to maturity__ is the single rate that reproduces one security's market price from its promised cash flows. For a coupon security with market price $V_B$ and $N$ remaining payments, the yield to maturity $\bar{y}$ solves:

$$
V_B
=V_P\left(1+\frac{\bar{y}}{n}\right)^{-N}
+C\sum_{j=1}^{N}\left(1+\frac{\bar{y}}{n}\right)^{-j}.
$$

This is a root-finding problem: there is no closed-form solution for $\bar{y}$ in general, so it is computed numerically. Yield to maturity is useful for quoting and comparing securities, but it is one average rate over all of one security's payment dates. It does not describe the full term structure.

__Spot rates__ assign a rate to each payment date. The spot rate for maturity $t_j$ is the yield of a zero-coupon security that pays at $t_j$; equivalently, it fixes the discount factor for that one date. Writing $D(0,t_j)$ for the time-0 discount factor for a payment at $t_j$ (the same object as $\mathcal{D}_{j,0}^{-1}$, indexed by date rather than period), curve-based pricing values each payment at its own maturity:

$$
V_B=\sum_{j=1}^{N}X_jD(0,t_j).
$$

__Par yield__ is the coupon rate that makes a hypothetical coupon security trade exactly at face value under the current spot curve. Setting $V_B=V_P$ in the coupon-bond price and solving for the coupon rate gives:

$$
c_{\text{par}}
=n\cdot\frac{1-D(0,t_N)}{\sum_{j=1}^{N}D(0,t_j)}.
$$

The published Treasury yield curve is a par yield curve. Its level, slope, and curvature describe how discounting varies by maturity; they do not by themselves forecast future rates.

## Summary

The lecture stated Malkiel's five theorems, built the calculus that proves them, and closed with a complete yield curve. The same principle connects each step: the value of a fixed-income security depends on the timing of its cash flows and the rates used to discount them.

> __Key Takeaways:__
>
> * __Price and yield move in opposite directions:__ For fixed positive cash flows, price decreases as yield increases. Modified duration measures the local slope of the price-yield relationship, and convexity measures its curvature.
> * __Sensitivity depends on maturity, coupon, and yield level:__ Malkiel's theorems describe how maturity and coupon rate change price sensitivity under specific assumptions. The par restriction is what makes the percentage and dollar sensitivity rankings agree.
> * __One yield does not value every payment date:__ Yield to maturity summarizes one security with a single rate. A spot curve values each payment at its own maturity, and a par yield restates that curve as a coupon rate.

These tools extend beyond Treasury securities: duration and convexity estimate how the value of any fixed stream of promised payments responds to a small change in its yield.
