## Concept Review: Are Treasury Securities Really Risk-Free?
U.S. Treasury securities are commonly treated as low-default-risk nominal-dollar benchmarks because they are obligations of the U.S. government. The label is conditional: a fixed nominal payment can still expose its holder to interest-rate, inflation, liquidity, reinvestment, and operational risk. Sovereign securities from different issuers must be evaluated under their own legal, monetary, and credit arrangements.

Are Treasury securities risk-free? No. Of the risks just listed, two deserve particular attention here:
* __Default risk__: In this scenario, the issuer of the bill/note/bond does not pay back the debt, i.e., they default. The United States has never repudiated its debt, although the record is not literally spotless: an operational failure in 1979 delayed payment on roughly `120 million USD` of maturing Treasury bills, an episode some authors classify as a technical default. Many other sovereigns have defaulted outright in modern times. For example, [Greece restructured its privately held government debt in 2012](https://en.wikipedia.org/wiki/Greek_government-debt_crisis), an exchange that the rating agencies and the credit-derivatives market both treated as a default event, and [Russia defaulted in 1998](https://en.wikipedia.org/wiki/1998_Russian_financial_crisis). Defaulting makes borrowing more expensive (if not impossible) because the issuer has _poor credit_. For U.S. Treasury securities, the default risk is __low__.
* __Interest-rate risk__: If required yields change, the secondary-market value of fixed promised cash flows changes. Selling before maturity can realize that gain or loss, while institutions that mark positions to market can be affected even without a sale. The magnitude is not universally `high` or `low`; it depends on remaining maturity, coupon, yield, embedded options, and the size and direction of the rate shock.

Let's look at another recent case study that demonstrates _interest rate risk_.


### Case Study: Interest Rate Risk and Silicon Valley Bank (SVB)
Silicon Valley Bank collapsed in March 2023 after rising interest rates eroded the value of its long‐duration bond portfolio. When a sudden run on deposits by tech clients occurred, the rapid outflow of funds left the bank unable to meet liquidity needs, leading to its closure by regulators.
* __What happened__? Silicon Valley Bank (SVB) invested heavily in long-duration U.S. government securities while interest rates were low. As the Federal Reserve raised rates in 2022–2023, the value of these bonds fell, leading to significant unrealized losses. In early March 2023, many uninsured tech-startup depositors withdrew funds. When SVB announced a capital raise to cover its realized bond losses, panic ensued, resulting in bank failure. 
* __What was the aftermath?__ The [Federal Deposit Insurance Corporation (FDIC)](https://www.fdic.gov) took control of SVB on March 10, 2023. With roughly `209 billion USD` in total assets, SVB was at that moment the second largest failure of an FDIC-insured bank on record; it now ranks third, behind Washington Mutual in 2008 and First Republic Bank, which failed about seven weeks later on May 1, 2023. To prevent a broader banking crisis, the U.S. government stepped in to protect all depositors, including those with uninsured balances. The [U.S. Federal Reserve](https://www.federalreserve.gov) also introduced emergency lending facilities to provide liquidity to banks facing similar issues.

## Zero-Coupon Treasury Bill

Let's start with the simplest case: a Treasury bill that makes no interest payments, and just returns your money plus profit at maturity.

> **Why start here?** T-bills are the "hydrogen atom" of finance: simple enough to understand completely, but containing all the fundamental principles that apply to more complex bonds.

Previously, we showed that the price of a zero-coupon Treasury bill is given by: $V_{B} = \mathcal{D}_{T,0}^{-1}(y;n)\;V_{P}$, where $V_{P}$ is the face (par) value of the bill, and $\mathcal{D}_{T,0}^{-1}(y;n)$ is the (inverse) discrete multiperiod discount factor for compounding events $0\rightarrow nT$. This is given by: $\mathcal{D}_{T,0}^{-1}(y;n) = \left(1+y/n\right)^{-nT}$, where $n$ is the number of compounding periods per year, and $T$ is the time to maturity in years. The per-period factor $1+y/n$ carries the yield, so every discount factor below is differentiated with the chain rule: $d\left[\left(1+y/n\right)^{-k}\right]/dy = -(k/n)\left(1+y/n\right)^{-k-1}$ for fixed $k$. The $1/n$ comes from differentiating $1+y/n$, and the extra factor of $\left(1+y/n\right)^{-1}$ from the power rule. Setting $k=nT$ gives the $T/(1+y/n)$ in the bill sensitivity below; for a coupon bond, the exponents $N$ and $j$ stay as weights inside the bracket once the common $1/\left[n(1+y/n)\right]$ is factored out. The nominal annual yield $y$ is the rate we use to discount the bill's single promised payment, and throughout this lecture we assume it is positive, $y>0$, as it has been for the U.S. Treasury data we use. Several results below fail without that assumption, and we flag each place where it does real work.

> **Intuition**: You pay less than 100 USD today to get exactly 100 USD back later. The factor $\mathcal{D}_{T,0}^{-1}(y;n)$ tells you exactly how much less we pay based on current interest rates and time to maturity.

### Deriving the Price Sensitivity
Now comes the key question: **How does the T-bill price change when market conditions change?**

Three things could in principle move the price of a fixed-income security: the yield $y$, the annual coupon rate $c$, and the time to maturity $T$. For a bill the coupon channel is closed by the contract itself, which fixes $c=0$ permanently, so $dc=0$ and only two channels remain. The bill price $V_{B}(y,T) = V_{P}\left(1+y/n\right)^{-nT}$ is a smooth function of those two variables, so we can take its total differential directly:
$$
\begin{aligned} 
dV_{B} &= \frac{\partial V_{B}}{\partial y}\,dy+\frac{\partial V_{B}}{\partial T}\,dT\\ &= -\underbrace{V_{P}\left(1+y/n\right)^{-nT}\left(\frac{T}{1+y/n}\right)}_{V_{B,y}}\;dy - \underbrace{V_{P}\left(1+y/n\right)^{-nT}\bigl(n\log(1+y/n)\bigr)}_{V_{B,T}}\;dT\\ &= -V_{B,y}\;dy - V_{B,T}\;dT\quad\blacksquare 
\end{aligned}
$$

> **Key insight, stated carefully**: it is tempting to say the coupon channel drops out because a bill pays no coupons, so its coupon derivative must be zero. That is the wrong reason. Holding the payment schedule fixed, the coupon-bearing family of the next section has $\partial V_{B}/\partial{c} = (V_{P}/n)\sum_{j}\mathcal{D}_{j,0}^{-1}(y;n)$, which is strictly positive even when evaluated at $c=0$. The channel is closed because the *contract* fixes the coupon rate at zero, so $dc=0$. A bill is not a security that is insensitive to coupons; it is a security whose coupon cannot move.

### Is the Maturity Channel Legitimate?
A bill promises a single payment, so its price depends on $T$ only through the exponent $nT$: there is no schedule of intermediate payments for a change in $T$ to add or remove. The partial derivative $\partial V_{B}/\partial{T}$ is therefore well defined here. But two different experiments hide behind the symbol $dT$, and we should keep them apart:
* __Changing the contractual maturity__: comparing a 6-month bill with a 12-month bill compares two securities issued on different terms. For a bill this is still a smooth comparison, because only the exponent changes. For a coupon security it is not, because changing the maturity adds or removes payments from a discrete schedule.
* __The passage of time__: holding one bill, calendar time $t$ advances while the remaining maturity falls, so $dT = -dt$ along the position. For a bill this is smooth over the entire holding period. For a coupon security it is smooth only between coupon dates, and it breaks each time a coupon leaves the remaining schedule.

Both distinctions return when we price coupon-bearing notes and bonds below.

### The Two Sensitivity Magnitudes
The term $V_{B,y}$ is the magnitude of the price sensitivity to a change in the yield, and $V_{B,T}$ is the magnitude of the sensitivity to a change in the time to maturity. Read these as magnitudes, not as the derivatives themselves: the derivatives $\partial{V_{B}}/\partial{y}$ and $\partial{V_{B}}/\partial{T}$ are the negatives of these quantities, which is why the minus signs appear explicitly in the total differential above. Under our standing assumption $y>0$, both magnitudes are positive:
$$
\begin{aligned} 
V_{B,y} &= \underbrace{V_{P}\left(1+y/n\right)^{-nT}}_{=\;V_{B}\;>\;0}\;\underbrace{\left(\frac{T}{1+y/n}\right)}_{>\;0\;\;\text{for}\;T>0}\\ 
V_{B,T} &= \underbrace{V_{P}\left(1+y/n\right)^{-nT}}_{=\;V_{B}\;>\;0}\;\underbrace{\Bigl(n\,\log\left(1+y/n\right)\Bigr)}_{=\;g_{y}\;>\;0\;\;\text{only if}\;y>0} 
\end{aligned}
$$
where $g_{y}:=n\log\left(1+y/n\right)$ is the continuously compounded equivalent of the nominal yield $y$ that we defined in L2a. The first magnitude does not require $y>0$; it remains positive whenever $1+y/n>0$, equivalently $y>-n$, because then $V_{B}>0$ and $T/(1+y/n)>0$ for $T>0$. The second one does, because $g_{y}$ carries the sign of $y$. At $y=0$ we get $g_{y}=0$ and maturity stops mattering: every discount factor equals one and the bill is worth par at any $T$. For a negative yield, i.e., $-n<y<0$, we would have $g_{y}<0$, a longer bill would be worth **more**, and the bill would trade **above** par, so the "pay less than 100 USD today" picture above would fail as well. U.S. Treasury yields have been positive over the sample period we use in this course, but negative nominal yields have occurred in other sovereign markets, so this is a condition to state rather than to assume silently.

### What This Says to Us

Both magnitudes are positive, so each channel moves the price in a definite direction.

**Decreasing in the yield**: holding maturity fixed, a higher yield discounts the face value more heavily, so $V_{B}\downarrow$ as $y\uparrow$.

> **Example**: you buy a 52-week bill at auction and the yield moves from 2% to 3%. The secondary-market price is now **lower** than you paid.

**Decreasing in the time to maturity**: holding $y>0$ fixed, a longer bill is worth less, so $V_{B}\downarrow$ as $T\uparrow$. As you *hold* the bill, though, the remaining time falls ($dT<0$).

> **Example**: six months into that same bill, with the yield unchanged, the price is **higher** than you paid, because fewer periods remain to discount. At $y=0$ the price would not move at all.

## Coupon-Bearing Treasury Notes and Bonds

Previously, we showed that the price of a coupon-bearing Treasury note (bond) with $N=nT$ remaining payment dates is given by:

$$V_{B} = \mathcal{D}_{N,0}^{-1}(y;n)\cdot V_{P} + C\cdot\sum_{j=1}^{N}\mathcal{D}_{j,0}^{-1}(y;n)$$

where $C\equiv(c/n)\cdot V_{P}$ is a coupon payment, $c$ is the annual coupon rate, and $\mathcal{D}_{j,0}^{-1}(y;n) = \left(1+y/n\right)^{-j}$. For a fixed cash-flow schedule, price sensitivities to yield and coupon are well defined:

$$
\begin{align*}
    \frac{\partial V_B}{\partial y} &= -\frac{1}{n(1+y/n)}\left[NV_P\,\mathcal{D}_{N,0}^{-1}(y;n)+C\sum_{j=1}^{N}j\,\mathcal{D}_{j,0}^{-1}(y;n)\right] < 0,\\
    \frac{\partial V_B}{\partial c} &= \frac{V_P}{n}\sum_{j=1}^{N}\mathcal{D}_{j,0}^{-1}(y;n) > 0.
\end{align*}
$$

The first derivative is negative because every promised cash flow is positive; the second is positive because increasing the contractual coupon adds cash flow. That second derivative is the one we just used to close the coupon channel for a bill: it is strictly positive even at $c=0$, so a bill's coupon term drops out only because $dc=0$. Remaining maturity is different, exactly as we flagged for the bill: $N$ is a discrete schedule index, and changing it adds or removes cash flows. Treating $T$ as a smooth variable while also using a finite sum with upper limit $nT$ is not a valid total differential without specifying a continuous extension. Maturity comparisons should instead price two complete schedules or compare their duration and convexity.

### What This Says to Us

Three channels, and only two of them have a definite sign.

**Decreasing in the yield**: a higher yield discounts both the coupon stream and the principal more heavily, so $V_{B}\downarrow$ as $y\uparrow$, holding $c$ and $T$ fixed.

> **Example**: you hold a 10-year note paying 3% coupons, bought when comparable securities yielded 3%. If market yields rise to 4%, its price falls: a new investor can earn 4% elsewhere, and will buy yours only at a price that matches.

**Increasing in the coupon rate, but not proportional to it**: a larger coupon means more cash flow to the holder, so $V_{B}\uparrow$ as $c\uparrow$, holding $y$ and $T$ fixed. The dependence is affine, not proportional: $\partial{V_{B}}/\partial{c}$ does not depend on $c$, and the discounted par payment $\mathcal{D}_{N,0}^{-1}(y;n)\,V_{P}$ survives even at $c=0$.

> **Example**: of two 10-year notes, one paying 2% coupons and one paying 4%, the 4% note is worth more, but **not** twice as much: doubling $c$ doubles the coupon stream and leaves the par payment untouched.

**Maturity does not determine the price direction by itself**: at a fixed coupon and yield, adding maturity adds coupons but postpones principal. A __discount bond__ ($c<y$, below par) drifts up toward par as maturity shortens, a __premium bond__ ($c>y$, above par) drifts down toward par, and a __par bond__ ($c=y$) stays at par on a coupon date. Malkiel's maturity result is about _sensitivity to a yield change_, not direction: otherwise comparable longer-maturity bonds are generally more yield-sensitive, with the increment diminishing at long maturities.

> **Example**: a 30-year 4% bond held at a constant 5% yield sits below par. Five years on, with 25 years left and the yield still 5%, it is closer to par and so worth more, apart from coupon-date and accrued-interest effects.

## Duration and Convexity

The derivatives above answer *which direction* the price moves when the yield changes. Risk management also needs *how much*, in units that can be compared across securities of different sizes. We get both by writing our securities in a single form and then normalizing the yield derivatives by the price.

Everything in this section applies to an __option-free__ security, meaning one whose promised cash flows are fixed in advance and do not change when yields change. Conventional fixed-rate, noncallable Treasury bills, notes, and bonds qualify. Callable bonds, inflation-indexed securities, and mortgage-backed securities do not, because a rate move can change the cash flows themselves.

Collect the promised payments into a fixed schedule. Let $N$ be the number of remaining compounding periods, and let $X_{j}\geq{0}$ be the dollar cash flow paid at period index $j\in\{1,\dots,N\}$, with at least one $X_{j}>0$. A coupon note pays $X_{j}=C$ at each coupon date and $X_{N}=C+V_{P}$ at maturity. A bill pays nothing until maturity, so $X_{j}=0$ for $j<N$ and $X_{N}=V_{P}$. The price is given by:
$$
V_{B}(y) = \sum_{j=1}^{N}X_{j}\,\mathcal{D}_{j,0}^{-1}(y;n)
$$
This is the same expression we have already written twice, once for the bill and once for the note. Only the cash-flow vector $X_{j}$ changes.

### Two Derivatives, One Shape
Differentiating the price twice with respect to the yield $y$, holding the cash flows $X_{j}$ and the schedule length $N$ fixed, gives:
$$
\frac{dV_{B}}{dy} = -\frac{1}{n}\sum_{j=1}^{N}j\,X_{j}\left(1+y/n\right)^{-j-1}<0,\qquad\frac{d^{2}V_{B}}{dy^{2}} = \frac{1}{n^{2}}\sum_{j=1}^{N}j\,(j+1)\,X_{j}\left(1+y/n\right)^{-j-2}>0
$$
where the strict signs follow from $1+y/n>0$ and from at least one $X_{j}$ being strictly positive: no term of either sum can be negative, and at least one term of each is positive. So the price-yield curve is __decreasing__ and __strictly convex__. That second fact is called __positive convexity__, and it is the property that makes the price response asymmetric. We use it in a moment.

### The Two Risk Measures
Dividing each derivative by the price removes the dependence on the size of the position and leaves two local risk measures. The __modified duration__ $D_{\mathrm{mod}}(y)$ and the __convexity__ $K(y)$ are given by:
$$
D_{\mathrm{mod}}(y) := -\frac{1}{V_{B}(y)}\frac{dV_{B}}{dy},\qquad K(y) := \frac{1}{V_{B}(y)}\frac{d^{2}V_{B}}{dy^{2}}
$$
where $V_{B}(y)>0$ is the price at yield $y$. Both are positive, by the signs established above. When the yield $y$ is quoted as a decimal per year, $D_{\mathrm{mod}}$ has units of years and $K$ has units of years squared.

> __Modified is not Macaulay__: the __Macaulay duration__ $D_{\mathrm{mac}}$ is the present-value-weighted average time to payment, and the two are related by $D_{\mathrm{mod}} = D_{\mathrm{mac}}/\left(1+y/n\right)$. For a zero-coupon bill every dollar arrives at maturity, so $D_{\mathrm{mac}} = T$ exactly and $D_{\mathrm{mod}} = T/(1+y/n)$. Compare that with the bill sensitivity we derived earlier: $V_{B,y}/V_{B} = T/(1+y/n)$. The duration machinery reproduces the earlier result.

### Using Them: the Second-Order Price Approximation
A second-order Taylor expansion of $V_{B}(y)$ around the current yield, divided by the price, turns the two measures into a repricing estimate. For a yield shock $\Delta{y}$, the approximate fractional price change is given by:
$$
\frac{\Delta V_{B}}{V_{B}}\approx-D_{\mathrm{mod}}\,\Delta{y}+\frac{1}{2}\,K\,\left(\Delta{y}\right)^{2}
$$
where $\Delta{V_{B}}\equiv{V_{B}(y+\Delta{y})-V_{B}(y)}$ is the price change. The duration term is the local slope, i.e., the straight-line estimate. The convexity term corrects that straight line for curvature, and because $K>0$ the correction is positive for a shock of either sign. This is a conditional statement rather than a forecast: it says what the price does *given* an assumed yield move, and says nothing about whether that move will happen.

> __The basis-point convention__: this formula takes $\Delta{y}$ as a decimal change in the annual yield. One basis point is $10^{-4}$, so a 25-basis-point shock is $\Delta{y}=0.0025$ and a 50-basis-point shock is $\Delta{y}=0.0050$. Note that a 50-basis-point change in the Fed's policy rate does **not** imply $\Delta{y}=0.0050$ for a given Treasury security: the curve can move by different amounts at different maturities, a distinction we take up in the next section.

With that convention fixed, the two terms are easiest to compare with numbers.

> __Example__: a note with $D_{\mathrm{mod}}=7.2$ years and $K=65$ years$^{2}$ faces a $+50$ basis point shock to its own yield, so $\Delta{y}=0.0050$. The duration term contributes $-7.2\times{0.0050}=-0.0360$ and the convexity term contributes $\tfrac{1}{2}\times{65}\times(0.0050)^{2}=+0.0008$. The estimate is a $-3.52\%$ price change, where the straight-line estimate alone would have said $-3.60\%$.

Both measures are **local**: they are derivatives evaluated at the current yield, so the approximation degrades as $\Delta{y}$ grows. For the multiple-percentage-point rate moves of 2022 and 2023 that hurt SVB, the honest calculation is to reprice the entire cash-flow schedule at the new yield rather than to extrapolate from duration.

## Connection to Malkiel's Bond Pricing Theorems

With duration and convexity in hand, we can state Burton Malkiel's bond pricing theorems and say *why* each one holds. Two ground rules matter. First, every theorem assumes an option-free security with fixed positive cash flows and holds the remaining contract terms fixed. Second, Malkiel's maturity theorems compare bonds **selling at par**, i.e., $c=y$, and measure price changes from that starting point. That restriction is not a technicality, and we show below what it buys us.

The par case has a clean closed form. Substituting $C=(c/n)V_{P}$ into the coupon-bond price and summing the geometric series gives:
$$
\frac{V_{B}}{V_{P}} = \frac{c}{y}+\left(1-\frac{c}{y}\right)\left(1+y/n\right)^{-N},\qquad{N}=nT
$$
which returns $V_{B}=V_{P}$ whenever $c=y$, as it must. Differentiating with respect to $y$ at fixed $c$, and only then setting $c=y$, gives the modified duration of a par bond:
$$
D_{\mathrm{mod}}(N) = \frac{1-\left(1+y/n\right)^{-N}}{y}
$$
The order of those two steps matters: setting $c=y$ first would differentiate along the par locus, where the price is pinned at $V_{P}$, and would return zero. This single expression drives Theorems 2 and 3. Assume $y>0$ throughout.

* __Theorem 1__ (Bond prices move inversely to bond yields): this is the sign $dV_{B}/dy<0$ established above. Every promised payment is positive, so raising $y$ discounts each of them more heavily. Equivalently, $D_{\mathrm{mod}}>0$.
* __Theorem 2__ (Longer maturity increases yield sensitivity): from the par-bond formula, $D_{\mathrm{mod}}(N)=(1-\left(1+y/n\right)^{-N})/y$ is strictly increasing in $N$, since $\left(1+y/n\right)^{-N}$ shrinks. The dollar sensitivity moves the same way, because a par bond starts at $V_{B}=V_{P}$ whatever its maturity, so $-dV_{B}/dy = V_{P}\,D_{\mathrm{mod}}(N)$ also increases. Holding the starting price at par guarantees that the percentage and dollar rankings agree.
* __Theorem 3__ (The sensitivity gain diminishes as maturity lengthens): extending the schedule from $N$ to $N+1$ periods raises modified duration by:
$$
D_{\mathrm{mod}}(N+1)-D_{\mathrm{mod}}(N) = \frac{\left(1+y/n\right)^{-N}-\left(1+y/n\right)^{-N-1}}{y} = \frac{\left(1+y/n\right)^{-N-1}}{n}
$$
  Successive increments shrink by the constant factor $\left(1+y/n\right)^{-1}<1$, so they decay geometrically: each extra period of maturity buys strictly less sensitivity than the one before. Modified duration therefore climbs toward the ceiling $1/y$, which is the modified duration of a __perpetuity__, a hypothetical security that pays its coupon forever and never repays principal.
* __Theorem 4__ (Price movements are asymmetric): a yield decrease of $\Delta{y}$ raises the price by more than an equal yield increase lowers it, measured from the same starting yield. The reason is **positive convexity**, $d^{2}V_{B}/dy^{2}>0$, which we established above. In the second-order approximation the duration term $-D_{\mathrm{mod}}\Delta{y}$ flips sign with the shock, while the convexity term $\tfrac{1}{2}K(\Delta{y})^{2}$ is positive for a shock of either sign: it adds to the gain and subtracts from the loss. Nonlinearity alone would not be enough to conclude this, since a concave price-yield curve would produce the opposite asymmetry.
* __Theorem 5__ (Higher coupon bonds are less sensitive to yield changes): this is a statement about **percentage** sensitivity, and it needs at least two remaining payment dates, $N\geq{2}$. A larger coupon moves a larger share of present value to earlier dates, which lowers the present-value-weighted average payment time and therefore lowers $D_{\mathrm{mod}}$. In dollars the comparison runs the other way: a higher-coupon bond has more cash flow and a higher price, so its $-dV_{B}/dy$ is larger, not smaller. The $N=1$ case is the excluded edge: with everything paid on one date, $D_{\mathrm{mod}}=1/\left[n(1+y/n)\right]$ no matter how large the coupon is.

> __What the par restriction is doing__: step outside it and the percentage and dollar statements can part company. For a zero-coupon bill, $D_{\mathrm{mod}}=T/(1+y/n)$ grows without bound, but the dollar sensitivity $-dV_{B}/dy = V_{P}\,T\left(1+y/n\right)^{-nT-1}$ is a product of a growing factor $T/(1+y/n)$ and a shrinking factor $V_{B}=V_{P}\left(1+y/n\right)^{-nT}$. It therefore rises, peaks, and falls: for $V_{P}=100$, $n=1$, and $y=5\%$ it peaks near $T\approx{20.5}$ years at about $718$ USD per unit of yield, then declines to about $541$ by $T=40$ years. A 40-year bill is far more sensitive than a 20-year bill in percentage terms and *less* sensitive in dollars. The two rankings happen to agree for short maturities, but only the par comparison guarantees it.

Our framework therefore does more than restate Malkiel: it tells us which of his statements are about percentages, which are about dollars, which require curvature rather than slope, and which comparison has to be held fixed for any of it to be true.

## From One Yield to the Yield Curve

Yield to maturity is the single internal rate that reproduces one security's price from its promised cash flows. It is useful for quoting and comparison, but it is not the full term structure. A **spot rate** applies to one payment date, so a spot curve supplies the discount factors used to price each cash flow. A **par yield** is the coupon rate that makes a hypothetical security trade at par under that curve.

For payments $X_j$ at times $t_j$, the curve-based price is
$$V_B=\sum_{j=1}^{N}X_jD(0,t_j),$$
where $D(0,t_j)$ is the time-0 discount factor for maturity $t_j$. Curve level, slope, and curvature describe how discounting varies by maturity; they do not, by themselves, identify a unique forecast of future short rates.

## Summary
This lecture developed the core tools used to describe and manage the interest-rate risk of fixed-income securities.

> __Key Takeaways:__
>
> * __Bond prices decrease as yields increase:__ For a fixed stream of positive cash flows, the price-yield relationship is decreasing and convex.
> * __Duration and convexity measure local price sensitivity:__ Duration gives the first-order response to a yield change, while convexity improves the approximation by accounting for curvature.
> * __One yield is not a yield curve:__ Yield to maturity summarizes one security with a single rate; spot rates and discount factors value cash flows at specific maturities, and par yields are implied by that term structure.

Together, price-yield geometry, duration, convexity, and the yield curve form a practical fixed-income risk toolkit.
