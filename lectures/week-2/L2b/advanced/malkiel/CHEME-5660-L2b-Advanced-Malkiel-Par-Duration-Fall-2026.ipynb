# Advanced: Par-Bond Duration and Malkiel's Maturity Theorems

The L2b lecture states Malkiel's five bond-pricing theorems and proves Theorems 1, 4, and 5 using duration and convexity. Theorems 2 and 3, the maturity theorems, are argued there qualitatively. This note supplies the quantitative version: we derive the closed-form price of a coupon bond, compute the modified duration of a par bond, and show that par-bond duration increases with maturity at a diminishing rate. We close with a zero-coupon counterexample showing why the par restriction matters.

> __Learning Objectives:__
>
> By the end of this note, you will be able to:
>
> * __Derive the closed-form price of a coupon bond:__ Sum the discounted coupon stream as a geometric series and combine it with the discounted principal. Read the premium, discount, and par cases directly from the resulting expression.
> * __Compute the modified duration of a par bond:__ Differentiate the price with respect to yield while holding the coupon rate fixed, and only then impose the par condition. Explain why reversing that order compares different bonds instead of measuring one bond's sensitivity.
> * __Prove Malkiel's maturity theorems:__ Show that each added payment period increases a par bond's duration by a positive amount that shrinks geometrically toward the perpetuity limit. Explain why dropping the par restriction lets percentage and dollar sensitivity rank maturities differently.

Throughout, $V_P$ is the face value, $y>0$ is the nominal annual yield with $n$ compounding periods per year, $c$ is the annual coupon rate, $N=nT$ is the number of remaining payment dates, and $C=(c/n)V_P$ is the coupon paid each period.

___

## The Closed-Form Price of a Coupon Bond

The price of a coupon bond with $N$ remaining payments is the discounted principal plus the discounted coupon stream, and is given by:

$$
V_B
=V_P\left(1+\frac{y}{n}\right)^{-N}
+C\sum_{j=1}^{N}\left(1+\frac{y}{n}\right)^{-j}.
$$

The coupon sum is geometric. With ratio $x=(1+y/n)^{-1}$, the finite geometric sum is given by:

$$
\sum_{j=1}^{N}x^{j}
=\frac{x\left(1-x^{N}\right)}{1-x}
=\frac{n}{y}\left[1-\left(1+\frac{y}{n}\right)^{-N}\right],
$$

where the second equality substitutes $x=(1+y/n)^{-1}$ and simplifies $1-x=(y/n)/(1+y/n)$. Substituting the sum into the price, using $C=(c/n)V_P$, and dividing by the face value gives the closed form:

$$
\frac{V_B}{V_P}
=\frac{c}{y}
+\left(1-\frac{c}{y}\right)
\left(1+\frac{y}{n}\right)^{-N}.
$$

Rearranging gives $V_B/V_P=1+(c/y-1)\left[1-(1+y/n)^{-N}\right]$, which recovers the premium, discount, and par vocabulary from L2a: the price sits above face value when $c>y$, below face value when $c<y$, and exactly at face value, for every maturity, when $c=y$.

___

## Modified Duration of a Par Bond

Malkiel's maturity theorems compare bonds that begin at par, so $c=y>0$. The order of operations matters here: we differentiate the price with respect to $y$ while holding $c$ fixed, and only then evaluate at $c=y$. Setting $c=y$ before differentiating would compare different par bonds, whose prices all equal face value, instead of measuring how one bond's price responds to a yield change.

Differentiating the closed form with respect to $y$ at fixed $c$ gives:

$$
\frac{1}{V_P}\frac{\partial V_B}{\partial y}
=-\frac{c}{y^{2}}\left[1-\left(1+\frac{y}{n}\right)^{-N}\right]
-\left(1-\frac{c}{y}\right)\frac{N}{n}\left(1+\frac{y}{n}\right)^{-N-1}.
$$

Evaluating at $c=y$ removes the second term and reduces the prefactor of the first term to $1/y$. The par bond's price is $V_B=V_P$, so the modified duration of a par bond is given by:

$$
D_{\mathrm{mod}}(N)
=-\frac{1}{V_B}\frac{\partial V_B}{\partial y}\bigg|_{c=y}
=\frac{1-\left(1+y/n\right)^{-N}}{y}.\quad\blacksquare
$$

___

## Theorems 2 and 3, Quantitatively

__Theorem 2 (longer maturity increases yield sensitivity):__ Adding one period to a par bond changes its modified duration by:

$$
D_{\mathrm{mod}}(N+1)-D_{\mathrm{mod}}(N)
=\frac{\left(1+y/n\right)^{-N}-\left(1+y/n\right)^{-N-1}}{y}
=\frac{\left(1+y/n\right)^{-N-1}}{n},
$$

which is positive for every $N$: a longer par bond always has greater modified duration. Because every bond in the comparison starts at face value $V_P$, the percentage and dollar sensitivity rankings agree. $\blacksquare$

__Theorem 3 (the sensitivity gain diminishes as maturity lengthens):__ Successive increments shrink by the factor $(1+y/n)^{-1}<1$, so each additional period contributes less duration than the previous one. In the limit, the par-bond duration approaches:

$$
\lim_{N\to\infty}D_{\mathrm{mod}}(N)=\frac{1}{y},
$$

the modified duration of a perpetuity. Duration therefore increases with maturity but saturates. $\blacksquare$

___

## Outside the Par Comparison: A Zero-Coupon Counterexample

Outside the par comparison, percentage and dollar sensitivity can have different maturity rankings, and a zero-coupon bond provides the counterexample. Its modified duration is given by:

$$
D_{\mathrm{mod}}=\frac{T}{1+y/n},
$$

which increases without bound as $T$ grows. Its dollar sensitivity is given by:

$$
-\frac{dV_B}{dy}
=V_PT\left(1+\frac{y}{n}\right)^{-nT-1}.
$$

The dollar expression multiplies a growing duration term by a shrinking discounted price, so it cannot grow forever. Writing the power as an exponential and setting the $T$-derivative to zero locates the peak at:

$$
T^{\star}=\frac{1}{n\log\left(1+y/n\right)}.
$$

For annual compounding ($n=1$) and $y=5\%$, the peak falls at $T^{\star}=1/\log(1.05)\approx 20.5$ years: beyond roughly twenty years, a longer bond has greater percentage sensitivity but smaller dollar sensitivity. The par restriction rules this divergence out by holding the starting price fixed at face value across maturities, which is why the percentage and dollar rankings agree in Malkiel's maturity theorems.

## Summary

This note made Malkiel's maturity theorems quantitative: it derived the closed-form price of a coupon bond, computed the modified duration of a par bond, and traced how that duration grows with maturity.

> __Key Takeaways:__
>
> * __Par bonds admit closed forms:__ The coupon-bond price collapses to a two-term expression in the coupon rate, the yield, and the number of remaining payments. Evaluating its yield derivative at the par condition gives a compact formula for the par bond's modified duration.
> * __Duration rises with maturity at a diminishing rate:__ Each added payment period contributes a positive duration increment, and successive increments shrink geometrically toward the perpetuity limit. These two facts are Malkiel's Theorems 2 and 3.
> * __The par restriction matters:__ Starting every bond in the comparison at face value aligns the percentage and dollar sensitivity rankings. A zero-coupon bond shows the rankings separating once maturity passes the dollar-sensitivity peak.

These results replace the lecture's qualitative argument for Theorems 2 and 3 with formulas you can evaluate and reuse.
