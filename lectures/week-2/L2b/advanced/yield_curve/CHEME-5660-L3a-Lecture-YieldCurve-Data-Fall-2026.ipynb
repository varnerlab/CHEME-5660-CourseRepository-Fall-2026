## Motivation
On the secondary market, Treasury securities are traded at prices that reflect the market's expectations for future interest rates, inflation, and economic growth. The market value of a security is its current price, but quotation conventions often express that price as a yield to maturity (YTM) rather than in dollars.

> __How are Treasury securities quoted?__ Quotation depends on the instrument. Bills are commonly reported with bank-discount and investment rates, while notes and bonds use price and bond-equivalent yield conventions; secondary coupon-security prices also distinguish clean from accrued-interest-adjusted dirty price. Yield to maturity is the single discount rate that reproduces a security's price under its cash-flow and compounding assumptions. A quoted yield reflects current prices but is not, by itself, a forecast of future short rates or a guaranteed holding-period return.

## Term Structure of Interest Rates
The term structure of interest rates describes the relationship between the interest rates of Treasury securities and their maturities. 
> __Term structure of interest rates__ is the relationship between the __yields__ on fixed income debt securities of different maturities but similar credit quality, observed at a specific point in time. It typically illustrates how interest rates vary with the length of time until a bond's maturity.

To understand the term structure of interest rates, we need to understand a few key concepts. Let's start with the __yield__ and the __spot rate__.

___

## Investment Yield for Zero-Coupon Instruments
The __investment yield__ is the return on an investment, expressed as a percentage of the investment's cost. Suppose we are looking at zero-coupon Treasury securities, e.g., T-bills that do not pay periodic interest but are sold at a discount to their face value. 
> __Investment yield__: The investment yield on a zero-coupon bond, e.g., T-bill is the difference between the purchase price and the face value, divided by the purchase price, multiplied by a factor to annualize the yield.

The __annualized investment yield__ of a zero-coupon instrument with $t$ days to maturity (units: days), face (par) value $V_{P}$ (units: USD), and purchase price $V_{B,t}$ at time $t$ (units: USD) is given by:
$$
\begin{align*}
Y(t) &= \left(\frac{V_{P} - V_{B,t}}{V_{B,t}}\right) \times \left(\frac{365}{t}\right)\\
&= \left(\frac{V_{P}}{V_{B,t}} - 1\right) \times \left(\frac{365}{t}\right)\quad\text{let}\;T = \frac{t}{365}\\
& = \underbrace{\left(\frac{V_{P}}{V_{B,t}} - 1\right)}_{\text{HPR}} \times \underbrace{\left(\frac{1}{T}\right)}_{\text{annualize}}\quad\blacksquare\\
\end{align*}
$$
where $\text{HPR}$ is the __holding period return__ and the $1/T$ term is an __annualization factor__. 

> __Convention note:__ This simple-interest formula is the classroom convention used throughout this material. The official Treasury investment-rate calculation matches the simple-interest form only for bills with six months or less to maturity; longer bills use a coupon-equivalent formula. See [31 CFR Part 356, Appendix B](https://www.treasurydirect.gov/files/laws-and-regulations/auction-regulations-uoc/31-cfr-part-356.pdf).

### Spot Rate
Next, let's consider the __spot rate__.
> __Spot rate__: The spot rate is the annualized interest rate for borrowing or lending money for a specific time period, with compounding. For zero-coupon bonds, the spot rate is the rate that, when compounded, equates the present value of the face value to the current market price.

**Key Distinction**: 
- **Investment Yield** ($Y$): A simple annualized return based on buying at discount and receiving face value
- **Spot Rate** ($y$): The compounded interest rate used in present value calculations

To better understand the relationship between the yield and the spot (discount) rate, we can rewrite the yield as:
$$
\begin{align*}
Y &= \left(\frac{V_{P}}{V_{B}} - 1\right) \times \left(\frac{1}{T}\right)\quad\text{substitute}\;V_{B}=\mathcal{D}_{nT,0}^{-1}(y)\;V_{P}\\
Y & = \left(\frac{V_{P}}{\mathcal{D}_{nT,0}^{-1}(y)\;V_{P}} - 1\right)\times \left(\frac{1}{T}\right)\\
Y & = \left(\frac{1}{\mathcal{D}_{nT,0}^{-1}(y)} - 1\right)\times \left(\frac{1}{T}\right)\\
Y & = \left(\mathcal{D}_{nT,0}(y) - 1\right)\times \left(\frac{1}{T}\right)\quad\text{solve for the discount factor}\\
\mathcal{D}_{nT,0}(y) &= 1 + Y\;T\quad\text{use}\;\mathcal{D}_{nT,0}(y)=\left(1+y/n\right)^{nT}\\
\left(1+y/n\right)^{nT} & = 1 + Y\;T\quad\text{solve for}\;y\\
y & = n\left[\left(1 + Y\;T\right)^{1/(nT)} - 1\right]\quad\blacksquare
\end{align*}
$$
This derivation shows the relationship between the **investment yield** $Y$ (an annualized return calculation), 
the time to maturity $T$ (in years), the compounding frequency $n$ (number of compounding periods per year),
and the **spot rate** $y$ (the discount rate used to price a zero-coupon instrument at auction).

> __Example__: For a 52-week zero-coupon Treasury Bill with semiannual compounding ($n=2$, $T=1$), the relationship between the discount rate and the investment yield can be expressed as:
>$$
\boxed{
y = 2\left[\left(1 + Y\right)^{1/2} - 1\right] \quad\text{for}\;n=2,\;T=1\quad\text{(52-week zero-coupon)}}
$$
>where $Y$ is the annualized investment yield and $y$ is the corresponding __spot rate__ used in zero-coupon pricing calculations.

> __Yield to Maturity (YTM)__ represents the total return an investor earns if the note (bond) is held until maturity, every promised payment is received, and each coupon is reinvested at the same yield. It takes into account both the bond's current market price and its future cash flows (coupon payments and face value at maturity).

The YTM for a coupon-bearing note (bond) is calculated by computing the net present value (NPV) of the bond's __remaining__ cash flows, which include periodic coupon payments and the face value at maturity. The YTM is the _discount rate_ that makes the NPV of these cash flows equal to zero, where we include the purchase price of the bond as a cash flow event at time $t=0$:
$$
\underbrace{\texttt{NPV}(T_{\text{rem}}, \bar{Y}, \bar{c})}_{=\,0} = \mathcal{D}^{-1}_{nT_{\text{rem}},0}(\bar{Y})\cdot{V_{P}}+C\cdot\sum_{j=1}^{n{T_{\text{rem}}}}\mathcal{D}^{-1}_{j,0}(\bar{Y}) - V_{B} = 0
$$
where $C$ is the coupon payment, $V_{P}$ is the face value of the bond, $V_{B}$ is the current market price of the bond, $T_{\text{rem}}$ is the remaining time to maturity (in years), and $\bar{Y}$ is the yield to maturity (written $\bar{y}$ in the required L2b lecture). 

> __Auction__: At auction, the YTM is equal to the discount rate used to compute the present value of the bond's cash flows. This is because $T = T_{\text{rem}}$ at the time of auction, and the bond is priced to yield the YTM at that time. 
>
>__Secondary Market__: In the secondary market, the YTM is the discount rate that equates the present value of the bond's future cash flows to its current market price, given the remaining time to maturity $T_{\text{rem}}$ and the bond's coupon payments. The YTM reflects current market conditions, e.g., interest rates and perceived credit risk, and the time remaining until the bond matures.

The YTM can be computed using various numerical methods; however, let's look at a particularly simple method, the __Secant Method__.

### Estimation of YTM using the Secant Method
The secant method is a numerical technique used to find the root of a function of a single variable, which in this case is the YTM. Consider the general case of finding $x\in\mathbb{R}$ such that $f(x) = 0$.  

Let's look at some pseudo-code to illustrate the secant method for estimating the YTM.

__Initialization:__ Given two initial guesses $x_0$ and $x_1$, a tolerance level $\epsilon>0$, a maximum number of iterations $\texttt{maxiter}$, a convergence flag $\texttt{converged}\gets\texttt{false}$ and an iteration counter $i\gets{1}$.

> __Parameter Selection Rules of Thumb:__
>
> **Tolerance ($\epsilon$)**: For financial calculations like YTM, a tolerance of $\epsilon = 10^{-6}$ to $10^{-8}$ is typically appropriate. This provides accuracy to about 6-8 decimal places, which is sufficient for most practical applications. For YTM calculations, $\epsilon = 10^{-6}$ corresponds to about 0.0001% accuracy in the yield estimate.
>
>**Maximum iterations ($\texttt{maxiter}$)**: For well-behaved functions like the NPV equation for YTM, the secant method typically converges within 10-20 iterations. A safe choice is $\texttt{maxiter} = 50$ to $100$. If convergence hasn't occurred by then, it usually indicates a problem with the initial guesses or the function itself.
>
> **Initial guesses**: For YTM calculations, reasonable starting values are often $x_0 = 0.01$ (1%) and $x_1 = 0.10$ (10%), which bracket typical bond yields. The secant method is generally robust to the choice of initial guesses as long as they're not too close together.

While not $\texttt{converged}$ and $i\leq\texttt{maxiter}$ __do__:
1. Compute the next estimate $x_{i+1}$ using the secant formula:
   $$
   x_{i+1} = x_i - f(x_i) \cdot \frac{x_i - x_{i-1}}{f(x_i) - f(x_{i-1})}
   $$
   where $f(x_i)$ and $f(x_{i-1})$ are the function values at the current and previous estimates, respectively.
2. Check for convergence: if $|x_{i+1} - x_i| < \epsilon$, set $\texttt{converged}\gets\texttt{true}$. Otherwise, continue iterating. If the maximum number of iterations is reached without convergence, return the best solution found so far and warn the user.
3. Update for next iteration: set $x_{i-1} \gets x_i$, $x_i \gets x_{i+1}$, and increment $i \gets i + 1$.

__Convergence Properties for YTM Calculations:__

The secant method exhibits **superlinear convergence** (meaning the error decreases faster than linearly with each iteration) with convergence order approximately 1.618 (the golden ratio; this measures how rapidly the accuracy improves). This makes it faster than simple linear methods but slower than Newton's method (which has quadratic convergence order 2.0).

> __Why secant works__: The NPV function for bond pricing is smooth and monotonically decreasing, so the secant method typically converges within a few iterations from reasonable initial guesses. Convergence is not guaranteed, however: poor guesses can overshoot or make the secant denominator nearly zero, which is why the tolerance and iteration limit above matter. 

While Newton's method converges faster, the secant method achieves good convergence without requiring derivative calculations, making it a more attractive option for YTM calculations.

## Yield Curve
The yield curve is a graphical representation of the relationship between the yields of coupon-bearing Treasury securities and their maturities. It is believed to have some power to predict future economic conditions, such as recessions or expansions.

### Par Yield Curve Rates

When people talk about the yield curve, they typically mean the par yield curve. The __par yield__ for a given maturity is the coupon rate that makes a newly issued (standard) coupon-bearing bond trade exactly at par (face) value when priced with the current set of discount factors (spot rates). The par yield curve is the function that maps maturity $T$ to the par coupon rate $\bar c_{\text{par}}(T)$.

> __Intuitively:__ A bond is priced at par if its present value $V_{B}$ equals its (future) face value $V_{P}$. For a given set of spot (discount) factors, we can solve for the coupon rate that achieves this equality; that coupon rate is the par yield.

Let $V_{P}$ denote the face (par) value (normalize $V_{P}=1$ for simplicity), and let $n$ be the number of coupon payments per year. In general, each payment date carries its own spot rate, so each period $j$ has its own discount factor; the required L2b lecture writes this date-indexed factor as $D(0,t_j)$. To keep the notation light, we write $\mathcal{D}^{-1}_{j,0}(y)=(1+y/n)^{-j}$ below, which corresponds to a flat spot curve at rate $y$; every step goes through unchanged with maturity-specific rates. For a bond with maturity $T$ (years) there are $nT$ coupon payments.

The price of a coupon-bearing bond with coupon rate $\bar c$ (annualized) is:

$$
V_B(\bar c,T) 
= C\sum_{j=1}^{nT}\mathcal{D}^{-1}_{j,0}(y) + V_{P}\,\mathcal{D}^{-1}_{nT,0}(y),
$$

where the coupon payment per period is $C=(\bar c/n)\,V_{P}$. The _par yield_ $\bar c_{\text{par}}(T)$ satisfies $V_B(\bar c_{\text{par}},T)=V_{P}$. Setting $V_{P}=1$ and solving for $\bar c_{\text{par}}(T)$ gives:

$$
1 = \frac{\bar c_{\text{par}}(T)}{n}\sum_{j=1}^{nT}\mathcal{D}^{-1}_{j,0}(y) + \mathcal{D}^{-1}_{nT,0}(y)
$$

and therefore:

$$
\boxed{\displaystyle \bar c_{\text{par}}(T) 
= n\cdot\frac{1 - \mathcal{D}^{-1}_{nT,0}(y)}{\sum_{j=1}^{nT}\mathcal{D}^{-1}_{j,0}(y)}.}
$$

How should we interpret this expression for the par yield?

> __Interpretation__: The numerator $1-\mathcal{D}^{-1}_{nT,0}$ measures the fraction of value that is _not_ paid by the final principal payment (i.e., the portion that must be supplied by coupons), while the denominator sums the present value of the periodic coupon payments. Their ratio (times $n$) is the coupon rate needed to make the coupon payments plus discounted principal equal to par.

Ok, so par yields are a hypothetical construct. What do we use these values for, and what value do we use for the discount factors $\mathcal{D}^{-1}_{j,0}(y)$ (particularly, what is the $y$)?

In practice, the calculation runs in the other direction. The Treasury fits discount factors to quoted prices of recently auctioned securities and publishes the resulting par yields directly, so the par yield curve is the form in which the market's discounting reaches the public. The example below therefore works with published par yields rather than constructing them from spot rates.

## Summary
This material introduced the term structure of interest rates and the yield measures used to describe Treasury securities.

> __Key Takeaways:__
>
> * __Different securities require different yield concepts:__ A spot rate applies to a zero-coupon payment at one maturity, while yield to maturity is the single internal rate that reproduces a coupon security's price. Each measure carries its own cash-flow and compounding assumptions.
> * __A yield curve organizes rates by maturity:__ Spot, par, and yield-to-maturity curves describe related but distinct quantities and should not be treated as interchangeable. Always identify which curve a quoted rate comes from before using it.
> * __Curve shape is informative but not deterministic:__ Level, slope, and curvature summarize market prices across maturities. Interpreting them as forecasts requires additional economic assumptions.

Yield curves convert a cross section of Treasury prices into a compact view of maturity-dependent discounting.
