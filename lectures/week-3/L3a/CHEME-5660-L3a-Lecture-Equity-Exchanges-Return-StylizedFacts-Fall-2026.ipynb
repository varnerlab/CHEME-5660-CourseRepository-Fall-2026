## Exchanges, Stocks, and ETFs
Before measuring equity growth rates, let's discuss the marketplace in which the securities are traded. An __exchange__ is a trading venue, while __stocks__ and __exchange-traded funds (ETFs)__ are example securities whose prices are discovered through trading on the exchange. 

### Exchanges Connect Buyers and Sellers
An exchange is an organized venue that matches orders to buy and sell securities. By bringing many orders together, exchanges support __price discovery__ (the process by which trading establishes a market price) and __liquidity__ (the ability to trade without moving that price excessively). Exchanges also impose listing standards and trading rules. After a trade is executed, separate clearing and settlement infrastructure completes the transfer of securities and funds.

U.S. equities trade across more than a dozen competing exchanges. By [matched share volume](https://www.cboe.com/us/equities/market_share/?data_type=shares), the largest exchange-operator groups are typically [NYSE](https://www.nyse.com/), [Nasdaq](https://www.nasdaq.com/), and [Cboe](https://www.cboe.com/), although their market shares change through time. Each group operates multiple trading venues. For example, Cboe operates the BZX, BYX, EDGA, and EDGX exchanges. 

The [Securities and Exchange Commission (SEC)](https://www.sec.gov) maintains [the current list of registered national securities exchanges](https://www.sec.gov/about/divisions-offices/division-trading-markets/national-securities-exchanges), including equities, options, and other specialized venues. 

### Orders Are Instructions to Trade
An __order__ is an instruction to trade: it names the security, the direction (buy or sell), the quantity, and how the price you may be willing to transact at is to be set. The common order types differ in how they trade off immediacy against price control:

* __Market order__: A market order seeks execution as soon as possible against the best prices available when it reaches the market. Execution is likely, but the exact price is not guaranteed because a large order may consume several price levels. We use this type of order when immediacy matters more than price control.
* __Limit order__: A limit order specifies a limit price: a buy limit order executes only at the limit price or lower, and a sell limit order only at the limit price or higher. The price is controlled, but execution is not guaranteed. Limit orders that do not execute immediately rest on the venue's __order book__, the collection of outstanding buy and sell limit orders for a security. The highest resting buy price is the venue's displayed __bid__, the lowest resting sell price is its displayed __ask__, and the difference between them is the __bid-ask spread__. An arriving market order to buy begins trading against the best ask, while one to sell begins trading against the best bid.
* __Stop and stop-limit orders__: Stop orders are conditional instructions that become market orders when the price reaches a specified level. Stop-limit orders are similar, but they become limit orders at the trigger level. 

The [Types of Orders page at Investor.gov](https://www.investor.gov/introduction-investing/investing-basics/how-stock-markets-work/types-orders) defines these and other variations.

The [limit-order book notebook](advanced/order_book/CHEME-5660-L3a-Advanced-LimitOrderBook-Mechanics-Fall-2026.ipynb) in the Optional Advanced Material section develops how a book processes arriving orders. Because each venue keeps its own order book, each venue displays its own bid and ask. But which price is the "right" price? The answer is that there is no single "right" price, but there is a common benchmark: the __National Best Bid and Offer (NBBO)__. 

The __National Best Bid and Offer (NBBO)__ is the highest displayed bid (buy) and lowest displayed ask (sell) across the national market system. See [Regulation NMS](https://www.sec.gov/rules-regulations/2005/06/regulation-nms) for order protection and [FINRA Rule 5310](https://www.finra.org/rules-guidance/rulebooks/finra-rules/5310) for the best-execution standard applied to customer orders.

### Stocks and ETFs Are Claims on Firms and Portfolios
A stock, also called a __share__ or __equity security__, represents fractional ownership in a corporation. Shareholders may receive voting rights and dividends (cash payments per share), have limited liability for the corporation's debts, and hold a residual claim on its assets after creditors are paid. The economic reward from holding a stock can therefore come from two channels: cash distributions (dividends) and changes in the share price.

Where a stock is a claim on a single corporation, an exchange-traded fund (ETF) pools assets into a portfolio and issues shares that trade on an exchange throughout the day like a single named security. Each ETF share represents an interest in that portfolio.
* __Broad-market ETFs__: Funds such as [SPY](https://finance.yahoo.com/quote/SPY), which tracks the Standard & Poor's 500 index, hold hundreds of stocks and so provide diversified exposure. Its price series moves with the index it tracks.
* __Concentrated and leveraged ETFs__: An ETF does not guarantee diversification. A sector ETF focuses on one industry, and a single-stock ETF provides exposure to one stock. Leveraged or inverse ETFs use derivatives to target a multiple, or the negative, of a stock's or index's daily move. The fund's exposure determines its risk.
* __Market price and NAV__: An ETF has both a market price and a __net asset value (NAV)__ per share: the fund's assets minus its liabilities, divided by the number of shares outstanding. The two are usually close but may not be identical, so an ETF can trade at a premium or discount to NAV.

For our class, the essential point is the same for both security types: a stock or an ETF produces an observable sequence of market prices.

## Reward
Investment decisions involve a trade-off between expected risk and reward. Greater risk does not guarantee greater realized reward. Instead, investors generally require greater expected compensation for bearing additional uncertainty. We begin with one component of realized reward: price appreciation.

We measure the rate of __price growth__ using the continuously compounded growth rate (CCGR), denoted by $g$ (units: inverse years). We compute $g$ from a daily, weekly, or monthly price series. When that series excludes dividends and other cash distributions, $g$ measures price growth rather than total investment return.

> __Continuously Compounded Growth Rate (CCGR)__
>
> Let's assume the share price of firm $i$ is governed by a continuously compounded expression of the form:
>$$
\begin{align*}
S^{(i)}_{j} &= S^{(i)}_{j-1}\;\underbrace{\exp\left(g^{(i)}_{j,j-1}\Delta{t}_{j}\right)}_{\text{accumulation factor}} \\
\end{align*}
$$
> where $S^{(i)}_{j-1}$ and $S^{(i)}_{j}$ are consecutive share prices and $\Delta t_j=t_j-t_{j-1}$ is the consecutive time-step length in years. The value $g^{(i)}_{j,j-1}$ (units: inverse years) is the continuously compounded growth rate, while the corresponding dimensionless log return is $r^{(i)}_{j,j-1}=(\Delta t_j)g^{(i)}_{j,j-1}=\ln(S^{(i)}_j/S^{(i)}_{j-1})$.

We can estimate the growth rate $g$ (and a closely related quantity called the excess growth rate, or $\bar{g}$) from the stock price at time $t-1$ and $t$:
$$
\begin{align*}
S_{t} &= S_{t-1} \cdot e^{g\Delta{t}}\quad\Longrightarrow\text{solve for}\;g\\
g &= \underbrace{\left(\frac{1}{\Delta{t}}\right)\ln\left(\frac{S_{t}}{S_{t-1}}\right)}_{\text{growth rate}}\quad\blacksquare \\
\end{align*}
$$
The sign of the growth rate $g$ indicates whether the stock price increased or decreased over the time period $\Delta{t}$:
> __Sign of the growth rate__
> * If $g > 0$, the stock price _increased_, i.e., $S_{t} > S_{t-1}$ over the time period $\Delta{t}$.
> * If $g < 0$, the stock price _decreased_, i.e., $S_{t} < S_{t-1}$ over the time period $\Delta{t}$.
> * If $g = 0$, the stock price _did not change_, i.e., $S_{t} = S_{t-1}$ over the time period $\Delta{t}$.
>
> Because the return $r=(\Delta t)g$ and the time step $\Delta t>0$, the return $r$ and the growth rate $g$ have the same sign.

Often, we want the growth rate _relative to a benchmark_ (e.g., a risk-free asset). In this case, we can subtract the benchmark growth rate from the stock's growth rate to get the __excess growth rate__. The __excess growth rate__ $\bar{g}$ is given by:
$$
\begin{align*}
\bar{g} &= g-g_y \\
&= \underbrace{\left(\frac{1}{\Delta{t}}\right)\ln\left(\frac{S_{t}}{S_{t-1}}\right) - g_y}_{\text{excess growth rate}}\quad\blacksquare \\
\end{align*}
$$
where $\bar{g}$ is the continuously compounded excess growth rate and $g_y$ is the continuously compounded benchmark growth rate corresponding to yield $y$. Both have units of inverse years. Before examining estimation methods, consider this example:

> __Example__
>
> Suppose a stock price changes from $S_0$ to $S_1$ over one year ($\Delta t=1$). We compare its continuously compounded growth rate with that of a benchmark investment growing continuously at rate $g_y$ over the same period.
>
> * If $\bar{g} > 0$, the stock price grew faster than the alternative benchmark.
> * If $\bar{g} < 0$, the stock price grew more slowly than the alternative benchmark.
> * If $\bar{g} = 0$, the stock price and alternative benchmark had the same continuously compounded growth rate.
>
> Thus, the excess growth rate $\bar g$ compares the stock's growth rate with a stated benchmark over the same period.

### Algorithm
Let's examine pseudocode for estimating the continuously compounded growth rate $g$ and excess growth rate $\bar{g}$ from a time series of prices $\left\{S_1,S_2,\ldots,S_T\right\}$. From $T$ daily, hourly, or otherwise equally spaced prices, we obtain $T-1$ growth-rate observations.

__Initialization:__ Given a price dataset $\mathcal{D}_{i} = \left\{S^{(i)}_1,S^{(i)}_2,\ldots,S^{(i)}_T\right\}$ for firm $i$, the time step $\Delta{t}$ in years, and the benchmark growth rate $g_y$ in inverse years.

For each time step $t = 2,3,\ldots,T$ __do__:
1. __Lookup__ the share price at time $t$ from dataset $\mathcal{D}_{i}$: $S^{(i)}_t\gets\mathcal{D}_{i}[t]$.
2. __Lookup__ the share price at time $t - 1$ for firm $i$ from dataset $\mathcal{D}_{i}$: $S^{(i)}_{t-1}\gets\mathcal{D}_{i}[t - 1]$.
3. __Compute__ the continuously compounded excess growth rate $\bar{g}$:
   $$\bar{g}_t^{(i)} \gets \left(\frac{1}{\Delta{t}}\right)\ln\left(\frac{S^{(i)}_t}{S^{(i)}_{t-1}}\right) - g_y$$

   > __Alternative:__
   > To compute the growth rate $g$ instead of the excess growth rate $\bar{g}$, set $g_y=0$.
4. __Store__ the continuously compounded excess growth rate $\bar{g}_t^{(i)}$ in a list or array for firm $i$.

The (excess) growth-rate samples $\left\{\bar{g}_2^{(i)},\bar{g}_3^{(i)},\ldots,\bar{g}_T^{(i)}\right\}$ can then be used to analyze the stock's statistical behavior.

## Risk
The growth rate describes realized price growth, whereas risk concerns the uncertainty surrounding future outcomes. Standard deviation provides a useful measure of dispersion, but its name, units, and time horizon must be stated explicitly.

> __Volatility conventions__
> 
> For equally spaced observations, the same sample dispersion can be reported as __growth-rate volatility__, $\sigma_g=\operatorname{std}(g_t)$, or __interval log-return volatility__, $\sigma_r(\Delta t)=\operatorname{std}(r_t)$. Because $r_t=(\Delta t)g_t$, they satisfy the exact relation $\sigma_r(\Delta t)=(\Delta t)\sigma_g$. When $\Delta t$ is measured in years, $\sigma_g$ has units of inverse years, whereas $\sigma_r(\Delta t)$ is dimensionless and refers to the return over the stated observation interval, e.g., one day, one week, or one month. 

### Growth-Rate, Log-Return, and Annualized Volatility
Suppose $T$ equally spaced prices produce the $T-1$ growth rates $\left\{g_2,g_3,\ldots,g_T\right\}$ and interval log returns $r_t=(\Delta t)g_t$. Let $g^{\prime}$ and $r^{\prime}$ denote their sample means. We define the two equivalent fixed-interval dispersion measures and the conventional annualized log-return volatility by:
$$
\boxed{
\begin{aligned}
\sigma_g &= \underbrace{\sqrt{\frac{1}{T-2}\sum_{t=2}^{T}\left(g_t-g^{\prime}\right)^2}}_{\text{growth-rate volatility}} \qquad [\mathrm{yr}^{-1}],\\
\sigma_r(\Delta t) &= \underbrace{\sqrt{\frac{1}{T-2}\sum_{t=2}^{T}\left(r_t-r^{\prime}\right)^2}}_{\text{interval log-return volatility}}=(\Delta t)\sigma_g \qquad [\text{dimensionless}],\\
\sigma_{\mathrm{ann}} &= \underbrace{\frac{\sigma_r(\Delta t)}{\sqrt{\Delta t}}}_{\text{annualized log-return volatility}}=\sqrt{\Delta t}\,\sigma_g \qquad [\mathrm{yr}^{-1/2}].
\end{aligned}}
$$
For a time step of length $\Delta t$, log-return volatility equals growth-rate volatility multiplied by $\Delta t$: $\sigma_r(\Delta t)=(\Delta t)\sigma_g$. To report volatility on an annual scale, we often divide by $\sqrt{\Delta t}$. This shortcut assumes that returns from different time steps are unrelated and have roughly the same variability. Real markets do not always behave this way (large changes often occur close together) so annualized volatility is an estimate, not an exact one-year value.

### Other Risk Measures
Standard deviation is only one risk measure. Other measures ask sharper questions: how large the loss can get, how far returns fall below a target or a past peak, and how far the return strays from a benchmark:

* __How large the loss can get__: Value at Risk (VaR) is a loss level the portfolio passes only with a stated probability over a stated horizon, for example a 5% chance of losing more than X USD in one day. Conditional Value at Risk (CVaR), also called Expected Shortfall, is the average loss on the days that pass the VaR level. VaR marks where the tail starts, and CVaR says how bad it gets inside the tail.
* __Downside and drawdown__: Downside deviation measures the size of shortfalls below a target return. Deviations are measured from that target; returns at or above it contribute zero shortfall. Maximum drawdown is the largest drop from a past peak to a later low. It depends on the order the returns arrive in, not just their spread.
* __Risk against a benchmark__: Beta measures how much an asset moves when the market benchmark moves. That sensitivity is called systematic risk, and diversification cannot remove it. Tracking error is the standard deviation of the gap between an investment's return and its benchmark's return, and the value $\beta$ returns in later lectures and examples.

Which measure fits depends on the decision: what kind of loss matters, over what horizon, and against which benchmark. We use standard deviation most often because it is easy to work with, and later lectures bring back several of these alternatives.

## Stylized Facts of Equity Growth Rates
Analysts and investors examine growth rates and returns for recurring statistical properties known as __stylized facts__. We focus on three widely documented patterns:

* __Heavy-tailed distributions__: Equity returns place more probability far from the center than a fitted Normal distribution, so extreme price movements occur more frequently than the Normal benchmark predicts.
* __Little linear autocorrelation in signed returns__: Past signed returns usually provide little linear information about future signed returns beyond short lags. This does not establish independence or rule out nonlinear dependence.
* __Volatility clustering__: Large absolute returns tend to be followed by other large absolute returns, while small absolute returns tend to be followed by other small absolute returns. Market activity therefore alternates between persistent high- and low-volatility periods.

These are not the only stylized facts of equity returns, but they are among the most widely recognized and studied. For more details, see the following references:
* Cont, R. (2001). Empirical properties of asset returns: stylized facts and statistical issues. *Quantitative Finance*, 1(2), 223-236. [Link](https://doi.org/10.1080/713665670)
* Mandelbrot, B. (1963). The Variation of Certain Speculative Prices. The Journal of Business, 36(4), 394–419. [Link](http://www.jstor.org/stable/2350970)

Let's look at each of these stylized facts in more detail.

### Heavy-Tailed Growth-Rate Distributions
Large positive and negative growth rates occur more often than a Normal model predicts. A model with tails that are too thin underestimates the chance of extreme market days.

The Normal, Laplace, and Student-$t$ distributions provide three useful comparisons:

| Distribution | Tail behavior and practical meaning |
|:--|:--|
| Normal | Its tails decrease roughly like $\exp(-x^2)$, so a Normal distribution assigns relatively little probability to extreme growth rates. It is our thin-tailed benchmark. |
| Laplace | Its tails decrease roughly like $\exp(-\lvert x\rvert)$. The Laplace distribution has a sharper center and gives large growth rates more probability than the Normal, but its tails still decrease exponentially. |
| Student-$t$ | Its tails decrease as a power law. Smaller degrees of freedom $\nu$ produce more extreme growth rates; for this model, the power-law tail index is $\alpha=\nu$. |

At a fixed time step, growth rates and log returns differ only by a scale factor. The table applies to both.

Laplace can fit the center and moderately large growth rates well. If the far tails follow a power law, Student-$t$ may be a better choice for extreme events. We compare empirical fits and check tail-index stability before choosing a model.

These distributions describe the sizes and frequencies of growth rates. They do not describe how large movements cluster through time.

To define power-law behavior formally, let $X=|g|$ denote the magnitude of a growth rate. The random variable $X$ has a __regularly varying tail__ when its survival probability satisfies:
$$
\mathbb{P}(X > x) \sim L(x)\,x^{-\alpha}, \quad \text{as } x \to \infty,
$$
where $\alpha>0$ is the __tail index__. The notation $\sim$ means that the ratio of the two sides approaches one as $x$ becomes large. The factor $L(x)$ is a __slowly varying function__, meaning:
$$
\lim_{x \to \infty}\frac{L(tx)}{L(x)}=1 \qquad \text{for every } t>0.
$$
The factor $L(x)$ does not change the leading power-law rate $x^{-\alpha}$. Student-$t$ tails are regularly varying; Normal and Laplace tails are not.

> __Why the tail index matters__: Smaller $\alpha$ means slower tail decay. For Student-$t$, $\alpha=\nu$.
>
> * If $\alpha<2$, the model has no finite variance. Sample volatility is still computable, but extreme observations can change it sharply.
> * If $2<\alpha<4$, variance is finite but the fourth moment is infinite. Volatility can stabilize, but kurtosis, which depends on fourth powers of deviations from the mean, has no finite population value.
>
> The companion example compares fitted $\nu$ with Hill estimates across several tail cutoffs.

#### Hill Estimator of the Tail Index
The Hill estimator uses the largest observed magnitudes to estimate the power-law tail index $\alpha$.

Under the regularly varying-tail model, let $X_1,X_2,\dots,X_n$ be independent observations of $|g|$, sorted from largest to smallest as $X_{(1)}\ge X_{(2)}\ge\cdots\ge X_{(n)}$. Choose $k<n$, the number of extreme observations to include. The Hill estimate of the tail index $\alpha$ is then given by:
$$
\boxed{
\hat{\alpha}_k = \left[ \frac{1}{k} \sum_{i=1}^{k} \ln\left(\frac{X_{(i)}}{X_{(k+1)}}\right) \right]^{-1}.
}
$$

Too few tail observations give a noisy estimate; too many include observations outside the far tail. We plot $\hat{\alpha}_k$ across several $k$ values and look for a stable region.

### Autocorrelation
Autocorrelation measures linear dependence between observations separated by a lag. For the $T-1$ growth rates $\{g_2,g_3,\ldots,g_T\}$ of one firm, define the empirical autocovariance:
$$
\hat{R}(k) = \frac{1}{T-1}\sum_{t=2}^{T-k}\left(g_t - g^{\prime}\right)\left(g_{t+k} - g^{\prime}\right)\quad\;k=0,1,\ldots,T-2
$$
Here, $k$ is the lag in time steps and $g^{\prime}$ is the sample mean. The autocorrelation function (ACF) normalizes the autocovariance by its value at zero lag:
$$
\boxed{
\rho(k) = \frac{\hat{R}(k)}{\hat{R}(0)}\quad\;k=0,1,\ldots,T-2
}
$$
With this normalization (the standard biased convention, which uses the full sample size at every lag), $|\rho(k)|\leq1$ for all $k$, and $\rho(0)=1$ by definition. For many short-memory stationary processes, the ACF decays toward zero as $k$ increases, indicating weaker linear association between observations farther apart in time.

> __Random-Walk Benchmark:__
>
> Burton Malkiel's book [*A Random Walk Down Wall Street*](https://en.wikipedia.org/wiki/A_Random_Walk_Down_Wall_Street) popularized the idea that past price movements provide little help in predicting the next price movement.
>
> * The ACF lets us examine this idea using data. For equally spaced observations, the random-walk benchmark suggests that past growth rates have little linear relationship with future growth rates. We therefore expect $\rho(k)$ to be close to zero for lags $k>0$.
> * An ACF near zero does not mean that growth rates are completely independent. The sizes of the movements may still cluster through time, as we discuss next.
>
> Thus, the direction of the next price movement may be hard to predict, while its size may still be related to the sizes of recent movements. This leads to our next stylized fact: volatility clustering.

### Volatility Clustering
Volatility clustering refers to the tendency for large absolute returns to be followed by other large absolute returns and for small absolute returns to be followed by other small absolute returns, creating persistent periods of high and low volatility.

The daily growth-rate series of SPY and AMD from the companion example dataset show the pattern directly: the shaded March-2020 weeks are turbulent in both series, while the isolated April-2016 spike in AMD is a single large jump, not a cluster.

__Why volatility clustering matters:__ A model that ignores time-varying volatility can misestimate risk measures such as Value at Risk, derivative values, hedge ratios, and portfolio allocations. Modeling clustering can improve risk forecasts and decisions that depend on changing market uncertainty.

Volatility clustering concerns the size of price movements, not their direction. We therefore compute the ACF of the growth-rate magnitudes $\bigl|g_t^{(i)}\bigr|$. Let firm $i$ have growth-rate series $\bigl\{g_{2}^{(i)},\,g_{3}^{(i)},\,\dots,\,g_{T}^{(i)}\bigr\}$, and let:
$$
\overline{|g|_{i}}
\;=\;\frac{1}{T-1}\sum_{t=2}^T\bigl|g_t^{(i)}\bigr|
$$
be the sample mean of the growth-rate magnitudes. We can then define the normalized empirical autocorrelation of the growth-rate magnitudes at lag $k$ as:

$$
\boxed{
\hat{\rho}_i(k)
=\frac{\displaystyle
   \sum_{t=2}^{\,T-k}
     \Bigl[\bigl|g_t^{(i)}\bigr|-\overline{|g|_{i}}\Bigr]
     \,\Bigl[\bigl|g_{t+k}^{(i)}\bigr|-\overline{|g|_{i}}\Bigr]
}{\displaystyle
   \sum_{t=2}^{\,T}
     \Bigl[\bigl|g_t^{(i)}\bigr|-\overline{|g|_{i}}\Bigr]^{2}
}\,.}
$$

Volatility clustering is observed when $\hat{\rho}_i(k)>0$ for many $k$, with values that fade toward zero only slowly (and not necessarily monotonically) as $k$ grows.

## Summary
This lecture introduced equity securities, continuously compounded growth rates, and recurring empirical patterns in financial return data.

> __Key Takeaways:__
>
> * __The price depends on how and where you trade:__ Each venue keeps its own order book, so what an investor pays depends on the order type sent and the venue that fills it. The NBBO is the benchmark used to compare quotes across venues, and every trade feeds the price series we analyze.
> * __The growth-rate series gives both reward and risk:__ The growth rate says how fast the price grew per year, the log return says how much it moved over one interval, and the excess growth rate compares that growth against a benchmark. Volatility is the spread of those same numbers, so every value needs a stated horizon, and annualizing by the square root of time is a convention rather than a law.
> * __Real returns are not independent or Normal:__ Large moves occur more often than a Normal distribution predicts. Signed returns show little linear autocorrelation, while return magnitudes remain correlated across time.
