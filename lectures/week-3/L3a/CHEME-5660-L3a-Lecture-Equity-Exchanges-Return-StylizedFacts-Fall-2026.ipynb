## Exchanges, Stocks, and ETFs
Before measuring equity growth rates, we need to distinguish the marketplace from the securities traded there. An __exchange__ is a trading venue; __stocks__ and __exchange-traded funds (ETFs)__ are securities whose prices are discovered through trading.

### Exchanges Connect Buyers and Sellers
An exchange is an organized venue that matches orders to buy and sell securities. By bringing many orders together, exchanges support __price discovery__ (the process by which trading establishes a market price) and __liquidity__ (the ability to trade without moving that price excessively). Exchanges also impose listing standards and trading rules. After a trade is executed, separate clearing and settlement infrastructure completes the transfer of securities and funds.


U.S. equities trade across more than a dozen competing exchanges, the three largest being the [New York Stock Exchange (NYSE)](https://www.nyse.com/), [NASDAQ](https://www.nasdaq.com/), and [Cboe BZX Exchange](https://www.cboe.com/); the SEC maintains [the current list of registered national securities exchanges](https://www.sec.gov/about/divisions-offices/division-trading-markets/national-securities-exchanges), which also includes options-only venues. A broker may route an order to one venue or divide it across several venues, depending on available prices, liquidity, and the order's instructions. 

### Orders Are Instructions to Trade
An __order__ is the instruction a broker submits on an investor's behalf: it names the security, the direction (buy or sell), the quantity, and how the price may be set. The common order types differ in how they trade off immediacy against price control:

* __Market order__: A market order executes as soon as possible at the best price currently available. Execution is nearly certain, but the price at which it occurs is not. We use this type of order when we want to trade quickly and are willing to accept an uncertain price.
* __Limit order__: A limit order specifies a limit price: a buy limit order executes only at the limit price or lower, and a sell limit order only at the limit price or higher. The price is controlled, but execution is not guaranteed. Limit orders that do not execute immediately rest on the venue's __order book__, the collection of outstanding buy and sell limit orders for a security; the highest resting buy price is the venue's displayed __bid__, the lowest resting sell price is its displayed __ask__, and the difference between them is the __bid-ask spread__. An arriving market order to buy trades at the ask, and one to sell trades at the bid.
* __Stop and stop-limit orders__: Stop orders are conditional instructions that become market orders when the price reaches a specified level. Stop-limit orders are similar, but they become limit orders at the trigger level. 

The [Types of Orders page at Investor.gov](https://www.investor.gov/introduction-investing/investing-basics/how-stock-markets-work/types-orders) defines these and other variations.

The limit-order book notebook in the Optional Advanced Material section develops how a book processes arriving orders. Because each venue keeps its own order book, each venue displays its own bid and ask. But which price is the "right" price? The answer is that there is no single "right" price, but there is a common benchmark: the __National Best Bid and Offer (NBBO)__. 

The __National Best Bid and Offer (NBBO)__ reports the highest displayed bid (buy) and the lowest displayed ask (sell) available through the national market system. The NBBO is more than a summary statistic: a broker must fill a customer order at the national best price or better, so a buyer pays no more than the NBBO ask and a seller receives no less than the NBBO bid, whichever venue executes the trade.

### Stocks and ETFs Are Claims on Firms and Portfolios
A stock, also called a __share__ or __equity security__, represents fractional ownership in a corporation. Shareholders may receive voting rights and dividends, have limited liability for the corporation's debts, and hold a residual claim on its assets after creditors are paid. The economic reward from holding a stock can therefore come from two channels: cash distributions (dividends) and changes in the share price.

Where a stock is a claim on a single corporation, an Exchange-Traded Fund (ETF) pools assets into a portfolio and issues shares that trade on an exchange throughout the day like a single named security. Each ETF share represents an interest in that portfolio. Broad-market funds such as [SPY](https://finance.yahoo.com/quote/SPY), which tracks the Standard & Poor's 500 index, can provide diversified exposure, but diversification is not guaranteed: sector, leveraged, inverse, and single-stock ETFs may be highly concentrated.

An ETF has both a market price and a __net asset value (NAV)__ per share: the fund's assets minus its liabilities, divided by the number of shares outstanding. The two are usually close but may not be identical, so an ETF can trade at a premium or discount to NAV. For our class, the essential point is the same for both security types: a stock or an ETF produces an observable sequence of market prices.

## Reward
At the core of every investment is the trade-off between risk and reward. The higher the risk, the higher the potential reward, but also the higher the potential loss. __This is no free lunch!__ But how do we measure this risk and reward in practice?

We measure investment __reward__ using the continuously compounded growth rate (CCGR), denoted by $g$ (units: inverse years). We compute $g$ from a closing-price series, e.g., the daily, weekly, or monthly closing prices; dividends and other cash distributions are typically not included.

> __Continuously Compounded Growth Rate (CCGR)__
>
> Let's assume the share price of firm $i$ is governed by a continuously compounded expression of the form:
>$$
\begin{align*}
S^{(i)}_{j} &= S^{(i)}_{j-1}\;\exp\left(g^{(i)}_{j,j-1}\Delta{t}_{j}\right)
\end{align*}
$$
> where $S^{(i)}_{j-1}$ and $S^{(i)}_{j}$ are consecutive share prices and $\Delta t_j=t_j-t_{j-1}$ is the time-step length in years. The value $g^{(i)}_{j,j-1}$ (units: inverse years) is the continuously compounded growth rate, while the corresponding dimensionless log return is $r^{(i)}_{j,j-1}=(\Delta t_j)g^{(i)}_{j,j-1}=\ln(S^{(i)}_j/S^{(i)}_{j-1})$.

How does growth rate $g$ differ from log return $r$, and why do we prefer working with the growth rate?

> __Why use continuous growth rate?__ 
> 
> Unlike log returns, the continuously compounded growth rate (CCGR) is annualized by default since our convention is to measure time in years, so growth over different horizons can be compared on a common scale. 
> 
> Further, when we need to aggregate over multiple periods, we convert back to log returns, which add across adjacent intervals, e.g., the total log return is the time-weighted sum $\sum_{j}(\Delta{t}_{j})\,g_{j}$ of the growth rates.

We can estimate the growth rate $g$ (and a closely related quantity called the excess growth rate, or $\bar{g}$) from the stock price at time $t-1$ and $t$:
$$
\begin{align*}
S_{t} &= S_{t-1} \cdot e^{g\Delta{t}}\quad\Longrightarrow\text{solve for}\;g\\
g &= \underbrace{\left(\frac{1}{\Delta{t}}\right)\ln\left(\frac{S_{t}}{S_{t-1}}\right)}_{\text{growth rate}}\quad\blacksquare \\
\end{align*}
$$
The sign of the growth rate $g$ indicates whether the stock price increased or decreased over the time period $\Delta{t}$:
> __Sign of the growth rate__
> * If $g > 0$, the stock price _increased_, i.e., $S_{t} > S_{t-1}$ over the time period $\Delta{t}$.
> * If $g < 0$, the stock price _decreased_, i.e., $S_{t} < S_{t-1}$ over the time period $\Delta{t}$.
> * If $g = 0$, the stock price _did not change_, i.e., $S_{t} = S_{t-1}$ over the time period $\Delta{t}$.
>
> Because the return $r=(\Delta t)g$ and the time step $\Delta t>0$, the return $r$ and the growth rate $g$ have the same sign.

Often, we want the growth rate _relative to a benchmark_ (e.g., a risk-free asset). In this case, we can subtract the benchmark growth rate from the stock's growth rate to get the __excess growth rate__. The __excess growth rate__ $\bar{g}$ is given by:
$$
\begin{align*}
\bar{g} &= g-g_f \\
&= \underbrace{\left(\frac{1}{\Delta{t}}\right)\ln\left(\frac{S_{t}}{S_{t-1}}\right) - g_f}_{\text{excess growth rate}}\quad\blacksquare \\
\end{align*}
$$
where $\bar{g}$ is the continuously compounded excess growth rate and $g_f$ is the continuously compounded benchmark growth rate. Both have units of inverse years. Before examining estimation methods, consider this example:

> __Example__
>
> Suppose we invest 100 USD in a stock and hold it for one year ($\Delta t=1$). We compare it with a benchmark investment that grows continuously at rate $g_f$ over the same period. At the end of the year, the stock price is $S_1$ USD/share.
>
> * If $\bar{g} > 0$, the stock price grew faster than the alternative benchmark.
> * If $\bar{g} < 0$, the stock price grew more slowly than the alternative benchmark.
> * If $\bar{g} = 0$, the stock price and alternative benchmark had the same continuously compounded growth rate.
>
> Thus, the excess growth rate $\bar g$ compares the stock's growth rate with a stated benchmark over the same period.

### Algorithm
Let's examine some pseudocode to estimate the continuously compounded growth rate $g$ and the excess growth rate $\bar{g}$ from a time series of stock prices $\left\{S_1,S_2,\ldots,S_T\right\}$. If we have $T$ days (minutes, hours, etc.) of stock prices, then we can estimate $T-1$ continuously compounded growth rate $g$ values.

__Initialization:__ Given a price dataset $\mathcal{D}_{i} = \left\{S^{(i)}_1,S^{(i)}_2,\ldots,S^{(i)}_T\right\}$ for firm $i$, the time step $\Delta{t}$ in years, and the benchmark growth rate $g_f$ in inverse years.

For each time step $t = 2,3,\ldots,T$ __do__:
1. __Lookup__ the share price at time $t$ from dataset $\mathcal{D}_{i}$: $S^{(i)}_t\gets\mathcal{D}_{i}[t]$.
2. __Lookup__ the share price at time $t - 1$ for firm $i$ from dataset $\mathcal{D}_{i}$: $S^{(i)}_{t-1}\gets\mathcal{D}_{i}[t - 1]$.
3. __Compute__ the continuously compounded excess growth rate $\bar{g}$:
   $$\bar{g}_t^{(i)} \gets \left(\frac{1}{\Delta{t}}\right)\ln\left(\frac{S^{(i)}_t}{S^{(i)}_{t-1}}\right) - g_f$$

   > __Alternative:__
   > To compute the growth rate $g$ instead of the excess growth rate $\bar{g}$, set $g_f=0$.
4. __Store__ the continuously compounded excess growth rate $\bar{g}_t^{(i)}$ in a list or array for firm $i$.

The (excess) growth-rate samples $\left\{\bar{g}_2^{(i)},\bar{g}_3^{(i)},\ldots,\bar{g}_T^{(i)}\right\}$ can then be used to analyze the stock's statistical behavior.

## Risk
The growth rate of investment is the reward, now let's consider the risk. The risk of an investment is the uncertainty associated with the potential for loss or gain. 

> __Risk model__ 
> 
> We can quantify risk using a variety of statistical measures, but the most common is the __standard deviation__ of the continuously compounded growth rate $g$, which we denote as $\sigma_{g}$ (units: inverse years). The standard deviation is a measure of the dispersion or variability of the returns around their mean. 
> 
> The standard deviation measures the uncertainty of the investment's growth rate. A higher standard deviation indicates a higher level of risk, as the returns are more spread out and less predictable. This has a special name in finance: __volatility__. 

### Volatility of the Growth Rate
Volatility is a measure of the magnitude of price fluctuations over time, and it is often used as a proxy for risk in financial markets.
Suppose we have $T$ price observations. From these observations, we can construct a time series of length $T-1$ of continuously compounded growth rates $\left\{g_2,g_3,\ldots,g_{T}\right\}$ for a stock. The sample standard deviation of the continuously compounded growth rate (our risk measure), is then given by:
$$
\boxed{
\begin{align*}
\sigma_{g} &= \underbrace{\sqrt{\frac{1}{T - 2}\sum_{t=2}^{T}\left(g_t - g^{\prime}\right)^2}}_{\text{volatility of the growth rate}} \\
\end{align*}}
$$
where $g^{\prime} = \frac{1}{T-1}\sum_{t=2}^{T}g_t$ is the sample mean of the $T-1$ growth observations. The denominator $T-2$ represents the degrees of freedom for the sample standard deviation: $(T-1) - 1 = T-2$. We'll typically compute the standard deviation using [the `std(...)` function exported by the `Statistics.jl` package](https://docs.julialang.org/en/v1/stdlib/Statistics/#Statistics.std).

### Other Risk Measures
While standard deviation (volatility) is the most commonly used risk measure, it's important to note that risk can be quantified in many ways, each capturing different aspects of uncertainty:

* __Value at Risk (VaR)__: The loss threshold that is exceeded with only a stated probability over a given horizon (e.g., a 5% chance of losing more than X USD in one day).
* __Conditional Value at Risk (CVaR)__: Also known as Expected Shortfall, it measures the average loss beyond the VaR threshold, providing information about tail risk.
* __Maximum Drawdown__: The largest peak-to-trough decline in portfolio value, measuring the worst-case scenario an investor would have experienced.
* __Beta__: A measure of systematic risk that quantifies how much an asset's returns move relative to the overall market. We'll for sure see $\beta$ in __many__ future lectures and exercises.
* __Downside Deviation__: Similar to standard deviation but only considers negative returns, focusing on the risk of losses rather than overall volatility.
* __Tracking Error__: The standard deviation of the difference between an investment's returns and its benchmark, measuring relative risk.

The choice of risk measure depends on the specific application, investor preferences, and the aspects of risk that are most relevant to the investment decision. In this course, we'll primarily focus on standard deviation due to its mathematical tractability, but we'll see a few of these alternative risk measures in future lectures.

## Stylized Facts of Equity Growth Rates
Analysts and investors gain insights into market behavior, risk, and investment opportunities by examining growth rates (or returns) and their statistical properties (stylized facts). While several stylized facts have been developed, let’s consider the following:

* __Heavy (also called fat) tailed distribution__: Stock returns often exhibit a distribution with fatter tails than would be expected under a normal distribution. This means that extreme price movements are more likely than would be predicted by a normal distribution.
* __Absence of Autocorrelation__: Autocorrelation refers to the tendency of stock returns to correlate with past returns over time. Autocorrelation suggests predictability in returns, which traders could exploit. On the other hand, if prices follow a random walk, the steps of that walk (the returns) are uncorrelated with one another. We expect prices to look like a random walk with occasional jumps.
* __Volatility clustering__: Stock returns tend to be more volatile during specific periods and less volatile during others. This phenomenon is known as volatility clustering, suggesting that large price movements are more likely to be followed by other large moves, and other small moves follow small moves.

These are not the only stylized facts of equity returns, but they are among the most widely recognized and studied. For more details, see the following references:
* Cont, R. (2001). Empirical properties of asset returns: stylized facts and statistical issues. *Quantitative Finance*, 1(2), 223-236. [Link](https://doi.org/10.1080/713665670)
* Mandelbrot, B. (1963). The Variation of Certain Speculative Prices. The Journal of Business, 36(4), 394–419. [Link](http://www.jstor.org/stable/2350970)

Let's look at each of these stylized facts in more detail.

### Heavy Tailed Return Distributions
Heavy‐tailed distributions have probability of extreme deviations that decays more slowly than exponentially, making large shocks rarer but much more likely than under a normal distribution. 

> __Black swan events__: Heavy tails in stock‐return distributions make extreme jumps far more frequent than normal distributions predict. These _black swans_, such as the 1987 crash, 2008 crisis, the COVID-19 pandemic, or April 2025 tariff sell-offs, arise from larger systemic failures, or sudden news, earnings surprises, or geopolitical shocks.

A **heavy-tailed distribution** has tails that are not exponentially bounded. For random variable $X$, tail probabilities decay more slowly than exponential:
$$
\mathbb{P}(|X| > x) \sim L(x)\,x^{-\alpha}, \quad \text{as } x \to \infty,
$$
where $\alpha > 0$ is the **tail index** (controlling decay rate) and $L(x)$ is a **slowly varying function** at infinity: $\lim_{x \to \infty} \frac{L(tx)}{L(x)} = 1$ for all $t > 0$.

> __Tail index__: 
> The tail index $\alpha$ is a key parameter in characterizing the heaviness of the tail. It determines how quickly the tail probabilities decay, and it describes tails that decay like a power law; a heavy-tailed distribution without a power-law tail, such as the lognormal, is not summarized by a single $\alpha$. 
> A smaller $\alpha$ indicates a heavier tail, while a larger $\alpha$ indicates a lighter tail. The classical thresholds describe which moments of the distribution exist:
> * If $\alpha < 1$, the mean does not exist, so single extreme events dominate long-run averages; the Cauchy distribution sits at this boundary with $\alpha = 1$ and has no mean.
> * If $\alpha < 2$, the variance is infinite, so volatility estimates never settle down as the sample grows.
> * If $\alpha < 4$, the fourth moment is infinite, and kurtosis-based diagnostics are dominated by the most extreme observations.
> * The thresholds are strict: exactly at $\alpha = 1$, $2$, or $4$, existence of the corresponding moment depends on finer tail detail. For any $\alpha$, a power-law tail places more probability on extreme events than a Normal model; a larger $\alpha$ only pushes the difference farther into the tail.

#### Hill's estimator of the tail index
Hill's estimator is a method for estimating the tail index $\alpha$ of a heavy-tailed distribution. The estimator requires positive observations, so for signed growth rates we apply it to the magnitude series $|g|$ (as the companion example does) or to one tail at a time.
Let $X_1, X_2, \dots, X_n$ be i.i.d. realizations of a heavy-tailed random variable $X$, and let $X_{(1)} \ge X_{(2)} \ge \cdots \ge X_{(n)}$ denote the **order statistics** (sorted in decreasing order). Fix $k < n$, the number of **upper order statistics** to use (typically with $k \ll n$), then the **Hill estimator** of the **tail index $\alpha$** is:
$$
\boxed{
\hat{\alpha}_k = \left[ \frac{1}{k} \sum_{i=1}^{k} \ln\left(\frac{X_{(i)}}{X_{(k+1)}}\right) \right]^{-1}.
}
$$

The choice of $k$ is crucial, as it balances bias and variance in the estimate; we'll typically look at a range of $k$ values and plot the estimates to find a stable region.

### Autocorrelation
Autocorrelation is a key concept in time series analysis, particularly in the context of financial returns. Suppose we have the $T-1$ length growth rate series $\left\{g_2^{(i)},g_3^{(i)},\ldots,g_T^{(i)}\right\}$ for firm $i$. The (empirical) autocorrelation sequence (where we neglect the firm superscript for simplicity) is given by:
$$
\hat{R}(k) = \frac{1}{T-1}\sum_{t=2}^{T-k}\left(g_t - g^{\prime}\right)\left(g_{t+k} - g^{\prime}\right)\quad\;k=0,1,\ldots,T-2
$$
where $k$ is the lag (units: time steps) and $g^{\prime}$ is the mean of the continuously compounded growth rates over the $T - 1$ samples. The autocorrelation function (ACF), which is the normalized autocorrelation sequence, is given by:
$$
\boxed{
\rho(k) = \frac{\hat{R}(k)}{\hat{R}(0)}\quad\;k=0,1,\ldots,T-2
}
$$
The ACF has some nice properties: with this normalization (the standard biased convention, which divides by the full sample size at every lag) $|\rho(k)| \leq 1$ for all $k$, and $\rho(0) = 1$ by definition. Additionally, for many short-memory stationary processes the ACF decays to zero as $k$ increases, indicating that past values have less influence on future values over time.

> __Random Walk:__ 
> 
> In the famous book [A Random Walk Down Wall Street](https://en.wikipedia.org/wiki/A_Random_Walk_Down_Wall_Street), Burton Malkiel argued that stock prices follow a random walk, meaning that past price movements do not predict future price movements. 
> 
> Thus, the autocorrelation function (ACF) of stock returns would be close to zero for all lags $k > 0$. Near-zero autocorrelation means past returns give little linear help in predicting future returns; it does not rule out other forms of dependence, such as the volatility clustering we discuss next.

If the random walk hypothesis holds, then the autocorrelation function (ACF) of the continuously compounded growth rates $\left\{g_2^{(i)},g_3^{(i)},\ldots,g_T^{(i)}\right\}$ should be close to zero for all lags $k > 0$ and all firms $i$. In other words, there should be no significant correlation between past and future returns.

### Volatility Clustering
Volatility clustering refers to the tendency for large (in absolute value) returns to be followed by large returns, and small returns by small ones, creating persistent periods of high and low volatility.

The daily growth-rate series of SPY and AMD from the companion example dataset show the pattern directly: the shaded March-2020 weeks are turbulent in both series, while the isolated April-2016 spike in AMD is a single large jump, not a cluster.

    <center>
        <img src="figs/Fig-L3a-AMD-Volatility-Clustering.svg" width="1000" alt="Daily continuously compounded growth rates of SPY (top) and AMD (bottom) from 2014 through 2024, with a shaded band over March 2020 where large-magnitude days of both signs cluster in both series, and a label marking AMD's isolated April 2016 single large jump"/>
    </center>
</div>

> __Why volatility clustering matters:__ Dynamic risk must be modeled to avoid misestimated risk measures (like Value at Risk), derivative prices, and portfolio allocations. Incorporating clustering improves risk forecasts, option valuations, hedging strategies, and capital allocation decisions.


Let firm $i$ have growth rate series: $\bigl\{g_{2}^{(i)},\,g_{3}^{(i)},\,\dots,\,g_{T}^{(i)}\bigr\},$
and let
$$
\overline{(g^2)_{i}} 
\;=\;\frac{1}{T-1}\sum_{t=2}^T\bigl(g_t^{(i)}\bigr)^2
$$
be the sample mean of the squared growth rates. Define the normalized empirical autocorrelation of squared growth rates at lag $k$ as

$$
\boxed{
\hat{\rho}_i(k)
=\frac{\displaystyle
   \overbrace{\sum_{t=2}^{\,T-k}
     \Bigl[\bigl(g_t^{(i)}\bigr)^2-\overline{(g^2)_{i}}\Bigr]
     \,\Bigl[\bigl(g_{t+k}^{(i)}\bigr)^2-\overline{(g^2)_{i}}\Bigr]
   }^{\hat R_i(k)\;\text{with squared growth rates}}
}{\displaystyle
   \underbrace{\sum_{t=2}^{\,T}
     \Bigl[\bigl(g_t^{(i)}\bigr)^2-\overline{(g^2)_{i}}\Bigr]^{2}
   }_{\hat R_i(0)\;\text{with squared growth rates}}
}\,.}
$$

Volatility clustering is observed when $\hat{\rho}_i(k)>0$ for many $k$, with values that fade toward zero only slowly (and not necessarily monotonically) as $k$ grows.

> * __Stocks and ETFs trade through market infrastructure:__ Exchanges support price discovery and liquidity under listing and trading rules, while separate clearing and settlement infrastructure completes each trade.
> * __Growth rates annualize compounded price change:__ The continuously compounded growth rate is the log price ratio divided by the time-step length, while the corresponding log return is dimensionless and adds across adjacent time intervals.
> * __Equity returns exhibit stylized facts:__ Return distributions have heavy tails, raw returns usually have little linear autocorrelation, and the magnitude of returns displays persistent volatility clustering.

These stylized facts provide empirical tests that useful equity-price models should confront.
