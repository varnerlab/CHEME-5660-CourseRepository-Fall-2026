## One-Step Growth-Rate Derivation
Let $S_{j-1}>0$ be the share price before step $j$, let $\Delta t>0$ be the step length in years, and let $u>d>0$ be the constant up and down factors. All probabilities and moments below use the real-world rule $\mathbb{P}$ defined by $p$, where $0<p<1$.

### Encode the Price Move
Define the move indicator $X_j$ as:
$$
X_j=\begin{cases}
1, & \text{up with probability }p,\\
0, & \text{down with probability }1-p,
\end{cases}
$$
Assume the indicators $X_1,X_2,\ldots$ are independent. Each indicator follows $\operatorname{Bernoulli}(p)$, with $\mathbb{E}_{\mathbb{P}}[X_j]=p$ and $\operatorname{Var}_{\mathbb{P}}(X_j)=p(1-p)$. The one-step price factor can be written as:
$$
\frac{S_j}{S_{j-1}}=u^{X_j}d^{1-X_j}=d\left(\frac{u}{d}\right)^{X_j}.
$$
When $X_j=1$, this expression equals $u$, and when $X_j=0$, it equals $d$.

### Derive the Growth-Rate Law
In the notation of L3a, write $g_j\equiv g_{j,j-1}$ for the one-step continuously compounded growth rate. The continuously compounded price model gives:
$$
S_j=S_{j-1}\exp(g_j\Delta t)
\qquad\Longrightarrow\qquad
g_j=\frac{1}{\Delta t}\ln\left(\frac{S_j}{S_{j-1}}\right).
$$
Substituting the binomial price factor and expanding the logarithm gives:
$$
\begin{aligned}
g_j
&=\frac{1}{\Delta t}\ln\left(u^{X_j}d^{1-X_j}\right)\\
&=\frac{1}{\Delta t}\left[\ln\left(u^{X_j}\right)+\ln\left(d^{1-X_j}\right)\right]\\
&=\frac{1}{\Delta t}\left[X_j\ln u+(1-X_j)\ln d\right]\\
&=\frac{1}{\Delta t}\left[\ln d+X_j\ln\left(\frac{u}{d}\right)\right].
\end{aligned}
$$
Thus, the one-step growth-rate law is:
$$
\boxed{g_j=\frac{\ln d+X_j\ln(u/d)}{\Delta t}}.
$$
The growth rate is therefore an affine transformation of a Bernoulli random variable.

### Derive the Mean and Variance
Apply linearity of expectation to the growth-rate law:
$$
\begin{aligned}
\mathbb{E}_{\mathbb{P}}[g_j]
&=\mathbb{E}_{\mathbb{P}}\left[\frac{\ln d+X_j\ln(u/d)}{\Delta t}\right]\\
&=\frac{1}{\Delta t}\left[\ln d+\mathbb{E}_{\mathbb{P}}[X_j]\ln\left(\frac{u}{d}\right)\right]\\
&=\frac{1}{\Delta t}\left[\ln d+p\ln\left(\frac{u}{d}\right)\right]\\
&=\boxed{\frac{p\ln u+(1-p)\ln d}{\Delta t}}.
\end{aligned}
$$
For the variance, use $\operatorname{Var}(aX+b)=a^2\operatorname{Var}(X)$ for constants $a$ and $b$:
$$
\begin{aligned}
\operatorname{Var}_{\mathbb{P}}(g_j)
&=\operatorname{Var}_{\mathbb{P}}\left(\frac{\ln d}{\Delta t}+\frac{\ln(u/d)}{\Delta t}X_j\right)\\
&=\left[\frac{\ln(u/d)}{\Delta t}\right]^2\operatorname{Var}_{\mathbb{P}}(X_j)\\
&=\boxed{\frac{p(1-p)}{(\Delta t)^2}\left[\ln\left(\frac{u}{d}\right)\right]^2}.
\end{aligned}
$$
Because $0<p<1$ and $u>d$, this variance is positive.

## Multiple-Step Return Derivation
We now extend the one-step result over $t\geq1$ equal time steps. The work runs in three stages: count the up moves, multiply the step factors to reach the terminal price and its cumulative return, then take the moments of that return. The count and the moments use the independence assumed above, while the pathwise price identity holds along any single path and needs no such assumption.

### Derive the Up-Move Count
Define the number of up moves through step $t$ as $K_t=\sum_{j=1}^{t}X_j$. Because the indicators are independent Bernoulli variables, $K_t\sim\operatorname{Binomial}(t,p)$. Its mean follows from linearity of expectation:
$$
\begin{aligned}
\mathbb{E}_{\mathbb{P}}[K_t]
&=\mathbb{E}_{\mathbb{P}}\left[\sum_{j=1}^{t}X_j\right]\\
&=\sum_{j=1}^{t}\mathbb{E}_{\mathbb{P}}[X_j]\\
&=\sum_{j=1}^{t}p=tp.
\end{aligned}
$$
Independence removes all cross-covariance terms from the variance calculation:
$$
\begin{aligned}
\operatorname{Var}_{\mathbb{P}}(K_t)
&=\operatorname{Var}_{\mathbb{P}}\left(\sum_{j=1}^{t}X_j\right)\\
&=\sum_{j=1}^{t}\operatorname{Var}_{\mathbb{P}}(X_j)\\
&=\sum_{j=1}^{t}p(1-p)=tp(1-p).
\end{aligned}
$$

### Derive the Terminal Price and Cumulative Return
Multiply the one-step price factors along a path:
$$
\begin{aligned}
\frac{S_t}{S_0}
&=\prod_{j=1}^{t}u^{X_j}d^{1-X_j}\\
&=u^{\sum_{j=1}^{t}X_j}d^{\sum_{j=1}^{t}(1-X_j)}\\
&=u^{K_t}d^{t-K_t}.
\end{aligned}
$$
Thus, the terminal share price is $S_t=S_0u^{K_t}d^{t-K_t}$. Define the dimensionless cumulative log return from step $0$ through step $t$ as $r_{t,0}=\ln(S_t/S_0)$. Substituting the terminal share price and expanding the logarithm gives:
$$
\begin{aligned}
r_{t,0}
&=\ln\left(u^{K_t}d^{t-K_t}\right)\\
&=K_t\ln u+(t-K_t)\ln d\\
&=t\ln d+K_t(\ln u-\ln d)\\
&=\boxed{t\ln d+K_t\ln\left(\frac{u}{d}\right)}.
\end{aligned}
$$

### Derive the Cumulative-Return Moments
Take the expectation of the affine expression for $r_{t,0}$:
$$
\begin{aligned}
\mathbb{E}_{\mathbb{P}}[r_{t,0}]
&=\mathbb{E}_{\mathbb{P}}\left[t\ln d+K_t\ln\left(\frac{u}{d}\right)\right]\\
&=t\ln d+\mathbb{E}_{\mathbb{P}}[K_t]\ln\left(\frac{u}{d}\right)\\
&=t\ln d+tp\ln\left(\frac{u}{d}\right)\\
&=\boxed{t\left[p\ln u+(1-p)\ln d\right]}.
\end{aligned}
$$
The constant term $t\ln d$ does not contribute to the variance. Therefore:
$$
\begin{aligned}
\operatorname{Var}_{\mathbb{P}}(r_{t,0})
&=\operatorname{Var}_{\mathbb{P}}\left[K_t\ln\left(\frac{u}{d}\right)\right]\\
&=\left[\ln\left(\frac{u}{d}\right)\right]^2\operatorname{Var}_{\mathbb{P}}(K_t)\\
&=\boxed{tp(1-p)\left[\ln\left(\frac{u}{d}\right)\right]^2}.
\end{aligned}
$$

### Distinguish a Sum from an Average Growth Rate
The cumulative log return is related to the sum of the one-step growth rates by:
$$
\sum_{j=1}^{t}g_j=\frac{1}{\Delta t}\sum_{j=1}^{t}\ln\left(\frac{S_j}{S_{j-1}}\right)
=\frac{1}{\Delta t}\ln\left(\frac{S_t}{S_0}\right)
=\frac{r_{t,0}}{\Delta t}.
$$
The average growth rate over the full horizon is instead $\bar g_{t,0}=t^{-1}\sum_{j=1}^{t}g_j=r_{t,0}/(t\Delta t)$. Its moments follow by scaling the cumulative-return moments:
$$
\begin{aligned}
\mathbb{E}_{\mathbb{P}}[\bar g_{t,0}]
&=\frac{1}{t\Delta t}\mathbb{E}_{\mathbb{P}}[r_{t,0}]
=\frac{p\ln u+(1-p)\ln d}{\Delta t},\\
\operatorname{Var}_{\mathbb{P}}(\bar g_{t,0})
&=\frac{1}{t^2(\Delta t)^2}\operatorname{Var}_{\mathbb{P}}(r_{t,0})
=\frac{p(1-p)}{t(\Delta t)^2}\left[\ln\left(\frac{u}{d}\right)\right]^2.
\end{aligned}
$$
The sum $\sum_{j=1}^{t}g_j$ and the average $\bar g_{t,0}$ are different quantities. Keeping the factor $t$ explicit prevents the cumulative return from being mistaken for an average annualized rate.

## Stylized-Fact Derivations
The return laws above let us derive the three conclusions stated in the lecture. We take them in the order L3a introduced them: first the shape of the distribution, then dependence in the signed growth rates, then dependence in their magnitudes. Bounded one-step growth rates rule out heavy tails without requiring independence. Independence gives the Normal limit and the temporal-dependence results.

### Heavy Tails
The one-step growth rate takes only the two values $\ln(d)/\Delta t$ and $\ln(u)/\Delta t$. Its distribution is bounded, so it cannot have heavy tails. This conclusion holds regardless of dependence between moves.

### Normal Limit of Standardized Cumulative Returns

For the cumulative return, subtract its mean from the affine expression derived above:
$$
\begin{aligned}
r_{t,0}-\mathbb{E}_{\mathbb{P}}[r_{t,0}]
&=t\ln d+K_t\ln\left(\frac{u}{d}\right)
-t\left[\ln d+p\ln\left(\frac{u}{d}\right)\right]\\
&=(K_t-tp)\ln\left(\frac{u}{d}\right).
\end{aligned}
$$
Because $u>d$, we have $\ln(u/d)>0$. Divide by the standard deviation of $r_{t,0}$ to obtain:
$$
\begin{aligned}
\frac{r_{t,0}-\mathbb{E}_{\mathbb{P}}[r_{t,0}]}
{\sqrt{\operatorname{Var}_{\mathbb{P}}(r_{t,0})}}
&=\frac{(K_t-tp)\ln(u/d)}
{\sqrt{tp(1-p)}\,\ln(u/d)}\\
&=\frac{K_t-tp}{\sqrt{tp(1-p)}}.
\end{aligned}
$$
The right-hand side is the standardized binomial up-move count. Therefore, the Central Limit Theorem gives:
$$
\frac{r_{t,0}-\mathbb{E}_{\mathbb{P}}[r_{t,0}]}
{\sqrt{\operatorname{Var}_{\mathbb{P}}(r_{t,0})}}
\xrightarrow{\mathrm{d}}\mathcal{N}(0,1)
\qquad\text{as }t\rightarrow\infty.
$$
Here, $\xrightarrow{\mathrm{d}}$ denotes convergence in distribution and $\mathcal{N}(0,1)$ is the standard Normal distribution.
This is a Normal limit for the standardized cumulative return. The CLT alone does not rule out heavy-tailed one-step returns: independent returns with heavy tails and finite variance can also satisfy it. For this lattice, bounded one-step support is what rules out heavy tails.

### One-Step Autocorrelation
Define $a=\ln(u/d)/\Delta t$ and $b=\ln d/\Delta t$. The one-step growth-rate law becomes $g_j=b+aX_j$. For any integer lag $\tau\geq1$, expand the covariance as:
$$
\begin{aligned}
\operatorname{Cov}_{\mathbb{P}}(g_j,g_{j+\tau})
&=\operatorname{Cov}_{\mathbb{P}}(b+aX_j,b+aX_{j+\tau})\\
&=a^2\operatorname{Cov}_{\mathbb{P}}(X_j,X_{j+\tau})\\
&=0.
\end{aligned}
$$
The last equality follows because $X_j$ and $X_{j+\tau}$ are independent. Their growth-rate variances are both $a^2p(1-p)>0$, so the autocorrelation is:
$$
\begin{aligned}
\rho_g(\tau)
&=\frac{\operatorname{Cov}_{\mathbb{P}}(g_j,g_{j+\tau})}
{\sqrt{\operatorname{Var}_{\mathbb{P}}(g_j)}
\sqrt{\operatorname{Var}_{\mathbb{P}}(g_{j+\tau})}}\\
&=\frac{0}{\sqrt{a^2p(1-p)}\sqrt{a^2p(1-p)}}\\
&=\boxed{0\qquad\text{for every }\tau\geq1}.
\end{aligned}
$$
The model therefore reproduces zero linear autocorrelation in one-step growth rates.

### Nested Cumulative Windows
Let $\ell_j=\ln(S_j/S_{j-1})=(\Delta t)g_j$ be the one-step log return, and let $\sigma_{\ell}^2=\operatorname{Var}_{\mathbb{P}}(\ell_j)>0$. For $t\geq1$ and $\tau\geq1$, define the additional return after step $t$ as $A_{t,\tau}=\sum_{j=t+1}^{t+\tau}\ell_j$. The two nested cumulative returns can then be written as:
$$
r_{t,0}=\sum_{j=1}^{t}\ell_j,
\qquad
r_{t+\tau,0}=r_{t,0}+A_{t,\tau}.
$$
The random variables $r_{t,0}$ and $A_{t,\tau}$ contain disjoint one-step returns and are independent. Their covariance is therefore zero. Use this fact to expand the covariance between the nested returns:
$$
\begin{aligned}
\operatorname{Cov}_{\mathbb{P}}(r_{t,0},r_{t+\tau,0})
&=\operatorname{Cov}_{\mathbb{P}}(r_{t,0},r_{t,0}+A_{t,\tau})\\
&=\operatorname{Var}_{\mathbb{P}}(r_{t,0})
+\operatorname{Cov}_{\mathbb{P}}(r_{t,0},A_{t,\tau})\\
&=t\sigma_{\ell}^2+0
=t\sigma_{\ell}^2.
\end{aligned}
$$
The two return variances are $\operatorname{Var}_{\mathbb{P}}(r_{t,0})=t\sigma_{\ell}^2$ and $\operatorname{Var}_{\mathbb{P}}(r_{t+\tau,0})=(t+\tau)\sigma_{\ell}^2$. Substitute these expressions into the correlation formula:
$$
\begin{aligned}
\operatorname{Corr}_{\mathbb{P}}(r_{t,0},r_{t+\tau,0})
&=\frac{t\sigma_{\ell}^2}
{\sqrt{t\sigma_{\ell}^2}\sqrt{(t+\tau)\sigma_{\ell}^2}}\\
&=\boxed{\sqrt{\frac{t}{t+\tau}}}.
\end{aligned}
$$
The correlation comes from the shared first $t$ steps. It does not imply that one-step growth rates are predictable. Returns over non-overlapping windows depend on disjoint move indicators and remain independent.

### Volatility Clustering
Volatility clustering appears as positive dependence between the magnitudes of growth rates. Define the magnitude $M_j=|g_j|$. For any integer lag $\tau\geq1$, the growth rates $g_j$ and $g_{j+\tau}$ are independent, so the transformed variables $M_j$ and $M_{j+\tau}$ are also independent. Expand their covariance:
$$
\begin{aligned}
\operatorname{Cov}_{\mathbb{P}}(M_j,M_{j+\tau})
&=\mathbb{E}_{\mathbb{P}}[M_jM_{j+\tau}]
-\mathbb{E}_{\mathbb{P}}[M_j]\mathbb{E}_{\mathbb{P}}[M_{j+\tau}]\\
&=\mathbb{E}_{\mathbb{P}}[M_j]\mathbb{E}_{\mathbb{P}}[M_{j+\tau}]
-\mathbb{E}_{\mathbb{P}}[M_j]\mathbb{E}_{\mathbb{P}}[M_{j+\tau}]\\
&=0.
\end{aligned}
$$
The one-step distribution also stays fixed, so its variance is the same at every step:
$$
\operatorname{Var}_{\mathbb{P}}(g_j)
=\frac{p(1-p)}{(\Delta t)^2}\left[\ln\left(\frac{u}{d}\right)\right]^2
\qquad\text{for every }j.
$$
Thus, the constant-parameter lattice has neither time-varying one-step variance nor dependence between growth-rate magnitudes. It cannot produce volatility clustering.

## Summary
Starting from the Bernoulli move indicator, we derived the return distribution and its time-series properties step by step.

> __Key Takeaways:__
>
> * __One-step growth rates are transformed Bernoulli variables:__ Expanding the logarithm of the price factor gives an affine function of $X_j$. Linearity of expectation and the variance scaling rule then give the one-step moments.
> * __Multi-step returns are transformed binomial variables:__ Multiplying the step factors makes the terminal price a function of the up-move count $K_t$. The binomial moments then give the cumulative-return and average-growth-rate moments.
> * __Independence has different consequences for different windows:__ It produces zero covariance for distinct one-step growth rates and their magnitudes, but nested cumulative returns remain correlated because they share observations. The model therefore captures zero signed autocorrelation but misses heavy tails and volatility clustering.

The correlation between overlapping cumulative returns is caused by shared observations, not by predictability in the one-step returns.
