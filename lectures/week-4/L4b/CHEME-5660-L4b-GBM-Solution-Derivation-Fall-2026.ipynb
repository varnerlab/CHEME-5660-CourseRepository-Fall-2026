# Advanced: Derivation of the Geometric Brownian Motion Model Solution
This example derives the exact solution of the geometric Brownian motion (GBM) model. We first define a Wiener process and state Itô's lemma, then apply the lemma to the log price.

> __Learning Objectives:__
>
> By the end of this example, you should be able to:
>
> * **Define the driving noise:** State the initial value, independent-increment property, and normal increment law of a Wiener process.
> * **Apply Itô's lemma:** Transform the GBM price equation by computing the derivatives of the log-price function.
> * **Derive the exact GBM transition:** Obtain both the continuous-time solution and the exact transition on a discrete observation grid.

Let's get started!

Geometric Brownian motion is defined by the stochastic differential equation:
$$
dS_t = \mu S_t dt + \sigma S_t dW_t,
$$
where $\mu\in\mathbb R$ is the constant arithmetic drift, $\sigma>0$ is the constant volatility, and $dW_t$ is a Wiener-process increment. We reserve $g$ for an observed continuously compounded growth rate, such as the one-step rate $g_{j}=(1/\Delta t)\ln(S_{t_{j}}/S_{t_{j-1}})$ used in the lecture.

> __Wiener Process:__
> 
> A Wiener process $\{W(t):0\leq t\leq T\}$ is a continuous, real-valued stochastic process with the following properties:
>
> * The process starts at zero: $W(0)=0$ with probability one.
> * The increments $W(t_{1})-W(t_{0}),\ldots,W(t_{k})-W(t_{k-1})$ are independent for any $0\leq t_{0}<t_{1}<\cdots<t_{k}\leq T$.
> * For $0\leq s<t\leq T$, the increment law is $W(t)-W(s)\sim\mathcal N(0,t-s)$. Equivalently, $W(t)-W(s)$ has the same distribution as $\sqrt{t-s}\,Z$, where $Z\sim\mathcal N(0,1)$.
> 

The independent normal increments determine how GBM accumulates random price changes over time.

## Itô's Lemma
Itô's lemma is the chain rule for a smooth function of an Itô process.

> __Itô's Lemma:__
>
> Let the stochastic process $X(t)$ satisfy the stochastic differential equation:
> $$
> \begin{align*}
> dX &= a\left(X(t),t\right)dt + b\left(X(t),t\right)dW(t),
> \end{align*}
> $$
> where $W(t)$ is a one-dimensional Wiener process, $dW(t)$ denotes its increment, and $a$ and $b$ are functions of $X(t)$ and $t$.
>
> Let $Y(t)=\phi(t,X(t))$, where $\phi$ is once continuously differentiable in $t$ and twice continuously differentiable in its state argument $x$. Then $Y(t)$ satisfies:
> $$
> \begin{align*}
> dY &= \left(\frac{\partial\phi}{\partial t}+a\frac{\partial\phi}{\partial x}+\frac{b^{2}}{2}\frac{\partial^{2}\phi}{\partial x^{2}}\right)dt+b\frac{\partial\phi}{\partial x}dW(t).
> \end{align*}
> $$

___


## Derivation of the GBM Model Solution
Starting from $S(t_0)>0$, the GBM price process has drift and diffusion coefficients proportional to the current price:
$$
\begin{align*}
dS(t) &= \mu S(t) dt + \sigma S(t) dW(t).
\end{align*}
$$
To solve this equation, set $\phi(S)=\ln S$, $a(S,t)=\mu S$, and $b(S,t)=\sigma S$. Because $S(t)>0$, the needed derivatives are $\partial_t\phi=0$, $\partial_S\phi=1/S$, and $\partial^2_S\phi=-1/S^2$. Itô's lemma then gives:
$$
\begin{align*}
d\ln S(t) &= \left(0+\mu S(t)\frac{1}{S(t)}+\frac{\sigma^2S(t)^2}{2}\left[-\frac{1}{S(t)^2}\right]\right)dt+\sigma S(t)\frac{1}{S(t)}dW(t)\\
&=\left(\mu-\frac{\sigma^2}{2}\right)dt+\sigma dW(t).
\end{align*}
$$
We integrate both sides of the equation from $t_{0}$ to $t$:
$$
\begin{align*}
\int_{t_{0}}^{t}d\ln S(s) &= \int_{t_{0}}^{t}\left(\mu - \frac{\sigma^{2}}{2}\right)ds + \sigma\int_{t_{0}}^{t}dW(s).
\end{align*}
$$
The Wiener integral of the constant integrand is its increment, $\int_{t_0}^{t}dW(s)=W(t)-W(t_0)$. Evaluating both integrals gives:
$$
\begin{align*}
\ln\left(\frac{S_{t}}{S_{t_{0}}}\right) &= \left(\mu - \frac{\sigma^{2}}{2}\right)\left(t - t_{0}\right) + \sigma\left[W(t) - W(t_0)\right].
\end{align*}
$$
Exponentiating gives the pathwise solution. At a fixed $t$, the Wiener increment has the same distribution as $\sqrt{t-t_0}Z$ for $Z\sim\mathcal N(0,1)$, so the solution and its fixed-time distribution can be written as follows:
$$
\boxed{
\begin{align*}
S_{t} &= S_{t_{0}}\exp\left[\left(\mu - \frac{\sigma^{2}}{2}\right)(t-t_{0})+\sigma(W(t)-W(t_0))\right]\\
&\overset{d}{=}S_{t_{0}}\exp\left[\left(\mu - \frac{\sigma^{2}}{2}\right)(t-t_{0})+\sigma\sqrt{t-t_{0}}Z\right],\quad t\geq t_0,\quad Z\sim\mathcal N(0,1).
\end{align*}}
$$

The equality marked $\overset{d}{=}$ is equality in distribution at one fixed time; the Wiener-increment expression defines the process across times. Setting $t_{0}=0$ and $S_{t_{0}}=S_{0}$ gives the form used in the lecture.

For observations on the grid $t_{j}=j\Delta t$, applying the solution over $[t_{j-1},t_j]$ gives the exact one-step transition used in the lecture:
$$
\boxed{
\begin{align*}
S_{t_{j}} &= S_{t_{j-1}}\cdot\exp\left[\left(\mu - \frac{\sigma^{2}}{2}\right)\Delta t + \sigma\sqrt{\Delta t} \cdot Z_{j}\right]\qquad\text{for }j=1,2,\ldots,
\end{align*}}
$$
where the $Z_{j}\sim\mathcal{N}(0,1)$ are independent standard normal random variables, one per step, because Wiener increments over disjoint intervals are independent. Dividing the log-price change over a step by the elapsed time gives the one-step growth rate $g_{j}=\mu-\sigma^2/2+(\sigma/\sqrt{\Delta t})Z_{j}$, so $\mathbb{E}[g_{j}]=\mu-\sigma^2/2$: the mean growth rate sits below the arithmetic drift by the half-variance correction.

## Summary
In this notebook, we derived the closed-form solution of the geometric Brownian motion model by applying Itô's Lemma to the logarithm of the share price.

> __Key Takeaways:__
>
> * **Itô's Lemma is the stochastic chain rule:** Because Brownian squared increments contribute at first order, transforming a stochastic differential equation picks up a second-derivative term that ordinary calculus would miss.
> * **The log price is a Brownian motion with drift:** Applying the lemma to the logarithm of the price removes the state dependence, and integrating gives a normal log price ratio, so the price itself is lognormal.
> * **The half-variance correction is a consequence, not a convention:** The mean growth rate sits below the arithmetic drift by half the variance, and the exact one-step transition follows from the same solution applied over a single step.

The exact solution is what makes GBM simulable step by step and gives the lecture its closed-form target probability.
