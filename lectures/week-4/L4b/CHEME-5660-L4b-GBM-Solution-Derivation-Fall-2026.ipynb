Geometric Brownian Motion is defined by the following stochastic differential equation (SDE):
$$
dS_t = \mu S_t dt + \sigma S_t dW_t,
$$
where $\mu$ is the arithmetic drift term and $\sigma$ is the diffusion term. We reserve $g$ for an observed continuously compounded growth rate, e.g., the one-step growth rate $g_{j}=(1/\Delta t)\ln(S_{t_{j}}/S_{t_{j-1}})$ of the lecture. The noise term $dW_t$ is a Wiener process (or Brownian motion) increment. Let's first define what a Wiener process is.

> __Wiener Process__
> 
> A Wiener process is a real-valued continuous-time stochastic process named after Norbert Wiener for his study of one-dimensional Brownian motion. A Wiener process is a continuous one-dimensional stochastic process $\left\{W\left(t\right), 0\leq{t}\leq{T}\right\}$ with the following properties:
>
> * The process starts at zero: $W\left(0\right) = 0$ with probability $1$
> * The increments $\left\{W(t_{1}) - W(t_{0}),\dots, W(t_{k}) - W(t_{k-1})\right\}$ are independent for any $k$ and $0\leq{t_{0}}< t_{1} < \dots < t_{k} \leq{T}$
> * The increment $(W(t) - W(s))\sim~\mathcal{N}\left(0,t-s\right)$ for any $0\leq{s}< t \leq{T}$, where $\mathcal{N}\left(0,t-s\right)$ denotes a normally distributed random variable with mean $0$ and variance $t - s$. Thus, we can write $W(t) - W(s) = \sqrt{t-s}\;Z$, where $Z\sim\mathcal{N}(0,1)$ is a standard normal random variable.
> 

Thus, a Wiener process is a continuous-time stochastic noise process with independent, normally distributed increments. Next, we need to introduce Itô's Lemma.
___

## Itô's Lemma
Itô's Lemma, developed by K. Itô in 1951, is an analog of the Taylor series for stochastic systems.
Let the random variable $X(t)$ be governed by the stochastic differential equation:
$$
\begin{align*}
dX &= a\left(X(t),t\right)dt + b\left(X(t),t\right)dW(t)\\
\end{align*}
$$
where $dW(t)$ is a one-dimensional Wiener process and $a$ and $b$ are functions of $X(t)$ and $t$. 
Let $Y(t) = \phi\left(t,X(t)\right)$ be twice differentiable with respect to $X(t)$, 
and once differentiable with respect to $t$. Then, $Y(t)$ is governed by the equation:
$$
\begin{align*}
dY &= \left(\frac{\partial{Y}}{\partial{t}}+a\frac{\partial{Y}}{\partial{X}}+\frac{b^{2}}{2}\frac{\partial^{2}{Y}}{\partial{X}^{2}}\right)dt+b\left(\frac{\partial{Y}}{\partial{X}}\right)dW(t)
\end{align*}
$$
___

## Derivation of the GBM Model Solution
Geometric Brownian motion (GBM) is a stochastic differential equation describing the share price $S(t)$ 
as a random walk with drift and volatility proportional to $S(t)$:
$$
\begin{align*}
dS(t) &= \mu S(t) dt + \sigma S(t) dW(t)
\end{align*}
$$
Here, $\mu\in\mathbb{R}$ denotes the constant arithmetic drift, $\sigma>0$ is a constant __volatility__ parameter, and $dW$ represents a Wiener process (randomness). To solve this SDE, we apply Itô's Lemma. Let $Y = \ln(S)$, $a(S,t) = \mu S$, and $b(S,t) = \sigma S$. Then, we can write:
$$
\begin{align*}
dY &= \left(\frac{\partial{Y}}{\partial{t}}+a\frac{\partial{Y}}{\partial{X}}+\frac{b^{2}}{2}\frac{\partial^{2}{Y}}{\partial{X}^{2}}\right)dt+b\left(\frac{\partial{Y}}{\partial{X}}\right)dW(t)\\
d\ln(S) &= \left(\frac{\partial{\ln(S)}}{\partial{t}}+a\frac{\partial{\ln(S)}}{\partial{S}}+\frac{b^{2}}{2}\frac{\partial^{2}{\ln(S)}}{\partial{S}^{2}}\right)dt+b\left(\frac{\partial{\ln(S)}}{\partial{S}}\right)dW(t)\\
&= \left(0 + \mu S \cdot \frac{1}{S} + \frac{(\sigma S)^2}{2} \cdot \left(-\frac{1}{S^2}\right)\right)dt + \sigma S \cdot \frac{1}{S} \cdot dW(t)\\
&= \left(\mu - \frac{\sigma^{2}}{2}\right)dt + \sigma dW(t)\\
\end{align*}
$$
We integrate both sides of the equation from $t_{0}$ to $t$:
$$
\begin{align*}
\int_{t_{0}}^{t}d\ln(S) &= \int_{t_{0}}^{t}\left(\mu - \frac{\sigma^{2}}{2}\right)dt + \int_{t_{0}}^{t}\sigma dW(t)\\
& = \left(\mu - \frac{\sigma^{2}}{2}\right)\int_{t_{0}}^{t}dt + \sigma\int_{t_{0}}^{t} dW(t)\\
\end{align*}
$$
The challenging part of this integral is the stochastic integral $\int_{t_{0}}^{t} dW(t)$. By the properties of the Wiener process, we know that $W(t) - W(t_0) \sim \mathcal{N}(0, t-t_0)$, which gives us the distribution of the integral. Evaluating the left-hand side integral, we obtain:
$$
\begin{align*}
\ln\left(\frac{S_{t}}{S_{t_{0}}}\right) &= \left(\mu - \frac{\sigma^{2}}{2}\right)\left(t - t_{0}\right) + \sigma\underbrace{\left[W(t) - W(t_0)\right]}_{= \mathcal{N}(0, t-t_0)}
\end{align*}
$$
Since $W(t) - W(t_0) \sim \mathcal{N}(0, t-t_0)$, we can write this as $\sqrt{t-t_0} \cdot Z$ where $Z\sim\mathcal{N}(0,1)$ is a standard normal random variable. Taking the exponential of both sides, we obtain the continuous-time GBM model solution:
$$
\boxed{
\begin{align*}
S_{t} &= S_{t_{0}}\cdot\exp\left[\left(\mu - \frac{\sigma^{2}}{2}\right)\left(t - t_{0}\right) + \sigma\sqrt{t-t_{0}} \cdot Z\right]\quad\text{for }t\geq{t_{0}},\;Z\sim\mathcal{N}(0,1)
\end{align*}}
$$

This is the solution at a fixed time $t$: the single standard normal $Z$ stands in for the increment $W(t)-W(t_{0})$ at that one time, and the pathwise object is the Wiener increment itself. Setting $t_{0}=0$ and $S_{t_{0}}=S_{0}$ gives the form used in the lecture.

However, sometimes we are interested in equally spaced discrete time steps, e.g., daily closing prices or average daily prices. On the grid $t_{j} = j\Delta{t}$ for $j=0,1,2,\ldots$, applying the solution over a single step (with $t_{0}\rightarrow t_{j-1}$ and $t\rightarrow t_{j}$) gives the exact one-step transition used in the lecture:
$$
\boxed{
\begin{align*}
S_{t_{j}} &= S_{t_{j-1}}\cdot\exp\left[\left(\mu - \frac{\sigma^{2}}{2}\right)\Delta t + \sigma\sqrt{\Delta t} \cdot Z_{j}\right]
\end{align*}\quad\text{for }j=1,2,\ldots}
$$
where the $Z_{j}\sim\mathcal{N}(0,1)$ are independent standard normal random variables, one per step, because Wiener increments over disjoint intervals are independent. Dividing the log-price change over a step by the elapsed time gives the one-step growth rate $g_{j}=\mu-\sigma^2/2+(\sigma/\sqrt{\Delta t})Z_{j}$, so $\mathbb{E}[g_{j}]=\mu-\sigma^2/2$: the mean growth rate sits below the arithmetic drift by the half-variance correction.

## Summary
In this notebook, we derived the closed-form solution of the geometric Brownian motion model by applying Itô's Lemma to the logarithm of the share price.

> __Key Takeaways:__
>
> * **Itô's Lemma is the stochastic chain rule:** Because Brownian squared increments contribute at first order, transforming a stochastic differential equation picks up a second-derivative term that ordinary calculus would miss.
> * **The log price is a Brownian motion with drift:** Applying the lemma to the logarithm of the price removes the state dependence, and integrating gives a normal log price ratio, so the price itself is lognormal.
> * **The half-variance correction is a consequence, not a convention:** The mean growth rate sits below the arithmetic drift by half the variance, and the exact one-step transition follows from the same solution applied over a single step.

The exact solution is what makes GBM simulable step by step and gives the lecture its closed-form target probability.
