# Advanced: Derivation of the Geometric Brownian Motion Model Solution
In the lecture, we used the exact solution of the geometric Brownian motion (GBM) model to simulate share prices and calculate target probabilities. Where does this solution come from, and why does its growth term contain the correction $\sigma^2/2$?

> __Learning Objectives:__
>
> By the end of this example, you should be able to:
>
> * **Describe the driving noise:** State the defining properties of a Wiener process and explain how its increments depend on elapsed time.
> * **Apply Itô’s lemma:** Transform the price equation into an equation for the log price and identify the origin of the half-variance correction.
> * **Derive the exact solution:** Integrate the log-price equation and obtain the one-step price transition used in the lecture’s simulations.

In this notebook, we apply Itô’s lemma—the stochastic counterpart of the chain rule—to the logarithm of the share price and derive the solution step by step.

Let's get started!


___
## Task 1: Define the Price Model and Driving Noise

In this task, we state the share-price model and recall the properties of the Wiener process that drives its random fluctuations. Starting from $S(0)=S_0>0$, geometric Brownian motion satisfies:

$$
dS(t)=\mu S(t)\,dt+\sigma S(t)\,dW(t).
$$

Here, $\mu\in\mathbb R$ is the constant arithmetic drift, $\sigma>0$ is the constant volatility, and $dW(t)$ represents a Wiener-process increment. We measure time in years, so $\mu$ has units of $\text{year}^{-1}$ and $\sigma$ has units of $\text{year}^{-1/2}$.

> __Wiener Process:__
> 
> A Wiener process $\{W(t):0\leq t\leq T\}$ is a continuous, real-valued stochastic process with the following properties:
>
> * The process starts at zero: $W(0)=0$ with probability one.
> * The increments $W(t_{1})-W(t_{0}),\ldots,W(t_{k})-W(t_{k-1})$ are independent for any $0\leq t_{0}<t_{1}<\cdots<t_{k}\leq T$.
> * For $0\leq s<t\leq T$, the increment law is $W(t)-W(s)\sim\mathcal N(0,t-s)$. Equivalently, $W(t)-W(s)$ has the same distribution as $\sqrt{t-s}\,Z$, where $Z\sim\mathcal N(0,1)$.
>

To solve the price equation, we will work with the logarithm of the share price. Let's introduce Itô’s lemma, which tells us how to transform a stochastic differential equation.

___


## Task 2: Introduce Itô’s Lemma

In this task, we introduce the rule needed to transform the price equation into an equation for the log price. Itô’s lemma extends the ordinary chain rule to functions of stochastic processes driven by Wiener noise.

> __Itô's Lemma:__
>
> Let the stochastic process $X(t)$ satisfy the stochastic differential equation:
> $$
> \begin{align*}
> dX &= a\left(X(t),t\right)dt + b\left(X(t),t\right)dW(t),
> \end{align*}
> $$
> where $W(t)$ is a one-dimensional Wiener process, $dW(t)$ denotes its increment, and $a$ and $b$ are the drift and diffusion coefficients, respectively, which may depend on $X(t)$ and $t$.
>
> Let $Y(t)=\phi(t,X(t))$, where $\phi$ is once continuously differentiable in $t$ and twice continuously differentiable in its state argument $x$. Then $Y(t)$ satisfies:
> $$
> \begin{align*}
> dY &= \left(\frac{\partial\phi}{\partial t}+a\frac{\partial\phi}{\partial x}+\frac{b^{2}}{2}\frac{\partial^{2}\phi}{\partial x^{2}}\right)dt+b\frac{\partial\phi}{\partial x}dW(t).
> \end{align*}
> $$

The additional term $\frac{b^2}{2}\frac{\partial^2\phi}{\partial x^2}$ accounts for the contribution of squared Wiener increments, which accumulate over time. For the logarithm, the second derivative is negative, producing the half-variance correction. 

Let's apply the lemma to the share-price equation and work through that calculation.

___

## Task 3: Derive the Exact Price Solution

In this task, we derive the exact share-price solution and the one-step transition used in the lecture. Recall the price equation:

$$
dS(t)=\mu S(t)\,dt+\sigma S(t)\,dW(t).
$$

Both coefficients depend on the current price. Taking the logarithm will remove this dependence, leaving an equation we can integrate directly. For positive prices, set $\phi(t,S)=\ln S$, with derivatives:

$$
\frac{\partial\phi}{\partial t}=0,
\qquad
\frac{\partial\phi}{\partial S}=\frac{1}{S},
\qquad
\frac{\partial^2\phi}{\partial S^2}=-\frac{1}{S^2}.
$$

Substituting these derivatives and the coefficients $a(S,t)=\mu S$ and $b(S,t)=\sigma S$ into Itô’s lemma gives:

$$
\begin{aligned}
d\ln S(t)
&=\left(\mu S(t)\frac{1}{S(t)}
+\frac{\sigma^2S(t)^2}{2}
\left[-\frac{1}{S(t)^2}\right]\right)dt
+\sigma S(t)\frac{1}{S(t)}\,dW(t)\\
&=\left(\mu-\frac{\sigma^2}{2}\right)dt+\sigma\,dW(t).
\end{aligned}
$$

The price factors cancel, leaving constant drift and diffusion coefficients for the log price. Its drift is the mean growth rate $\mu_g=\mu-\sigma^2/2$ introduced in the lecture. Let's integrate this equation to obtain the solution.

Integrating from $t_0$ to $t$, with $t\geq t_0$, gives:

$$
\int_{t_0}^{t}d\ln S(s)
=\int_{t_0}^{t}\mu_g\,ds
+\sigma\int_{t_0}^{t}dW(s).
$$

The left-hand side is the change in log price, $\ln S(t)-\ln S(t_0)$. Because $\mu_g$ is constant, the drift integral is $\mu_g(t-t_0)$. The stochastic integral of the constant integrand $1$ is the Wiener increment:

$$
\int_{t_0}^{t}dW(s)=W(t)-W(t_0).
$$

Combining these results gives:

$$
\ln\!\left(\frac{S(t)}{S(t_0)}\right)
=\mu_g(t-t_0)+\sigma[W(t)-W(t_0)].
$$

The log return therefore consists of mean growth accumulated over the interval and a random contribution from the Wiener increment. Exponentiating will recover the share price.

Exponentiating and multiplying by $S(t_0)$ gives the exact price solution:

$$
\boxed{
S(t)=S(t_0)\exp\!\left[
\mu_g(t-t_0)+\sigma[W(t)-W(t_0)]
\right].
}
$$

The exponential is strictly positive, so a positive initial price remains positive.

To recover the lecture’s formula, set $t_0=0$, $S(0)=S_0$, and consider a fixed future time $T>0$. Since $W(0)=0$ and $W(T)\sim\mathcal N(0,T)$, the standardized increment is:

$$
Z=\frac{W(T)}{\sqrt T}\sim\mathcal N(0,1).
$$

Substituting $W(T)=\sqrt T\,Z$ and $\mu_g=\mu-\sigma^2/2$ gives:

$$
\boxed{
S_T=S_0\exp\!\left[
\left(\mu-\frac{\sigma^2}{2}\right)T+\sigma\sqrt T\,Z
\right],
\qquad Z\sim\mathcal N(0,1).
}
$$

Thus, the log return is normally distributed, and the price at time $T$ is lognormally distributed. Let's now apply the solution between successive observation times to obtain the one-step transition.

Consider observation times $t_j=j\Delta t$, for $j=0,\ldots,N$, with $\Delta t>0$ measured in years and $T=N\Delta t$. Over each interval, define the standardized Wiener increment:

$$
Z_j=\frac{W(t_j)-W(t_{j-1})}{\sqrt{\Delta t}},
\qquad j=1,\ldots,N.
$$

The Wiener increments are modeled as independent over disjoint intervals, so $Z_1,\ldots,Z_N$ are independent standard normal random variables. Applying the price solution over $[t_{j-1},t_j]$ gives:

$$
\boxed{
S_{t_j}=S_{t_{j-1}}\exp\!\left[
\left(\mu-\frac{\sigma^2}{2}\right)\Delta t
+\sigma\sqrt{\Delta t}\,Z_j
\right].
}
$$

For constant $\mu$ and $\sigma$, this transition gives exact model prices at the observation times and supplies the update used in the lecture’s simulations.

Taking the logarithm of the price ratio and dividing by $\Delta t$ gives the continuously compounded growth rate:

$$
g_j\equiv\frac{1}{\Delta t}
\ln\!\left(\frac{S_{t_j}}{S_{t_{j-1}}}\right)
=\mu_g+\frac{\sigma}{\sqrt{\Delta t}}Z_j.
$$

Since each shock has mean zero, $\mathbb E[g_j]=\mu_g=\mu-\sigma^2/2$. Thus, averaging observed growth rates estimates mean growth; recovering the arithmetic drift requires adding the half-variance correction.

## Summary

In this notebook, we derived the exact solution of the geometric Brownian motion model by applying Itô’s lemma to the logarithm of the share price.

> __Key Takeaways:__
>
> * **Origin of the half-variance correction:** We applied Itô’s lemma to the log price and obtained the mean growth rate $\mu_g=\mu-\sigma^2/2$. The second-derivative term explains why mean growth differs from the arithmetic drift $\mu$.
> * **Exact price solution:** We integrated the log-price equation and exponentiated to recover the share price. The solution shows that prices remain positive and are lognormally distributed at a fixed future time.
> * **Exact one-step transition:** We used independent Wiener increments to derive the price update between successive observations. This connects the continuous-time model to the discrete price-path simulations used in the lecture.

These results provide the mathematical basis for the lecture’s simulations and calculations of the probability of exceeding a target return.
