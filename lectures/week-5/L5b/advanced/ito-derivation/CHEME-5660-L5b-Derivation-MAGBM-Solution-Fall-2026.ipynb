# Derivation: The Multiple Asset GBM Model Solution
The [L5b lecture](../../CHEME-5660-L5b-Lecture-MultipleAsset-GBM-Fall-2026.ipynb) states the mean growth rate $\mu_{g,i}=\mu_i-C_{ii}/2$ and the exact one-step transition for the multiple asset geometric Brownian motion (GBM) model. Where do these results come from, and why do the covariances between assets not appear in the drift correction?

> __Learning Objectives:__
>
> By the end of this derivation, you should be able to:
>
> * **Describe the driving noise:** State the multiplication rules for the increments of independent Wiener processes and explain why accumulated products of increments from different processes vanish.
> * **Apply Itô's lemma with several noise sources:** Transform the price equation for one asset into an equation for its log price and identify the origin of the half-variance correction $C_{ii}/2$.
> * **Derive the exact solution:** Integrate the log-price equation, obtain the one-step transition driven by a shared shock vector, and compute the covariance of the one-step growth rates.

In this derivation, we extend the single asset argument from L4b to $M$ assets driven by a shared vector of independent Wiener increments. The new ingredient is the rule for products of increments from different processes.

Let's get started!

___

## The Price Model and the Driving Noise
In this section, we restate the multiple asset price model and record the properties of its driving noise. Consider the portfolio $\mathcal{P}=\{1,2,\ldots,M\}$ of $M\geq2$ assets from the lecture. The share price of asset $i\in\mathcal{P}$ satisfies:

$$
dS_i(t)=\mu_iS_i(t)\,dt+S_i(t)\sum_{\ell=1}^{M}A_{i\ell}\,dW_\ell(t),
\qquad S_i(0)>0.
$$

Here, $\mu_i$ is the constant drift of asset $i$ (units: inverse years), $\mathbf{A}\in\mathbb{R}^{M\times M}$ is the constant loading matrix (units: inverse square-root years) with $\mathbf{A}\mathbf{A}^{\top}=\mathbf{C}$, the covariance rate matrix (units: inverse years), and $W_1(t),\ldots,W_M(t)$ are standard Wiener processes with the properties we used in L4b: increments over disjoint intervals are independent, and an increment over an interval of length $\Delta t$ is normally distributed with mean zero and variance $\Delta t$. The multiple asset model adds one assumption: the $M$ processes are __mutually independent__.

Itô's lemma keeps terms through order $dt$ and discards terms of higher order. With several noise sources, we need to know which products of increments contribute at order $dt$.

> __Multiplication rules for Wiener increments:__
>
> Let $dW_\ell$ and $dW_m$ be increments of $W_\ell$ and $W_m$ over the same short interval of length $dt$. In Itô calculus, products of increments are replaced according to the rules:
>
> $$
> dW_\ell\,dW_m=\delta_{\ell m}\,dt,
> \qquad
> dW_\ell\,dt=0,
> \qquad
> dt\,dt=0,
> $$
>
> where $\delta_{\ell m}$ is the Kronecker delta, equal to one when $\ell=m$ and zero otherwise. These rules are shorthand for limits of sums over a time grid.

Why do these rules hold? Consider a grid $t_k=k\Delta t$ of $N$ steps of length $\Delta t=T/N$ on $[0,T]$, and let $\Delta W_{k,\ell}=W_\ell(t_k)-W_\ell(t_{k-1})$ be the increment of $W_\ell$ over step $k$. A single squared increment $\Delta W_{k,\ell}^2$ has mean $\Delta t$ and standard deviation $\sqrt{2}\,\Delta t$, so its fluctuations are as large as its mean; the rules describe sums over the grid. Because increments over disjoint steps are independent, the sum of the squared increments has mean and variance given by:

$$
\mathbb{E}\left[\sum_{k=1}^{N}\Delta W_{k,\ell}^2\right]=N\Delta t=T,
\qquad
\operatorname{Var}\left[\sum_{k=1}^{N}\Delta W_{k,\ell}^2\right]=2N\Delta t^2=2T\Delta t.
$$

As $\Delta t\to0$ with $T$ fixed, the variance goes to zero, so accumulated squared increments behave like elapsed time. This is the first rule when $\ell=m$.

For $\ell\neq m$, the two increments are independent with mean zero, so their product has mean zero and variance $\Delta t^2$. Summing over the grid gives:

$$
\mathbb{E}\left[\sum_{k=1}^{N}\Delta W_{k,\ell}\Delta W_{k,m}\right]=0,
\qquad
\operatorname{Var}\left[\sum_{k=1}^{N}\Delta W_{k,\ell}\Delta W_{k,m}\right]=N\Delta t^2=T\Delta t.
$$

The variance again goes to zero, so accumulated products of increments from different Wiener processes contribute nothing at order $dt$. The remaining rules follow because $\sum_k\Delta W_{k,\ell}\Delta t=\Delta t\,[W_\ell(T)-W_\ell(0)]$ and $\sum_k\Delta t^2=T\Delta t$, both of which vanish as $\Delta t\to0$. Independence is what makes the cross products vanish; our model keeps the Wiener processes independent and puts all the correlation between assets into $\mathbf{A}$.

Let's now state Itô's lemma for a process driven by several independent Wiener processes.

## Itô's Lemma with Several Noise Sources
In this section, we extend the version of Itô's lemma from L4b to a scalar process driven by $M$ independent Wiener processes. The statement is the same as before, except that the squared diffusion coefficient in the second-order term is replaced by a sum of squared coefficients, and the noise term becomes a sum over the $M$ processes.

> __Itô's Lemma (several noise sources):__
>
> Let the scalar process $X(t)$ satisfy the stochastic differential equation:
>
> $$
> dX=a\left(X(t),t\right)dt+\sum_{\ell=1}^{M}b_\ell\left(X(t),t\right)dW_\ell(t),
> $$
>
> where $W_1(t),\ldots,W_M(t)$ are independent standard Wiener processes, and the drift $a$ and the diffusion coefficients $b_1,\ldots,b_M$ may depend on $X(t)$ and $t$.
>
> Let $Y(t)=\phi(t,X(t))$, where $\phi$ is once continuously differentiable in $t$ and twice continuously differentiable in its state argument $x$. Then $Y(t)$ satisfies:
>
> $$
> dY=\left(\frac{\partial\phi}{\partial t}+a\frac{\partial\phi}{\partial x}+\frac{1}{2}\left[\sum_{\ell=1}^{M}b_\ell^{2}\right]\frac{\partial^{2}\phi}{\partial x^{2}}\right)dt+\frac{\partial\phi}{\partial x}\sum_{\ell=1}^{M}b_\ell\,dW_\ell(t).
> $$

With a single noise source, $M=1$, the bracketed sum reduces to $b_1^2$ and we recover the L4b statement. Where does the sum of squares come from? Expanding $\phi$ to second order in the increment $dX$ gives:

$$
dY=\frac{\partial\phi}{\partial t}\,dt+\frac{\partial\phi}{\partial x}\,dX+\frac{1}{2}\frac{\partial^{2}\phi}{\partial x^{2}}\,(dX)^2+\cdots
$$

The first-order term contributes $a\,\partial\phi/\partial x\,dt$ and the noise term of the lemma. The second-order term needs the multiplication rules. Substituting the equation for $dX$ and keeping only products that survive at order $dt$ gives:

$$
(dX)^2
=\left(\sum_{\ell=1}^{M}b_\ell\,dW_\ell\right)^2
=\sum_{\ell=1}^{M}\sum_{m=1}^{M}b_\ell b_m\,dW_\ell\,dW_m
=\sum_{\ell=1}^{M}b_\ell^{2}\,dt.
$$

The rule $dW_\ell\,dW_m=\delta_{\ell m}\,dt$ removes every term with $\ell\neq m$ from the double sum and leaves the $M$ diagonal terms. Thus, the quantity that replaces $b^2$ from the single noise case is the __total squared diffusion__ $\sum_\ell b_\ell^2$; the products of different coefficients do not survive in the second-order term.

Let's apply the lemma to the log price of asset $i$.

___

## The Exact Solution and the One-Step Transition
In this section, we derive the log-price equation, integrate it on the time grid, and recover the transition and growth-rate covariance stated in the lecture. Fix an asset $i\in\mathcal{P}$. Comparing the price equation with the general form in the lemma gives the coefficients:

$$
a\left(S_i,t\right)=\mu_iS_i,
\qquad
b_\ell\left(S_i,t\right)=A_{i\ell}S_i,
\qquad \ell=1,2,\ldots,M.
$$

As in L4b, taking the logarithm removes the dependence on the current price. For positive prices, set $\phi(t,S_i)=\ln S_i$, with derivatives:

$$
\frac{\partial\phi}{\partial t}=0,
\qquad
\frac{\partial\phi}{\partial S_i}=\frac{1}{S_i},
\qquad
\frac{\partial^2\phi}{\partial S_i^2}=-\frac{1}{S_i^2}.
$$

Substituting the derivatives and coefficients into Itô's lemma gives:

$$
\begin{aligned}
d\ln S_i(t)
&=\left(\mu_iS_i(t)\frac{1}{S_i(t)}
+\frac{1}{2}\left[\sum_{\ell=1}^{M}A_{i\ell}^{2}S_i(t)^2\right]
\left[-\frac{1}{S_i(t)^2}\right]\right)dt
+\frac{1}{S_i(t)}\sum_{\ell=1}^{M}A_{i\ell}S_i(t)\,dW_\ell(t)\\
&=\left(\mu_i-\frac{1}{2}\sum_{\ell=1}^{M}A_{i\ell}^{2}\right)dt
+\sum_{\ell=1}^{M}A_{i\ell}\,dW_\ell(t).
\end{aligned}
$$

The price factors cancel, leaving constant coefficients for the log price. The sum of squares in the drift is the $(i,i)$ entry of $\mathbf{A}\mathbf{A}^{\top}$, the squared volatility of asset $i$:

$$
\sum_{\ell=1}^{M}A_{i\ell}^{2}=\left(\mathbf{A}\mathbf{A}^{\top}\right)_{ii}=C_{ii}=\sigma_i^2.
$$

Thus, the drift of the log price is the mean growth rate from the lecture, $\mu_{g,i}=\mu_i-C_{ii}/2$, and the log-price equation is given by:

$$
d\ln S_i(t)=\mu_{g,i}\,dt+\sum_{\ell=1}^{M}A_{i\ell}\,dW_\ell(t).
$$

This answers our opening question. The correction for asset $i$ is half its own variance rate $C_{ii}$; the covariances in $\mathbf{C}$ describe how its noise moves with other assets' noise, but they never enter its own second-order term.

### Integrating on the time grid
As in the lecture, define the time grid $t_k=k\Delta t$ for $k=0,1,\ldots,N$, with $\Delta t>0$ measured in years and $T=N\Delta t$. Because $\mu_{g,i}$ and the loadings are constant, integrating the log-price equation from $t_{k-1}$ to $t_k$ gives:

$$
\ln\!\left(\frac{S_i(t_k)}{S_i(t_{k-1})}\right)
=\mu_{g,i}\Delta t
+\sum_{\ell=1}^{M}A_{i\ell}\left[W_\ell(t_k)-W_\ell(t_{k-1})\right].
$$

Following L4b, standardize each increment by its standard deviation $\sqrt{\Delta t}$:

$$
Z_{k,\ell}=\frac{W_\ell(t_k)-W_\ell(t_{k-1})}{\sqrt{\Delta t}},
\qquad \ell=1,2,\ldots,M,\quad k=1,2,\ldots,N.
$$

Collect the $M$ standardized increments for step $k$ into the vector $\mathbf{Z}_k=(Z_{k,1},\ldots,Z_{k,M})^{\top}$. Its components are independent standard normal random variables because they come from different Wiener processes, so $\mathbf{Z}_k\sim\mathcal{N}(\mathbf{0},\mathbf{I}_M)$, and the vectors $\mathbf{Z}_1,\ldots,\mathbf{Z}_N$ are independent because they come from disjoint time intervals. These are the two independence properties the lecture uses when it draws the shocks.

Substituting $W_\ell(t_k)-W_\ell(t_{k-1})=\sqrt{\Delta t}\,Z_{k,\ell}$, the noise term becomes $\sqrt{\Delta t}\sum_\ell A_{i\ell}Z_{k,\ell}=\sqrt{\Delta t}\,(\mathbf{A}\mathbf{Z}_k)_i$. Exponentiating and multiplying by $S_i(t_{k-1})$ recovers the exact one-step transition stated in the lecture:

$$
\boxed{
S_i(t_k)
=S_i(t_{k-1})\exp\!\left[\mu_{g,i}\Delta t
+\sqrt{\Delta t}\,(\mathbf{A}\mathbf{Z}_k)_i\right],
\qquad i\in\mathcal{P},\quad k=1,2,\ldots,N.
}
$$

Prices at a single fixed horizon $T$ follow from one step of length $T$ starting at $t_0=0$, with a single shock vector $\mathbf{Z}\sim\mathcal{N}(\mathbf{0},\mathbf{I}_M)$.

### The covariance of the one-step growth rates
Dividing the log-price change over step $k$ by $\Delta t$ gives the growth rate of asset $i$, $g_{k,i}=\frac{1}{\Delta t}\ln(S_i(t_k)/S_i(t_{k-1}))$. Collect the $M$ growth rates into the vector $\mathbf{g}_k$; we keep the step index $k$ that the lecture suppresses when it writes $\mathbf{g}$. In vector form, the integrated log-price equation reads:

$$
\mathbf{g}_k=\boldsymbol{\mu}_g+\frac{1}{\sqrt{\Delta t}}\,\mathbf{A}\mathbf{Z}_k,
$$

where $\boldsymbol{\mu}_g$ collects the mean growth rates. Because $\mathbf{Z}_k$ has mean zero and identity covariance, the growth rates have mean and covariance given by:

$$
\begin{aligned}
\mathbb{E}[\mathbf{g}_k]&=\boldsymbol{\mu}_g,\\
\operatorname{Cov}(\mathbf{g}_k)
&=\frac{1}{\Delta t}\,\mathbf{A}\operatorname{Cov}(\mathbf{Z}_k)\mathbf{A}^{\top}
=\frac{1}{\Delta t}\,\mathbf{A}\mathbf{A}^{\top}
=\frac{\mathbf{C}}{\Delta t}.
\end{aligned}
$$

The components of $\mathbf{g}_k$ are the growth rates $g_k^{(i)}$ of the lecture's empirical covariance section, and $\operatorname{Cov}(\mathbf{g}_k)=\mathbf{C}/\Delta t$ is the scaling relation the lecture uses to estimate $\mathbf{C}$ from data.

## Summary
In this derivation, we extended the single asset GBM solution to $M$ correlated assets by applying Itô's lemma to the log price of each asset, with the noise supplied by a shared vector of independent Wiener processes.

> __Key Takeaways:__
>
> * **Products of independent increments:** We justified the multiplication rules $dW_\ell\,dW_m=\delta_{\ell m}\,dt$ by showing that sums of squared increments over a grid converge to elapsed time while sums of cross products converge to zero. As a result, the second-order term in Itô's lemma contains only the sum of squared diffusion coefficients.
>
> * **Origin of the half-variance correction:** We applied the lemma to $\ln S_i$ and found that the correction for asset $i$ is half its total squared diffusion, $\tfrac12\sum_\ell A_{i\ell}^2=C_{ii}/2$. The correction to each asset's drift therefore depends on its own variance rate, with no contribution from its covariances with other assets.
>
> * **Exact transition and covariance:** We integrated the log-price equation over one step, standardized the Wiener increments into the shock vector $\mathbf{Z}_k$, and recovered the boxed transition from the lecture. In vector form, the growth rates have covariance $\mathbf{C}/\Delta t$, the scaling relation used to estimate the covariance rate from data.

These results provide the mathematical basis for the multiple asset simulations and the covariance estimation developed in the lecture.
