# Derivation: Global Minimum-Variance Portfolio Weights
The [L6a lecture](../../CHEME-5660-L6a-Lecture-MAGBM-Data-Portfolios-Fall-2026.ipynb) gives the global minimum-variance weights for a portfolio that allows short positions. Here, we derive those weights using a Lagrange multiplier and calculate the minimum variance.

> __Learning Objectives:__
>
> By the end of this derivation, you should be able to:
>
> * __Formulate the problem:__ Express growth-rate variance and the full-investment constraint in terms of the portfolio weights and covariance matrix.
> * __Derive the weights:__ Use a Lagrange multiplier and the budget constraint to obtain the global minimum-variance allocation.
> * __Interpret the solution:__ Calculate the minimum variance and explain why the solution depends on covariance rather than expected growth.

Let's get started!

___


## The Constrained Variance Problem
Let $\mathbf{w}=[w_1,\ldots,w_M]^{\top}$ contain the fractions of our initial investment assigned to $M$ risky assets. The weighted growth rate $g_p=\mathbf{w}^{\top}\mathbf{g}$ has variance $\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}$. Growth rates are measured in inverse years, so the covariance entries have units of inverse years squared.

We require the weights to sum to one and allow negative weights for short positions. There is no growth target. The lecture explains the idealized financing assumptions behind these positions.

For this derivation, assume the growth-rate covariance matrix $\mathbf{\Sigma}_g$ is symmetric and positive definite. Let $\mathbf{1}\in\mathbb{R}^M$ be the vector of ones, so $\mathbf{1}^{\top}\mathbf{w}=1$ expresses the budget constraint. The optimization problem is given by:
$$
\begin{aligned}
\underset{\mathbf{w}\in\mathbb{R}^M}{\operatorname{minimize}}\quad
&\frac{1}{2}\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}\\
\text{subject to}\quad &\mathbf{1}^{\top}\mathbf{w}=1.
\end{aligned}
$$
The factor of one half simplifies the derivative and does not change the minimizing weights.

___


## Deriving the Weights
A Lagrange multiplier lets us minimize the variance while enforcing the budget constraint.

> __Derivation:__
>
> We introduce a scalar multiplier $\lambda$ for the budget constraint and form the Lagrangian:
>
> $$\mathcal{L}(\mathbf{w},\lambda) =\frac{1}{2}\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w} -\lambda\left(\mathbf{1}^{\top}\mathbf{w}-1\right).$$
>
> Differentiating with respect to the weights and setting the gradient equal to zero gives:
>
> $$\nabla_{\mathbf{w}}\mathcal{L} =\mathbf{\Sigma}_g\mathbf{w}-\lambda\mathbf{1} =\mathbf{0}.$$
>
> Because $\mathbf{\Sigma}_g$ is positive definite, it has an inverse. Multiplying by that inverse gives the weights in terms of $\lambda$:
>
> $$\mathbf{w}=\lambda\mathbf{\Sigma}_g^{-1}\mathbf{1}.$$
>
> We determine $\lambda$ by requiring these weights to satisfy the budget constraint. The inverse of a positive-definite matrix is also positive definite, so $\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1}>0$ and we can divide by it:
>
> $$\begin{aligned} 1&=\mathbf{1}^{\top}\mathbf{w} =\lambda\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1},\\ \lambda&=\frac{1}{\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1}}. \end{aligned}$$
>
> Substituting this value into the weight expression gives:
>
> $$\boxed{ \mathbf{w}_{\mathrm{GMV}} =\frac{\mathbf{\Sigma}_g^{-1}\mathbf{1}} {\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1}}. }$$
>
> The denominator scales the weights so they sum to one. Positive definiteness also makes the variance objective strictly convex, so these weights give the unique minimum under the budget constraint.

We now calculate the variance achieved by these weights.

___


## Minimum Variance and Interpretation
What variance does this allocation achieve? Using $\mathbf{\Sigma}_g\mathbf{w}_{\mathrm{GMV}}=\lambda\mathbf{1}$ and $\mathbf{w}_{\mathrm{GMV}}^{\top}\mathbf{1}=1$, the minimum variance is given by:
$$
\boxed{
\begin{aligned}
\sigma_{g,\mathrm{GMV}}^2
&=\mathbf{w}_{\mathrm{GMV}}^{\top}\mathbf{\Sigma}_g\mathbf{w}_{\mathrm{GMV}}\\
&=\lambda\mathbf{w}_{\mathrm{GMV}}^{\top}\mathbf{1}
=\frac{1}{\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1}}.
\end{aligned}
}
$$
Its square root is the GMV portfolio's growth-rate standard deviation, in inverse years. The weights depend only on the covariance matrix because we have imposed no expected-growth requirement. When working from data, we use the estimated covariance $\hat{\mathbf{\Sigma}}_g$ in these expressions, provided it is positive definite.

___


## Summary
We minimized portfolio growth-rate variance subject to a full-investment constraint, allowing short positions.

> __Key Takeaways:__
>
> * **A covariance-based allocation:** We formulated the GMV problem without an expected-growth target. As a result, the solution requires the covariance matrix but no mean-growth estimates.
>
> * **The budget constraint:** We used a Lagrange multiplier to derive the weights, then normalized them to satisfy the full-investment constraint. Negative weights remain admissible.
>
> * **The minimum variance:** We substituted the weights into the variance expression and obtained the smallest achievable variance, the reciprocal of the sum of all entries of the inverse covariance matrix. Positive definiteness gives a unique solution.
