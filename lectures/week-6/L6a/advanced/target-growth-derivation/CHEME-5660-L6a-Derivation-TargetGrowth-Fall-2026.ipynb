# Derivation: Minimum-Variance Weights at an Exact Growth Target
The [GMV derivation](../gmv-derivation/CHEME-5660-L6a-Derivation-GMV-Fall-2026.ipynb) minimizes portfolio variance subject to a full-investment constraint. Here, we add an exact target for estimated expected growth and derive the analytical weights and minimum variance given in the [L6a lecture](../../CHEME-5660-L6a-Lecture-MAGBM-Data-Portfolios-Fall-2026.ipynb).

> __Learning Objectives:__
>
> By the end of this derivation, you should be able to:
>
> * __Formulate the problem:__ Express the budget and exact-growth constraints in terms of portfolio weights and sample-mean growth rates.
> * __Derive the weights:__ Use two Lagrange multipliers and the allocation constraints to obtain the unique minimum-variance portfolio at a given growth target.
> * __Interpret the solution:__ Calculate the minimum variance, recover the GMV portfolio as a special case, and explain how the target determines the additional variance.

Let's get started!

___


## The Target-Growth Problem
Let $\mathbf{w}=[w_1,\ldots,w_M]^{\top}\in\mathbb{R}^M$ contain the fractions of our initial investment assigned to $M\geq2$ risky assets. We allow negative weights for short positions. Let $\mathbf{1}\in\mathbb{R}^M$ be the vector of ones, so $\mathbf{1}^{\top}\mathbf{w}=1$ expresses the budget constraint.

The vector $\mathbf{g}^{\prime}\in\mathbb{R}^M$ contains the asset sample-mean growth rates, and $\mathbf{\Sigma}_g\in\mathbb{R}^{M\times M}$ is their growth-rate covariance matrix. Estimated portfolio growth is $(\mathbf{g}^{\prime})^{\top}\mathbf{w}$, and portfolio variance is $\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}$. Growth rates have units of inverse years, and covariance entries have units of inverse years squared.

For this derivation, assume $\mathbf{\Sigma}_g$ is symmetric and positive definite and the sample-mean growth rates are not all equal. For an exact growth target $g_{\star}$, in inverse years, the optimization problem is given by:

$$
\begin{aligned}
\underset{\mathbf{w}\in\mathbb{R}^M}{\operatorname{minimize}}\quad
&\frac{1}{2}\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}\\
\text{subject to}\quad
&\mathbf{1}^{\top}\mathbf{w}=1,\\
&(\mathbf{g}^{\prime})^{\top}\mathbf{w}=g_{\star}.
\end{aligned}
$$

The factor of one half simplifies differentiation and does not change the minimizing weights. The two constraints are independent because $\mathbf{g}^{\prime}$ is not a multiple of $\mathbf{1}$.

___


## Two Lagrange Multipliers
We extend the GMV calculation by introducing a second multiplier for the growth constraint.

> __Derivation:__
>
> Let $\lambda$ be the scalar multiplier for the budget constraint and $\gamma$ the scalar multiplier for the growth constraint. The Lagrangian is given by:
>
> $$\begin{aligned}
> \mathcal{L}(\mathbf{w},\lambda,\gamma)
> &=\frac{1}{2}\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}
> -\lambda(\mathbf{1}^{\top}\mathbf{w}-1)\\
> &\quad-\gamma\big((\mathbf{g}^{\prime})^{\top}\mathbf{w}-g_{\star}\big).
> \end{aligned}$$
>
> Differentiating with respect to the weights and setting the gradient equal to zero gives:
>
> $$\nabla_{\mathbf{w}}\mathcal{L}
> =\mathbf{\Sigma}_g\mathbf{w}-\lambda\mathbf{1}-\gamma\mathbf{g}^{\prime}
> =\mathbf{0}.$$
>
> Positive definiteness makes the covariance invertible, so the weights in terms of the two multipliers are given by:
>
> $$\mathbf{w}
> =\lambda\mathbf{\Sigma}_g^{-1}\mathbf{1}
> +\gamma\mathbf{\Sigma}_g^{-1}\mathbf{g}^{\prime}.$$

The budget and growth constraints determine $\lambda$ and $\gamma$.

___


## Solving for the Multipliers
To find $\lambda$ and $\gamma$, we substitute the weight expression into both constraints. The lecture's four scalar coefficients let us write the result compactly:

$$
\begin{aligned}
a&=\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1},
&b&=\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{g}^{\prime},\\
c&=(\mathbf{g}^{\prime})^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{g}^{\prime},
&d&=ac-b^2.
\end{aligned}
$$

The two constraints become:

$$
\begin{aligned}
1&=\mathbf{1}^{\top}\mathbf{w}=a\lambda+b\gamma,\\
g_{\star}&=(\mathbf{g}^{\prime})^{\top}\mathbf{w}=b\lambda+c\gamma.
\end{aligned}
$$

The coefficient of $\lambda$ in the second equation is also $b$ because $\mathbf{\Sigma}_g^{-1}$ is symmetric. Thus, we have a two-by-two linear system:

$$
\begin{bmatrix}a&b\\b&c\end{bmatrix}
\begin{bmatrix}\lambda\\\gamma\end{bmatrix}
=\begin{bmatrix}1\\g_{\star}\end{bmatrix}.
$$

Before we invert the coefficient matrix, we check that its determinant $d$ is nonzero.

> __Why is the system invertible?__
>
> The inverse covariance is positive definite, so $a>0$. The vector $\mathbf{g}^{\prime}-(b/a)\mathbf{1}$ is nonzero because the sample means are not all equal. Its quadratic form is therefore positive:
>
> $$\begin{aligned}
> \left(\mathbf{g}^{\prime}-\frac{b}{a}\mathbf{1}\right)^{\top}
> \mathbf{\Sigma}_g^{-1}
> \left(\mathbf{g}^{\prime}-\frac{b}{a}\mathbf{1}\right)
> &=c-2\frac{b^2}{a}+\frac{b^2}{a}\\
> &=c-\frac{b^2}{a}=\frac{d}{a}>0.
> \end{aligned}$$
>
> Since $a>0$, the determinant $d=ac-b^2$ is positive.

Inverting the coefficient matrix gives the two multipliers:

$$
\begin{bmatrix}\lambda\\\gamma\end{bmatrix}
=\frac{1}{d}
\begin{bmatrix}c&-b\\-b&a\end{bmatrix}
\begin{bmatrix}1\\g_{\star}\end{bmatrix}
=\begin{bmatrix}(c-bg_{\star})/d\\(ag_{\star}-b)/d\end{bmatrix}.
$$

___


## Optimal Weights and Minimum Variance
Substituting the two multipliers into the weight expression gives the minimizing allocation $\mathbf{w}_{\star}(g_{\star})$:

$$
\boxed{
\mathbf{w}_{\star}(g_{\star})
=\frac{c-bg_{\star}}{d}\,\mathbf{\Sigma}_g^{-1}\mathbf{1}
+\frac{ag_{\star}-b}{d}\,\mathbf{\Sigma}_g^{-1}\mathbf{g}^{\prime}.
}
$$

These weights satisfy both constraints by construction. Positive definiteness makes the variance strictly convex, so they give the unique minimum for each target. The allocation now depends on both the covariance matrix and the sample-mean growth rates.

What variance do these weights achieve? At the solution, stationarity gives $\mathbf{\Sigma}_g\mathbf{w}_{\star}=\lambda\mathbf{1}+\gamma\mathbf{g}^{\prime}$. Multiplying by $\mathbf{w}_{\star}^{\top}$ and using the two constraints gives:

$$
\begin{aligned}
\sigma_g^2(g_{\star})
&=\mathbf{w}_{\star}^{\top}\mathbf{\Sigma}_g\mathbf{w}_{\star}\\
&=\lambda\mathbf{w}_{\star}^{\top}\mathbf{1}
+\gamma\mathbf{w}_{\star}^{\top}\mathbf{g}^{\prime}\\
&=\lambda+\gamma g_{\star}\\
&=\frac{c-bg_{\star}}{d}+\frac{ag_{\star}-b}{d}g_{\star}.
\end{aligned}
$$

Combining the terms gives the minimum variance:

$$
\boxed{\sigma_g^2(g_{\star})=\frac{ag_{\star}^2-2bg_{\star}+c}{d}.}
$$

This variance has units of inverse years squared. Its square root is the smallest growth-rate standard deviation achievable at the target.

___


## Recovering the GMV Portfolio
The GMV portfolio has weights $\mathbf{w}_{\mathrm{GMV}}=\mathbf{\Sigma}_g^{-1}\mathbf{1}/a$. Its estimated expected growth rate is given by:

$$
(\mathbf{g}^{\prime})^{\top}\mathbf{w}_{\mathrm{GMV}}
=\frac{(\mathbf{g}^{\prime})^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1}}{a}
=\frac{b}{a}.
$$

When we set $g_{\star}=b/a$, the multiplier formulas reduce to:

$$
\gamma=\frac{a(b/a)-b}{d}=0,
\qquad
\lambda=\frac{c-b(b/a)}{d}=\frac{ac-b^2}{ad}=\frac{1}{a}.
$$

The growth term in the weight expression vanishes, and we recover the GMV allocation:

$$
\boxed{\mathbf{w}_{\star}(b/a)=\frac{\mathbf{\Sigma}_g^{-1}\mathbf{1}}{a}
=\mathbf{w}_{\mathrm{GMV}}.}
$$

We can also complete the square in the variance expression, using $c-b^2/a=d/a$:

$$
\begin{aligned}
\sigma_g^2(g_{\star})
&=\frac{a(g_{\star}-b/a)^2+c-b^2/a}{d}\\
&=\underbrace{\frac{1}{a}}_{\text{GMV variance}}
+\underbrace{\frac{a}{d}\left(g_{\star}-\frac{b}{a}\right)^2}_{\text{extra variance for the growth target}}.
\end{aligned}
$$

Any exact target different from the GMV growth rate requires additional variance. Varying $g_{\star}$ traces both branches of the minimum-variance frontier. Targets $g_{\star}\geq b/a$ give its efficient upper branch.

___


## Summary
We extended the GMV calculation to minimize portfolio growth-rate variance while meeting an exact growth target, allowing short positions.

> __Key Takeaways:__
>
> * **Two allocation constraints:** We added an exact growth target to the budget constraint. This connects the allocation to the sample-mean growth rates as well as the covariance matrix.
>
> * **Analytical portfolio weights:** We used two Lagrange multipliers and solved a two-by-two system to obtain the weights. Positive definiteness and unequal asset means made that system invertible and gave a unique minimizing allocation.
>
> * **The variance cost of a target:** We derived the minimum variance and recovered the GMV portfolio when the target matched its expected growth. Completing the square showed why any other exact target requires additional variance.
