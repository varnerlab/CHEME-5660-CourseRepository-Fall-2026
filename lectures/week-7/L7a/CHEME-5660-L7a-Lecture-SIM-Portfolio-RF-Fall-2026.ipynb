## Concept Review: Minimum-Variance Portfolios with SIM Inputs
Recall that the single index model of L6b writes the growth rate of each risky asset $i$ in the chosen set $\mathcal{P}$ as a market term plus a residual:

$$
g_{i} = \alpha_{i} + \beta_{i}\,g_{M} + \varepsilon_{i}
$$

where $g_{M}$ is the market growth rate. The market term carries the movement that asset $i$ shares with the other assets, and the residual $\varepsilon_{i}$ carries the variation left over. The model assumes that residuals have zero mean, are uncorrelated with the market, and are uncorrelated across assets.

Since $\mathbb{E}[\varepsilon_{i}]=0$, the residual drops out of the mean, so $\mathbb{E}[g_{i}]=\alpha_{i}+\beta_{i}\,\mathbb{E}[g_{M}]$. On the diagonal of the covariance matrix, the variance of asset $i$ is given by:

$$
\begin{align*}
\text{Var}\left(g_{i}\right) &= \text{Var}\left(\alpha_{i}+\beta_{i}\,g_{M}+\varepsilon_{i}\right) &&\quad\text{substitute the SIM}\\
&= \beta_{i}^{2}\,\text{Var}\left(g_{M}\right)+2\beta_{i}\,\underbrace{\text{Cov}\left(g_{M},\varepsilon_{i}\right)}_{=0}+\text{Var}\left(\varepsilon_{i}\right) &&\quad\text{expand}\\
&= \beta_{i}^{2}\,\sigma^{2}_{g,M}+\sigma^{2}_{g,\varepsilon,i} &&\quad\text{second assumption}
\end{align*}
$$

where $\sigma^{2}_{g,M}$ and $\sigma^{2}_{g,\varepsilon,i}$ are the market and residual variances. Off the diagonal, for two different assets $i\neq j$, taking the covariance of their SIMs gives:

$$
\begin{align*}
\text{Cov}\left(g_{i},g_{j}\right) &= \text{Cov}\left(\alpha_{i}+\beta_{i}\,g_{M}+\varepsilon_{i},\;\alpha_{j}+\beta_{j}\,g_{M}+\varepsilon_{j}\right) &&\quad\text{substitute both SIMs}\\
&= \beta_{i}\beta_{j}\,\text{Var}\left(g_{M}\right)+\beta_{i}\,\underbrace{\text{Cov}\left(g_{M},\varepsilon_{j}\right)}_{=0}+\beta_{j}\,\underbrace{\text{Cov}\left(\varepsilon_{i},g_{M}\right)}_{=0}+\underbrace{\text{Cov}\left(\varepsilon_{i},\varepsilon_{j}\right)}_{=0} &&\quad\text{expand}\\
&= \beta_{i}\beta_{j}\,\sigma^{2}_{g,M} &&\quad\text{second and third assumptions}
\end{align*}
$$

Collecting the betas in $\boldsymbol{\beta}$ and placing these entries in a matrix shows where each term goes:

$$
\begin{aligned}
\mathbf{\Sigma}_{g}
&=\begin{bmatrix}
\beta_{1}^{2}\,\sigma^{2}_{g,M}+\sigma^{2}_{g,\varepsilon,1} & \beta_{1}\beta_{2}\,\sigma^{2}_{g,M} & \cdots\\
\beta_{2}\beta_{1}\,\sigma^{2}_{g,M} & \beta_{2}^{2}\,\sigma^{2}_{g,M}+\sigma^{2}_{g,\varepsilon,2} & \cdots\\
\vdots & \vdots & \ddots
\end{bmatrix}
&&\quad\text{place the entries}\\
&=\sigma^{2}_{g,M}\begin{bmatrix}
\beta_{1}^{2} & \beta_{1}\beta_{2} & \cdots\\
\beta_{2}\beta_{1} & \beta_{2}^{2} & \cdots\\
\vdots & \vdots & \ddots
\end{bmatrix}
+\begin{bmatrix}
\sigma^{2}_{g,\varepsilon,1} & 0 & \cdots\\
0 & \sigma^{2}_{g,\varepsilon,2} & \cdots\\
\vdots & \vdots & \ddots
\end{bmatrix}
&&\quad\text{split each entry}\\
&=\underbrace{\sigma^{2}_{g,M}\,\boldsymbol{\beta}\boldsymbol{\beta}^{\top}}_{\text{market}}+\underbrace{\mathbf{D}_{g}}_{\text{residual}}
&&\quad\text{name the two matrices}
\end{aligned}
$$

The first matrix is $\boldsymbol{\beta}\boldsymbol{\beta}^{\top}$, whose $(i,j)$ entry is $\beta_{i}\beta_{j}$, so the market term fills every entry. The second matrix is $\mathbf{D}_{g}=\text{diag}\left(\sigma^{2}_{g,\varepsilon,1},\dots,\sigma^{2}_{g,\varepsilon,|\mathcal{P}|}\right)$, which holds only the residual variances. Each diagonal entry of $\mathbf{\Sigma}_{g}$ has a market part $\beta_{i}^{2}\,\sigma^{2}_{g,M}$ from $\sigma^{2}_{g,M}\,\boldsymbol{\beta}\boldsymbol{\beta}^{\top}$ and a residual part $\sigma^{2}_{g,\varepsilon,i}$ from $\mathbf{D}_{g}$.

Replacing these population quantities with their training-data estimates, and collecting the intercepts and betas in $\hat{\boldsymbol{\alpha}}$ and $\hat{\boldsymbol{\beta}}$, gives the estimated portfolio inputs:

$$
\boxed{\begin{aligned}
\hat{\mathbf{g}}_{\mathrm{SIM}} &= \hat{\boldsymbol{\alpha}}+\hat{\boldsymbol{\beta}}\,g'_M &&\quad\text{mean growth rates}\\
\hat{\mathbf{\Sigma}}_{g,\mathrm{SIM}} &= \underbrace{s^2_{g,M}\hat{\boldsymbol{\beta}}\hat{\boldsymbol{\beta}}^\top}_{\text{market}}+\underbrace{\hat{\mathbf{D}}_g}_{\text{residual}} &&\quad\text{growth-rate covariance}.
\end{aligned}}
$$

Here, $g'_M$ and $s^2_{g,M}$ are the sample mean and variance of the market growth rate. The diagonal matrix $\hat{\mathbf{D}}_g$ holds the estimated residual variances $s^2_{g,\varepsilon,i}$. Growth rates and intercepts have units of inverse years, betas are dimensionless, and covariance entries have units of inverse years squared.

### The long-only problem with SIM inputs
For fixed weights $\mathbf{w}$, the portfolio growth rate $g_{p}=\mathbf{w}^{\top}\mathbf{g}$ has the variance from L6a. Substituting the SIM entries of $\mathbf{\Sigma}_{g}$ gives:

$$
\begin{align*}
\text{Var}\left(g_{p}\right) &= \sum_{i\in\mathcal{P}}\sum_{j\in\mathcal{P}}w_{i}w_{j}\,\text{Cov}\left(g_{i},g_{j}\right) &&\quad\text{portfolio variance (L6a)}\\
&= \sum_{i\in\mathcal{P}}w_{i}^{2}\left(\beta_{i}^{2}\,\sigma^{2}_{g,M}+\sigma^{2}_{g,\varepsilon,i}\right)+\sum_{i\in\mathcal{P}}\sum_{j\neq i}w_{i}w_{j}\,\beta_{i}\beta_{j}\,\sigma^{2}_{g,M} &&\quad\text{diagonal and off-diagonal entries}\\
&= \sigma^{2}_{g,M}\sum_{i\in\mathcal{P}}\sum_{j\in\mathcal{P}}\left(w_{i}\beta_{i}\right)\left(w_{j}\beta_{j}\right)+\sum_{i\in\mathcal{P}}w_{i}^{2}\,\sigma^{2}_{g,\varepsilon,i} &&\quad\text{collect the market terms}\\
&= \sigma^{2}_{g,M}\Bigl(\sum_{i\in\mathcal{P}}w_{i}\beta_{i}\Bigr)^{2}+\sum_{i\in\mathcal{P}}w_{i}^{2}\,\sigma^{2}_{g,\varepsilon,i} &&\quad\text{the double sum is a square}\\
&= \underbrace{\beta_{p}^{2}\,\sigma^{2}_{g,M}}_{\text{market risk}}+\underbrace{\sum_{i\in\mathcal{P}}w_{i}^{2}\,\sigma^{2}_{g,\varepsilon,i}}_{\text{residual risk}} &&\quad\text{portfolio beta }\beta_{p}=\textstyle\sum_{i}w_{i}\beta_{i}
\end{align*}
$$

Market risk depends on the weights only through the portfolio beta $\beta_{p}$. Spreading the weights evenly over many assets whose residuals are uncorrelated and of similar size drives residual risk toward zero. With the estimated inputs, this split is the objective of the long-only allocation problem, where $\mathbf{1}$ is the vector of ones:

> __Long-only minimum-variance portfolio with SIM inputs:__
>
> __Setup.__ Choose the weights $\mathbf{w}\in\mathbb{R}^{|\mathcal{P}|}$, where $w_{i}$ is the initial fraction invested in asset $i$. Given the SIM inputs and a floor $g_{\star}$ on estimated expected growth, in inverse years, we solve:
>
> $$\boxed{\begin{aligned} \underset{\mathbf{w}}{\operatorname{minimize}}\quad &\underbrace{s^{2}_{g,M}\left(\hat{\boldsymbol{\beta}}^{\top}\mathbf{w}\right)^{2}}_{\text{market risk}}+\underbrace{\sum_{i\in\mathcal{P}}s^{2}_{g,\varepsilon,i}\,w_{i}^{2}}_{\text{residual risk}} &&\quad\text{minimize }\mathbf{w}^{\top}\hat{\mathbf{\Sigma}}_{g,\text{SIM}}\mathbf{w}\\ \text{subject to}\quad &\hat{\mathbf{g}}_{\text{SIM}}^{\top}\mathbf{w}\geq g_{\star} &&\quad\text{meet or exceed the growth target}\\ &\mathbf{1}^{\top}\mathbf{w}=1 &&\quad\text{weights sum to one}\\ &0\leq w_{i}\leq1,\quad i\in\mathcal{P} &&\quad\text{exclude short positions}. \end{aligned}}\tag{SIM-1}$$
>
> __Solution.__ This is the convex quadratic program LO-1 of L6a with a different covariance matrix. Sweeping $g_{\star}$ upward from the growth rate of the long-only global minimum-variance (GMV) portfolio traces the efficient frontier.

## Adding a Risk-Free Asset
So far, every asset was risky. A __risk-free asset__ has a known growth rate $g_{f}$ (inverse years) and zero variance over the holding period, such as a Treasury bill held to maturity. The figure shows what the risk-free asset adds:

    <center>
        <img src="figs/capital-allocation-line/cal.svg" width="760" alt="Capital allocation line from the risk-free asset at zero standard deviation, tangent to the red efficient frontier at portfolio T. The segment to T represents lending with 0 less than w_f less than 1. The extension beyond T represents borrowing with w_f less than 0. The risky global minimum-variance portfolio and dominated lower branch are also shown."/>
    </center>
</div>

* __Capital allocation line (CAL):__ Mixing the risk-free asset with a risky portfolio gives a straight line from $(0,g_{f})$ through that portfolio.
* __Tangent portfolio $\mathcal{T}$:__ The steepest such line touches the efficient frontier at $\mathcal{T}$.
* __Lending and borrowing:__ Points before $\mathcal{T}$ lend at $g_{f}$, and points beyond it borrow at $g_{f}$.

### The capital allocation line
Put a fraction $w_{f}$ of wealth in the risk-free asset and $1-w_{f}$ in a risky portfolio $p$ with standard deviation $\sigma_{g,p}>0$. The resulting __complete portfolio__ $c$ has:

$$
\begin{align*}
\mathbb{E}[g_{c}] &= w_{f}\,g_{f}+(1-w_{f})\,\mathbb{E}[g_{p}] &&\quad\text{expectation is linear}\\
&= g_{f}+(1-w_{f})\left(\mathbb{E}[g_{p}]-g_{f}\right) &&\quad\text{write }w_{f}=1-(1-w_{f})\\
\sigma^{2}_{g,c} &= w_{f}^{2}\,\underbrace{\text{Var}\left(g_{f}\right)}_{=0}+2w_{f}(1-w_{f})\,\underbrace{\text{Cov}\left(g_{f},g_{p}\right)}_{=0}+(1-w_{f})^{2}\,\sigma^{2}_{g,p} &&\quad\text{expand the variance}\\
&= (1-w_{f})^{2}\,\sigma^{2}_{g,p} &&\quad g_{f}\text{ is constant}\\
\sigma_{g,c} &= |1-w_{f}|\,\sigma_{g,p} &&\quad\text{take the square root}
\end{align*}
$$

For $w_{f}\leq1$, substituting $1-w_{f}=\sigma_{g,c}/\sigma_{g,p}$ gives the __capital allocation line__:

$$
\boxed{
\mathbb{E}[g_{c}] = g_{f}+\underbrace{\frac{\mathbb{E}[g_{p}]-g_{f}}{\sigma_{g,p}}}_{\text{Sharpe ratio }\mathrm{SR}_{p}}\,\sigma_{g,c}
}
$$

The slope $\mathrm{SR}_{p}$ is the expected excess growth per unit of growth-rate standard deviation. We lend for $0<w_{f}<1$, hold $p$ alone for $w_{f}=0$, and borrow for $w_{f}<0$.

> __Which Sharpe ratio?__ Each growth rate is a one-day log return divided by $\Delta t$. That division scales the excess mean and the standard deviation alike, so the slope is the one-day Sharpe ratio. Under L5b's model of independent increments with constant parameters, the annualized ratio with $\tau=1$ year is given by:
>
> $$\mathrm{SR}_{p}^{\mathrm{annual}}=\sqrt{\frac{\tau}{\Delta t}}\,\mathrm{SR}_{p}$$
>
> For $\Delta t=1/252$ years, the factor is $\sqrt{252}\approx15.9$, and both versions rank portfolios the same way ([Sharpe, 1994](https://web.stanford.edu/~wfsharpe/art/sr/SR.htm)).

Which risky portfolio gives the steepest line?

### The tangent portfolio
The __tangent portfolio__ $\mathcal{T}$ maximizes the Sharpe ratio. With short positions allowed, it has a closed form:

> __Tangent portfolio:__
>
> __Setup.__ Let $\boldsymbol{\mu}_{g}$ hold the expected growth rates and $\mathbf{\Sigma}_{g}$ be their positive-definite covariance matrix. Let the expected growth of the GMV portfolio with short positions allowed exceed $g_{f}$, and define the excess-growth vector $\mathbf{e}=\boldsymbol{\mu}_{g}-g_{f}\mathbf{1}$. We choose fully invested weights $\mathbf{w}$ to solve:
>
> $$\underset{\mathbf{w}}{\operatorname{maximize}}\quad\frac{\mathbf{e}^{\top}\mathbf{w}}{\sqrt{\mathbf{w}^{\top}\mathbf{\Sigma}_{g}\mathbf{w}}}\qquad\text{subject to}\quad\mathbf{1}^{\top}\mathbf{w}=1\tag{T-1}$$
>
> __Derivation sketch.__ Multiplying $\mathbf{w}$ by any $c>0$ leaves the Sharpe ratio unchanged, so we can set its gradient to zero and fix the scale afterward. With $\sigma=\sqrt{\mathbf{w}^{\top}\mathbf{\Sigma}_{g}\mathbf{w}}$, this gives:
>
> $$\begin{aligned} \nabla\mathrm{SR}(\mathbf{w}) &= \frac{\mathbf{e}}{\sigma}-\frac{\mathbf{e}^{\top}\mathbf{w}}{\sigma^{3}}\,\mathbf{\Sigma}_{g}\mathbf{w}=\mathbf{0} &&\quad\text{quotient rule, }\nabla\sigma=\mathbf{\Sigma}_{g}\mathbf{w}/\sigma\\ \mathbf{\Sigma}_{g}\mathbf{w} &= \frac{\mathbf{w}^{\top}\mathbf{\Sigma}_{g}\mathbf{w}}{\mathbf{e}^{\top}\mathbf{w}}\,\mathbf{e} &&\quad\text{solve for }\mathbf{\Sigma}_{g}\mathbf{w} \end{aligned}\tag{T-2}$$
>
> __Solution.__ The weights point in the direction $\mathbf{\Sigma}_{g}^{-1}\mathbf{e}$. Requiring them to sum to one gives:
>
> $$\boxed{\mathbf{w}_{\mathcal{T}}=\frac{\mathbf{\Sigma}_{g}^{-1}\left(\boldsymbol{\mu}_{g}-g_{f}\mathbf{1}\right)}{\mathbf{1}^{\top}\mathbf{\Sigma}_{g}^{-1}\left(\boldsymbol{\mu}_{g}-g_{f}\mathbf{1}\right)}}\tag{T-3}$$
>
> The [tangent derivation companion](advanced/tangent-derivation/CHEME-5660-L7a-Derivation-Tangent-Fall-2026.ipynb) gives the full calculation, including why the GMV growth rate must exceed $g_{f}$.

When T-3 has negative weights, the long-only tangent portfolio has no closed form. The risky and risk-free problem below finds it numerically.

### Two-fund separation
The steepest CAL separates choosing a risky portfolio from choosing how much risk to take:

> __Two-fund separation:__ Mean-variance investors with the same inputs who can lend and borrow freely at $g_{f}$ hold the __same risky fund__ $\mathcal{T}$, combined with the risk-free asset. Their risk preferences set $w_{f}$. This is Tobin's separation result ([Tobin, 1958](https://doi.org/10.2307/2296205)).

Real investors face limits on how they finance their positions:

* __Long-only weights:__ The risky fund is the long-only maximum-Sharpe portfolio, and investors still choose $w_{f}$.
* __No borrowing ($w_{f}\geq0$):__ The efficient boundary follows the CAL up to $\mathcal{T}$, then the risky efficient frontier. Investors who want more growth than $\mathcal{T}$ must hold a different risky portfolio.
* __Costlier borrowing:__ A borrowing rate above $g_{f}$ gives borrowers a flatter line through a different risky portfolio.

Next, we put the SIM estimates into the risky and risk-free problem.

### Risky and risk-free allocation with SIM inputs
With SIM inputs, the risky and risk-free problem is given by:

> __Risky and risk-free portfolio with SIM inputs:__
>
> __Setup.__ Choose the risky weights $\mathbf{w}\in\mathbb{R}^{|\mathcal{P}|}$ and put the rest, $w_{f}=1-\mathbf{1}^{\top}\mathbf{w}$, in the risk-free asset. For a target $g_{\star}>g_{f}$, in inverse years, we solve:
>
> $$\boxed{\begin{aligned} \underset{\mathbf{w}}{\operatorname{minimize}}\quad &\mathbf{w}^{\top}\hat{\mathbf{\Sigma}}_{g,\text{SIM}}\mathbf{w} &&\quad\text{minimize growth-rate variance}\\ \text{subject to}\quad &\hat{\mathbf{g}}_{\text{SIM}}^{\top}\mathbf{w}+\underbrace{\left(1-\mathbf{1}^{\top}\mathbf{w}\right)}_{w_{f}}g_{f}\geq g_{\star} &&\quad\text{meet or exceed the growth target}\\ &0\leq w_{i}\leq1,\quad i\in\mathcal{P} &&\quad\text{exclude short positions}. \end{aligned}}\tag{SIM-2}$$
>
> The fraction $w_{f}$ has no sign constraint, so $w_{f}<0$ means borrowing at $g_{f}$.
>
> __Solution.__ Eliminating $w_{f}$ gives an excess-growth constraint, $\left(\hat{\mathbf{g}}_{\text{SIM}}-g_{f}\mathbf{1}\right)^{\top}\mathbf{w}\geq g_{\star}-g_{f}$. Scaling $\mathbf{w}$ by $c>0$ scales the excess growth by $c$ and the variance by $c^{2}$. So, until a bound $w_{i}\leq1$ binds, every target selects the same risky direction:
>
> $$\boxed{\mathbf{w}(g_{\star})=(1-w_{f})\,\mathbf{w}_{\mathcal{T}},\qquad 1-w_{f}=\frac{g_{\star}-g_{f}}{\hat{\mathbf{g}}_{\text{SIM}}^{\top}\mathbf{w}_{\mathcal{T}}-g_{f}}}\tag{SIM-3}$$
>
> where the long-only tangent portfolio $\mathbf{w}_{\mathcal{T}}=\mathbf{w}/\mathbf{1}^{\top}\mathbf{w}$ is the solution at any target, rescaled to sum to one. Targets below $\hat{\mathbf{g}}_{\text{SIM}}^{\top}\mathbf{w}_{\mathcal{T}}$ lend, and targets above it borrow.

## Summary

In this lecture, we compared minimum-variance portfolios built from SIM and data-driven inputs, then added a risk-free asset. The tangent portfolio and capital allocation line let us separate the choice of risky holdings from the amount invested in them.

> __Key Takeaways:__
>
> * __The covariance model changes the allocation:__ We recalled that the SIM keeps the sample means but sets the residual covariances to zero. Under the sample covariance, which the data-driven weights minimize, the SIM weights carry at least as much risk at every growth floor.
>
> * __The tangent portfolio sets the steepest capital allocation line:__ We combined a risky portfolio with a risk-free asset and identified the Sharpe ratio as the line's slope. Maximizing that slope selects the tangent portfolio.
>
> * __Lending and borrowing set the complete portfolio's risk:__ Investors with the same inputs who can lend and borrow freely at the risk-free rate hold the same risky fund, choosing only the risk-free fraction. We found that fund with SIM inputs, noted how financing limits can change it, and tested complete portfolios against an index fund on 2025 prices.

We can now take any set of firms, build their SIM inputs, and choose a complete portfolio that matches a client's tolerance for risk.
