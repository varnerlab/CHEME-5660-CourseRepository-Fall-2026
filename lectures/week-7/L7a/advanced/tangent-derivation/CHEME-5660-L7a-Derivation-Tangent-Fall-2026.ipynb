# Derivation: Tangent Portfolio Weights and the Capital Allocation Line
The [L7a lecture](../../CHEME-5660-L7a-Lecture-SIM-Portfolio-RF-Fall-2026.ipynb) gives the tangent portfolio's weights when short positions are allowed. Here, we derive them, show that they give the largest Sharpe ratio, and connect them to the risky and risk-free problem.

> __Learning Objectives:__
>
> By the end of this derivation, you should be able to:
>
> * __Formulate the problem:__ Express the Sharpe ratio in terms of the portfolio weights and explain why multiplying the weights by a positive number leaves it unchanged.
> * __Derive the weights:__ Use the gradient of the Sharpe ratio to obtain the tangent weights and explain the lecture's condition on the global minimum-variance portfolio.
> * __Connect to the risk-free problem:__ Show that the tangent weights maximize the Sharpe ratio and that the same risky mix stays optimal when we add a risk-free asset.

Let's get started!

___

## The Sharpe-Ratio Problem
Let $\mathbf{w}=[w_1,\ldots,w_M]^{\top}$ contain the fractions of our initial investment assigned to $M\geq2$ risky assets, with short positions allowed. The vector $\boldsymbol{\mu}_g$ holds the expected growth rates, in inverse years, and $\mathbf{\Sigma}_g$ is their symmetric, positive-definite covariance matrix, in inverse years squared. Let $g_f$ be the risk-free growth rate and $\mathbf{1}$ the vector of ones. As in the lecture, the excess-growth vector is $\mathbf{e}=\boldsymbol{\mu}_g-g_f\mathbf{1}$. We assume the expected growth rates are not all equal, so the risky frontier has two branches and $\mathbf{e}\neq\mathbf{0}$.

For fully invested weights, $\mathbf{1}^{\top}\mathbf{w}=1$, so we can write $g_f=g_f\mathbf{1}^{\top}\mathbf{w}$. The expected excess growth is then $\boldsymbol{\mu}_g^{\top}\mathbf{w}-g_f=\mathbf{e}^{\top}\mathbf{w}$. The lecture's problem (T-1) is given by:

$$
\begin{aligned}
\underset{\mathbf{w}\in\mathbb{R}^M}{\operatorname{maximize}}\quad
&\mathrm{SR}(\mathbf{w})=\frac{\mathbf{e}^{\top}\mathbf{w}}{\sqrt{\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}}}\\
\text{subject to}\quad &\mathbf{1}^{\top}\mathbf{w}=1.
\end{aligned}
$$

__Scaling the weights.__ For any number $c>0$, the ratio is unchanged:

$$
\mathrm{SR}(c\,\mathbf{w})=\frac{c\,\mathbf{e}^{\top}\mathbf{w}}{c\sqrt{\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}}}=\mathrm{SR}(\mathbf{w}).
$$

The Sharpe ratio depends only on the direction of $\mathbf{w}$. So we find the best direction first and then check whether a positive scale makes it sum to one.

___

## Deriving the Weights
As in the L6a GMV derivation, we use a Lagrange multiplier for the budget constraint.

> __Derivation:__
>
> Let $\sigma=\sqrt{\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}}$ be the portfolio's growth-rate standard deviation. It is positive for $\mathbf{w}\neq\mathbf{0}$, and its gradient is $\nabla\sigma=\mathbf{\Sigma}_g\mathbf{w}/\sigma$. The quotient rule gives the gradient of the Sharpe ratio:
>
> $$\nabla\mathrm{SR}(\mathbf{w})=\frac{\mathbf{e}}{\sigma}-\frac{\mathbf{e}^{\top}\mathbf{w}}{\sigma^{3}}\,\mathbf{\Sigma}_g\mathbf{w}.$$
>
> A stationary point of the Lagrangian $\mathcal{L}(\mathbf{w},\lambda)=\mathrm{SR}(\mathbf{w})-\lambda(\mathbf{1}^{\top}\mathbf{w}-1)$ satisfies $\nabla\mathrm{SR}(\mathbf{w})=\lambda\mathbf{1}$. Multiplying both sides by $\mathbf{w}^{\top}$ gives:
>
> $$\begin{aligned}
> \lambda\,\mathbf{w}^{\top}\mathbf{1}
> &=\frac{\mathbf{e}^{\top}\mathbf{w}}{\sigma}-\frac{\mathbf{e}^{\top}\mathbf{w}}{\sigma^{3}}\,\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w} &&\quad\text{substitute the gradient}\\
> &=0. &&\quad\text{use }\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}=\sigma^{2}
> \end{aligned}$$
>
> Since $\mathbf{w}^{\top}\mathbf{1}=1$, the multiplier is $\lambda=0$. The budget constraint only fixes the scale, which the Sharpe ratio ignores. Setting the gradient to zero gives the lecture's condition (T-2):
>
> $$\mathbf{\Sigma}_g\mathbf{w}=k\,\mathbf{e},\qquad k=\frac{\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}}{\mathbf{e}^{\top}\mathbf{w}}.$$
>
> We can divide by $\mathbf{e}^{\top}\mathbf{w}$ because it is not zero at a stationary point: if it were, the gradient would be $\mathbf{e}/\sigma\neq\mathbf{0}$. Because $k$ is a scalar, every stationary point has the form $\mathbf{w}=k\,\mathbf{\Sigma}_g^{-1}\mathbf{e}$. The budget constraint gives $1=k\,\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}$. When $\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}\neq0$, solving for $k$ gives the tangent weights:
>
> $$\boxed{\mathbf{w}_{\mathcal{T}}=\frac{\mathbf{\Sigma}_g^{-1}\mathbf{e}}{\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}}.}$$

The sign of the denominator decides whether these weights maximize the Sharpe ratio.

___

## The Denominator and the GMV Condition
Why does the lecture's Setup assume that the global minimum-variance (GMV) portfolio's expected growth exceeds $g_f$? Using the GMV weights from L6a, $\mathbf{w}_{\mathrm{GMV}}=\mathbf{\Sigma}_g^{-1}\mathbf{1}/(\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1})$, the denominator factors as:

$$
\begin{aligned}
\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}
&=\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\boldsymbol{\mu}_g-g_f\,\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1} &&\quad\text{substitute }\mathbf{e}\\
&=\left(\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1}\right)\left(\frac{\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\boldsymbol{\mu}_g}{\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1}}-g_f\right) &&\quad\text{factor out }\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1}\\
&=\left(\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{1}\right)\left(\boldsymbol{\mu}_g^{\top}\mathbf{w}_{\mathrm{GMV}}-g_f\right). &&\quad\mathbf{\Sigma}_g^{-1}\text{ is symmetric}
\end{aligned}
$$

The first factor is positive because $\mathbf{\Sigma}_g^{-1}$ is positive definite. So the denominator has the sign of the GMV portfolio's expected excess growth:

* __GMV growth above $g_f$:__ The tangent weights are a positive multiple of $\mathbf{\Sigma}_g^{-1}\mathbf{e}$, and the tangent portfolio's expected excess growth is $\mathbb{E}[g_{\mathcal{T}}]-g_f=\mathbf{e}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}/\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}>0$. The capital allocation line touches the efficient upper branch, as in the lecture's figure.
* __GMV growth at or below $g_f$:__ If GMV growth equals $g_f$, the denominator is zero, and no fully invested portfolio is a multiple of $\mathbf{\Sigma}_g^{-1}\mathbf{e}$. If GMV growth is below $g_f$, the denominator is negative, so normalizing flips the direction. The formula then gives the fully invested portfolio with the smallest Sharpe ratio, on the lower branch. In both cases, no fully invested portfolio has the largest Sharpe ratio, because a maximizer would have to be a stationary point.

___

## The Largest Sharpe Ratio
Is the stationary point a maximum? Let $\mathbf{\Sigma}_g^{1/2}$ be the symmetric, positive-definite square root of $\mathbf{\Sigma}_g$. For any $\mathbf{w}\neq\mathbf{0}$, the Cauchy–Schwarz inequality bounds the expected excess growth:

$$
\begin{aligned}
\mathbf{e}^{\top}\mathbf{w}
&=\left(\mathbf{\Sigma}_g^{-1/2}\mathbf{e}\right)^{\top}\left(\mathbf{\Sigma}_g^{1/2}\mathbf{w}\right) &&\quad\text{insert the identity }\mathbf{I}=\mathbf{\Sigma}_g^{-1/2}\mathbf{\Sigma}_g^{1/2}\\
&\leq\left\lVert\mathbf{\Sigma}_g^{-1/2}\mathbf{e}\right\rVert_2\left\lVert\mathbf{\Sigma}_g^{1/2}\mathbf{w}\right\rVert_2 &&\quad\text{Cauchy–Schwarz}\\
&=\sqrt{\mathbf{e}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}}\,\sqrt{\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}}. &&\quad\text{expand both norms}
\end{aligned}
$$

Dividing by the standard deviation gives $\mathrm{SR}(\mathbf{w})\leq\sqrt{\mathbf{e}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}}$ for every portfolio. Equality holds exactly when $\mathbf{w}=c\,\mathbf{\Sigma}_g^{-1}\mathbf{e}$ for some $c>0$. Under the GMV condition, the tangent weights are such a multiple, so they attain the bound:

$$
\boxed{\mathrm{SR}_{\mathcal{T}}=\max_{\mathbf{w}\neq\mathbf{0}}\,\mathrm{SR}(\mathbf{w})=\sqrt{\mathbf{e}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}}.}
$$

The maximum is positive because $\mathbf{e}\neq\mathbf{0}$. It is the slope of the steepest capital allocation line.

## The Same Risky Mix with a Risk-Free Asset
Now we add the risk-free asset, as in the lecture's risky and risk-free problem. The risky weights no longer need to sum to one, and the rest of our wealth, $w_f=1-\mathbf{1}^{\top}\mathbf{w}$, is lent or borrowed at $g_f$. The complete portfolio has expected growth $\mathbb{E}[g_c]=g_f+\mathbf{e}^{\top}\mathbf{w}$ and variance $\sigma^{2}_{g,c}=\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}$. At the minimum, the lecture's growth floor binds, since scaling the weights down lowers the variance. So for a target $g_{\star}>g_f$, in inverse years, we solve:

$$
\begin{aligned}
\underset{\mathbf{w}\in\mathbb{R}^M}{\operatorname{minimize}}\quad
&\frac{1}{2}\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}\\
\text{subject to}\quad &\mathbf{e}^{\top}\mathbf{w}=g_{\star}-g_f.
\end{aligned}
$$

> __Derivation:__
>
> With a multiplier $\gamma$ for the growth constraint, the Lagrangian is $\mathcal{L}(\mathbf{w},\gamma)=\frac{1}{2}\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}-\gamma\left(\mathbf{e}^{\top}\mathbf{w}-g_{\star}+g_f\right)$. Setting its gradient to zero gives $\mathbf{\Sigma}_g\mathbf{w}=\gamma\,\mathbf{e}$, so $\mathbf{w}=\gamma\,\mathbf{\Sigma}_g^{-1}\mathbf{e}$. Substituting into the growth constraint gives $\gamma$ and the weights:
>
> $$\gamma=\frac{g_{\star}-g_f}{\mathbf{e}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}}>0,\qquad\boxed{\mathbf{w}(g_{\star})=\gamma\,\mathbf{\Sigma}_g^{-1}\mathbf{e}.}$$
>
> The variance is strictly convex, so these weights give the unique minimum.

Every target gives a positive multiple of $\mathbf{\Sigma}_g^{-1}\mathbf{e}$, the tangent direction. Under the GMV condition, the risky weights are $(1-w_f)\,\mathbf{w}_{\mathcal{T}}$, and the risky fraction is given by:

$$
\begin{aligned}
1-w_f=\mathbf{1}^{\top}\mathbf{w}(g_{\star})
&=\frac{(g_{\star}-g_f)\,\mathbf{1}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}}{\mathbf{e}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}} &&\quad\text{sum the weights}\\
&=\frac{g_{\star}-g_f}{\mathbb{E}[g_{\mathcal{T}}]-g_f}. &&\quad\text{use }\mathbb{E}[g_{\mathcal{T}}]-g_f\text{ from above}
\end{aligned}
$$

The target changes only the split between the risk-free asset and the tangent portfolio, as two-fund separation says. Using stationarity and the growth constraint, the smallest variance at each target is given by:

$$
\begin{aligned}
\sigma^{2}_{g,c}=\mathbf{w}^{\top}\mathbf{\Sigma}_g\mathbf{w}
&=\gamma\,\mathbf{w}^{\top}\mathbf{e} &&\quad\text{use }\mathbf{\Sigma}_g\mathbf{w}=\gamma\,\mathbf{e}\\
&=\gamma\,(g_{\star}-g_f) &&\quad\text{growth constraint}\\
&=\frac{(g_{\star}-g_f)^{2}}{\mathbf{e}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}}. &&\quad\text{substitute }\gamma
\end{aligned}
$$

Taking the square root and solving for the target gives the capital allocation line with the largest Sharpe ratio:

$$
\boxed{g_{\star}=g_f+\underbrace{\sqrt{\mathbf{e}^{\top}\mathbf{\Sigma}_g^{-1}\mathbf{e}}}_{\mathrm{SR}_{\mathcal{T}}}\,\sigma_{g,c}.}
$$

Long-only bounds can change the risky direction. The lecture's scaling argument still applies: every feasible target above $g_f$ uses the same long-only direction until an upper bound binds.

___

## Summary
We derived the tangent weights, showed that they give the largest Sharpe ratio, and connected them to the risky and risk-free problem.

> __Key Takeaways:__
>
> * **One direction for the risky weights:** We showed that multiplying the weights by a positive number leaves the Sharpe ratio unchanged, so the budget constraint only sets their scale. The gradient then gave a single direction, which we scaled to sum to one.
>
> * **The GMV condition:** We factored the normalizing denominator into a positive term and the GMV portfolio's expected excess growth. When that excess growth is positive, the tangent portfolio's Sharpe ratio equals the Cauchy–Schwarz bound, the largest possible value.
>
> * **The same risky fund with a risk-free asset:** With lending and borrowing allowed, the minimum-variance risky weights at every target above the risk-free rate were a multiple of the tangent weights. The risky mix stayed fixed, and the minimum-risk portfolios traced the capital allocation line.
