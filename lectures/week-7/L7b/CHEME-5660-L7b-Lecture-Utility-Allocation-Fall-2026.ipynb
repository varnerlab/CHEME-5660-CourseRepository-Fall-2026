## Concept Review: The Capital Allocation Line
Recall from L7a that a __complete portfolio__ $c$ puts a fraction $w_{f}$ of wealth in the risk-free asset and $1-w_{f}$ in the tangent portfolio $\mathcal{T}$. Its expected growth rate and growth-rate variance are given by:

$$
\begin{align*}
\mathbb{E}[g_{c}] &= g_{f}+(1-w_{f})\left(\mathbb{E}[g_{\mathcal{T}}]-g_{f}\right) &&\quad\text{expectation is linear}\\
\sigma^{2}_{g,c} &= (1-w_{f})^{2}\,\sigma^{2}_{g,\mathcal{T}} &&\quad\text{the risk-free asset adds no variance}
\end{align*}
$$

where $g_{f}$ is the risk-free growth rate. The tangent portfolio has expected growth rate $\mathbb{E}[g_{\mathcal{T}}]$ and growth-rate standard deviation $\sigma_{g,\mathcal{T}}$, all in inverse years. Lending ($0<w_{f}<1$) and borrowing ($w_{f}<0$) move the complete portfolio along the capital allocation line, whose slope is the Sharpe ratio of $\mathcal{T}$.

By two-fund separation, investors with the same inputs who can lend and borrow freely at $g_{f}$ hold the same risky fund $\mathcal{T}$. Their risk preferences set only $w_{f}$, and the L6b client interview asked for $w_{f}$ directly. Which $w_{f}$ should a given investor choose?

## Utility Functions and Rational Choice
Utility functions are mathematical expressions that describe a decision-maker's preferences for different choices, outcomes, or states of the world. A utility function gives a number, the __utility__, for each option. Utility is a proxy for satisfaction, so a larger utility is always better than a smaller utility:

> __Utility function:__ A decision-maker (agent) chooses from a set of alternatives $\mathcal{X}=\{x_{1},\dots,x_{n}\}$, e.g., bundles of goods, portfolios, or states of the world. A utility function $U:\mathcal{X}\rightarrow\mathbb{R}$ assigns each alternative a real number, in units of _utils_, that ranks the agent's preferences:
>
> $$
> U(x_{i})>U(x_{j})\;\Longleftrightarrow\;x_{i}\succ x_{j}\qquad\text{and}\qquad U(x_{i})=U(x_{j})\;\Longleftrightarrow\;x_{i}\sim x_{j}
> $$
>
> where $x_{i}\succ x_{j}$ means the agent strictly prefers $x_{i}$ to $x_{j}$, and $x_{i}\sim x_{j}$ means the agent is indifferent between them. Utility functions are __ordinal__: they rank alternatives but do not measure how much better one alternative is than another. Thus, any order-preserving transformation of $U$, e.g., $\ln U$ for positive $U$, describes the same preferences.

Rational choice theory is built on one idea:

> __Key idea:__ When faced with competing alternatives, a rational decision-maker always chooses the alternative with the largest utility.

Utility functions, and their parameters, are subjective and differ from one decision-maker to another. Five common forms for a bundle of goods $\mathbf{x}=(x_{1},\dots,x_{n})$ with $x_{i}\geq 0$ and parameters $\alpha_{i}>0$ are:

| Utility function | Form | Models |
|:--|:--|:--|
| Linear | $U(\mathbf{x})=\sum_{i}\alpha_{i}x_{i}$ | goods that are perfect substitutes |
| Logarithmic | $U(\mathbf{x})=\ln\left(\sum_{i}\alpha_{i}x_{i}+\beta\right)$, with $\beta>0$ | diminishing satisfaction from more goods |
| Cobb–Douglas | $U(\mathbf{x})=\prod_{i}x_{i}^{\alpha_{i}}$ | goods held together |
| Leontief | $U(\mathbf{x})=\min\left(x_{1}/\alpha_{1},\dots,x_{n}/\alpha_{n}\right)$ | perfect complements, used in fixed proportions |
| CES | $U(\mathbf{x})=\left(\sum_{i}\alpha_{i}x_{i}^{(\eta-1)/\eta}\right)^{\eta/(\eta-1)}$, with $\eta>0$, $\eta\neq 1$ | substitution set by the elasticity $\eta$ |

We use the Cobb–Douglas and CES utilities later in this lecture. The constant elasticity of substitution (CES) utility moves between three of the other forms as its elasticity $\eta$ changes: linear as $\eta\to\infty$, Cobb–Douglas as $\eta\to 1$ (with $\sum_{i}\alpha_{i}=1$), and Leontief as $\eta\to 0$.

### Risk and Risk Aversion
So far, every choice had a sure outcome. Investing does not. When outcomes are uncertain, a rational decision-maker ranks choices by the __expected utility__ $\mathbb{E}[U(W)]$, the probability-weighted average of the utility of each wealth outcome $W$ ([the von Neumann–Morgenstern theorem](https://en.wikipedia.org/wiki/Von_Neumann%E2%80%93Morgenstern_utility_theorem)). The shape of the utility of wealth $U(W)$ encodes the decision-maker's attitude toward risk. Compare a risky outcome $W$ with its expected value $\mathbb{E}(W)$ received for certain:

| Curvature of $U$ | Risky versus certain | Risk attitude |
|:--|:--|:--|
| Concave ($U''<0$) | $\mathbb{E}[U(W)]\leq U(\mathbb{E}(W))$ | risk averse |
| Linear ($U''=0$) | $\mathbb{E}[U(W)]=U(\mathbb{E}(W))$ | risk neutral |
| Convex ($U''>0$) | $\mathbb{E}[U(W)]\geq U(\mathbb{E}(W))$ | risk seeking |

The __certainty equivalent__ (CE) is the guaranteed wealth with the same utility as the risky outcome, $U(\text{CE})=\mathbb{E}[U(W)]$. The __risk premium__ is the expected wealth a decision-maker gives up to remove the risk:

$$
\text{RP} = \mathbb{E}(W)-\text{CE}
$$

For a risk-averse decision-maker, $\text{CE}\leq\mathbb{E}(W)$ and $\text{RP}\geq 0$.

__How risk averse is a decision-maker?__ Arrow and [Pratt (1964)](https://doi.org/10.2307/1913738) turned the curvature of $U$ into a number:

> __Arrow–Pratt measures:__ Let $U(w)$ be a twice-differentiable utility of wealth $w$ with $U'(w)>0$. The coefficients of absolute and relative risk aversion are given by:
>
> $$
> r(w) = -\frac{U''(w)}{U'(w)}\qquad\text{and}\qquad\bar{r}(w) = w\,r(w)
> $$
>
> A risk-averse decision-maker has $r(w)>0$, a risk-neutral decision-maker has $r(w)=0$, and a risk-seeking decision-maker has $r(w)<0$.

For three common utilities of wealth, the measures are:

| Utility | $U(w)$ | $r(w)$ | $\bar{r}(w)$ |
|:--|:--|:--|:--|
| Logarithmic | $\ln w$ | $1/w$ | $1$ |
| Linear | $a\,w+b$, with $a>0$ | $0$ | $0$ |
| Power | $w^{\tau}$, with $\tau>0$ | $(1-\tau)/w$ | $1-\tau$ |

The logarithmic and power utilities have constant relative risk aversion. Their attitude toward a risk sized as a fraction of wealth does not depend on how wealthy the decision-maker is.

Why does this matter for a portfolio? For a small risk, Pratt showed that the risk premium is approximately $\tfrac{1}{2}\,r(\mathbb{E}(W))\,\text{Var}(W)$. Measured as a fraction of wealth, and for small returns, the certainty-equivalent return is then approximately the expected return minus $\bar{r}/2$ times the variance of the return. That is the form of the mean-variance utility in the next section, where the coefficient $A$ plays the role of the relative risk aversion $\bar{r}$.

## Mean-Variance Utility on the Capital Allocation Line
For a complete portfolio, the classic mean-variance utility rewards expected growth and charges for risk in proportion to the growth-rate variance:

> __Mean-variance utility:__ For a risk-aversion coefficient $A>0$ (dimensionless), the utility of the complete portfolio $c$ is given by:
>
> $$
> U(w_{f}) = \underbrace{\mathbb{E}[g_{c}]}_{\text{reward}} - \underbrace{\frac{A}{2}\,\Delta t\,\sigma^{2}_{g,c}}_{\text{risk penalty}}
> $$
>
> where $\Delta t$ is the time step (one trading day, in years), so $\Delta t\,\sigma^{2}_{g,c}$ is the squared volatility of the complete portfolio. As in L5a, this converts the growth-rate variance to the scale of the GBM volatility, so both terms have units of inverse years. A larger $A$ charges more for the same risk.

Along the capital allocation line, both terms depend only on $w_{f}$. Substituting the Concept Review expressions and differentiating gives:

$$
\begin{align*}
U(w_{f}) &= g_{f}+(1-w_{f})\left(\mathbb{E}[g_{\mathcal{T}}]-g_{f}\right)-\frac{A}{2}\,(1-w_{f})^{2}\,\Delta t\,\sigma^{2}_{g,\mathcal{T}} &&\quad\text{substitute the line}\\
\frac{dU}{dw_{f}} &= -\left(\mathbb{E}[g_{\mathcal{T}}]-g_{f}\right)+A\,(1-w_{f})\,\Delta t\,\sigma^{2}_{g,\mathcal{T}} &&\quad\text{differentiate}\\
0 &= -\left(\mathbb{E}[g_{\mathcal{T}}]-g_{f}\right)+A\,(1-w_{f}^{\star})\,\Delta t\,\sigma^{2}_{g,\mathcal{T}} &&\quad\text{first-order condition}
\end{align*}
$$

Solving for the risky fraction gives the __optimal complete portfolio__:

$$
\boxed{1-w_{f}^{\star} = \frac{\mathbb{E}[g_{\mathcal{T}}]-g_{f}}{A\,\Delta t\,\sigma^{2}_{g,\mathcal{T}}}}
$$

The second derivative, $-A\,\Delta t\,\sigma^{2}_{g,\mathcal{T}}$, is negative, so this is the maximum. The risky fraction is the excess growth of $\mathcal{T}$ divided by its risk penalty, so doubling $A$ halves it. The investor with $A_{\mathcal{T}}=(\mathbb{E}[g_{\mathcal{T}}]-g_{f})/(\Delta t\,\sigma^{2}_{g,\mathcal{T}})$ holds $\mathcal{T}$ alone. Investors with $A>A_{\mathcal{T}}$ lend, and investors with $A<A_{\mathcal{T}}$ borrow.

__Where does the optimum sit on the line?__ An __indifference curve__ joins the portfolios with the same utility. Holding $U$ fixed gives $\mathbb{E}[g_{c}]=U+\frac{A}{2}\,\Delta t\,\sigma^{2}_{g,c}$, a parabola that opens upward in the risk-return plane. Higher curves mean higher utility, and the optimal complete portfolio is where the highest curve the investor can reach touches the line.

The same result runs backward. A client who picks a fraction $w_{f}$ reveals the coefficient that makes it optimal:

$$
A = \frac{\mathbb{E}[g_{\mathcal{T}}]-g_{f}}{(1-w_{f})\,\Delta t\,\sigma^{2}_{g,\mathcal{T}}}
$$

## Utility over Holdings
Instead of choosing one risky fund and a fraction, we can ask a question from consumer choice theory. Given a budget and today's prices, which holdings does the investor prefer most? The investor chooses share counts that maximize a utility function subject to the budget:

> __Maximum-utility problem:__ At time $t$, given $|\mathcal{P}|$ assets with prices $S_{i}(t)$ (USD per share) and wealth $W_{\mathcal{P}}(t)$ (USD), choose the share counts $n_{i}\geq 0$ that solve:
>
> $$
> \operatorname*{maximize}_{n_{1},\ldots,n_{|\mathcal{P}|}}\quad U(n_{1},\ldots,n_{|\mathcal{P}|})\qquad\text{subject to}\quad\sum_{i\in\mathcal{P}}n_{i}\,S_{i}(t) = W_{\mathcal{P}}(t)
> $$
>
> where $U$ encodes the investor's preferences over the shares held.

This is a __holding utility__, a preference over the shares held today at today's prices. Unlike the mean-variance utility, it has no variance term. Risk enters through the preference weights of the next section, which depend on the SIM betas and the state of the market.

### Cobb–Douglas utility
The workhorse is the Cobb–Douglas utility, whose exponents are the __preference weights__ $\gamma_{i}\in(-1,1)$. The investor maximizes the product of powers over every asset, subject to the budget and a small floor of $n_{\min}>0$ shares on each holding:

$$
\boxed{\begin{aligned}
\operatorname*{maximize}_{n_{1},\ldots,n_{|\mathcal{P}|}}\quad & U(n_{1},\ldots,n_{|\mathcal{P}|}) = \prod_{i\in\mathcal{P}}n_{i}^{\gamma_{i}}\\
\text{subject to}\quad & \sum_{i\in\mathcal{P}}n_{i}\,S_{i}(t) = W_{\mathcal{P}}(t) &&\quad\text{(budget constraint)}\\
& n_{i}\geq n_{\min}\quad\forall\,i\in\mathcal{P} &&\quad\text{(share floor)}
\end{aligned}}
$$

The floor keeps the utility finite, because for $\gamma_{i}<0$ the term $n_{i}^{\gamma_{i}}$ grows without bound as $n_{i}\to 0$. The sign of $\gamma_{i}$ splits the assets in two. The __preferred set__ $\mathcal{A}^{+}=\{i:\gamma_{i}>0\}$ is the basket, and the rest form the __non-preferred set__ $\mathcal{A}^{-}=\{i:\gamma_{i}\leq 0\}$.

Only the ranking matters, so $\ln U$ has the same maximizer as $U$. We set the floors first, then split the remaining budget with a Lagrange multiplier $\lambda$, assuming each preferred share count stays above its floor:

$$
\begin{align*}
\frac{\partial\ln U}{\partial n_{k}} &= \frac{\gamma_{k}}{n_{k}}\leq 0 &&\quad k\in\mathcal{A}^{-}\text{, so more shares never raise }\ln U\\
n_{k}^{\star} &= n_{\min} &&\quad\text{spend every dollar above the floor on }\mathcal{A}^{+}\\
\sum_{i\in\mathcal{A}^{+}}n_{i}S_{i}(t)+n_{\min}\sum_{k\in\mathcal{A}^{-}}S_{k}(t) &= W_{\mathcal{P}}(t) &&\quad\text{budget with the floors substituted}\\
\sum_{i\in\mathcal{A}^{+}}n_{i}S_{i}(t) &= W_{\mathcal{P}}(t)-n_{\min}\sum_{k\in\mathcal{A}^{-}}S_{k}(t)\equiv W_{\text{adj}} &&\quad\text{net budget for }\mathcal{A}^{+}\text{, assumed positive}\\
\mathcal{L} &= \sum_{i\in\mathcal{A}^{+}}\gamma_{i}\ln n_{i}-\lambda\Bigl(\sum_{i\in\mathcal{A}^{+}}n_{i}S_{i}(t)-W_{\text{adj}}\Bigr) &&\quad\text{drop constant floor terms, add budget}\\
0 &= \frac{\gamma_{i}}{n_{i}}-\lambda\,S_{i}(t) &&\quad\text{differentiate with respect to }n_{i}\\
n_{i}S_{i}(t) &= \gamma_{i}/\lambda &&\quad\text{dollars in proportion to }\gamma_{i}\\
W_{\text{adj}} &= \sum_{j\in\mathcal{A}^{+}}\gamma_{j}/\lambda &&\quad\text{sum over the preferred set}\\
1/\lambda &= W_{\text{adj}}\Big/\sum_{j\in\mathcal{A}^{+}}\gamma_{j} &&\quad\text{solve for the multiplier}
\end{align*}
$$

Substituting $1/\lambda$ into $n_{i}S_{i}(t)=\gamma_{i}/\lambda$ gives the __Cobb–Douglas allocation__:

$$
\boxed{n_{i}^{\star} = \underbrace{\frac{\gamma_{i}}{\sum_{j\in\mathcal{A}^{+}}\gamma_{j}}}_{\text{preference share}}\cdot\underbrace{\frac{W_{\text{adj}}}{S_{i}(t)}}_{\text{budget in shares}}\qquad i\in\mathcal{A}^{+},\qquad n_{i}^{\star}=n_{\min}\quad i\in\mathcal{A}^{-}}
$$

If the formula gives a preferred asset fewer than $n_{\min}$ shares, we pin that asset at its floor, subtract its cost from $W_{\text{adj}}$, and solve again with the rest.

With the floors fixed, every remaining $\gamma_{i}$ is positive, so $\ln U$ is strictly concave and the maximum is unique. The dollar weight of a preferred asset within $W_{\text{adj}}$ is its normalized preference weight $\gamma_{i}/\sum_{j}\gamma_{j}$, independent of prices. If the preferred set is empty, the allocation is the floors plus cash.

### CES utility
The Cobb–Douglas allocator always spends on every preferred asset in proportion to $\gamma_{i}$. The constant elasticity of substitution (CES) utility adds one parameter, the elasticity $\eta>0$. It controls how willingly the allocator trades one preferred asset for another. For $\eta\neq 1$, the CES utility is given by:

$$
U_{\text{CES}}(n_{1},\ldots,n_{|\mathcal{P}|}) = \left(\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{(\eta-1)/\eta}\right)^{\eta/(\eta-1)}
$$

Its budget-constrained maximizer is also closed form, with the same floor check. On the preferred set, the share counts are proportional to $(\gamma_{i}/S_{i}(t))^{\eta}$, where $\gamma_{i}/S_{i}(t)$ is the _bang for the buck_ of consumer theory:

$$
\boxed{n_{i}^{\star} = W_{\text{adj}}\;\frac{\left(\gamma_{i}/S_{i}(t)\right)^{\eta}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)\left(\gamma_{j}/S_{j}(t)\right)^{\eta}}\qquad i\in\mathcal{A}^{+}}
$$

The elasticity is a __concentration dial__ with three limits:

* __Unit elasticity ($\eta\to 1$):__ The allocation tends to the Cobb–Douglas allocation.
* __Large elasticity ($\eta\to\infty$):__ Every other asset falls to its floor, and the rest of the budget goes to the asset with the largest $\gamma_{i}/S_{i}(t)$, when that asset is unique.
* __Small elasticity ($\eta\to 0^{+}$):__ The allocator buys equal share counts of every preferred asset, whatever its $\gamma_{i}$ or price.

It also shows that for $\eta\neq 1$ the CES dollar weights depend on the price level of each share, while the Cobb–Douglas weights do not. Where do the preference weights come from?

## Preference Weights from the Single Index Model
Each trading day, we compute the preference weights $\gamma_{i}(t)$ from the L6b SIM parameters $(\alpha_{i},\beta_{i})$ and two market inputs. The __recent market growth rate__ $\tilde{g}_{M,t}$ plays the role of the SIM's expected market growth. The __market-state signal__ $\xi_{t}$ says whether the market has recently been rising or falling.

> __Preference model:__ For each asset $i$ on day $t$, the preference weight is given by:
>
> $$
> \boxed{\gamma_{i}(t) = \tanh\left(\frac{\alpha_{i}}{\beta_{i}^{\xi_{t}}} + \beta_{i}^{1-\xi_{t}}\,\tilde{g}_{M,t}\right)}
> $$
>
> The hyperbolic tangent keeps the weight in $(-1,1)$. Its argument equals $(\alpha_{i}+\beta_{i}\tilde{g}_{M,t})/\beta_{i}^{\xi_{t}}$, the SIM's expected growth rate of asset $i$ in today's market divided by $\beta_{i}^{\xi_{t}}$. We read this growth rate, in inverse years, as a pure number at a one-year horizon. The model assumes $\beta_{i}>0$ for every asset, because $\beta_{i}$ is raised to a real power.

The model answers the two questions of the allocation separately:

* __Which assets belong in the basket?__ Since $\beta_{i}^{\xi_{t}}>0$ and $\tanh$ keeps the sign, $\gamma_{i}(t)>0$ exactly when $\alpha_{i}+\beta_{i}\tilde{g}_{M,t}>0$, that is, when $\tilde{g}_{M,t}>-\alpha_{i}/\beta_{i}$. A strong market pulls in assets with a small negative $\alpha_{i}$ and a large $\beta_{i}$, and a weak market keeps only the assets with the lowest thresholds. When every $\gamma_{i}(t)\leq 0$, the allocator holds the floors plus cash.
* __How much of each?__ The magnitude of a positive $\gamma_{i}(t)$ sets its share of the budget, and the signal tilts the magnitudes. A bearish signal ($\xi_{t}>0$) divides by $\beta_{i}^{\xi_{t}}$, which is above one for high-beta assets and below one for low-beta assets. That tilts the budget toward low-beta names. A bullish signal ($\xi_{t}<0$) tilts it toward high-beta names.

Both market inputs come from the SPY price through exponential moving averages (EMAs). For a price series $S_{t}$ and a window of $L$ trading days, the EMA of L5a is $\bar{S}_{t}=\omega\,S_{t}+(1-\omega)\,\bar{S}_{t-1}$ with $\omega=2/(L+1)$:

> __Market inputs:__ With a short window $L_{\text{short}}=21$ days, a long window $L_{\text{long}}=63$ days, and a dimensionless gain $G>0$, the market-state signal is the scaled crossover of the two price EMAs:
>
> $$
> \xi_{t} = -G\left(\frac{\bar{S}^{\text{short}}_{t}}{\bar{S}^{\text{long}}_{t}}-1\right)
> $$
>
> A falling market (short EMA below the long one) gives $\xi_{t}>0$, bearish, and a rising market gives $\xi_{t}<0$, bullish. We choose $G$ on the training data so that $|\xi_{t}|\leq 1$ on 99% of training days. The recent market growth rate $\tilde{g}_{M,t}$ is the EMA, with a window of $L_{\text{growth}}=10$ days, of the daily market growth rates $g_{M,t}=\ln(S_{t}/S_{t-1})/\Delta t$ (inverse years).

Both inputs on day $t$ use prices through day $t-1$, so a decision never uses a price it could not have known. The recent market growth swings by about one per year, several times larger than the intercepts. These swings move assets across their thresholds, so the basket responds to the market.

In this lecture, we used utility functions to turn an investor's preferences into a portfolio. We first chose a point on the capital allocation line with mean-variance utility. We then wrote preferences for each asset and solved the Cobb–Douglas and CES allocation problems, with preference weights from the single index model.
