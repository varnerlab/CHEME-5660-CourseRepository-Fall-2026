## The CES Problem
Recall the setting of the lecture. On trading day $t$, asset $i$ trades at $S_{i}(t)$ (USD per share), and the investor holds $n_{i}$ shares. The preferred set $\mathcal{A}^{+}=\{i:\gamma_{i}>0\}$ holds the assets with preference weights $\gamma_{i}\in(0,1)$. The non-preferred assets sit at the floor $n_{\min}$, which leaves the net budget $W_{\text{adj}}=W_{\mathcal{P}}(t)-n_{\min}\sum_{k\in\mathcal{A}^{-}}S_{k}(t)$ (USD) for the preferred set.

To keep the algebra light, we write the CES exponent as $\rho\equiv(\eta-1)/\eta$. This $\rho$ is a local shorthand, not the correlation of L5b. The CES problem over the vector $\mathbf{n}$ of preferred share counts is given by:

$$
\boxed{\begin{aligned}
\operatorname*{maximize}_{\mathbf{n}}\quad & U_{\text{CES}}(\mathbf{n}) = \Bigl(\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{\rho}\Bigr)^{1/\rho}\\
\text{subject to}\quad & \sum_{i\in\mathcal{A}^{+}}n_{i}\,S_{i}(t) = W_{\text{adj}} &&\quad\text{(net budget)}\\
& n_{i}\geq n_{\min}\quad\forall\,i\in\mathcal{A}^{+} &&\quad\text{(share floor)}
\end{aligned}}
$$

where $\eta>0$, $\eta\neq 1$, is the elasticity of substitution. As in the lecture, we assume no preferred floor binds, so the floor drops out of the Lagrangian.
___

## The Closed-Form Optimum
Let $\Psi=\sum_{j\in\mathcal{A}^{+}}\gamma_{j}\,n_{j}^{\rho}$, so that $U_{\text{CES}}=\Psi^{1/\rho}$. With a multiplier $\lambda$ on the net budget, as in the lecture's Cobb–Douglas derivation, the first-order conditions give:

$$
\begin{align*}
\mathcal{L} &= \Psi^{1/\rho}-\lambda\Bigl(\sum_{i\in\mathcal{A}^{+}}n_{i}S_{i}(t)-W_{\text{adj}}\Bigr) &&\quad\text{add the budget}\\
0 &= \Psi^{1/\rho-1}\,\gamma_{i}\,n_{i}^{\rho-1}-\lambda\,S_{i}(t) &&\quad\text{differentiate with respect to }n_{i}\text{ (chain rule)}\\
\frac{\gamma_{i}\,n_{i}^{\rho-1}}{\gamma_{j}\,n_{j}^{\rho-1}} &= \frac{S_{i}(t)}{S_{j}(t)} &&\quad\text{divide the conditions for }i\text{ and }j\\
\Bigl(\frac{n_{i}}{n_{j}}\Bigr)^{-1/\eta} &= \frac{S_{i}(t)/\gamma_{i}}{S_{j}(t)/\gamma_{j}} &&\quad\text{use }\rho-1=-1/\eta\\
n_{i} &= c\,\Bigl(\frac{\gamma_{i}}{S_{i}(t)}\Bigr)^{\eta} &&\quad\text{raise to the power }{-\eta}\text{, with }c>0\text{ common to all }i\\
c\sum_{j\in\mathcal{A}^{+}}S_{j}(t)\Bigl(\frac{\gamma_{j}}{S_{j}(t)}\Bigr)^{\eta} &= W_{\text{adj}} &&\quad\text{substitute into the budget}
\end{align*}
$$

Since $\rho<1$ for every finite $\eta>0$, the CES utility is concave for positive share counts. The budget is linear, so this stationary point is the maximum. Solving for $c$ gives the lecture's CES allocation:

$$
\boxed{n_{i}^{\star} = W_{\text{adj}}\;\frac{\left(\gamma_{i}/S_{i}(t)\right)^{\eta}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)\left(\gamma_{j}/S_{j}(t)\right)^{\eta}}\qquad i\in\mathcal{A}^{+}\quad\blacksquare}
$$

The elasticity is the exponent on each asset's bang for the buck $\gamma_{i}/S_{i}(t)$, so a larger $\eta$ magnifies the differences between assets.

The dollar weights $w_{i}^{\star}=S_{i}(t)\,n_{i}^{\star}/W_{\text{adj}}$, the fractions of the net budget, are given by:

$$
\boxed{w_{i}^{\star} = \frac{S_{i}(t)^{1-\eta}\,\gamma_{i}^{\eta}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)^{1-\eta}\,\gamma_{j}^{\eta}}}
$$

When no floor binds, [the `allocate_ces(...)` function](https://varnerlab.org/CHEME-5660-CourseRepository-Fall-2026/dev/adaptive/#VLQuantitativeFinancePackage.allocate_ces) in the course package returns this allocation.

## The Three Limits
The lecture calls the elasticity a __concentration dial__. Each of its three limits sends the exponent $\rho=(\eta-1)/\eta$ to an extreme: $\rho\to 0$ as $\eta\to 1$, $\rho\to 1$ as $\eta\to\infty$, and $\rho\to-\infty$ as $\eta\to 0^{+}$.

### Unit elasticity recovers Cobb–Douglas ($\eta\to 1$)
For $\eta\to 1$ we have $\rho\to 0$. Scaling every $\gamma_{i}$ by the same $c>0$ multiplies $U_{\text{CES}}$ by $c^{1/\rho}$, which leaves the optimizer unchanged. So we normalize the weights to $\sum_{i\in\mathcal{A}^{+}}\gamma_{i}=1$, as the lecture does for this limit.

Without the normalization, the sum inside the parentheses tends to $\sum_{i}\gamma_{i}\neq 1$, and the two one-sided limits are $0$ and $\infty$. With it, the limit has the form $1^{\infty}$, so we take the logarithm and apply L'Hôpital's rule in $\rho$:

$$
\begin{align*}
\ln U_{\text{CES}} &= \frac{\ln\bigl(\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{\rho}\bigr)}{\rho} &&\quad\text{numerator and denominator vanish as }\rho\to 0\\
\lim_{\rho\to 0}\ln U_{\text{CES}} &= \left.\frac{\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{\rho}\ln n_{i}}{\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{\rho}}\right|_{\rho=0} &&\quad\text{differentiate both in }\rho\text{ (L'Hôpital)}\\
&= \sum_{i\in\mathcal{A}^{+}}\gamma_{i}\ln n_{i} &&\quad\text{use }n_{i}^{0}=1\text{ and }\textstyle\sum_{i}\gamma_{i}=1
\end{align*}
$$

Exponentiating recovers the Cobb–Douglas utility on the preferred set:

$$
\boxed{\lim_{\eta\to 1}U_{\text{CES}}(\mathbf{n}) = \prod_{i\in\mathcal{A}^{+}}n_{i}^{\gamma_{i}}\quad\blacksquare}
$$

The optimum needs no normalization. Setting $\eta=1$ in the dollar weights gives:

$$
w_{i}^{\star} = \frac{S_{i}(t)^{0}\,\gamma_{i}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)^{0}\,\gamma_{j}} = \frac{\gamma_{i}}{\sum_{j\in\mathcal{A}^{+}}\gamma_{j}}
$$

These are the lecture's Cobb–Douglas weights, which do not depend on prices.

### Large elasticity concentrates the budget ($\eta\to\infty$)
For $\eta\to\infty$ we have $\rho\to 1$, and the CES utility becomes linear in the share counts:

$$
\boxed{\lim_{\eta\to\infty}U_{\text{CES}}(\mathbf{n}) = \sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}}
$$

To find the optimum, write $r_{i}=\gamma_{i}/S_{i}(t)$ for the bang for the buck of asset $i$. Assume a unique asset $i^{\star}$ attains the largest ratio $r^{\star}=\max_{i\in\mathcal{A}^{+}}r_{i}$. Dividing the numerator and denominator of the closed form by $(r^{\star})^{\eta}$ gives:

$$
n_{i}^{\star} = W_{\text{adj}}\;\frac{(r_{i}/r^{\star})^{\eta}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)\,(r_{j}/r^{\star})^{\eta}}
$$

For $i\neq i^{\star}$, the ratio $r_{i}/r^{\star}<1$, so $(r_{i}/r^{\star})^{\eta}\to 0$, while the $i^{\star}$ term stays at one. Without floors, the whole budget goes to the best asset:

$$
\boxed{\lim_{\eta\to\infty}n_{i}^{\star} = \begin{cases}W_{\text{adj}}/S_{i^{\star}}(t) & i = i^{\star}\\ 0 & \text{otherwise}\end{cases}\quad\blacksquare}
$$

This is the linear-programming corner solution. With the lecture's floors, the other preferred assets stop at $n_{\min}$, and $i^{\star}$ receives the rest of the net budget. If several assets tie for $r^{\star}$, the limit buys equal share counts of the tied assets.

### Small elasticity equalizes share counts ($\eta\to 0^{+}$)
For $\eta\to 0^{+}$ we have $\rho\to-\infty$. Factoring out the smallest share count $m=\min_{i\in\mathcal{A}^{+}}n_{i}$ gives:

$$
U_{\text{CES}}(\mathbf{n}) = m\,\Bigl(\sum_{j\in\mathcal{A}^{+}}\gamma_{j}\Bigl(\frac{n_{j}}{m}\Bigr)^{\rho}\Bigr)^{1/\rho}
$$

As $\rho\to-\infty$, each ratio with $n_{j}>m$ gives $(n_{j}/m)^{\rho}\to 0$, so the sum tends to the total weight of the assets that hold exactly $m$ shares. That total is positive, and its power $1/\rho$ tends to one:

$$
\boxed{\lim_{\eta\to 0^{+}}U_{\text{CES}}(\mathbf{n}) = \min_{i\in\mathcal{A}^{+}}n_{i}}
$$

This is the lecture's Leontief utility with every $\alpha_{i}=1$. The investor wants the preferred assets in equal share counts, and the preference weights drop out. The optimum follows from the closed form, because $(\gamma_{i}/S_{i}(t))^{0}=1$ for every preferred asset:

$$
\boxed{\lim_{\eta\to 0^{+}}n_{i}^{\star} = \frac{W_{\text{adj}}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)}\qquad i\in\mathcal{A}^{+}\quad\blacksquare}
$$

The dollar weights are then $w_{i}^{\star}=S_{i}(t)/\sum_{j}S_{j}(t)$, so the most expensive share gets the most dollars. The Dow Jones Industrial Average weights its stocks the same way.

## Log-Linear Utility Has the Cobb–Douglas Optimizer
The log-linear utility is the logarithm of the Cobb–Douglas utility $U_{\text{CD}}(\mathbf{n})=\prod_{i\in\mathcal{A}^{+}}n_{i}^{\gamma_{i}}$:

$$
U_{\log}(\mathbf{n}) = \ln U_{\text{CD}}(\mathbf{n}) = \sum_{i\in\mathcal{A}^{+}}\gamma_{i}\ln n_{i}
$$

The logarithm is strictly increasing on the positive reals, so $U_{\log}(\mathbf{n}_{1})>U_{\log}(\mathbf{n}_{2})$ exactly when $U_{\text{CD}}(\mathbf{n}_{1})>U_{\text{CD}}(\mathbf{n}_{2})$. The two utilities rank every pair of feasible portfolios the same way, so they have the same maximizer under the same budget, the Cobb–Douglas allocation. The lecture used this fact when it maximized $\ln U$ instead of $U$.

What changes is the __value__ of the utility, which becomes a sum with one term per asset. A sum is convenient when utility values are averaged across scenarios or accumulated as a reward by a learning algorithm.
___

## The CES Weights Depend on Share Prices
The Cobb–Douglas dollar weights $\gamma_{i}/\sum_{j}\gamma_{j}$ contain no prices, so they do not depend on how a company denominates its shares. The CES dollar weights $w_{i}^{\star}\propto S_{i}(t)^{1-\eta}\,\gamma_{i}^{\eta}$ do whenever $\eta\neq 1$. A two-for-one stock split halves $S_{i}(t)$ and multiplies the asset's term by $2^{\eta-1}$, although nothing economic has happened. For $\eta>1$, the allocator moves dollars into the asset after the split, and for $\eta<1$, it moves dollars out.

The reason is that the CES utility is written over share counts, which are an accounting unit. Written over dollar holdings $x_{i}=n_{i}S_{i}(t)$ instead, the same derivation with every price set to one gives $x_{i}^{\star}\propto\gamma_{i}^{\eta}$, which no split can change.
___

## Summary
In this notebook, we derived the closed-form CES optimum from the budget-constrained Lagrangian and took its three limits. We also showed that log-linear utility has the Cobb–Douglas optimizer and that the CES weights depend on the price of each share.

> __Key Takeaways:__
>
> * __One formula, three regimes:__ We derived a single closed-form CES optimum whose shape is set by the elasticity. Unit elasticity recovered Cobb–Douglas, a large elasticity concentrated the budget on the best bang for the buck, and a small one bought equal share counts.
> * __Log-linear utility is Cobb–Douglas in another form:__ We showed that the logarithm preserves the ranking of portfolios, so the two utilities share one optimizer. Only the utility values change, and they add across assets.
> * __CES weights depend on share prices:__ We found that for any elasticity other than one, a stock split changes the CES dollar weights, because the utility is written over share counts. Writing the utility over dollar holdings removes this dependence.
