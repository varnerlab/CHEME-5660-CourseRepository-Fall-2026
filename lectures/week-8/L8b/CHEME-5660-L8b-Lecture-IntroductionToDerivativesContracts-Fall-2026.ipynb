## Concept Review: From Assets to Contracts Written on Assets
Units 1 and 2 priced and allocated assets that we hold directly. We discounted Treasury cash flows (L2a, L2b), modeled equity growth rates as random variables (L3a, L4b), and built portfolios from those growth rates: minimum-variance and efficient-frontier allocations (L5b), single index model estimates of the firm-specific growth $\alpha_{i}$ and the relative risk $\beta_{i}$ (L6a, L6b), a utility-based rebalancing engine (L7a), and an online estimator that lets $\alpha_{i}$ and $\beta_{i}$ move as data arrive (L7b).

> __What carries forward:__ the price of the underlying asset and the models we built for it. Let $S_{t}\geq{0}$ denote the price of the underlying in USD/share at time $t$, so that $S_{0}$ is its price today and $S_{T}$ is its price at some future date $T$. Everything in Unit 3 is written on top of this price.

What changes is the object we are holding. A derivative is not an asset with a growth rate of its own; it is a __contract__ between two counterparties whose cash flows are a function of someone else's price. Two consequences shape the entire unit:

> * __The payoff is contractual, not statistical.__ Given the terminal price $S_{T}$, the amount the contract pays is a formula written into the contract. No estimation enters that step. Uncertainty enters only through $S_{T}$ itself.
> * __The two sides are not symmetric.__ One counterparty holds a _right_ and the other carries an _obligation_. The premium is the price of that asymmetry, and it is paid whether or not the right is ever used.

Today we read the contract and compute what it pays at expiration. Assigning a fair value to the contract _before_ expiration requires a pricing model, which is the subject of L9a.

## What Is a Derivative?
A __derivative__ is a financial contract between a buyer and a seller whose cash flows depend on an __underlying__ quantity, such as an equity price, an interest rate, a currency, a commodity, an index, or a credit event. The contract is an agreement about a future transaction; its value comes from the performance of the underlying, not from any productive activity of its own.

Let's watch a short video to understand the concept of derivatives better: [Derivatives Explained in 2 Minutes in Basic English | 2023](https://www.yout-ube.com/watch?v=B0jeGk5GJX0).

The common contract families differ in _who is obligated to do what_, and in _where they trade_:

| Contract | Obligation structure | Venue |
|---|---|---|
| __Futures__ | Both sides are obligated to transact the underlying at a set price on a future date | Exchange-traded, standardized, centrally cleared |
| __Forwards__ | Both sides are obligated, as with futures | Over-the-counter, terms negotiated bilaterally |
| __Swaps__ | Both sides are obligated to exchange two streams of cash flows | Over-the-counter |
| __Options__ | The buyer holds a _right_; only the seller carries an _obligation_ | Both: listed equity options are exchange-traded and centrally cleared, and an over-the-counter market also exists |

> __Why options are different:__ futures, forwards, and swaps obligate _both_ counterparties. An option obligates only one. That asymmetry is the whole subject of this lecture: it is what makes the payoff a kinked function rather than a straight line, and it is what the buyer pays a premium to acquire.

We focus on __listed__ options contracts for the rest of Unit 3, which is the market the course data comes from. The largest U.S. options exchange is the [Chicago Board Options Exchange (CBOE)](https://www.cboe.com).

## Reading an Option Contract
An option description is incomplete without its terms. Before any payoff can be computed, seven quantities must be pinned down:

| Symbol | Term | Units | Fixed by |
|---|---|---|---|
| $-$ | Underlying asset | $-$ | the contract |
| $K>0$ | __Strike price__, the price at which the transaction happens if the right is used | USD/share | the contract |
| $T>0$ | __Expiration__, the last date on which the right exists | years | the contract |
| $-$ | __Exercise style__, European or American | $-$ | the contract |
| $q\in\mathbb{N}$ | __Deliverable multiplier__, shares controlled per contract | shares/contract | the contract |
| $\theta\in\{-1,+1\}$ | __Position direction__, $+1$ long (bought) and $-1$ short (sold) | $-$ | your trade |
| $C_{0},P_{0}\geq{0}$ | __Premium__ of the call and of the put, paid when the position opens | USD/share | the market |

Of these, only the premium is a market quantity: it is quoted, it moves, and L9a is about where it comes from. The rest are written into the contract or chosen by you when you trade.

> __Two terms that are easy to skip:__
> * __Exercise style.__ A __European-style__ option may be exercised only at $T$. An __American-style__ option may be exercised at any time on or before $T$. Today we compute cash flows _at expiration_, where the two styles behave identically; the distinction affects the premium, and L9a and L9b take it up.
> * __The multiplier $q$.__ Standard U.S. equity options use $q = 100$ shares/contract, but adjusted and nonstandard contracts differ, and the contract specification always controls. Every per-share quantity below is multiplied by $q$ and by the number of contracts to reach position-level dollars. Getting $q$ wrong is a factor-of-100 error, not a rounding error.

The two contract types are the __call__ and the __put__. A call gives its long holder the right to _buy_ the underlying at $K$; a put gives its long holder the right to _sell_ at $K$. In both cases the short counterparty carries the matching obligation and collects the premium for doing so.

### What is the business case for each of the four positions?
Every option trade is one of four positions: long or short, call or put.

> __Long call.__ The buyer gains from upside in the underlying without buying the shares outright, and can acquire shares at $K$ instead of the market price. Risk is defined: the most that can be lost is the premium paid.
>
> __Short call.__ The seller collects the premium and takes on the obligation to deliver shares at $K$ if assigned. An _uncovered_ (naked) short call has no finite bound on its loss, because the underlying price has no finite upper bound. Sellers therefore usually already hold the required shares, which makes the position a __covered call__ (L11a).
>
> __Long put.__ The buyer gains from downside in the underlying without short selling, and can sell shares at $K$ instead of the market price. Risk is again defined at the premium paid.
>
> __Short put.__ The seller collects the premium and takes on the obligation to buy shares at $K$ if assigned. The loss is bounded, but the bound is large: the worst case is the underlying going to zero. Sellers typically hold enough cash to buy the shares, which makes the position a __cash-secured put__ (L11a).

## Expiration Payoff Is Contractual
Fix the notation for the positive part of a real number: for any $x\in\mathbb{R}$, write $x^{+}\equiv\max(x,0)$. Then the __payoff per share at expiration__ to the _long_ holder of a call and of a put is given by:
$$
\boxed{
\begin{align*}
h_{c}(S_{T};K) &= \left(S_{T}-K\right)^{+} \quad&\text{call payoff}\\
h_{p}(S_{T};K) &= \left(K-S_{T}\right)^{+} \quad&\text{put payoff}
\end{align*}}
$$
where $S_{T}\geq{0}$ is the price of the underlying at expiration in USD/share, and $K>0$ is the strike in USD/share. Both are measured in USD/share.

At expiration, these two formulas are the whole of the contract's promise, and each is a statement about a choice:
* The __call__ holder exercises only when buying at $K$ beats buying at $S_{T}$, that is when $S_{T}>K$; otherwise the right is abandoned and the payoff is $0$.
* The __put__ holder exercises only when selling at $K$ beats selling at $S_{T}$, that is when $S_{T}<K$; otherwise the right is abandoned and the payoff is $0$.

Because the holder may always decline to exercise, both payoffs are nonnegative. That is the mathematical signature of holding a right rather than an obligation.

> __Moneyness at expiration:__ a call is __in the money__ when $S_{T}>K$, __at the money__ when $S_{T}=K$, and __out of the money__ when $S_{T}<K$; a put reverses the inequalities. Moneyness describes exercise value relative to the strike. It does __not__ by itself say whether the position made money, because the premium was paid or received separately.

### The short side is the mirror image
Let $\theta\in\{-1,+1\}$ denote the position direction, $+1$ for long and $-1$ for short. Ignoring counterparty default, the writer receives exactly what the holder pays, so the signed payoff per share of either contract type is given by:
$$
\begin{align*}
V_{T}^{\,c,\theta}(S_{T}) &= \theta\,h_{c}(S_{T};K)\\
V_{T}^{\,p,\theta}(S_{T}) &= \theta\,h_{p}(S_{T};K)
\end{align*}
$$
where $V_{T}^{\,\cdot,\theta}$ denotes the signed payoff of the position at expiration in USD/share.

This symmetry in the _payoff_ does not carry over to _risk_, and the reason is the domain. The price satisfies $S_{T}\geq{0}$, so $h_{p}\leq{K}$ is bounded above by the strike, while $h_{c}$ has no finite upper bound because $S_{T}$ has none. We return to this asymmetry below; it is the single most misquoted fact about options.

> __Where does the premium come from?__ Today the premium is an __input__: we read it from a quoted options chain. It is not arbitrary. Under a stated arbitrage-free model, the premium of a European contract is the expected payoff under a risk-neutral measure, discounted back to today. For an American contract the holder also chooses _when_ to exercise, so the value is the __supremum__ of that discounted expectation taken over all admissible exercise rules, called __stopping times__: rules that decide whether to exercise using only the information available at that moment and never the future. Since the American holder can always choose to imitate the European holder and wait until $T$, an American contract is worth at least its European counterpart. L9a derives both cases.

## Profit Requires a Cash-Flow Convention
Payoff belongs to the contract. __Profit belongs to the investor__, and computing it requires saying what we do about the premium, which was paid or received at a different time.

The classroom expiration diagram uses the simplest possible rule: subtract the opening premium from the terminal payoff without growing it through time. Under this __undiscounted cash-flow convention__, the profit per share of a call and of a put position is given by:
$$
\boxed{
\begin{align*}
\Pi_{T}^{\,c,\theta}(S_{T}) &= \theta\Bigl(\underbrace{h_{c}(S_{T};K)}_{\text{payoff at }T} - \underbrace{C_{0}}_{\text{premium at }0}\Bigr)\\
\Pi_{T}^{\,p,\theta}(S_{T}) &= \theta\Bigl(\underbrace{h_{p}(S_{T};K)}_{\text{payoff at }T} - \underbrace{P_{0}}_{\text{premium at }0}\Bigr)
\end{align*}}
$$
where $\Pi_{T}^{\,\cdot,\theta}$ is the profit per share at expiration in USD/share, $C_{0},P_{0}\geq{0}$ are the call and put premiums in USD/share, and $\theta\in\{-1,+1\}$ is the direction. For a long position the premium is an initial outflow; for a short position it is an initial inflow, which is why the same $\theta$ multiplies both terms.

> __Be honest about what this convention hides.__ The expression above mixes time-$0$ dollars with time-$T$ dollars, which every earlier lecture in this course was careful never to do. A time-$T$ economic profit would grow the premium forward at the financing rate; a time-$0$ net present value would instead discount the terminal payoff (L1b). Fees, commissions, and the bid-ask spread must also be added consistently. We use the undiscounted convention because it gets the __shape__ and the __breakeven__ right with the fewest moving parts, and because it is what every brokerage payoff diagram shows. It is a convention, not an accounting identity.

### Breakeven prices
The __breakeven__ is the terminal price at which profit is exactly zero. Setting $\Pi_{T}=0$ and solving in the region where the option is exercised gives, for the long call, $S_{T}-K-C_{0}=0$, and for the long put, $K-S_{T}-P_{0}=0$. Therefore the breakeven prices are given by:
$$
\begin{align*}
S_{\text{BE}}^{\,c} &= K + C_{0}\\
S_{\text{BE}}^{\,p} &= K - P_{0}
\end{align*}
$$
where $S_{\text{BE}}$ denotes the breakeven price of the underlying at expiration in USD/share.

Two observations follow immediately:
* __The breakeven is not the strike.__ The premium shifts the profit curve down (long) or up (short) without moving the kink at $K$. Buying a call that finishes in the money is not the same as making money.
* __Both directions share a breakeven.__ Short profits are the exact negatives of long profits under this convention, so the short call and short put break even at the same prices, $K+C_{0}$ and $K-P_{0}$. What differs is which side of that price each position wants.

If $P_{0}>K$ the algebraic put breakeven is negative and therefore lies outside the feasible domain $S_{T}\geq{0}$, meaning the long put position cannot break even at any reachable price.

## Loss Bounds and Direction Matter
With the payoff and the profit convention fixed, the extremes of each of the four positions follow by inspection. All quantities are __per share__; multiply by the deliverable $q$ and by the number of contracts to obtain position-level dollars.

| Position | Maximum profit | Maximum loss | Breakeven |
|---|---|---|---|
| Long call | unbounded as $S_{T}\to\infty$ | $C_{0}$ | $K+C_{0}$ |
| Short call | $C_{0}$ | unbounded as $S_{T}\to\infty$ | $K+C_{0}$ |
| Long put | $K-P_{0}$, attained at $S_{T}=0$ | $P_{0}$ | $K-P_{0}$ |
| Short put | $P_{0}$ | $K-P_{0}$, attained at $S_{T}=0$ | $K-P_{0}$ |

Read the two rows for puts against the two rows for calls. The difference is not a convention; it comes entirely from the domain of the underlying price.

> __Why the put entries are genuinely nonnegative.__ The table reports $K-P_{0}$ as a maximum profit and as a maximum loss, which only makes sense if $K-P_{0}\geq{0}$. It always is. A long put's payoff can never exceed $K$, and that payoff arrives at $T$, so the honest comparison discounts it rather than setting it against a time-$0$ premium directly. A writer of a __European__ put who sets aside $Ke^{-g_{y}T}$ today holds exactly $K$ at expiration, which covers any obligation that contract can create. A premium above that reserve would let someone sell the put, fund the reserve, and keep the difference with no risk, so $P_{0}\leq{K}e^{-g_{y}T}\leq{K}$ whenever the discount rate $g_{y}\geq{0}$, with $g_{y}$ the continuously compounded rate of L2a. An __American__ put needs the weaker argument, because assignment can arrive at any $t\leq{T}$ and a reserve that only reaches $K$ at $T$ would not cover it: hold the undiscounted $K$ from the start instead, which gives $P_{0}\leq{K}$ directly. That is all the table requires, and it is why the discounted bound is not claimed for the American contracts the examples build. The put entries are therefore true maxima rather than signed extremes that happen to be negative. Where a bound can go negative we say so explicitly, as with the composite positions below.

> __The asymmetry, stated correctly:__
> * An __uncovered short call__ has __no finite bound__ on its terminal loss, because $S_{T}$ has no finite upper bound. This is the position that deserves the word _unlimited_.
> * A __short put__ on an ordinary contract with a finite strike has a __finite__ maximum terminal loss of $K-P_{0}$ per share, attained when the underlying goes to zero. Calling this "unlimited downside" is __incorrect__. It can still be a very large loss: on a $K=220$ contract collecting $16.33$, the worst case is $203.67$ USD/share, which at $q=100$ is $20{,}367$ USD per contract. Large and bounded are not the same thing as unlimited, and the distinction matters when you size a position or set a margin requirement.

> __What the terminal bounds do not capture.__ These are expiration-date extremes under the undiscounted convention. Before expiration, margin requirements, early assignment, borrowing constraints, liquidity, and forced liquidation can make realized risk severe long before the terminal bound is approached. A position that is bounded at $T$ is not thereby safe at every $t<T$.

## Composite Positions Add Leg by Leg
Single contracts give four payoff shapes. Combining them gives many more, and the combination rule is the simplest one possible: __add__.

Let a composite position $\mathcal{C}$ contain __legs__ $i=1,\dots,d$, all written on the same underlying with the same expiration. Leg $i$ is described by:
* its __type__ $\tau_{i}\in\{c,p\}$, call or put,
* its __strike__ $K_{i}>0$ in USD/share,
* its __direction__ $\theta_{i}\in\{-1,+1\}$, short or long,
* its __copy number__ $n_{i}\in\mathbb{N}$, how many contracts,
* its __premium__ $P_{i}\geq{0}$ in USD/share, paid or received when the position opens.

Write the long per-share payoff of leg $i$ compactly as:
$$
\begin{align*}
h_{i}(S_{T}) = \begin{cases}
\left(S_{T}-K_{i}\right)^{+} & \tau_{i}=c\\
\left(K_{i}-S_{T}\right)^{+} & \tau_{i}=p
\end{cases}
\end{align*}
$$
Then the total payoff and the total profit of the composite position at expiration are given by:
$$
\boxed{
\begin{align*}
H_{\mathcal{C}}(S_{T}) &= \sum_{i=1}^{d}\underbrace{\theta_{i}}_{\text{direction}}\,\underbrace{n_{i}}_{\text{contracts}}\;h_{i}(S_{T})\\
\Pi_{\mathcal{C}}(S_{T}) &= \sum_{i=1}^{d}\theta_{i}\,n_{i}\Bigl[\,\underbrace{h_{i}(S_{T})}_{\text{payoff at }T}-\underbrace{P_{i}}_{\text{premium at }0}\Bigr]
\end{align*}}
$$
where $H_{\mathcal{C}}$ is the composite payoff and $\Pi_{\mathcal{C}}$ the composite profit, both __per share__ in USD/share, which is what the example notebooks compute and plot.

> __Converting to account dollars.__ Multiply leg $i$ by its __deliverable__ $q_{i}\in\mathbb{N}$ in shares/contract, so the position payoff is $\sum_{i}\theta_{i}n_{i}q_{i}h_{i}(S_{T})$. Standard U.S. equity options carry $q_{i}=100$. The deliverable is applied per leg rather than factored out because a corporate-action adjustment can change one leg's deliverable while the others stay standard.

### Why linearity is worth stating
The sum above looks like bookkeeping, but it has real consequences:
* __The kinks are the strikes.__ Each $h_{i}$ bends only at $K_{i}$, so the composite profit is piecewise linear, and a corner can appear only at a component strike. Strike $K$ is an actual corner exactly when the aggregate slope jump there, $\sum_{i:K_{i}=K}\theta_{i}n_{i}$, is nonzero; opposing legs at one strike cancel and leave the diagram straight. Reading a strategy diagram means reading off its corners.
* __The vertical offset is the net premium.__ The term $\sum_{i}\theta_{i}n_{i}P_{i}$ is the net __debit__ (paid) or __credit__ (received) at the open. It shifts the whole diagram without changing its shape.
* __Nonlinear payoffs are reachable from linear parts.__ Combining kinked pieces produces profit profiles that no single contract and no stock position can produce: capped on one side, floored on the other, or profitable only inside a band. That is why composite positions exist.

> __What the single diagram requires.__ Writing the position as one function of one variable $S_{T}$ is legitimate only when every leg is written on the __same underlying__ and the __same expiration__. A portfolio holding options on two different underlyings, or on two different dates, is a perfectly valid position; it simply is not describable by a single expiration diagram, and needs a multivariate or multi-date cash-flow representation instead. The deliverables $q_{i}$ need __not__ agree, which is why each is applied to its own leg when converting to account dollars.

The examples build the standard structures from these pieces: __vertical spreads__ (two legs of one type, different strikes, opposite directions), __straddles__ (a call and a put at the same strike, same direction), and __strangles__ (a call and a put at different strikes, same direction). L10b returns to composite positions to compute the probability that one finishes profitable and to hedge it.

## Summary
In this lecture, we translated option contract terms into expiration cash flows, and kept the contractual payoff separate from the investor's premium-dependent profit.

> __Key Takeaways:__
> 
> * __Payoff belongs to the contract.__ A long call pays $\left(S_{T}-K\right)^{+}$ and a long put pays $\left(K-S_{T}\right)^{+}$ per share; short positions reverse those cash flows exactly. No estimation enters this step, only the terminal price $S_{T}$ and the terms written into the contract.
> * __Profit needs a convention.__ Premium, financing time, multiplier $q$, fees, and direction $\theta$ all determine profit, so the strike, the moneyness, and the breakeven are three different things. The long call breaks even at $K+C_{0}$ and the long put at $K-P_{0}$, neither of which is the strike.
> * __Loss bounds are asymmetric, and only one side is unlimited.__ An uncovered short call has no finite terminal loss bound, because $S_{T}$ has no upper bound. A standard short put's maximum terminal loss is finite at $K-P_{0}$, reached when the underlying goes to zero. Large is not the same as unlimited.

Composite positions add leg by leg, $\Pi_{\mathcal{C}}=\sum_{i}\theta_{i}n_{i}\left[h_{i}(S_{T})-P_{i}\right]$ per share, which puts a corner at each strike whose aggregate slope jump $\sum_{i:K_{i}=K}\theta_{i}n_{i}$ is nonzero, and shifts the diagram by the net debit or credit. Next, L9a values European claims with Black-Scholes-Merton and American claims with a Cox-Ross-Rubinstein exercise lattice, which is where the premium we took as an input today actually comes from.
