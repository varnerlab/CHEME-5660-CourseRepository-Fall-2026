## Concept Review: Utility Functions
Utility functions are mathematical expressions that describe a decision-maker's preferences for different choices, outcomes, or states of the world. A utility function gives a number, the __utility__ (in units of _utils_), for each option. Utility is a proxy for satisfaction, so a larger utility is always better than a smaller utility:

> __Utility function:__ A decision-maker (agent) chooses from a set of alternatives $\mathcal{X}=\{x_{1},\dots,x_{n}\}$, e.g., bundles of goods, portfolios, or states of the world. A utility function $U:\mathcal{X}\rightarrow\mathbb{R}$ assigns each alternative a real number, in units of _utils_, that ranks the agent's preferences:
>
> $$
> U(x_{i})>U(x_{j})\;\Longleftrightarrow\;x_{i}\succ x_{j}\qquad\text{and}\qquad U(x_{i})=U(x_{j})\;\Longleftrightarrow\;x_{i}\sim x_{j}
> $$
>
> where $x_{i}\succ x_{j}$ means the agent strictly prefers $x_{i}$ to $x_{j}$, and $x_{i}\sim x_{j}$ means the agent is indifferent between them. Utility functions are __ordinal__: they rank alternatives but do not measure how much better one alternative is than another. Thus, any order-preserving transformation of $U$, e.g., $\ln U$ for positive $U$, describes the same preferences.

Rational choice theory is __built on one key idea (assumption):__ a rational decision-maker always chooses the alternative with the largest utility. This assumption is the foundation of many economic and financial models, including portfolio optimization.

Utility functions, and their parameters, are subjective and differ from one decision-maker to another. Five common forms for a bundle of goods $\mathbf{x}=(x_{1},\dots,x_{n})$ with $x_{i}\geq 0$ and parameters $\alpha_{i}>0$ are:

| Utility function | Form | Models |
|:--|:--|:--|
| Linear | $U(\mathbf{x})=\sum_{i}\alpha_{i}x_{i}$ | goods that are perfect substitutes |
| Logarithmic | $U(\mathbf{x})=\ln\left(\sum_{i}\alpha_{i}x_{i}+\beta\right)$, with $\beta>0$ | diminishing satisfaction from more goods |
| Cobb–Douglas | $U(\mathbf{x})=\prod_{i}x_{i}^{\alpha_{i}}$ | goods held together |
| Leontief | $U(\mathbf{x})=\min\left(x_{1}/\alpha_{1},\dots,x_{n}/\alpha_{n}\right)$ | perfect complements, used in fixed proportions |
| CES | $U(\mathbf{x})=\left(\sum_{i}\alpha_{i}x_{i}^{(\eta-1)/\eta}\right)^{\eta/(\eta-1)}$, with $\eta>0$, $\eta\neq 1$ | substitution set by the elasticity $\eta$ |

We use the Cobb–Douglas and CES utilities in the next section. The constant elasticity of substitution (CES) utility moves between three of the other forms as its elasticity $\eta$ changes: linear as $\eta\to\infty$, Cobb–Douglas as $\eta\to 1$ (with $\sum_{i}\alpha_{i}=1$), and Leontief as $\eta\to 0$.

## Utility over Holdings
An investor chooses share counts of assets in $\mathcal{P}$ to maximize a __holding utility__, which ranks alternative holdings. The value of the shares at today's prices, plus any cash, must equal the available wealth. The utility has no variance term. In our model, risk enters through preference weights based on SIM betas and the market state.


### Cobb–Douglas utility
Cobb–Douglas utility uses the __preference weights__ $\gamma_i\in(-1,1)$ as exponents. We maximize utility over every asset, subject to a budget and a floor of $n_{\min}>0$ shares:

$$
\boxed{\begin{aligned}
\operatorname*{maximize}_{n_{1},\ldots,n_{|\mathcal{P}|}}\quad & U(n_{1},\ldots,n_{|\mathcal{P}|}) = \prod_{i\in\mathcal{P}}n_{i}^{\gamma_{i}}\\
\text{subject to}\quad & \sum_{i\in\mathcal{P}}n_{i}\,S_{i}(t)+\text{cash}(t) = W_{\mathcal{P}}(t) &&\quad\text{(budget constraint)}\\
& n_{i}\geq n_{\min}\quad\forall\,i\in\mathcal{P} &&\quad\text{(share floor)}\\
& \text{cash}(t)\geq 0 &&\quad\text{(no borrowing)}
\end{aligned}}
$$

Here, $\mathcal{P}$ is the asset set and $n_i$ is the share count of asset $i$. At time $t$, prices $S_i(t)$ are in USD per share. Wealth $W_{\mathcal{P}}(t)$ and uninvested $\text{cash}(t)$ are in USD. Cash contributes no utility.

#### Solution
Assets with positive preference weights form the __preferred set__ $\mathcal{A}^{+}=\{i:\gamma_i>0\}$, which we call the __basket__. The rest form the __non-preferred set__ $\mathcal{A}^{-}=\{i:\gamma_i\leq0\}$. The positive floor keeps utility finite: for $\gamma_i<0$, $n_i^{\gamma_i}$ grows without bound as $n_i\to0$.

Assume positive prices, enough wealth to fund all floors, and a nonempty basket. Non-preferred assets stay at their floors and cash is zero. The budget left to allocate over the preferred holdings is given by:

$$
W_{\text{adj}}
=W_{\mathcal{P}}(t)-n_{\min}\sum_{k\in\mathcal{A}^{-}}S_k(t).
$$

When preferred holdings stay above their floors, the unique __Cobb–Douglas allocation__ is:

$$
\boxed{\begin{aligned}
n_i^\star
&=\underbrace{\frac{\gamma_i}{\sum_{j\in\mathcal{A}^{+}}\gamma_j}}_{\text{preference share}}
\cdot\underbrace{\frac{W_{\text{adj}}}{S_i(t)}}_{\text{budget in shares}}
&&\quad i\in\mathcal{A}^{+},\\
n_k^\star&=n_{\min}&&\quad k\in\mathcal{A}^{-}.
\end{aligned}}
$$

The [Cobb–Douglas derivation companion](advanced/cobb-douglas-derivation/CHEME-5660-L8b-Derivation-CobbDouglas-Fall-2026.ipynb) derives the allocation, proves the rule for binding floors, and works through a numerical example.


### CES utility
The constant elasticity of substitution (CES) utility adds one parameter to Cobb–Douglas, the elasticity $\eta>0$. The elasticity controls how willingly the allocator (us) trades one preferred asset for another. For $\eta\neq 1$, the CES utility is given by:

$$
U_{\text{CES}}(n_{1},\ldots,n_{|\mathcal{P}|}) = \left(\sum_{i\in\mathcal{A}^{+}}\gamma_{i}\,n_{i}^{(\eta-1)/\eta}\right)^{\eta/(\eta-1)}
$$

Its budget-constrained maximizer is also closed form, with the same floor check. Among preferred assets above their floors, share counts are proportional to $(\gamma_{i}/S_{i}(t))^{\eta}$, where $\gamma_{i}/S_{i}(t)$ is the _bang for the buck_ of consumer theory:

$$
\boxed{n_{i}^{\star} = W_{\text{adj}}\;\frac{\left(\gamma_{i}/S_{i}(t)\right)^{\eta}}{\sum_{j\in\mathcal{A}^{+}}S_{j}(t)\left(\gamma_{j}/S_{j}(t)\right)^{\eta}}\qquad i\in\mathcal{A}^{+}}
$$

The elasticity is a __concentration dial__ with three limits:

* __Unit elasticity ($\eta\to 1$):__ The allocation tends to the Cobb–Douglas allocation.
* __Large elasticity ($\eta\to\infty$):__ The asset with the largest $\gamma_{i}/S_{i}(t)$ gets the whole budget above the floors, when that asset is unique. Every other asset falls to its floor.
* __Small elasticity ($\eta\to 0^{+}$):__ The allocator buys equal share counts of every preferred asset, whatever its $\gamma_{i}$ or price.

The [CES companion](advanced/adaptive_utility/CHEME-5660-L8b-Advanced-CES-Limits-Fall-2026.ipynb) proves the three limits. It also shows that for $\eta\neq 1$ the CES dollar weights depend on the price level of each share. Where do the preference weights come from?

## Preference Weights from the Single Index Model
Each trading day, we compute the preference weights $\gamma_{i}(t)$ from the SIM parameters $(\alpha_{i},\beta_{i})$ and two market inputs. The __recent market growth rate__ $\tilde{g}_{M,t}$ plays the role of the SIM's expected market growth. The __market-state signal__ $\xi_{t}$ says whether the market has recently been rising or falling.

> __Preference model:__ For each asset $i$ on day $t$, the preference weight is given by:
>
> $$
> \boxed{\gamma_{i}(t) = \tanh\left(h\left[\frac{\alpha_{i}}{\beta_{i}^{\xi_{t}}} + \beta_{i}^{1-\xi_{t}}\,\tilde{g}_{M,t}\right]\right)}
> $$
>
> where $\tilde{g}_{i,t}=\alpha_{i}+\beta_{i}\tilde{g}_{M,t}$ is the SIM's expected growth rate of asset $i$ in today's market, in inverse years. The argument of the $\tanh$ is $h\,\tilde{g}_{i,t}/\beta_{i}^{\xi_{t}}$, and the horizon $h = 1$ year makes it dimensionless. The hyperbolic tangent keeps the weight in $(-1,1)$. The model assumes $\beta_{i}>0$ for every asset, because $\beta_{i}$ is raised to a real power.

We read the preference weight in two parts:

* __Sign, which assets are in the basket:__ The hyperbolic tangent satisfies $\tanh(x) < 0$ for $x < 0$, $\tanh(0) = 0$, and $\tanh(x) > 0$ for $x > 0$. Since $h > 0$ and $\beta_{i}^{\xi_{t}} > 0$, the sign of $\gamma_{i}(t)$ is the sign of $\tilde{g}_{i,t}$. Asset $i$ is preferred when $\tilde{g}_{i,t} > 0$, that is, when $\tilde{g}_{M,t}$ is above its threshold $-\alpha_{i}/\beta_{i}$. The preferred assets form the __basket__. As $\tilde{g}_{M,t}$ rises, more assets clear their thresholds and the basket grows. As $\tilde{g}_{M,t}$ falls, assets drop out, and only those with the lowest thresholds stay. If the basket is empty, the allocator holds the floors plus cash.
* __Magnitude, how much each asset gets:__ The magnitude of a positive $\gamma_{i}(t)$ sets its share of the budget. The signal $\xi_{t}$, scaled so that $-1 \leq \xi_{t} \leq 1$ on nearly every day, never changes a sign. It tilts the magnitudes toward low-beta assets when $\xi_{t} > 0$ (bearish), not at all when $\xi_{t} = 0$ (neutral), and toward high-beta assets when $\xi_{t} < 0$ (bullish). A bearish signal ($\xi_{t}>0$) divides by $\beta_{i}^{\xi_{t}}$, which is above one for high-beta assets and below one for low-beta assets.


### Market inputs from SPY
Both market inputs come from the SPY price through __exponential moving averages__ (EMAs). An EMA is a running average that gives recent prices the most weight. For a price series $S_{t}$ and a window of $L$ trading days, the EMA is given by:

$$
\bar{S}_{t} = \omega\,S_{t}+(1-\omega)\,\bar{S}_{t-1}\qquad\text{where}\qquad\omega=\frac{2}{L+1}
$$

> __Market inputs:__ With a short window $L_{\text{short}}=21$ days, a long window $L_{\text{long}}=63$ days, and a dimensionless gain $G>0$, the market-state signal is the scaled crossover of the two price EMAs:
>
> $$
> \xi_{t} = -G\left(\frac{\bar{S}^{\text{short}}_{t}}{\bar{S}^{\text{long}}_{t}}-1\right)
> $$
>
> The short EMA responds faster to price changes than the long EMA, so the ratio says where recent prices sit against the quarter's average:
>
> * $\bar{S}^{\text{short}}_{t} > \bar{S}^{\text{long}}_{t}$: the ratio is above one, so $\xi_{t} < 0$ and the signal is __bullish__. Recent prices sit above the quarter's average: the market has been rising.
> * $\bar{S}^{\text{short}}_{t} = \bar{S}^{\text{long}}_{t}$: $\xi_{t} = 0$ and the signal is __neutral__, so the preference weights carry no tilt. Recent prices match the quarter's average, which happens in a flat market or at the moment a trend reverses and the two EMAs cross.
> * $\bar{S}^{\text{short}}_{t} < \bar{S}^{\text{long}}_{t}$: the ratio is below one, so $\xi_{t} > 0$ and the signal is __bearish__. Recent prices sit below the quarter's average: the market has been falling.
>
> We choose $G$ on the training data so that $|\xi_{t}|\leq 1$ on 99% of training days. The recent market growth rate $\tilde{g}_{M,t}$ is the EMA, with a window of $L_{\text{growth}}=252$ days, of the daily market growth rates $g_{M,t}=\ln(S_{t}/S_{t-1})/\Delta t$ (inverse years).

Both inputs on day $t$ use prices through day $t-1$, so a decision never uses a price it could not have known.

## Mean-Variance Utility on the Capital Allocation Line
A __complete portfolio__ $c$ puts a fraction $w_{f}$ of wealth in the risk-free asset and $1-w_{f}$ in the tangent portfolio $\mathcal{T}$. Let $g_{c}$ and $g_{\mathcal{T}}$ be the growth rates of $c$ and $\mathcal{T}$, with standard deviations $\sigma_{g,c}$ and $\sigma_{g,\mathcal{T}}$. The risk-free asset grows at the rate $g_{f}$, and growth rates are in inverse years. The expected growth rate and variance of $c$ are given by:

$$
\begin{align*}
\mathbb{E}[g_{c}] &= g_{f}+(1-w_{f})\left(\mathbb{E}[g_{\mathcal{T}}]-g_{f}\right) &&\quad\text{expectation is linear}\\
\sigma^{2}_{g,c} &= (1-w_{f})^{2}\,\sigma^{2}_{g,\mathcal{T}} &&\quad\text{the risk-free asset adds no variance}
\end{align*}
$$

A negative $w_{f}$ means borrowing at $g_{f}$ to buy more of $\mathcal{T}$. Which $w_{f}$ should a given investor choose?

### Measuring risk aversion
Let $U(W)$ be the utility of an uncertain wealth $W$. A concave $U$ describes a risk-averse decision-maker. The __certainty equivalent__ (CE) is the sure wealth with the same utility as the risky outcome. The __risk premium__ $\mathrm{RP}=\mathbb{E}[W]-\mathrm{CE}$ is the expected wealth a decision-maker gives up to remove the risk.

__How risk averse is a decision-maker?__ The more concave the utility function $U$ is, the further the CE falls below $\mathbb{E}[W]$. Arrow and [Pratt (1964)](https://doi.org/10.2307/1913738) turned this curvature into a number, the __relative risk aversion__ $\bar{r}$.

Investors think in growth rates. Write the wealth after one trading day as $W=W_{0}e^{g_{c}\Delta t}$, where $W_{0}$ is today's wealth, $g_{c}$ is the growth rate of the complete portfolio, and $\Delta t=1/252$ year. For a smooth, increasing $U$ and small daily changes, $\bar{r}$ and the risk premium are given by:

$$
\begin{align*}
\bar r &= -W_0\frac{U''(W_0)}{U'(W_0)} &&\quad\text{relative risk aversion}\\
\mathrm{RP} &\approx -\frac{1}{2}\,\frac{U''(W_0)}{U'(W_0)}\,\mathrm{Var}(W) &&\quad\text{Pratt (1964), small risk}\\
\mathrm{Var}(W) &\approx W_0^{2}\,(\Delta t)^{2}\sigma^2_{g,c} &&\quad W\approx W_0(1+g_c\Delta t)\text{ for one day}\\
\frac{\mathrm{RP}}{W_0} &\approx \frac{\bar r}{2}\,(\Delta t)^2\sigma^2_{g,c} &&\quad\text{substitute, divide by }W_0
\end{align*}
$$

where $U'$ and $U''$ are the first and second derivatives of $U$ with respect to wealth. Because $U$ is increasing, $U'>0$, and because $U$ is concave, $U''<0$, so $\bar{r}>0$ for a risk-averse decision-maker.

The __certainty-equivalent growth rate__ $g_{\mathrm{CE}}$ is the sure growth rate the investor would accept in place of the risky $g_{c}$. It takes today's wealth to the CE in one trading day, just as $g_{c}$ takes it to $W$:

$$
\begin{align*}
\mathrm{CE} &= W_0\,e^{g_{\mathrm{CE}}\Delta t} &&\quad\text{definition of }g_{\mathrm{CE}}\\
g_{\mathrm{CE}} &= \frac{1}{\Delta t}\ln\!\left(\frac{\mathrm{CE}}{W_0}\right) &&\quad\text{take the log, divide by }\Delta t\\
\frac{W}{W_0} = e^{g_c\Delta t} &\approx 1+g_c\,\Delta t+\frac{1}{2}\,(g_c\,\Delta t)^2 &&\quad\text{expand }e^{x}\text{ to second order}\\
\frac{\mathbb E[W]}{W_0} &\approx 1+\mathbb E[g_c]\,\Delta t+\frac{1}{2}\,(\Delta t)^2\left(\mathbb E[g_c]^2+\sigma^2_{g,c}\right) &&\quad\text{take expectations, }\mathbb E[g_c^2]=\mathbb E[g_c]^2+\sigma^2_{g,c}\\
&\approx 1+\mathbb E[g_c]\,\Delta t+\frac{1}{2}\,(\Delta t)^2\sigma^2_{g,c} &&\quad\mathbb E[g_c]^2\ll\sigma^2_{g,c}\text{ for one day}\\
\frac{\mathrm{CE}}{W_0} &= \frac{\mathbb E[W]}{W_0}-\frac{\mathrm{RP}}{W_0} &&\quad\mathrm{RP}=\mathbb E[W]-\mathrm{CE}\text{, divide by }W_0\\
&\approx 1+\mathbb E[g_c]\,\Delta t+\frac{1}{2}\,(\Delta t)^2\sigma^2_{g,c}-\frac{\bar r}{2}\,(\Delta t)^2\sigma^2_{g,c} &&\quad\text{substitute }\mathbb E[W]/W_0\text{ and }\mathrm{RP}/W_0\\
&\approx 1+\mathbb E[g_c]\,\Delta t-\frac{\bar r-1}{2}\,(\Delta t)^2\sigma^2_{g,c} &&\quad\text{collect the }(\Delta t)^2\sigma^2_{g,c}\text{ terms}\\
g_{\mathrm{CE}} &\approx \mathbb E[g_c]-\frac{\bar r-1}{2}\,\Delta t\,\sigma^2_{g,c} &&\quad\ln(1+x)\approx x\text{, divide by }\Delta t
\end{align*}
$$

The variance term lowers $g_{\mathrm{CE}}$ only when $\bar{r}>1$, which we assume from here on.

### The optimal complete portfolio
Each risk-free fraction $w_{f}$ gives a different complete portfolio, with its own reward $\mathbb{E}[g_{c}]$ and risk $\sigma^{2}_{g,c}$, and so its own certainty-equivalent growth rate. The risk-averse investor wants the __largest sure growth rate__, so we choose $w_{f}$ to maximize the certainty-equivalent growth rate of the complete portfolio. 

Defining $A\equiv\bar{r}-1$, the certainty-equivalent growth rate $g_{\mathrm{CE}}$ is the __mean-variance utility__ $U(w_{f})$, which rises with expected growth (makes the investor happy) and falls with variance (makes the investor unhappy):

$$
\underset{w_f}{\operatorname{maximize}}\quad
U(w_f)=\mathbb E[g_c]-\frac{A}{2}\,\Delta t\,\sigma^2_{g,c}\approx g_{\mathrm{CE}},
$$

where the risk-aversion coefficient $A>0$ (dimensionless) sets how much the investor dislikes variance, and $\Delta t\,\sigma^{2}_{g,c}$ is the portfolio's squared volatility, in inverse years. We assume $\mathbb{E}[g_{\mathcal{T}}]>g_{f}$.

On the capital allocation line, $\mathbb{E}[g_{c}]$ and $\sigma^{2}_{g,c}$ depend only on $w_{f}$. Substituting them into $U(w_{f})$ and setting the derivative to zero at the optimal fraction $w_{f}^{\star}$ gives:

$$
\begin{align*}
U(w_f) &= g_f+(1-w_f)(\mathbb E[g_{\mathcal T}]-g_f)-\frac{A}{2}(1-w_f)^2\Delta t\,\sigma^2_{g,\mathcal T}
&&\quad\text{substitute the line}\\
\frac{dU}{dw_f} &= -(\mathbb E[g_{\mathcal T}]-g_f)+A(1-w_f)\Delta t\,\sigma^2_{g,\mathcal T}
&&\quad\text{differentiate}\\
0 &= -(\mathbb E[g_{\mathcal T}]-g_f)+A(1-w_f^\star)\Delta t\,\sigma^2_{g,\mathcal T}
&&\quad\text{first-order condition}.
\end{align*}
$$

Solving for the fraction in $\mathcal{T}$ gives the __optimal complete portfolio__:

$$
\boxed{1-w_f^\star=
\frac{\mathbb E[g_{\mathcal T}]-g_f}
{A\,\Delta t\,\sigma^2_{g,\mathcal T}}}.
$$

The second derivative, $-A\,\Delta t\,\sigma^{2}_{g,\mathcal{T}}$, is negative, so this is a maximum. The fraction in $\mathcal{T}$ rises with the excess growth $\mathbb{E}[g_{\mathcal{T}}]-g_{f}$ and falls as $A$ or $\sigma^{2}_{g,\mathcal{T}}$ grows. Doubling $A$ halves it.

__Where does the optimum sit on the line?__ An __indifference curve__ joins the portfolios with the same utility. Fixing the utility at a level $U$ gives:

$$
\mathbb E[g_c]=U+\frac{A}{2}\,\Delta t\,\sigma^2_{g,c}.
$$

Plotted against $\sigma_{g,c}$, each curve is a parabola that opens upward, and higher curves have higher utility. The optimal portfolio is where the highest curve the investor can reach touches the line.

The result also runs backward: solving the boxed formula for $A$ gives the risk aversion behind a client's chosen $w_{f}<1$.

__Can we measure $A$ without asking for $w_{f}$?__ [Barsky, Juster, Kimball, and Shapiro (1997)](https://doi.org/10.1162/003355397555280) offered people two jobs: a sure income, or equal chances of doubling it or cutting it by a fraction $\delta$. Suppose we pay a price $p$ today for a contract that pays 1 USD in $T$ years if a condition is true, and nothing if it is false. The condition is that the share price $S_{T}$ finishes at or above a target price $K$. The payoff (USD) is given by:

$$
V(S_{T}) = \begin{cases}
1 & \text{if } S_{T}\geq K\\
0 & \text{if } S_{T}<K
\end{cases}
$$

This contract is an abstract asset with two cash flows: we pay $p$ at $t=0$ and receive $V(S_{T})$ at $t=T$. Discounting the payoff at the benchmark growth rate $g_{y}$ gives the contract's net present value:

$$
\operatorname{NPV}_{0} = -p+\mathcal{D}^{-1}_{T,0}(g_{y})\,V(S_{T})\qquad\text{where}\qquad\mathcal{D}^{-1}_{T,0}(g_{y})=e^{-g_{y}T}
$$

The payoff is uncertain, so this NPV is random. It is $-p+e^{-g_{y}T}$ if the condition is true and $-p$ if it is false. A payoff that is either one or zero has an expected value equal to the probability that it is one. Under our beliefs about $S_{T}$, the expected NPV is given by:

$$
\mathbb{E}\left[\operatorname{NPV}_{0}\right] = -p+e^{-g_{y}T}\,\mathbb{P}(S_{T}\geq K)
$$

For a share that follows GBM, $\mathbb{P}(S_{T}\geq K)$ is the L5a terminal target probability $1-\Phi(z_{\star})$, because $S_{T}=K$ has probability zero. The bet has a positive expected NPV when its price is below its discounted expected payoff, $p<e^{-g_{y}T}\,\mathbb{P}(S_{T}\geq K)$.

__What would a risk-averse buyer pay?__ On its own, the payoff is a gamble. A risk-averse buyer values it at its certainty equivalent, which is below its expected value, so this buyer would pay less than $e^{-g_{y}T}\,\mathbb{P}(S_{T}\geq K)$. How much less depends on the buyer's risk aversion and on how many contracts the buyer holds.

Contracts whose payoffs depend on the future price of another asset are called __derivatives__. Each one trades at a market price that every buyer pays, yet buyers with different beliefs and different risk aversion value the same contract differently. What sets the price?
