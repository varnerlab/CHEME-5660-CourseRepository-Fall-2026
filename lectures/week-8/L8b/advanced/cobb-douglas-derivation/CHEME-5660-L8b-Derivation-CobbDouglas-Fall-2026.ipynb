## The Allocation Problem
Fix a trading day $t$. Let $\mathcal{P}$ be the finite set of assets, $S_i=S_i(t)>0$ the price of asset $i$ (USD per share), and $W=W_{\mathcal{P}}(t)$ the investor's wealth (USD). The decision variables are the fractional share counts $n_i$ and uninvested cash $c$ (USD). Each asset has a fixed preference weight $\gamma_i\in(-1,1)$ and must be held at or above the same floor $n_{\min}>0$ (shares). Cash contributes no utility.

The investor chooses holdings and cash by solving:

$$
\boxed{\begin{aligned}
\operatorname*{maximize}_{(n_i)_{i\in\mathcal{P}},\,c}\quad
&U(\mathbf n)=\prod_{i\in\mathcal{P}}n_i^{\gamma_i}\\
\text{subject to}\quad
&\sum_{i\in\mathcal{P}}n_iS_i+c=W &&\quad\text{(budget)}\\
&n_i\geq n_{\min}\quad i\in\mathcal{P} &&\quad\text{(share floors)}\\
&c\geq0 &&\quad\text{(no borrowing)}.
\end{aligned}}
$$

The floor has a cost even for an asset the investor does not prefer. Define the total cost of buying every floor by:

$$
C_{\min}=n_{\min}\sum_{i\in\mathcal{P}}S_i.
$$

The problem is feasible exactly when $W\geq C_{\min}$. At $W=C_{\min}$, the only feasible allocation holds every asset at its floor with zero cash. We develop the remaining cases for $W>C_{\min}$.


### Why non-preferred assets stay at their floors
Because every share count is positive, the product utility is positive and its logarithm is defined. The logarithm is strictly increasing, so it preserves the ranking of every pair of feasible allocations. It also turns the product into a sum:

$$
\ln U(\mathbf n)
=\ln\left(\prod_{i\in\mathcal{P}}n_i^{\gamma_i}\right)
=\sum_{i\in\mathcal{P}}\gamma_i\ln n_i.
$$

Partition the asset set according to the preference weights:

$$
\mathcal{A}^{+}=\{i\in\mathcal{P}:\gamma_i>0\},
\qquad
\mathcal{A}^{-}=\{i\in\mathcal{P}:\gamma_i\leq0\}.
$$

The derivative of log utility with respect to a share count is:

$$
\frac{\partial\ln U}{\partial n_i}=\frac{\gamma_i}{n_i}.
$$

Assume $\mathcal{A}^{+}$ is nonempty. If a non-preferred asset $k$ is held above its floor, sell a small dollar amount of it and invest the proceeds in a preferred asset. Reducing $n_k$ cannot lower log utility, and increasing the preferred holding strictly raises it. The budget is preserved, so the original allocation cannot be optimal. Every non-preferred holding must therefore equal its floor.

The same argument settles cash. If cash is positive, investing some of it in a preferred holding strictly raises log utility. Thus, an optimum with at least one preferred asset satisfies:

$$
n_k^\star=n_{\min}\quad k\in\mathcal{A}^{-},
\qquad c^\star=0.
$$

If $\mathcal{A}^{+}$ is empty, one optimum holds every asset at its floor and keeps $W-C_{\min}$ in cash. An asset with zero preference weight can also receive surplus dollars without changing utility, so this case need not have a unique optimum.


### The budget available to preferred holdings
With the non-preferred holdings and cash settled, substitute their values into the budget:

$$
\begin{align*}
\sum_{i\in\mathcal{A}^{+}}n_iS_i
+n_{\min}\sum_{k\in\mathcal{A}^{-}}S_k&=W,\\
\sum_{i\in\mathcal{A}^{+}}n_iS_i
&=W-n_{\min}\sum_{k\in\mathcal{A}^{-}}S_k
\equiv W_{\text{adj}}.
\end{align*}
$$

The adjusted budget $W_{\text{adj}}$ (USD) funds all preferred holdings, including their floors. The terms $\gamma_k\ln n_{\min}$ for non-preferred assets are constant, so define the remaining objective by:

$$
f(\mathbf n)=\sum_{i\in\mathcal{A}^{+}}\gamma_i\ln n_i.
$$

The reduced optimization problem is:

$$
\boxed{\begin{aligned}
\operatorname*{maximize}_{(n_i)_{i\in\mathcal{A}^{+}}}\quad
&f(\mathbf n)\\
\text{subject to}\quad
&\sum_{i\in\mathcal{A}^{+}}n_iS_i=W_{\text{adj}} &&\quad\text{(preferred budget)}\\
&n_i\geq n_{\min}\quad i\in\mathcal{A}^{+} &&\quad\text{(preferred floors)}.
\end{aligned}}
$$

This reduction preserves the optimum of the full problem whenever at least one asset is preferred.

## Allocation above the Share Floors
Assume every preferred holding lies strictly above its floor.

Introduce a scalar Lagrange multiplier $\lambda$ for the preferred budget. The Lagrangian is:

$$
\mathcal L(\mathbf n,\lambda)
=\sum_{i\in\mathcal{A}^{+}}\gamma_i\ln n_i
-\lambda\left(\sum_{i\in\mathcal{A}^{+}}n_iS_i-W_{\text{adj}}\right).
$$

A small increase or decrease in any preferred share count is allowed when that count is above its floor. The first-order conditions are therefore:

$$
\frac{\partial\mathcal L}{\partial n_i}
=\frac{\gamma_i}{n_i}-\lambda S_i=0,
\qquad i\in\mathcal{A}^{+}.
$$

Divide by the positive share price and rearrange to obtain:

$$
\frac{\gamma_i}{n_iS_i}=\lambda,
\qquad
n_iS_i=\frac{\gamma_i}{\lambda}.
$$

The first expression equates the marginal log utility per dollar across preferred holdings. If two such holdings had different marginal values, transferring a small dollar amount toward the larger one would increase the objective. The second expression shows that dollars are proportional to preference weights.


### Determine the multiplier from the budget
The multiplier is common to every preferred asset, so the budget determines its value. Substitute the dollar holdings into the budget and solve:

$$
\begin{align*}
W_{\text{adj}}
&=\sum_{i\in\mathcal{A}^{+}}n_iS_i
=\sum_{i\in\mathcal{A}^{+}}\frac{\gamma_i}{\lambda}
=\frac{1}{\lambda}\sum_{i\in\mathcal{A}^{+}}\gamma_i,\\
\lambda&=\frac{\sum_{j\in\mathcal{A}^{+}}\gamma_j}{W_{\text{adj}}}>0,\\
\frac{1}{\lambda}
&=\frac{W_{\text{adj}}}{\sum_{j\in\mathcal{A}^{+}}\gamma_j}.
\end{align*}
$$

The multiplier measures marginal log utility per additional dollar of budget. Substituting its reciprocal into the dollar holding and dividing by the price gives:

$$
\boxed{
n_i^\star
=\frac{\gamma_i}{\sum_{j\in\mathcal{A}^{+}}\gamma_j}
\frac{W_{\text{adj}}}{S_i},
\qquad i\in\mathcal{A}^{+}.
}
$$

The preferred budget is divided into dollar shares proportional to the positive preference weights. Prices convert those dollars into share counts.

__Why is this a maximum?__ The Hessian of the reduced log utility is diagonal:

$$
\nabla^2 f(\mathbf n)
=\operatorname{diag}\left(-\frac{\gamma_i}{n_i^2}\right)_{i\in\mathcal{A}^{+}}
\prec0.
$$

Every diagonal entry is negative, so the reduced objective is strictly concave. The feasible set is convex because the budget and share floors are linear. Thus, a feasible stationary point is the unique maximum. If the formula gives a holding exactly at its floor, it is still optimal; a holding below its floor requires enforcing the share-floor constraint.

## Binding Share Floors
When the proportional formula gives a holding fewer shares than its floor, the floor restricts the allocation. We include that constraint in the Lagrangian to determine which holdings remain at their floors.

### Optimality conditions with share floors
For each preferred asset, introduce a nonnegative multiplier $\nu_i$ on the constraint $n_i-n_{\min}\geq0$. With the same budget multiplier $\lambda$, the Lagrangian is:

$$
\begin{aligned}
\mathcal L(\mathbf n,\lambda,\boldsymbol{\nu})
={}&\sum_{i\in\mathcal{A}^{+}}\gamma_i\ln n_i
-\lambda\left(\sum_{i\in\mathcal{A}^{+}}n_iS_i-W_{\text{adj}}\right)\\
&+\sum_{i\in\mathcal{A}^{+}}\nu_i(n_i-n_{\min}).
\end{aligned}
$$

The Karush–Kuhn–Tucker optimality conditions require stationarity, a feasible allocation, nonnegative floor multipliers, and complementary slackness:

$$
\begin{align*}
\frac{\gamma_i}{n_i}-\lambda S_i+\nu_i&=0
&&\quad i\in\mathcal{A}^{+}\quad\text{(stationarity)},\\
\sum_{i\in\mathcal{A}^{+}}n_iS_i&=W_{\text{adj}}
&&\quad\text{(budget)},\\
n_i&\geq n_{\min}
&&\quad i\in\mathcal{A}^{+}\quad\text{(share floors)},\\
\nu_i&\geq0
&&\quad i\in\mathcal{A}^{+}\quad\text{(floor multipliers)},\\
\nu_i(n_i-n_{\min})&=0
&&\quad i\in\mathcal{A}^{+}\quad\text{(complementary slackness)}.
\end{align*}
$$

Complementary slackness says that a holding above its floor has $\nu_i=0$. Its stationarity condition equates marginal log utility per dollar with the budget multiplier. A holding at its floor may have $\nu_i>0$, because the investor would prefer to reduce it further.

The two cases give:

$$
\begin{aligned}
n_i>n_{\min}
&\quad\Longrightarrow\quad
\nu_i=0,\qquad n_i=\frac{\gamma_i}{\lambda S_i},\\
n_i=n_{\min}
&\quad\Longrightarrow\quad
\nu_i=\lambda S_i-\frac{\gamma_i}{n_{\min}}\geq0.
\end{aligned}
$$

The second line means that a holding at its floor has marginal log utility per dollar at most $\lambda$. When its marginal value is strictly smaller than the common value, moving some of its dollars into a holding above its floor would increase utility, but the floor prevents that transfer.

Combining both cases gives the allocation in terms of a single positive multiplier:

$$
\boxed{\begin{aligned}
n_i^\star&=\max\left\{n_{\min},\frac{\gamma_i}{\lambda S_i}\right\},\\
\sum_{i\in\mathcal{A}^{+}}S_i
\max\left\{n_{\min},\frac{\gamma_i}{\lambda S_i}\right\}
&=W_{\text{adj}}.
\end{aligned}}
$$

Because $W>C_{\min}$, at least one preferred holding is above its floor. The budget equation then determines a unique $\lambda>0$. The reduced objective is strictly concave, so a feasible allocation satisfying these conditions is its unique maximum. Together with the non-preferred floors and zero cash, it solves the full problem.


### Why pinning and solving again works
Let $\mathcal B\subseteq\mathcal{A}^{+}$ contain the preferred assets pinned at their floors, and let $\mathcal F=\mathcal{A}^{+}\setminus\mathcal B$ contain the remaining preferred assets. The budget $W_{\mathcal F}$ (USD) available to these remaining holdings is:

$$
W_{\mathcal F}
=W_{\text{adj}}-n_{\min}\sum_{k\in\mathcal B}S_k.
$$

For a proposed pinned set, equating marginal log utility per dollar over $\mathcal F$ and using its budget gives:

$$
\lambda=\frac{\sum_{j\in\mathcal F}\gamma_j}{W_{\mathcal F}},
\qquad
n_i=\frac{\gamma_i}{\sum_{j\in\mathcal F}\gamma_j}
\frac{W_{\mathcal F}}{S_i},
\qquad i\in\mathcal F.
$$

Start with no preferred holdings pinned. If every computed count meets its floor, the allocation satisfies the optimality conditions. Otherwise, let $\mathcal V\subset\mathcal F$ be the set of holdings whose computed counts fall below the floor. At least one remaining holding is above its floor, because the candidate holdings spend a budget that exceeds the cost of all their floors. For each violating holding:

$$
\frac{\gamma_i}{\lambda_{\text{old}}S_i}<n_{\min}
\quad\Longrightarrow\quad
\gamma_i<\lambda_{\text{old}}n_{\min}S_i,
\qquad i\in\mathcal V.
$$

Pin these holdings, subtract their floor costs from the available budget, and solve for the remaining holdings. Since the old free budget equals $\sum_{j\in\mathcal F}\gamma_j/\lambda_{\text{old}}$, the inequality above gives:

$$
\begin{align*}
\sum_{j\in\mathcal F\setminus\mathcal V}\gamma_j
&>\lambda_{\text{old}}
\left(W_{\mathcal F}-n_{\min}\sum_{i\in\mathcal V}S_i\right),\\
\lambda_{\text{new}}
&=\frac{\sum_{j\in\mathcal F\setminus\mathcal V}\gamma_j}
{W_{\mathcal F}-n_{\min}\sum_{i\in\mathcal V}S_i}
>\lambda_{\text{old}}.
\end{align*}
$$

The denominator is positive because the total budget can fund all floors and has a surplus. As $\lambda$ rises, the unconstrained count $\gamma_i/(\lambda S_i)$ falls. A holding already pinned therefore continues to satisfy its floor optimality condition. We can keep it pinned while checking the remaining holdings.

Each unsuccessful solve pins at least one additional preferred asset. At least one remains above its floor when the budget has a surplus, so at most $|\mathcal{A}^{+}|$ solves are needed. At termination, the free holdings satisfy stationarity with zero floor multipliers, and the pinned holdings satisfy it with nonnegative floor multipliers. The budget and complementary slackness also hold, which proves that the procedure gives the unique optimum.


### Example: two floor adjustments
Consider three preferred assets with a one-share floor and a budget of 150 USD. Their prices and preference weights are:

| Asset | Price (USD per share) | Preference weight |
|:--|--:|--:|
| 1 | 10 | 0.6 |
| 2 | 20 | 0.3 |
| 3 | 100 | 0.1 |

The floors cost 130 USD, so the problem is feasible and 20 USD can be invested above them. All assets are preferred, so the adjusted budget is the full 150 USD.

First divide the budget in proportion to the preference weights. The third holding falls below its floor. After pinning it at one share, 50 USD remains for the first two assets, and their next allocation makes the second holding fall below its floor:

| Solve | Budget for remaining holdings (USD) | Shares of asset 1 | Shares of asset 2 | Shares of asset 3 | Next action |
|:--|--:|--:|--:|--:|:--|
| 1 | 150 | 9 | 2.25 | 0.15 | Pin asset 3 at one share |
| 2 | 50 | 10/3 | 5/6 | 1 (pinned) | Pin asset 2 at one share |
| 3 | 30 | 3 | 1 (pinned) | 1 (pinned) | All floors satisfied |

The first two rows are candidate allocations whose floor violations must be corrected. The final allocation spends 30 USD, 20 USD, and 100 USD on the three assets. Only the first holding is above its floor, so its marginal log utility per dollar determines the multiplier:

$$
\lambda=\frac{0.6}{3\cdot10}=0.02.
$$

The floor multipliers follow from stationarity:

$$
\begin{align*}
\nu_1&=0,\\
\nu_2&=0.02(20)-\frac{0.3}{1}=0.1,\\
\nu_3&=0.02(100)-\frac{0.1}{1}=1.9.
\end{align*}
$$

All floor multipliers are nonnegative, and the only holding above its floor has a zero multiplier. The budget equals 150 USD, and complementary slackness holds. These checks establish that the final allocation is the unique maximum.

## Summary
We derived the Cobb–Douglas allocation from the full problem over share counts and cash. We then retained the share floors in the optimality conditions and showed why pinning a violating holding and solving again reaches the optimum.

> __Key Takeaways:__
>
> * __Preference signs settle part of the allocation:__ We showed that, when at least one asset is preferred, non-preferred assets stay at their floors and cash is fully invested. This left a reduced budget for preferred holdings.
> * __The budget determines proportional dollar holdings:__ We equated marginal log utility per dollar and used the budget to determine the common multiplier. The resulting dollar holdings were proportional to preference weights whenever the share floors allowed that split.
> * __Binding floors change the budget available to the remaining assets:__ We derived the floor optimality conditions and proved that pinning and solving again satisfies them after finitely many solves. The numerical example showed why correcting one floor can make another floor bind.
