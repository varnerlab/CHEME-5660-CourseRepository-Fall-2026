## Concept Review: A Bet on the Future
Suppose we pay a price $p$ today for a contract that pays 1 USD in $T$ years if a condition is true (here, an asset's share price $S(T)$ finishes at or above a target share price $K$), and nothing if it is false. How should we think about the price of this bet today?

The payoff (USD) of this bet at time $T$ is given by:

$$
V(S(T)) = \begin{cases}
1 & \text{if } S(T)\geq K\\
0 & \text{if } S(T)<K
\end{cases}
$$

This bet is an abstract asset with two cash flows: we pay $p$ at $t=0$ and receive $V(S(T))$ at $t=T$. Discounting the payoff at the benchmark growth rate $g_{y}$ gives the bet's net present value:

$$
\operatorname{NPV}_{0} = -p+\mathcal{D}^{-1}_{T,0}(g_{y})\,V(S(T))\qquad\text{where}\qquad\mathcal{D}^{-1}_{T,0}(g_{y})=e^{-g_{y}T}
$$

The payoff is uncertain, so this NPV is random. It is $-p+e^{-g_{y}T}$ if the condition is true and $-p$ if it is false. Under our beliefs about $S(T)$, the expected NPV is given by:

$$
\begin{align*}
\mathbb{E}\left[\operatorname{NPV}_{0}\right] &= \left(-p+e^{-g_{y}T}\right)\underbrace{\mathbb{P}(S(T)\geq K)}_{\text{true}}+(-p)\,\underbrace{\mathbb{P}(S(T)<K)}_{\text{false}} &&\quad\text{weight each NPV by its probability}\\
&= -p\,\underbrace{\left[\mathbb{P}(S(T)\geq K)+\mathbb{P}(S(T)<K)\right]}_{=1}+e^{-g_{y}T}\,\mathbb{P}(S(T)\geq K) &&\quad\text{collect the }p\text{ terms}\\
&= -p+e^{-g_{y}T}\,\underbrace{\mathbb{P}(S(T)\geq K)}_{\text{true}} &&\quad\text{the two probabilities sum to one}
\end{align*}
$$

The bet has a non-negative expected NPV when its price is at or below its discounted expected payoff, $p\leq e^{-g_{y}T}\,\mathbb{P}(S(T)\geq K)$. The bet is an example of a __derivative contract__: its payoff depends on the future price of another asset (or event). Each contract trades at a market price that every buyer pays, yet buyers with different beliefs value the same contract differently.

## What is a Derivative?
A derivative is a financial contract between a buyer and a seller that describes an agreement to transact an asset at a predetermined price on a specified future date. Thus, a derivative is an abstract asset whose cash flows depend on some underlying asset (or event).

Common types of derivatives include:
* __Futures, forwards, and swaps__ obligate both parties to transact. Futures and forwards are agreements to buy or sell an underlying asset at a predetermined price on a future date. Futures trade on exchanges with standard terms. Forwards trade over-the-counter (OTC), where the two parties negotiate the terms and each bears the other's default risk. Swaps exchange one asset's cash flows for another's and also trade OTC.
* __Options contracts__ give the contract buyer the right, but not the obligation, to buy (call option) or sell (put option) an asset at a specific price. Options are exchange-traded. The largest options exchange is the [Chicago Board Options Exchange (CBOE)](https://www.cboe.com).
* __Event contracts__ pay a fixed amount, usually one dollar, if a specified event occurs and nothing otherwise. [Exchanges such as Kalshi](https://kalshi.com) list them on economic data, weather, and financial index levels.

We will focus on options contracts in this lecture.

A __call__ option gives the owner the right, but not the obligation, to __buy__ 100 shares of an underlying asset (per contract) at a specific price, the strike price $K$, by a future date, the expiration. What is the business case for __long and short call options__?

> __Buyer (long)__. From the buyer's perspective, a long call contract allows:
> * The buyer to benefit from upside price movement of the underlying asset without purchasing the asset directly. Long calls have defined risk: the maximum loss for the buyer is the premium paid for the contract.
> * If the owner exercises the option, they can purchase shares at the strike price instead of the market price. 
>
> __Seller (short)__. From the seller's perspective, a short call contract offers:
> * The seller receives the contract premium for selling a call contract. Short call contracts have unlimited upside risk. Call options are often sold only by investors who already own the required number of shares as a covered call (100 shares per contract). 
> * Short call options obligate the seller to sell shares at the strike price $K$ if the option is exercised.

A __put__ option gives the owner the right, but not the obligation, to __sell__ 100 shares at the strike price $K$ by the expiration date. What is the business case for __long and short put options__?

> __Buyer (long)__. From the buyer's perspective, a long put contract allows:
> * The buyer to benefit from downside price movement of the underlying asset without selling it directly. Long puts have defined risk: the maximum loss for the buyer is the premium paid for the contract.
> * Long put options provide an opportunity to sell shares at the strike price $K$ instead of the market price $S$.
>
> __Seller (short)__. From the seller's perspective, a put contract offers:
> * The seller receives the contract premium for selling a put contract. Short put contracts have large but bounded downside risk: if the share price falls to zero, the seller loses the strike price minus the premium's future value per share. Put options are often sold only by investors with sufficient funds to purchase shares (100 shares per contract). This is a cash-secured put position.
> * Short put options obligate the seller to purchase shares at the strike price $K$ if the option is exercised.

The payoff per share of a single __call__ or __put__ contract at expiration (or when the contract is exercised) is:
$$
\boxed{
\begin{align*}
V_{c}(K,S(T)) & = \max\left(S(T) - K,~0\right) = \left(S(T) - K\right)^{+}\\
V_{p}(K,S(T)) & = \max\left(K - S(T),~0\right) = \left(K - S(T)\right)^{+}
\end{align*}}
$$
where $K$ denotes the strike price and $S(T)$ is the share price of the underlying asset at expiration (time $t=T$). The seller charges the buyer a premium per share, $\mathcal{P}_{c}(K,S(0))$ for a call and $\mathcal{P}_{p}(K,S(0))$ for a put.

The premium is paid at $t=0$ and the payoff arrives at $t=T$, so we compare them at expiration using the premium's future value, $\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}$, where $\mathcal{D}_{T,0}(g_{y})=e^{g_{y}T}$. The __buyer's profit per share__ for a single contract at expiration is the payoff minus the premium's future value:
$$
\begin{align*}
P_{c}^{\text{buyer}}(K,S(T)) & = V_{c}(K,S(T)) - \mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c}(K,S(0))\\
P_{p}^{\text{buyer}}(K,S(T)) & = V_{p}(K,S(T)) - \mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}(K,S(0))
\end{align*}
$$
A position's __orientation__ $\theta$ is $+1$ for the buyer (long) and $-1$ for the seller (short). Its profit per share is $\theta\cdot P^{\text{buyer}}$, so the seller's profit is $P^{\text{seller}}=(-1)\cdot P^{\text{buyer}}$, and $P^{\text{buyer}}+P^{\text{seller}}=0$ at every share price. Options are a __zero-sum game__: one party's gain is the other's loss.

Setting the buyer's profit to zero gives the __breakeven__ share price at expiration:
$$
\begin{align*}
S_{\text{BE}} & = K + \mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c}(K,S(0)) && \text{call}\\
S_{\text{BE}} & = K - \mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}(K,S(0)) && \text{put}
\end{align*}
$$
Above the call breakeven, and below the put breakeven, the buyer profits and the seller loses. Listed contracts often expire within weeks, so $\mathcal{D}_{T,0}(g_{y})$ is close to one (about 1.003 for 30 days at $g_{y}=4\%$). However, for longer-dated contracts, the accumulation factor can be significant: for example, a 1-year contract at $g_{y}=4\%$ has $\mathcal{D}_{T,0}(g_{y})=e^{0.04}\approx 1.041$.

## Pricing Options
What is an option's payoff worth today? The answer depends on when the holder can exercise:
* __European-style__ options can only be exercised on the expiration date.
* __American-style__ options can be exercised at any time before or on the expiration date. This creates extra risk for the contract seller, so American-style options cost at least as much as otherwise identical European-style options.
* __Bermudan-style__ options, a less common hybrid, can be exercised only on specified dates up to the expiration date.

Let's start with European-style options and compute their net present value.

### Options as Abstract Assets

A European option is an abstract asset with two cash flows: the buyer pays the premium $\mathcal{P}$ today and receives the payoff $V(K,S(T))\geq 0$ at expiration. Here, denote the payoff for a call contract by $V_{c}$ and for a put by $V_{p}$.

This is a share purchase with the sale price $S(T)$ replaced by the payoff, so the net present value at $t=0$ is:
$$
\begin{align*}
\operatorname{NPV}_{0} = -\mathcal{P} + \mathcal{D}^{-1}_{T,0}(g_{y})\cdot V(K,S(T))
\end{align*}
$$
Factoring out $\mathcal{D}^{-1}_{T,0}(g_{y})$ gives $\operatorname{NPV}_{0}=\mathcal{D}^{-1}_{T,0}(g_{y})\cdot P^{\text{buyer}}$: the NPV is the buyer's profit at expiration discounted to today, so it is zero at the breakeven share price.

### The European Premium
A Treasury's cash flows are known, so its price makes the NPV zero. An option's payoff is random, so no single premium makes its NPV zero in every outcome. Suppose instead that a __replicating portfolio__ of shares and the risk-free asset, traded over time, pays the option's payoff at expiration in every outcome. When the replicating trades have no transaction costs, any premium other than the portfolio's cost creates a risk-free profit:
* __Premium too low:__ Traders buy the option and sell a portfolio with the same payoff, built from shares and the risk-free asset.
* __Premium too high:__ Traders sell the option and buy that portfolio.
* __Premium at the portfolio's cost:__ Neither trade earns a risk-free profit, so the premium stays there.

We can write the portfolio's cost as an expected discounted payoff, computed with the __risk-neutral measure__ $\mathbb{Q}$, the probabilities under which the share's expected price grows at the risk-free rate $g_{y}$. Under $\mathbb{Q}$, the option's expected NPV is zero, so the premium of a European-style contract is given by:
$$
\boxed{
\begin{align*}
\mathbb{E}_{\mathbb{Q}}\bigl(\operatorname{NPV}_{0}\bigr) = 0
\quad\Longrightarrow\quad
\mathcal{P}^{\text{EU}} = \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\cdot V(K,S(T))\Bigr)
\end{align*}}
$$

__What does $\mathbb{Q}$ mean?__ It is a pricing rule, while the real-world probabilities $\mathbb{P}$ still describe how often outcomes occur:
* __The share:__ With no dividends, $\mathbb{E}_{\mathbb{Q}}\bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\,S(T)\bigr)=S(0)$, so the share also has zero expected NPV under $\mathbb{Q}$.
* __The option:__ Under $\mathbb{P}$, a call expires worthless with probability $\mathbb{P}(S(T)\leq K)$, from the terminal target formula, and the buyer then loses the whole premium.
* __The event contract:__ A contract that pays $V=\mathbb{1}\{\text{event}\}$, one dollar if the event occurs and nothing otherwise, has price $p=\mathbb{E}_{\mathbb{Q}}\bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{1}\{\text{event}\}\bigr)=\mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{Q}(\text{event})$. The pricing probability $\mathbb{Q}(\text{event})$ can differ from the real-world probability $\mathbb{P}(\text{event})$.

The bet in the Concept Review is the event contract on $S(T)\geq K$. Its market price is its discounted expected payoff with $\mathbb{Q}$ in place of $\mathbb{P}$, $p=\mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{Q}(S(T)\geq K)$. A general Kalshi event has no replicating portfolio, so choosing its pricing probability requires additional assumptions.

> __Why is the bet so cheap?__ Take a one-year bet that a share finishes at or above today's price. With 30% volatility, a 6% real-world mean growth rate, and a 4% risk-free rate, the bet pays off 58% of the time. A client with 50,000 USD and relative risk aversion 3 would pay 0.557 USD for it, yet the market price is only 0.474 USD. Where did the other 8 cents go?

### Black–Scholes–Merton Premiums
The price of a European option contract can be determined using the Black-Scholes-Merton (BSM) model.

> __Nobel Prize 1997:__ The 1997 Nobel Prize in Economic Sciences was awarded to Robert C. Merton and Myron S. Scholes for their work in developing the Black-Scholes-Merton model. Their contributions have had a profound impact on financial markets, risk management, and investment strategies worldwide. [See more here](https://www.nobelprize.org/prizes/economic-sciences/1997/summary/). Fischer Black was also a key contributor to the model but had passed away before the prize was awarded.

We assume that a non-dividend-paying share follows geometric Brownian motion with constant volatility $\sigma$, and that the continuously compounded risk-free rate $g_y$ is constant. Trading in the share and the risk-free asset is continuous and frictionless, with fractional holdings, short sales, and borrowing and lending at $g_y$ allowed. Under these assumptions, a trading strategy in the share and the risk-free asset replicates the European option's payoff.

__How does the replication work?__ Write the premium at time $t$, when the share price is $S$, as $\mathcal{P}(S,t)$. At expiration, it equals the payoff, $\mathcal{P}(S,T)=V(K,S)$. Under the BSM assumptions, the premium satisfies the __Black–Scholes–Merton equation__:
$$
\boxed{
\begin{align*}
\frac{\partial\mathcal{P}}{\partial t}+g_{y}\,S\,\frac{\partial\mathcal{P}}{\partial S}+\frac{1}{2}\,\sigma^{2}S^{2}\,\frac{\partial^{2}\mathcal{P}}{\partial S^{2}} = g_{y}\,\mathcal{P}
\end{align*}}
$$
The equation follows from a hedge, and its solution gives the premium:
* __Hedge:__ Hold one option and sell $\Delta=\partial\mathcal{P}/\partial S$ shares, the replicating portfolio's share holding. Over a short time interval, the random change in the option's price cancels the random change in the shares, so this portfolio is risk-free.
* __No arbitrage:__ A risk-free portfolio must grow at the risk-free rate $g_{y}$. Setting the hedged portfolio's growth equal to $g_{y}$ gives the equation. The share's real-world expected growth cancels in the hedge, so it appears in neither the equation nor the premium.
* __Solution:__ Solving the equation backward in time from the payoff gives today's premium, $\mathcal{P}(S(0),0)$. This solution equals the risk-neutral premium $\mathcal{P}^{\text{EU}}$.

The [contingent claims notebook](advanced/contingent_claims/CHEME-5660-L9a-Advanced-ContingentClaims-BSM-Fall-2026.ipynb) builds the replicating portfolio on a one-step lattice, derives the equation from the hedge, and shows that its solution is this expectation.

To compute the expectation under $\mathbb{Q}$, we can simulate the future share price using a [geometric Brownian motion model](https://en.wikipedia.org/wiki/Geometric_Brownian_motion) with drift $g_y$. Alternatively, we can use the [Black–Scholes pricing formula](https://en.wikipedia.org/wiki/Black–Scholes_model) for a European call and put option ($t = 0$):
$$
\boxed{
\begin{align*}
\mathcal{P}_{c}(K,S(0)) & = N(d_{+})\cdot{S}(0) - N(d_{-})\cdot{K}\cdot\mathcal{D}^{-1}_{T,0}(g_{y})\\
\mathcal{P}_{p}(K,S(0)) & = N(-d_{-})\cdot{K}\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) - N(-d_{+})\cdot{S}(0)
\end{align*}}
$$
where:
$$
\begin{align*}
d_{+} & = \frac{1}{\sigma\sqrt{T}}\left[\ln\left(\frac{S(0)}{K}\right) + \left(g_{y}+\frac{\sigma^{2}}{2}\right)T\right] \\
d_{-} & = d_{+} - \sigma\sqrt{T}
\end{align*}
$$
and $N(\dots)$ denotes the standard normal cumulative distribution function, $\sigma$ is the volatility of the underlying asset, $g_{y}$ is the risk-free rate, and $T$ is the time to expiration of the contract in years.

### Put–Call Parity
The call and put premiums are linked. For a call and a put with the same strike and expiration, the payoffs differ by $S(T)-K$ at every share price:
$$
\begin{align*}
V_{c}(K,S(T)) - V_{p}(K,S(T)) = \left(S(T)-K\right)^{+} - \left(K-S(T)\right)^{+} = S(T) - K
\end{align*}
$$
Subtracting the put premium condition from the call premium condition gives:
$$
\begin{align*}
\mathcal{P}_{c}(K,S(0)) - \mathcal{P}_{p}(K,S(0))
& = \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\cdot\bigl[V_{c}(K,S(T)) - V_{p}(K,S(T))\bigr]\Bigr) && \text{expectation is linear}\\
& = \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\cdot\bigl[S(T) - K\bigr]\Bigr) && \text{payoff identity}\\
& = S(0) - K\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) && \mathbb{E}_{\mathbb{Q}}\bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\,S(T)\bigr) = S(0)
\end{align*}
$$
This is __put–call parity__:
$$
\boxed{
\begin{align*}
\mathcal{P}_{c}(K,S(0)) - \mathcal{P}_{p}(K,S(0)) = S(0) - K\cdot\mathcal{D}^{-1}_{T,0}(g_{y})
\end{align*}}
$$
Parity uses only the payoff identity and the share's zero expected NPV, so it holds whatever model describes the share price under $\mathbb{Q}$.

> __What about American-style options?__ The holder can exercise at any time $\tau\leq T$, chosen using only the information available at that time. The premium is the largest expected payoff, discounted from $\tau$ back to today, over all such exercise times:
> $$
> \begin{align*}
> \mathcal{P}^{\text{AM}} = \sup_{\tau\leq T} \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{\tau,0}(g_{y})\cdot V(K,S(\tau))\Bigr) \geq \mathcal{P}^{\text{EU}}
> \end{align*}
> $$
> The American put has no closed-form formula like BSM, so its premium is usually computed on a binomial lattice.

## Summary
In this lecture, we've explored derivatives, a fundamental asset class used for risk management, hedging, and speculation.

> __Key Takeaways:__
> 
> * __Derivatives derive their value from underlying assets__ such as stocks, bonds, commodities, or currencies. Common types include futures, forwards, swaps, and options, each with different characteristics, maturities, and uses.
> * __Option payoffs are determined by strike prices and contract orientation__ (long/short calls or puts). Understanding payoff diagrams at expiration helps in analyzing option profitability and risk exposure.
> * __An option's premium is the price that makes its expected net present value zero__ under the risk-neutral measure. For European contracts, the Black–Scholes–Merton formulas compute this premium, and put–call parity links the call and put premiums.

Understanding derivative mechanics, payoff structures, and pricing is important for financial decision-making.
