## Concept Review: A Bet on the Future
Suppose we pay a price $p$ today for a contract that pays 1 USD in $T$ years if a condition is true (here, an asset's share price $S(T)$ finishes at or above a target share price $K$), and nothing if it is false. How should we think about the price of this bet today?

The payoff (USD) of this bet at time $T$ is given by:

$$
V(S(T)) = \begin{cases}
1 & \text{if } S(T)\geq K\\
0 & \text{if } S(T)<K
\end{cases}
$$

This bet is an abstract asset with two cash flows: we pay $p$ at $t=0$ and receive $V(S(T))$ at $t=T$. Discounting the payoff at the benchmark growth rate $g_{y}$ gives the bet's net present value:

$$
\operatorname{NPV}_{0} = -p+\mathcal{D}^{-1}_{T,0}(g_{y})\,V(S(T))\qquad\text{where}\qquad\mathcal{D}^{-1}_{T,0}(g_{y})=e^{-g_{y}T}
$$

The payoff is uncertain, so this NPV is random. It is $-p+e^{-g_{y}T}$ if the condition is true and $-p$ if it is false. Under our beliefs about $S(T)$, the expected NPV is given by:

$$
\begin{align*}
\mathbb{E}\left[\operatorname{NPV}_{0}\right] &= \left(-p+e^{-g_{y}T}\right)\underbrace{\mathbb{P}(S(T)\geq K)}_{\text{true}}+(-p)\,\underbrace{\mathbb{P}(S(T)<K)}_{\text{false}} &&\quad\text{weight each NPV by its probability}\\
&= -p\,\underbrace{\left[\mathbb{P}(S(T)\geq K)+\mathbb{P}(S(T)<K)\right]}_{=1}+e^{-g_{y}T}\,\mathbb{P}(S(T)\geq K) &&\quad\text{collect the }p\text{ terms}\\
&= -p+e^{-g_{y}T}\,\underbrace{\mathbb{P}(S(T)\geq K)}_{\text{true}} &&\quad\text{the two probabilities sum to one}
\end{align*}
$$

The bet has a non-negative expected NPV when its price is at or below its discounted expected payoff, $p\leq e^{-g_{y}T}\,\mathbb{P}(S(T)\geq K)$. The bet is an example of a __derivative contract__: its payoff depends on the future price of another asset (or event). Each contract trades at a market price that every buyer pays, yet buyers with different beliefs value the same contract differently.

## What is a Derivative?
A derivative is a financial contract between a buyer and a seller that describes an agreement to transact an asset at a predetermined price on a specified future date. Thus, a derivative is an abstract asset whose cash flows depend on some underlying asset (or event).

Common types of derivatives include:
* __Futures, forwards, and swaps__ obligate both parties to transact. Futures and forwards are agreements to buy or sell an underlying asset at a predetermined price on a future date. Futures trade on exchanges with standard terms. Forwards trade over-the-counter (OTC), where the two parties negotiate the terms and each bears the other's default risk. Swaps exchange one asset's cash flows for another's and also trade OTC.
* __Options contracts__ give the contract buyer the right, but not the obligation, to buy (call option) or sell (put option) an asset at a specific price. Options are exchange-traded. The largest options exchange is the [Chicago Board Options Exchange (Cboe)](https://www.cboe.com).
* __Event contracts__ pay a fixed amount, usually one dollar, if a specified event occurs and nothing otherwise. [Exchanges such as Kalshi](https://kalshi.com) list them on economic data, weather, and financial index levels.

We will focus on options contracts in this lecture.

A __call__ option gives the owner the right, but not the obligation, to __buy__ 100 shares of an underlying asset (per contract) at a specific price, the strike price $K$, by a future date, the expiration. What is the business case for __long and short call options__?

> __Buyer (long)__. From the buyer's perspective, a long call contract allows:
> * The buyer to benefit from upside price movement of the underlying asset without purchasing the asset directly. Long calls have defined risk: the maximum loss for the buyer is the premium paid for the contract.
> * If the owner exercises the option, they can purchase shares at the strike price instead of the market price. 
>
> __Seller (short)__. From the seller's perspective, a short call contract offers:
> * The seller receives the contract premium for selling a call contract. Short call contracts have unlimited upside risk. Call options are often sold only by investors who already own the required number of shares as a covered call (100 shares per contract). 
> * Short call options obligate the seller to sell shares at the strike price $K$ if the option is exercised.

A __put__ option gives the owner the right, but not the obligation, to __sell__ 100 shares at the strike price $K$ by the expiration date. What is the business case for __long and short put options__?

> __Buyer (long)__. From the buyer's perspective, a long put contract allows:
> * The buyer to benefit from downside price movement of the underlying asset without selling it directly. Long puts have defined risk: the maximum loss for the buyer is the premium paid for the contract.
> * Long put options provide an opportunity to sell shares at the strike price $K$ instead of the market price.
>
> __Seller (short)__. From the seller's perspective, a put contract offers:
> * The seller receives the contract premium for selling a put contract. Short put contracts have large but bounded downside risk: if the share price falls to zero, the seller loses the strike price minus the premium's future value per share. Put options are often sold only by investors with sufficient funds to purchase shares (100 shares per contract). This is a cash-secured put position.
> * Short put options obligate the seller to purchase shares at the strike price $K$ if the option is exercised.

The payoff per share of a single __call__ or __put__ contract at expiration (or when the contract is exercised) is:
$$
\boxed{
\begin{align*}
V_{c}(K,S(T)) & = \max\left(S(T) - K,~0\right) = \left(S(T) - K\right)^{+}\\
V_{p}(K,S(T)) & = \max\left(K - S(T),~0\right) = \left(K - S(T)\right)^{+}
\end{align*}}
$$
where $K$ denotes the strike price and $S(T)$ is the share price of the underlying asset at expiration (time $t=T$). The seller charges the buyer a premium per share, $\mathcal{P}_{c}(K,S(0))$ for a call and $\mathcal{P}_{p}(K,S(0))$ for a put.

The premium is paid at $t=0$ and the payoff arrives at $t=T$, so we compare them at expiration using the premium's future value, $\mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}$, where $\mathcal{D}_{T,0}(g_{y})=e^{g_{y}T}$. The __buyer's profit per share__ for a single contract at expiration is the payoff minus the premium's future value:
$$
\begin{align*}
P_{c}^{\text{buyer}}(K,S(T)) & = V_{c}(K,S(T)) - \mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c}(K,S(0))\\
P_{p}^{\text{buyer}}(K,S(T)) & = V_{p}(K,S(T)) - \mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}(K,S(0))
\end{align*}
$$
A position's __orientation__ $\theta$ is $+1$ for the buyer (long) and $-1$ for the seller (short). Its profit per share is $\theta\cdot P^{\text{buyer}}$, so the seller's profit is $P^{\text{seller}}=(-1)\cdot P^{\text{buyer}}$, and $P^{\text{buyer}}+P^{\text{seller}}=0$ at every share price. Options are a __zero-sum game__: one party's gain is the other's loss.

Setting the buyer's profit to zero gives the __breakeven__ share price $S^{\ddagger}$ at expiration:
$$
\begin{align*}
S^{\ddagger} & = K + \mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{c}(K,S(0)) && \text{call}\\
S^{\ddagger} & = K - \mathcal{D}_{T,0}(g_{y})\cdot\mathcal{P}_{p}(K,S(0)) && \text{put}
\end{align*}
$$
Above the call breakeven, and below the put breakeven, the buyer profits and the seller loses. Listed contracts often expire within weeks, so $\mathcal{D}_{T,0}(g_{y})$ is close to one (about 1.003 for 30 days at $g_{y}=4\%$). However, for longer-dated contracts, the accumulation factor can be significant: for example, a 1-year contract at $g_{y}=4\%$ has $\mathcal{D}_{T,0}(g_{y})=e^{0.04}\approx 1.041$.

## Pricing Put and Call Options Contracts

An option's price depends on the underlying asset's value, strike price, time to expiration, and exercise style.

__European-style__ options allow exercise only at expiration. __American-style__ options allow it any time through expiration. The less common __Bermudan-style__ options allow exercise on specified dates. The American option's added flexibility makes it worth at least as much as an otherwise identical European option.

Options on single stocks and on exchange-traded funds such as SPY are American-style. The European-style contracts we can trade are cash-settled index options, such as the S&P 500 options in the Cboe profile ([SPX](https://www.cboe.com/tradable-products/sp-500/spx-options), with SPXW for weekly expirations), the Mini-SPX options at one tenth the size ([XSP](https://www.cboe.com/tradable_products/sp_500/mini_spx_options/specifications)), the Nasdaq-100 options ([NDX](https://www.nasdaq.com/nasdaq-100-index-options)), and the Russell 2000 options ([RUT](https://www.cboe.com/tradable-products/ftse-russell/russell-2000-index-options)).

Let's start with European-style options and compute their net present value.

### Options as Abstract Assets

A European option is an abstract asset with two cash flows: the buyer pays the premium $\mathcal{P}$ today and receives the payoff $V(K,S(T))\geq 0$ at expiration.

This is a share purchase with the premium and the payoff in place of the purchase and sale prices, so the net present value at $t=0$ is:
$$
\begin{align*}
\operatorname{NPV}_{0} = -\mathcal{P} + \mathcal{D}^{-1}_{T,0}(g_{y})\cdot V(K,S(T))
\end{align*}
$$
Factoring out $\mathcal{D}^{-1}_{T,0}(g_{y})$ gives $\operatorname{NPV}_{0}=\mathcal{D}^{-1}_{T,0}(g_{y})\cdot P^{\text{buyer}}$: the NPV is the buyer's profit at expiration discounted to today, so it is zero at the breakeven share price.

### The European Premium
__How is the market price set?__ A call is a bet whose payoff grows with the share price. As with the bet in the Concept Review, we pay the premium $\mathcal{P}$ today. At expiration, the call pays $S(T)-K$ instead of 1 USD when $S(T)>K$, and nothing otherwise. Its NPV is $-\mathcal{P}+\mathcal{D}^{-1}_{T,0}(g_{y})\,(S(T)-K)$ if $S(T)>K$ and $-\mathcal{P}$ otherwise. Following the same steps as for the bet, the expected NPV is given by:
$$
\begin{align*}
\mathbb{E}\left[\operatorname{NPV}_{0}\right] &= \left(-\mathcal{P}+\mathcal{D}^{-1}_{T,0}(g_{y})\,\bar{V}\right)\mathbb{P}(S(T)>K)+(-\mathcal{P})\,\mathbb{P}(S(T)\leq K) &&\quad\text{weight each NPV by its probability}\\
&= -\mathcal{P}\,\underbrace{\left[\mathbb{P}(S(T)>K)+\mathbb{P}(S(T)\leq K)\right]}_{=1}+\mathcal{D}^{-1}_{T,0}(g_{y})\,\bar{V}\,\mathbb{P}(S(T)>K) &&\quad\text{collect the }\mathcal{P}\text{ terms}\\
&= -\mathcal{P}+\mathcal{D}^{-1}_{T,0}(g_{y})\,\mathbb{P}(S(T)>K)\,\bar{V} &&\quad\text{the two probabilities sum to one}
\end{align*}
$$
where $\bar{V}$ is the average payoff $S(T)-K$ over the outcomes with $S(T)>K$. The expected NPV is zero when the premium is the discounted chance of exercise times the average payoff when exercised, $\mathcal{P}=\mathcal{D}^{-1}_{T,0}(g_{y})\,\mathbb{P}(S(T)>K)\,\bar{V}$. 

__Whose probabilities?__ The probabilities $\mathbb{P}$ are our own beliefs about the future share price $S(T)$, so this premium is the most we would pay. A buyer with different beliefs computes a different premium. The market instead prices every option with one shared set of probabilities, the __risk-neutral pricing rule__ $\mathbb{Q}$, so there is one price.

Under $\mathbb{Q}$, the expected share price grows at the risk-free rate $g_{y}$. For geometric Brownian motion with no dividends, $\mathbb{Q}$ replaces the drift with $g_{y}$. Repeating the expected NPV calculation with $\mathbb{Q}$ in place of $\mathbb{P}$ gives:
$$
\begin{align*}
\mathbb{E}_{\mathbb{Q}}\left[\operatorname{NPV}_{0}\right] &= \left(-\mathcal{P}+\mathcal{D}^{-1}_{T,0}(g_{y})\,\bar{V}\right)\mathbb{Q}(S(T)>K)+(-\mathcal{P})\,\mathbb{Q}(S(T)\leq K) &&\quad\text{weight each NPV by its probability}\\
&= -\mathcal{P}\,\underbrace{\left[\mathbb{Q}(S(T)>K)+\mathbb{Q}(S(T)\leq K)\right]}_{=1} &&\quad\text{collect terms, the call pays 0 if }S(T)\leq K\\
&\qquad+\mathcal{D}^{-1}_{T,0}(g_{y})\,\underbrace{\left[\bar{V}\,\mathbb{Q}(S(T)>K)+0\cdot\mathbb{Q}(S(T)\leq K)\right]}_{=\mathbb{E}_{\mathbb{Q}}\left[V_{c}(K,S(T))\right]}\\
&= -\mathcal{P}+\mathcal{D}^{-1}_{T,0}(g_{y})\,\mathbb{E}_{\mathbb{Q}}\left[V_{c}(K,S(T))\right] &&\quad\text{the two probabilities sum to one}
\end{align*}
$$
where $\bar{V}$ is now the average payoff $S(T)-K$ over the outcomes with $S(T)>K$ under $\mathbb{Q}$. The second bracket is the expected payoff of the call under $\mathbb{Q}$: the payoff in each outcome, weighted by its probability. Setting $\mathbb{E}_{\mathbb{Q}}\left[\operatorname{NPV}_{0}\right]=0$ gives the market premium of the call, $\mathcal{P}_{c}=\mathcal{D}^{-1}_{T,0}(g_{y})\,\mathbb{E}_{\mathbb{Q}}\left[V_{c}(K,S(T))\right]$.

Repeating these steps for a put, which pays $K-S(T)$ when $S(T)<K$, gives the same result with $V_{p}$ in place of $V_{c}$. For either contract, with payoff $V(K,S(T))$, the premium of a European-style contract is the discounted expected payoff under $\mathbb{Q}$:
$$
\boxed{
\begin{align*}
\mathcal{P}^{\text{EU}} = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{E}_{\mathbb{Q}}\left[V(K,S(T))\right]
\end{align*}}
$$

The bet in the Concept Review is an __event contract__ on $S(T)\geq K$, which pays $V=\mathbb{1}\{\text{event}\}$, one dollar if the event occurs and nothing otherwise. Its market price is:
$$
\begin{align*}
p = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{E}_{\mathbb{Q}}\left[\mathbb{1}\{\text{event}\}\right] = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{Q}(\text{event})
\end{align*}
$$
The pricing probability $\mathbb{Q}(\text{event})$ can differ from the real-world probability $\mathbb{P}(\text{event})$. A general Kalshi event, such as an election, is not tied to a share price, so choosing its pricing probability requires additional assumptions.

__A Monte Carlo premium.__ Simulation gives a direct way to estimate the market premium. The payoff depends only on the share price at expiration, so we can draw $S(T)$ directly from the solution of geometric Brownian motion, with drift $g_{y}$. Let's outline some pseudocode for the Monte Carlo premium of a European contract.

__Initialize__: Given the share price $S(0)$, the strike price $K$, the payoff function $V(K,S)$ of the call or put, the time to expiration $T$ (units: years), the volatility $\sigma$, the risk-free rate $g_{y}$, and the number of simulated paths $M$.

1. For each path $i = 1,2,\dots,M$ __do__:
   - a. Draw a standard normal random number $Z_{i}$.
   - b. Compute the share price at expiration: $S_{i}(T) \gets S(0)\cdot\exp\Bigl[\left(g_{y}-\frac{\sigma^{2}}{2}\right)T + \sigma\sqrt{T}\,Z_{i}\Bigr]$.
   - c. Compute the payoff: $V_{i} \gets V(K,S_{i}(T))$.
2. The premium is the discounted average payoff: $\mathcal{P} \gets \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\frac{1}{M}\sum_{i=1}^{M}V_{i}$.

There is an alternative way to think about the premium, which the [contingent claims notebook](advanced/contingent_claims/CHEME-5660-L9a-Advanced-ContingentClaims-BSM-Fall-2026.ipynb) works through.

### Black–Scholes–Merton (BSM) Premiums
The price of a European option contract can be determined using the Black–Scholes–Merton (BSM) model.

> __Nobel Prize 1997:__ The 1997 Nobel Prize in Economic Sciences was awarded to Robert C. Merton and Myron S. Scholes for their work in developing the Black–Scholes–Merton model. Their contributions have had a profound impact on financial markets, risk management, and investment strategies worldwide. [See more here](https://www.nobelprize.org/prizes/economic-sciences/1997/summary/). Fischer Black was also a key contributor to the model but had passed away before the prize was awarded.

We assume that the price of a non-dividend-paying share follows geometric Brownian motion with constant volatility $\sigma$, and that the continuously compounded risk-free rate $g_y$ is constant.

__The Black–Scholes–Merton (BSM) equation.__ Write the premium at time $t$, when the share price is $S$, as $\mathcal{P}(S,t)$. At expiration, it equals the payoff, $\mathcal{P}(S,T)=V(K,S)$. Under the BSM assumptions, the premium satisfies the __Black–Scholes–Merton (BSM) equation__:
$$
\boxed{
\begin{align*}
\frac{\partial\mathcal{P}}{\partial t}+g_{y}\,S\,\frac{\partial\mathcal{P}}{\partial S}+\frac{1}{2}\,\sigma^{2}S^{2}\,\frac{\partial^{2}\mathcal{P}}{\partial S^{2}} = g_{y}\,\mathcal{P}
\end{align*}}
$$
Only risk-free rate $g_{y}$ and volatility $\sigma$ appear in the equation, not the real-world growth rate of the share price. Solving it backward in time from the payoff gives today's premium, $\mathcal{P}(S(0),0)$, and this solution equals the risk-neutral premium $\mathcal{P}^{\text{EU}}$ that the Monte Carlo simulation estimates. The [contingent claims notebook](advanced/contingent_claims/CHEME-5660-L9a-Advanced-ContingentClaims-BSM-Fall-2026.ipynb) derives the equation and shows that its solution is this expectation.

Instead of simulating, we can compute this premium with the [Black–Scholes pricing formula](https://en.wikipedia.org/wiki/Black–Scholes_model) for a European call and put option ($t = 0$):
$$
\boxed{
\begin{align*}
\mathcal{P}_{c}(K,S(0)) & = N(d_{+})\cdot{S}(0) - N(d_{-})\cdot{K}\cdot\mathcal{D}^{-1}_{T,0}(g_{y})\\
\mathcal{P}_{p}(K,S(0)) & = N(-d_{-})\cdot{K}\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) - N(-d_{+})\cdot{S}(0)
\end{align*}}
$$
where:
$$
\begin{align*}
d_{+} & = \frac{1}{\sigma\sqrt{T}}\left[\ln\left(\frac{S(0)}{K}\right) + \left(g_{y}+\frac{\sigma^{2}}{2}\right)T\right] \\
d_{-} & = d_{+} - \sigma\sqrt{T}
\end{align*}
$$
and $N(\dots)$ denotes the standard normal cumulative distribution function, $\sigma$ is the volatility of the underlying asset, $g_{y}$ is the risk-free rate, and $T$ is the time to expiration of the contract in years.

__Which volatility $\sigma$ should we use?__ If we knew the right volatility, the BSM model (or the Monte Carlo model) would return the observed market premium. 
However, in practice we observe the market premium and want to back out the volatility that makes the model match the market. This is called __implied volatility__.

> __Historical or implied volatility?__ There are two ways to put a number on $\sigma$, and the second also gives a quick range for the share price:
> * __Historical volatility:__ Previously, we estimated $\sigma$ from past share prices as $\sigma_{\mathrm{ann}}=\sqrt{\Delta t}\cdot\sigma_{g}$, where $\sigma_{g}$ is the standard deviation of the growth rates (units: $\mathrm{yr}^{-1}$) and $\Delta t$ is the time between price observations (units: years). This estimate looks backward.
> * __Implied volatility:__ The [implied volatility](https://en.wikipedia.org/wiki/Implied_volatility) (IV) is the market's forecast of a likely movement in the share price. The implied volatility is often equal to or larger than the historical volatility.
> * __How can we use it?__ The implied volatility represents a one-standard-deviation movement in the share price over the next year. Over a horizon of $T$ years, the size of that move scales with $\sqrt{T}$.
>
> Moving one standard deviation up or down in the solution of geometric Brownian motion gives a back-of-the-envelope range for the share price:
> $$
> \begin{align*}
> S(T) & = S(0)\exp\left[\mu_{g}T + \sigma\sqrt{T}\,Z\right] && \text{solution of geometric Brownian motion}\\
> S_{\pm} & = S(0)\exp\left[\mu_{g}T \pm \text{IV}\sqrt{T}\right] && \text{set } \sigma=\text{IV} \text{ and } Z=\pm 1\\
> & \approx S(0)\exp\left[\pm\,\text{IV}\sqrt{T}\right] && \mu_{g}T \text{ is small next to } \text{IV}\sqrt{T} \text{ for short } T\\
> & \approx S(0)\left(1 \pm \text{IV}\sqrt{T}\right) && e^{x}\approx 1+x \text{ for small } x
> \end{align*}
> $$
> where $\mu_{g}$ is the mean growth rate, $Z$ is a standard normal random variable, and $S_{\pm}$ is the share price one standard deviation above or below $S(0)$. For example, a share at $S(0)$ = 100 USD/share with IV = 40% and 30 days to expiration ($T$ = 30/365 years) has $\text{IV}\sqrt{T}$ = 0.115, so the range is about 88.5 to 111.5 USD/share. Under the model, the share price ends inside this range about 68% of the time.

Let's outline some pseudocode for backing out the implied volatility by bisection.

__Initialize__: Given the market premium $\mathcal{P}^{\text{market}}$, the share price $S(0)$, the strike price $K$, the time to expiration $T$ (units: years), the risk-free rate $g_{y}$, a volatility bracket $\sigma_{\text{lo}}<\sigma_{\text{hi}}$ whose BSM premiums satisfy $\mathcal{P}(\sigma_{\text{lo}})\leq\mathcal{P}^{\text{market}}\leq\mathcal{P}(\sigma_{\text{hi}})$, a tolerance $\epsilon>0$, a maximum number of iterations $\texttt{maxiter}$, and an iteration counter $i\gets 1$.

While $\sigma_{\text{hi}}-\sigma_{\text{lo}}>\epsilon$ and $i\leq\texttt{maxiter}$ __do__:
1. Compute the midpoint $\sigma_{\text{mid}}\gets(\sigma_{\text{lo}}+\sigma_{\text{hi}})/2$ and its BSM premium $\mathcal{P}(\sigma_{\text{mid}})$.
2. If $\mathcal{P}(\sigma_{\text{mid}})<\mathcal{P}^{\text{market}}$, __then__ $\sigma_{\text{lo}}\gets\sigma_{\text{mid}}$. Otherwise, $\sigma_{\text{hi}}\gets\sigma_{\text{mid}}$.
3. Update the iteration counter $i\gets i+1$.

The implied volatility is the midpoint of the final bracket, $\hat{\sigma}\gets(\sigma_{\text{lo}}+\sigma_{\text{hi}})/2$. The BSM premium rises with $\sigma$, so exactly one $\sigma$ matches the market premium, and each step halves the bracket around it. We use the BSM formula here rather than the Monte Carlo premium, because the Monte Carlo premium changes with the random draws and costs $M$ simulated paths at every step.

> __Advanced example__: The [SPXW volatility skew notebook](advanced/spxw_volatility_skew/CHEME-5660-L9a-Advanced-SPXW-Volatility-Skew-Fall-2026.ipynb) applies this bisection method to a chain of S&P 500 index options. Each strike and expiration has its own implied volatility, although the BSM model assumes one constant $\sigma$ (this is what is known as the "volatility smile" or "volatility skew"). To price a contract without a quote, or on a future date, we can predict its implied volatility from a surface fitted to today's quotes across strikes and expirations ([Sun et al., 2026](https://arxiv.org/abs/2605.13998)).

### Put–Call Parity for European Contracts
The call and put premiums are linked. For a European call and put with the same strike and expiration, the payoffs differ by $S(T)-K$ at every share price:
$$
\begin{align*}
V_{c}(K,S(T)) - V_{p}(K,S(T)) = \left(S(T)-K\right)^{+} - \left(K-S(T)\right)^{+} = S(T) - K
\end{align*}
$$
Subtracting the put premium from the call premium gives:
$$
\begin{align*}
\mathcal{P}_{c}(K,S(0)) - \mathcal{P}_{p}(K,S(0))
& = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{E}_{\mathbb{Q}}\left[V_{c}(K,S(T)) - V_{p}(K,S(T))\right] && \text{premium formula, expectation is linear}\\
& = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{E}_{\mathbb{Q}}\left[S(T) - K\right] && \text{payoff identity}\\
& = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\left(S(0)\,e^{g_{y}T} - K\right) && \text{expected share price grows at }g_{y}\text{ under }\mathbb{Q}\\
& = S(0) - K\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) && \mathcal{D}^{-1}_{T,0}(g_{y})=e^{-g_{y}T}
\end{align*}
$$
This is __put–call parity__:
$$
\boxed{
\begin{align*}
\mathcal{P}_{c}(K,S(0)) - \mathcal{P}_{p}(K,S(0)) = S(0) - K\cdot\mathcal{D}^{-1}_{T,0}(g_{y})
\end{align*}}
$$
__Put–call parity holds only for European contracts.__ The derivation compares the two payoffs on the expiration date, but an American holder can exercise before then. For American contracts, the difference between the call and put premiums is bounded, not fixed. For European contracts, parity uses only the payoff identity and the rule that the expected share price grows at $g_{y}$ under $\mathbb{Q}$, so it holds whatever model describes the share price under $\mathbb{Q}$.

In this example, we compute European call and put premiums by Monte Carlo simulation and with the Black–Scholes–Merton formulas. We then look at how the premium changes with strike price.

Now that we can compute the premium of a European option, let's look at the American option.

> __What about American-style options?__ The holder can exercise at any time $\tau\leq T$, chosen using only the information available at that time. The premium is the largest expected payoff, discounted from $\tau$ back to today, over all such exercise times:
> $$
> \begin{align*}
> \mathcal{P}^{\text{AM}} = \max_{\tau\leq T} \mathbb{E}_{\mathbb{Q}}\Bigl[\mathcal{D}^{-1}_{\tau,0}(g_{y})\cdot V(K,S(\tau))\Bigr] \geq \mathcal{P}^{\text{EU}}
> \end{align*}
> $$
> Many texts write $\sup$, the least upper bound, in place of $\max$ to cover contracts with no best exercise time. The American put has no closed-form formula like BSM, so its premium is usually computed on a binomial lattice.

## Summary
In this lecture, we've explored derivatives, a fundamental asset class used for risk management, hedging, and speculation.

> __Key Takeaways:__
> 
> * __Derivatives derive their value from underlying assets__ such as stocks, bonds, commodities, or currencies. Common types include futures, forwards, swaps, and options, each with different characteristics, maturities, and uses.

> * __An option's premium is the price that makes its expected net present value zero__ under the risk-neutral pricing rule. For European contracts, Monte Carlo simulation or the Black–Scholes–Merton formulas compute this premium, and put–call parity links the call and put premiums.

Understanding derivative mechanics, payoff structures, and pricing is important for financial decision-making.
