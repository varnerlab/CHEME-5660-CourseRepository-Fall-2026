## Replication on a One-Step Lattice
An abstract asset with random cash flows is described by what it pays in every state of the world. Suppose two assets pay the same amount in every state at every date. If their prices differed, we could buy the cheaper one and sell the other. The future cash flows would cancel in every state, and the price difference would be a profit today with no risk.

> __Law of one price:__ In an arbitrage-free market, two assets with identical cash flows in every state have the same price, and therefore the same NPV.

To price an option, we build a portfolio of assets whose prices we know, with the same cash flows as the option in every state. The option's premium is then the portfolio's cost.

Consider a one-step lattice for the share price. Over one step, the share price moves from $S(0)$ to $uS(0)$ (up) or $dS(0)$ (down), and one dollar in the risk-free asset grows to $R_f$. A claim pays $H_u$ in the up state and $H_d$ in the down state. Let's hold $\Delta$ shares and $B$ dollars in the risk-free asset, and choose them so that the portfolio pays what the claim pays in both states:
$$
\begin{align*}
\Delta\,uS(0) + B\,R_f &= H_u && \text{up state}\\
\Delta\,dS(0) + B\,R_f &= H_d && \text{down state}
\end{align*}
$$
Solving the two equations gives the holdings:
$$
\begin{align*}
\Delta &= \frac{H_u - H_d}{(u-d)\,S(0)} && \text{subtract down from up, }B\text{ cancels}\\
B &= \frac{uH_d - dH_u}{(u-d)\,R_f} && \text{substitute }\Delta\text{ into the down state}
\end{align*}
$$
By the law of one price, the claim's price $V_0$ is the portfolio's cost $\Delta S(0) + B$:
$$
\begin{align*}
V_0 &= \frac{H_u - H_d}{u-d} + \frac{uH_d - dH_u}{(u-d)\,R_f} && \text{cost of }\Delta\text{ shares and }B\text{ dollars}\\
&= \frac{1}{R_f}\left[\frac{R_f-d}{u-d}\,H_u + \frac{u-R_f}{u-d}\,H_d\right] && \text{common denominator, collect }H_u\text{ and }H_d\\
&= \frac{1}{R_f}\Bigl[q\,H_u + (1-q)\,H_d\Bigr], \qquad q=\frac{R_f-d}{u-d} && \text{so }1-q=\frac{u-R_f}{u-d}
\end{align*}
$$
This is the risk-neutral pricing rule, which we previously stated without a derivation. Three points follow:
* __The real-world up probability never appears.__ The portfolio matches the claim in both states, whatever their probabilities, so the price does not depend on them.
* __No arbitrage requires $d<R_f<u$.__ If $R_f\leq d$, borrowing to buy the share never loses and gains in the up state. If $R_f\geq u$, selling the share short and lending the proceeds never loses and gains in the down state. With $d<R_f<u$, the weight $q$ lies strictly between $0$ and $1$.
* __The share is priced by the same rule.__ With $H_u=uS(0)$ and $H_d=dS(0)$, the rule returns $S(0)$, because $qu+(1-q)d=R_f$.

### State Prices and Event Contracts
The pricing rule is linear in the payoffs. Let's define the __state prices__:
$$
\begin{align*}
\psi_u = \frac{q}{R_f}, \qquad \psi_d = \frac{1-q}{R_f}
\end{align*}
$$
The price of any claim is then the inner product of the state prices with its payoffs:
$$
\begin{align*}
V_0 = \psi_u H_u + \psi_d H_d = \bigl\langle \boldsymbol{\psi},\mathbf{H}\bigr\rangle
\end{align*}
$$
NPV is an inner product over dates, with each cash flow multiplied by its discount factor. A state price plays the same role across states: $\psi_u$ is the price today of one dollar paid only in the up state.

__Event contracts are the simplest claims.__ A YES contract on the event "the share moves up" pays one dollar in the up state and nothing otherwise, so its price is $\psi_u$. The NO contract pays one dollar in the down state, so its price is $\psi_d$. Holding both pays one dollar in every state, which is a zero-coupon bond:
$$
\begin{align*}
\psi_u + \psi_d = \frac{q + (1-q)}{R_f} = \frac{1}{R_f}
\end{align*}
$$
This is put–call parity in miniature: when two claims' payoffs add to a known amount, their prices add to that amount's present value.

## The Black–Scholes–Merton Formulas
The one-step rule extends to a lattice with $n$ steps of length $\Delta t=T/n$. We replicate the claim one step at a time, backward from expiration. The price is then the discounted expectation of the payoff under $\mathbb{Q}$, the law that moves up with probability $q$ at every step. Previously, an optional notebook showed that as the steps shrink, a binomial lattice approaches geometric Brownian motion. We state the result under $\mathbb{Q}$ without proof.

> __Proposition (lattice limit under $\mathbb{Q}$):__ Let $u=e^{\sigma\sqrt{\Delta t}}$, $d=1/u$, and $R_f=e^{g_{y}\Delta t}$. As $n\to\infty$, the distribution of the share price at expiration under $\mathbb{Q}$ approaches:
> $$
> S(T) = S(0)\exp\Bigl(\bigl(g_{y}-\tfrac{\sigma^{2}}{2}\bigr)T + \sigma\sqrt{T}\,Z\Bigr), \qquad Z\sim\mathcal{N}(0,1)
> $$
> and the price of a European claim with payoff $H(S(T))$ approaches $\mathcal{D}^{-1}_{T,0}(g_{y})\,\mathbb{E}_{\mathbb{Q}}\bigl(H(S(T))\bigr)$.

Under $\mathbb{P}$, the GBM model has the same form with mean growth rate $\mu_g$. Under $\mathbb{Q}$, the drift is the risk-free rate $g_{y}$, so the mean growth rate is $g_{y}-\sigma^{2}/2$.

### The Black–Scholes–Merton Equation
In continuous time, we can rebalance the replicating portfolio at every instant. Write the premium at time $t$, when the share price is $S$, as $\mathcal{P}(S,t)$. At expiration, the premium equals the payoff, $\mathcal{P}(S,T)=V(K,S)$. In the real world, the share follows geometric Brownian motion:
$$
\begin{align*}
dS(t)=\mu S(t)\,dt+\sigma S(t)\,dW(t)
\end{align*}
$$
where $\mu$ is the drift, so the mean growth rate is $\mu_{g}=\mu-\sigma^{2}/2$, and $dW(t)$ is a Wiener-process increment. To find how the premium changes over a short interval $dt$, we use Itô's lemma, the stochastic chain rule from the GBM solution derivation:

> __Itô's lemma:__ If $dX=a\,dt+b\,dW(t)$ and $Y(t)=f(t,X(t))$, then:
> $$
> dY=\left(\frac{\partial f}{\partial t}+a\frac{\partial f}{\partial x}+\frac{b^{2}}{2}\frac{\partial^{2} f}{\partial x^{2}}\right)dt+b\frac{\partial f}{\partial x}\,dW(t)
> $$

Let's hold one option and sell $\Delta$ shares, so the portfolio is worth $\Pi=\mathcal{P}-\Delta S$. We hold $\Delta$ fixed over the interval and apply Itô's lemma to $\mathcal{P}$ with $a=\mu S$ and $b=\sigma S$:
$$
\begin{align*}
d\Pi &= d\mathcal{P}-\Delta\,dS && \Delta\text{ held fixed over }dt\\
&= \Bigl(\frac{\partial\mathcal{P}}{\partial t}+\mu S\frac{\partial\mathcal{P}}{\partial S}+\frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}\mathcal{P}}{\partial S^{2}}-\mu S\,\Delta\Bigr)dt+\sigma S\Bigl(\frac{\partial\mathcal{P}}{\partial S}-\Delta\Bigr)dW(t) && \text{Itô's lemma, substitute }dS\\
&= \Bigl(\frac{\partial\mathcal{P}}{\partial t}+\frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}\mathcal{P}}{\partial S^{2}}\Bigr)dt && \text{choose }\Delta=\frac{\partial\mathcal{P}}{\partial S}
\end{align*}
$$
This choice of $\Delta$ is the continuous version of the one-step share holding $\Delta$, and it has three consequences:
* __The random term:__ The $dW(t)$ term is gone, so the hedged portfolio is risk-free over the interval.
* __The real-world drift:__ The terms with $\mu$ cancel, so the share's expected growth plays no role in the premium.
* __What is left:__ Only the premium's change with time and its curvature in $S$ remain.

By the law of one price, a risk-free portfolio must grow at the risk-free rate $g_{y}$:
$$
\begin{align*}
\Bigl(\frac{\partial\mathcal{P}}{\partial t}+\frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}\mathcal{P}}{\partial S^{2}}\Bigr)dt &= g_{y}\,\Pi\,dt && \text{no arbitrage}\\
&= g_{y}\Bigl(\mathcal{P}-S\frac{\partial\mathcal{P}}{\partial S}\Bigr)dt && \Pi=\mathcal{P}-\Delta S\text{ with }\Delta=\frac{\partial\mathcal{P}}{\partial S}
\end{align*}
$$
Dividing by $dt$ and moving the $g_{y}S\,\partial\mathcal{P}/\partial S$ term to the left gives the __Black–Scholes–Merton equation__:
$$
\boxed{
\begin{align*}
\frac{\partial\mathcal{P}}{\partial t}+g_{y}S\frac{\partial\mathcal{P}}{\partial S}+\frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}\mathcal{P}}{\partial S^{2}}=g_{y}\,\mathcal{P}, \qquad \mathcal{P}(S,T)=V(K,S)
\end{align*}}
$$
It holds for every share price $S>0$ and time $0\leq t<T$. The terminal condition is $(S-K)^{+}$ for a call and $(K-S)^{+}$ for a put.

### From the Equation to an Expectation
We solve the equation backward in time from the payoff, and the solution at $t=0$ is the premium. Let's show that this solution is the risk-neutral expectation from the lattice limit. Under $\mathbb{Q}$, the share grows at the risk-free rate:
$$
\begin{align*}
dS(t)=g_{y}S(t)\,dt+\sigma S(t)\,dW^{\mathbb{Q}}(t)
\end{align*}
$$
where $W^{\mathbb{Q}}(t)$ is a Wiener process under $\mathbb{Q}$. Let's discount the premium along a price path to today, $M(t)=\mathcal{D}^{-1}_{t,0}(g_{y})\cdot\mathcal{P}(S(t),t)$, with $\mathcal{D}^{-1}_{t,0}(g_{y})=e^{-g_{y}t}$. Itô's lemma with $f(t,S)=e^{-g_{y}t}\mathcal{P}(S,t)$, $a=g_{y}S$, and $b=\sigma S$ gives:
$$
\begin{align*}
dM &= e^{-g_{y}t}\Bigl(\frac{\partial\mathcal{P}}{\partial t}+g_{y}S\frac{\partial\mathcal{P}}{\partial S}+\frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}\mathcal{P}}{\partial S^{2}}-g_{y}\mathcal{P}\Bigr)dt+e^{-g_{y}t}\,\sigma S\frac{\partial\mathcal{P}}{\partial S}\,dW^{\mathbb{Q}}(t) && \text{Itô's lemma, product rule in }t\\
&= e^{-g_{y}t}\,\sigma S\frac{\partial\mathcal{P}}{\partial S}\,dW^{\mathbb{Q}}(t) && \text{the bracket is zero by the BSM equation}
\end{align*}
$$
The discounted premium has no $dt$ term, so it has no drift under $\mathbb{Q}$. At the two ends, $M(0)=\mathcal{P}(S(0),0)$ and $M(T)=\mathcal{D}^{-1}_{T,0}(g_{y})\cdot V(K,S(T))$, because the premium equals the payoff at expiration. Integrating from $0$ to $T$ and taking expectations gives:
$$
\begin{align*}
M(T)-M(0) &= \int_{0}^{T}e^{-g_{y}t}\,\sigma S(t)\frac{\partial\mathcal{P}}{\partial S}\,dW^{\mathbb{Q}}(t) && \text{integrate }dM\\
\mathbb{E}_{\mathbb{Q}}\bigl(M(T)\bigr)-M(0) &= 0 && \text{the stochastic integral has mean zero}\\
\mathcal{P}(S(0),0) &= \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\cdot V(K,S(T))\Bigr) && \text{substitute }M(0)\text{ and }M(T)
\end{align*}
$$
This is the lecture's European premium. The link between the equation and the expectation is the __[Feynman–Kac theorem](https://en.wikipedia.org/wiki/Feynman%E2%80%93Kac_formula)__. We assumed that the stochastic integral has mean zero. This holds for calls and puts, because $\partial\mathcal{P}/\partial S$ stays between $-1$ and $1$ and the share price has a finite variance.

Let's compute this expectation for a call by splitting its payoff into two simple claims. As a check, the resulting formula with $T$ replaced by the time remaining $T-t$ solves the BSM equation and approaches $(S-K)^{+}$ as $t\to T$.

### Digital Claims
A __digital__ (cash-or-nothing) claim pays one dollar at expiration if $S(T)>K$ and nothing otherwise. Its payoff is $\mathbb{1}\{S(T)>K\}$, the event contract on the share finishing above the strike. Taking logs of the lattice limit gives the probability of that event:
$$
\begin{align*}
\mathbb{Q}\bigl(S(T)>K\bigr)
&= \mathbb{Q}\Bigl(\bigl(g_{y}-\tfrac{\sigma^{2}}{2}\bigr)T + \sigma\sqrt{T}\,Z > \ln\tfrac{K}{S(0)}\Bigr) && \text{take logs}\\
&= \mathbb{Q}\bigl(Z > -d_{-}\bigr) && \text{solve for }Z\\
&= N(d_{-}) && \text{the standard normal is symmetric}
\end{align*}
$$
where $d_{-}=\bigl[\ln\frac{S(0)}{K}+\bigl(g_{y}-\frac{\sigma^{2}}{2}\bigr)T\bigr]/(\sigma\sqrt{T})$ is the lecture's $d_{-}$.

This is the terminal target probability. Under $\mathbb{P}$, the terminal target formula gives $\mathbb{P}(S(T)>K)=1-\Phi(z_\star)$ with mean growth rate $\mu_g$. Replacing $\mu_g$ by $g_{y}-\sigma^{2}/2$ gives $z_\star=-d_{-}$, and $N$ is the same standard normal CDF as $\Phi$. The price of the digital claim is therefore:
$$
\begin{align*}
\mathcal{D}^{-1}_{T,0}(g_{y})\cdot N(d_{-})
\end{align*}
$$

### Asset-or-Nothing Claims
An __asset-or-nothing__ claim pays the share price $S(T)$ at expiration if $S(T)>K$ and nothing otherwise. Write $m=\bigl(g_{y}-\tfrac{\sigma^{2}}{2}\bigr)T$ and $s=\sigma\sqrt{T}$, so that $S(T)=S(0)e^{m+sZ}$ and $S(T)>K$ exactly when $Z>-d_{-}$. With $\varphi$ the standard normal density:
$$
\begin{align*}
\mathbb{E}_{\mathbb{Q}}\bigl(S(T)\,\mathbb{1}\{S(T)>K\}\bigr)
&= S(0)\,e^{m}\int_{-d_{-}}^{\infty} e^{sz}\,\varphi(z)\,dz && \text{expectation over }Z\\
&= S(0)\,e^{m+s^{2}/2}\int_{-d_{-}}^{\infty}\varphi(z-s)\,dz && e^{sz}\varphi(z)=e^{s^{2}/2}\varphi(z-s)\text{, complete the square}\\
&= S(0)\,e^{g_{y}T}\,\mathbb{Q}\bigl(Z>-d_{-}-s\bigr) && m+\tfrac{s^{2}}{2}=g_{y}T\text{, shift the variable}\\
&= S(0)\,e^{g_{y}T}\,N(d_{+}) && d_{-}+\sigma\sqrt{T}=d_{+}\text{, symmetry}
\end{align*}
$$
Discounting by $\mathcal{D}^{-1}_{T,0}(g_{y})=e^{-g_{y}T}$, the price of the asset-or-nothing claim is $S(0)\,N(d_{+})$.

### The Call and the Put
The call payoff splits into the two claims we just priced:
$$
\begin{align*}
V_{c}(K,S(T)) = \max\bigl(S(T)-K,~0\bigr) = S(T)\,\mathbb{1}\{S(T)>K\} - K\cdot\mathbb{1}\{S(T)>K\}
\end{align*}
$$
The call is therefore one asset-or-nothing claim minus $K$ digital claims. Pricing each piece and adding, by linearity:
$$
\begin{align*}
\mathcal{P}_{c}(K,S(0))
&= \mathcal{D}^{-1}_{T,0}(g_{y})\Bigl[\mathbb{E}_{\mathbb{Q}}\bigl(S(T)\,\mathbb{1}\{S(T)>K\}\bigr) - K\,\mathbb{Q}\bigl(S(T)>K\bigr)\Bigr] && \text{expectation is linear}\\
&= e^{-g_{y}T}\Bigl[S(0)\,e^{g_{y}T}N(d_{+}) - K\,N(d_{-})\Bigr] && \text{asset-or-nothing and digital}\\
&= N(d_{+})\cdot S(0) - N(d_{-})\cdot K\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) && \text{the lecture's BSM call}
\end{align*}
$$
For the put, the lecture's put–call parity gives:
$$
\begin{align*}
\mathcal{P}_{p}(K,S(0))
&= \mathcal{P}_{c}(K,S(0)) - S(0) + K\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) && \text{put–call parity}\\
&= S(0)\bigl[N(d_{+})-1\bigr] + K\cdot\mathcal{D}^{-1}_{T,0}(g_{y})\bigl[1-N(d_{-})\bigr] && \text{substitute the call}\\
&= N(-d_{-})\cdot K\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) - N(-d_{+})\cdot S(0) && 1-N(x)=N(-x)
\end{align*}
$$
which is the lecture's BSM put.

### Reading $N(d_{-})$ and $N(d_{+})$
* __Exercise probability.__ $N(d_{-})$ is the $\mathbb{Q}$-probability that the call is exercised. It can differ from the real-world probability $\mathbb{P}(S(T)>K)$ that the terminal target formula computes with $\mu_g$.
* __Share weight.__ $N(d_{+})$ weights the share received at exercise. The asset-or-nothing price $S(0)\,N(d_{+})$ counts the share only in the states where it is delivered. $N(d_{+})$ is the probability of the same event under a different pricing measure, one that uses the share rather than the risk-free asset as its unit of account. Because $d_{+}>d_{-}$, it is the larger of the two.

## Certainty Equivalents and Market Prices
The bet in the lecture's Concept Review pays one dollar at expiration if $S(T)\geq K$ and nothing otherwise. Under geometric Brownian motion, the share finishes exactly at $K$ with probability zero, so this bet is the digital claim priced above. A buyer can also value the bet with a certainty equivalent. Do the buyer's price and the market price agree?

Let's work one case with $S(0)=K=100$ USD/share, $T=1$ year, volatility $\sigma=0.30$, and $g_{y}=0.04$ per year. In the real world, the share's mean growth rate is $\mu_{g}=0.06$ per year, while under $\mathbb{Q}$ it is $g_{y}-\sigma^{2}/2=-0.005$ per year. Repeating the digital-claim calculation with each mean growth rate gives the two probabilities of the event:

$$
\begin{align*}
\mathbb{P}\bigl(S(T)\geq K\bigr) &= N\!\left(\frac{\ln\frac{S(0)}{K}+\mu_{g}T}{\sigma\sqrt{T}}\right) = N(0.200) \approx 0.579 && \text{real-world mean growth rate}\\
\mathbb{Q}\bigl(S(T)\geq K\bigr) &= N\!\left(\frac{\ln\frac{S(0)}{K}+\bigl(g_{y}-\frac{\sigma^{2}}{2}\bigr)T}{\sigma\sqrt{T}}\right) \approx N(-0.017) \approx 0.493 && \text{risk-neutral mean growth rate, }N(d_{-})
\end{align*}
$$

Now consider a client with relative risk aversion $\bar{r}=3$ whose other wealth at expiration is a sure $w=50{,}000$ USD. Owning the bet with payoff $V$ makes the client's wealth $W=w+V$, with $\mathbb{E}_{\mathbb{P}}(V)=\mathbb{P}\bigl(S(T)\geq K\bigr)$. Pratt's small-risk result gives the client's risk premium for the bet:

$$
\begin{align*}
\mathrm{RP} &\approx -\frac{1}{2}\,\frac{U''(w)}{U'(w)}\,\mathrm{Var}(W) && \text{Pratt (1964), small risk}\\
&\approx \frac{\bar{r}}{2w}\,\mathrm{Var}(V) && \bar{r}=-w\,U''(w)/U'(w)\text{, and }w\text{ is sure}\\
&\approx \frac{\bar{r}}{2w}\,\mathbb{P}\bigl(S(T)\geq K\bigr)\Bigl[1-\mathbb{P}\bigl(S(T)\geq K\bigr)\Bigr] && \text{variance of a one-dollar bet}\\
&\approx \frac{3}{2\times 50{,}000}\times 0.579\times 0.421 \approx 7.3\times10^{-6}\ \text{USD} && \text{substitute the case}
\end{align*}
$$

The bet is worth a sure $\mathbb{E}_{\mathbb{P}}(V)-\mathrm{RP}$ to the client at expiration. Discounted to today, it is approximately the most the client would pay. The table compares it with the market price, for the Concept Review bet and for the opposite bet, which pays one dollar if $S(T)<K$:

| Bet | $\mathbb{P}$ (real world) | $\mathbb{Q}$ (risk neutral) | Client's RP (USD) | Client's value today (USD) | Market price (USD) |
|:--|--:|--:|--:|--:|--:|
| Up: pays if $S(T)\geq K$ | 0.579 | 0.493 | 0.000007 | 0.557 | 0.474 |
| Down: pays if $S(T)<K$ | 0.421 | 0.507 | 0.000007 | 0.404 | 0.487 |

The client values the up-bet at almost its full discounted expected payoff, because one dollar is tiny next to 50,000 USD. The market sells the up-bet for 8.3 cents less than the client's value, and it sells the down-bet for 8.3 cents more. The client's risk premium is less than a thousandth of a cent, so it cannot explain either gap.

### Marginal Utility and the Risk-Neutral Probability
Where do the gaps come from? Let's return to the one-step lattice, where the up and down states have real-world probabilities $\mathbb{P}(u)$ and $\mathbb{P}(d)=1-\mathbb{P}(u)$. An investor holds the amounts of the share and the risk-free asset that maximize expected utility. The investor ends the step with wealth $W$, equal to $W_{u}$ or $W_{d}$. Let the investor move a small amount $\varepsilon$ (USD) from the risk-free asset into a claim with price $V_{0}$ and payoff $H$, equal to $H_{u}$ or $H_{d}$. In state $s$, the $\varepsilon/V_{0}$ claims pay $\varepsilon H_{s}/V_{0}$, and the risk-free asset pays $\varepsilon R_{f}$ less. At the best holdings, no small move can raise expected utility, so its derivative at $\varepsilon=0$ is zero:

$$
\begin{align*}
0 &= \frac{d}{d\varepsilon}\,\mathbb{E}_{\mathbb{P}}\Bigl[U\Bigl(W+\varepsilon\bigl(\tfrac{H}{V_{0}}-R_{f}\bigr)\Bigr)\Bigr]_{\varepsilon=0} && \text{best holdings}\\
&= \mathbb{P}(u)\,U'(W_{u})\Bigl(\frac{H_{u}}{V_{0}}-R_{f}\Bigr)+\mathbb{P}(d)\,U'(W_{d})\Bigl(\frac{H_{d}}{V_{0}}-R_{f}\Bigr) && \text{chain rule in each state}\\
V_{0}\,R_{f}\,\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr) &= \mathbb{P}(u)\,U'(W_{u})\,H_{u}+\mathbb{P}(d)\,U'(W_{d})\,H_{d} && \text{multiply by }V_{0}\text{, rearrange}\\
V_{0} &= \frac{1}{R_{f}}\left[\frac{\mathbb{P}(u)\,U'(W_{u})}{\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr)}\,H_{u}+\frac{\mathbb{P}(d)\,U'(W_{d})}{\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr)}\,H_{d}\right] && \text{divide}
\end{align*}
$$

where $U'$ is the investor's marginal utility and $\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr)=\mathbb{P}(u)\,U'(W_{u})+\mathbb{P}(d)\,U'(W_{d})$ is its real-world average. Replication priced every claim on the lattice with the weights $q$ and $1-q$. Both rules hold for every claim, so the weights on $H_{u}$ must match:

$$
\boxed{
\begin{align*}
q = \mathbb{P}(u)\,\frac{U'(W_{u})}{\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr)}
\end{align*}}
$$

The risk-neutral probability is the real-world probability reweighted by marginal utility, and the state price $\psi_{u}=q/R_{f}$ is that weight discounted. Three points follow:
* __A bet that pays in the rich state costs less.__ An investor who holds shares is richer in the up state, $W_{u}>W_{d}$. Because $U$ is concave ($U''<0$), $U'(W_{u})$ is below the average $\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr)$, so $q<\mathbb{P}(u)$ and the up-bet sells below its discounted expected payoff. Its dollar arrives when the investor already has more dollars.
* __Insurance costs more than its discounted expected payoff.__ The same argument puts $U'(W_{d})$ above the average, so $1-q>\mathbb{P}(d)$ and the down-bet sells above its discounted expected payoff. It pays when wealth is low, which is when one more dollar is worth the most.
* __The certainty equivalent values the bet on its own.__ With sure other wealth, the client ends with $w$ or $w+1$ USD, so the client's marginal utility is almost the same in both states. The market weights states by the marginal utility of investors whose wealth rises and falls with the share. The gap comes from when the bet pays, not from the bet's own risk.

Replication fixed $q$ from $u$, $d$, and $R_{f}$ alone, so the BSM formulas never need $U$. The utility rule says what $q$ means. [Rubinstein (1976)](https://doi.org/10.2307/3003264) showed that the utility rule gives exactly the BSM formulas when the market prices claims like one investor with power utility. That investor's wealth and the share price must be jointly lognormal. Each buyer's preferences still matter, but they set how much of the bet to hold at the market price. The risk-aversion coefficient $A$ sets the risk-free fraction on the capital allocation line in the same way.

## Summary
We derived the Black–Scholes–Merton premiums from the law of one price. Replication gave the risk-neutral probability, and state prices turned pricing into an inner product over states. A continuous hedge gave the BSM equation, and two simple claims built the call. Marginal utility then explained why the market price of a bet differs from a client's certainty equivalent.

> __Key Takeaways:__
>
> * __Replication and state prices:__ We matched a claim's payoff in both lattice states with shares and the risk-free asset, and the portfolio's cost defined the risk-neutral probability. We then wrote every price as state prices times payoffs, so a YES and NO pair on one event costs the same as a zero-coupon bond.
> * __The BSM equation and two simple claims:__ We hedged an option with shares in continuous time and obtained the BSM equation, whose solution is the risk-neutral expectation of the discounted payoff. We then priced a digital claim and an asset-or-nothing claim, built the call from them, and used put–call parity for the put.
> * __The risk premium sits in the risk-neutral probability:__ We found a client's certainty equivalent for a one-dollar bet almost equal to its expected payoff, while the market price differed by 8 cents. We showed that the market weights each state by marginal utility, so a bet that pays when investors are rich costs less, and insurance costs more.

__Assumptions.__ The derivation assumes no arbitrage, frictionless trading in the share and the risk-free asset, a constant risk-free rate $g_{y}$ and volatility $\sigma$, no dividends, and the lattice limit stated above without proof. The hedge also assumes geometric Brownian motion, continuous rebalancing with $\Delta$ fixed over each short interval, and a premium smooth enough for Itô's lemma. The step to the expectation assumes the stochastic integral has mean zero. The last section adds investors with an increasing, concave utility $U$ who hold their best portfolio. They can buy or sell small amounts of any claim.
