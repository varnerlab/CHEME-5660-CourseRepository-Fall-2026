# Interesting Fact: Put–Call Parity for European Contracts
In the [lecture](CHEME-5660-L9b-Lecture-American-Derivatives-Intrinsic-Extrinsic-Value-Fall-2026.ipynb), we priced a European call or put as its discounted expected payoff under the risk-neutral pricing rule $\mathbb{Q}$:
$$
\begin{align*}
\mathcal{P}_{\star}(K,S(0)) = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{E}_{\mathbb{Q}}\left[V_{\star}(K,S(T))\right]\quad\text{where }\star\in\{c,p\}
\end{align*}
$$
where $V_{c}(K,S(T))=\max(S(T)-K,~0)$ and $V_{p}(K,S(T))=\max(K-S(T),~0)$ are the call and put payoffs at expiration $T$, $K$ is the strike price, $S(0)$ is today's share price, and $\mathcal{D}^{-1}_{T,0}(g_{y})=e^{-g_{y}T}$ discounts the payoff back to today at the risk-free rate $g_{y}$. How are the call and put premiums related?
___

## Derivation
The call and put premiums are linked. For a European call and put with the same strike and expiration, the payoffs differ by $S(T)-K$ at every share price:
$$
\begin{align*}
V_{c}(K,S(T)) - V_{p}(K,S(T)) & = \max\left(S(T)-K,~0\right) - \max\left(K-S(T),~0\right) && \text{payoff definitions}\\
& = \begin{cases}
\left(S(T)-K\right) - 0 & \text{if } S(T)>K\text{, the put pays 0}\\
0 - \left(K-S(T)\right) & \text{if } S(T)\leq K\text{, the call pays 0}
\end{cases} && \text{split at the strike price }K\\
& = S(T) - K && \text{simplify both cases}
\end{align*}
$$
Subtracting the put premium from the call premium gives:
$$
\begin{align*}
\mathcal{P}_{c}(K,S(0)) - \mathcal{P}_{p}(K,S(0))
& = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{E}_{\mathbb{Q}}\left[V_{c}(K,S(T))\right] && \text{premium formula for the call and the put}\\
& \qquad - \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{E}_{\mathbb{Q}}\left[V_{p}(K,S(T))\right]\\
& = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{E}_{\mathbb{Q}}\left[V_{c}(K,S(T)) - V_{p}(K,S(T))\right] && \text{expectation is linear}\\
& = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\mathbb{E}_{\mathbb{Q}}\left[S(T) - K\right] && \text{payoff identity}\\
& = \mathcal{D}^{-1}_{T,0}(g_{y})\cdot\left(S(0)\,e^{g_{y}T} - K\right) && \text{expected share price grows at }g_{y}\text{ under }\mathbb{Q}\\
& = S(0) - K\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) && \mathcal{D}^{-1}_{T,0}(g_{y})=e^{-g_{y}T}
\end{align*}
$$
This is __put–call parity__:
$$
\boxed{
\begin{align*}
\mathcal{P}_{c}(K,S(0)) - \mathcal{P}_{p}(K,S(0)) = S(0) - K\cdot\mathcal{D}^{-1}_{T,0}(g_{y})
\end{align*}}
$$
__Put–call parity holds only for European contracts.__ The derivation compares the two payoffs on the expiration date, but an American holder can exercise before then. For American contracts, the difference between the call and put premiums is bounded, not fixed.
