## Replication on a One-Step Lattice
An abstract asset is described by what it pays in each possible outcome, or __state__, at each date. Suppose two assets pay the same amount in every state at every date. If their prices differed, we could buy the cheaper one and sell the other. The future cash flows would cancel in every state, and the price difference would be a profit today with no risk, called an __arbitrage__.

> __Law of one price:__ In a market with no arbitrage, two assets with identical cash flows in every state have the same price, and therefore the same NPV.

To price an option, we build a portfolio of assets whose prices we know, with the same cash flows as the option in every state. This is called __replication__, and the option's premium is the portfolio's cost.

__A one-step example.__ The share price is 100 USD today and moves to 120 or 80 USD. The interest rate is zero. A call with strike price 100 USD pays 20 USD if the share price moves up and nothing if it moves down. Let's buy half a share for 50 USD, borrowing 40 USD of the cost:

| Holding | Share price moves up to 120 USD | Share price moves down to 80 USD |
| --- | --- | --- |
| Half a share | 60 | 40 |
| Repay the loan | $-40$ | $-40$ |
| Portfolio | 20 | 0 |
| Call payoff | 20 | 0 |

The portfolio pays what the call pays in both states, so the call costs what the portfolio costs, $50-40=10$ USD. For the put, which pays nothing up and 20 USD down, we sell half a share short for 50 USD and lend 60 USD, a cost of $60-50=10$ USD. The general solution below gives these holdings for any payoff.

__Is this a portfolio on the capital allocation line?__ The capital allocation line from L8b mixes the risk-free asset with one risky asset, and $w_{f}$ is the fraction of wealth in the risk-free asset. The call's portfolio has that form. It puts 50 USD in the share using 10 USD of its own money, so $w_{f}=-40/10=-4$, a borrowing position. What sets $w_{f}$ differs. On the capital allocation line, the risky asset is the tangent portfolio and the investor's risk aversion sets $w_{f}$. Here, the risky asset is the share and the call's payoff sets $w_{f}$.

On a one-step lattice, the share price moves from $S(0)$ to $uS(0)$ (up) or $dS(0)$ (down), and one dollar in the risk-free asset grows to $R_f$. A __claim__ is a contract whose payoff depends on the state. It pays $H_u$ in the up state and $H_d$ in the down state. Let's choose $\Delta$ shares and $B$ dollars in the risk-free asset so that the portfolio pays what the claim pays in both states:
$$
\begin{align*}
\Delta\,uS(0) + B\,R_f &= H_u && \text{up state}\\
\Delta\,dS(0) + B\,R_f &= H_d && \text{down state}
\end{align*}
$$
Solving the two equations gives the holdings:
$$
\begin{align*}
\Delta &= \frac{H_u - H_d}{(u-d)\,S(0)} && \text{subtract down from up, }B\text{ cancels}\\
B &= \frac{uH_d - dH_u}{(u-d)\,R_f} && \text{substitute }\Delta\text{ into the down state}
\end{align*}
$$
By the law of one price, the claim's price $V_0$ is the portfolio's cost $\Delta S(0) + B$:
$$
\begin{align*}
V_0 &= \frac{H_u - H_d}{u-d} + \frac{uH_d - dH_u}{(u-d)\,R_f} && \text{cost of }\Delta\text{ shares and }B\text{ dollars}\\
&= \frac{1}{R_f}\left[\frac{R_f-d}{u-d}\,H_u + \frac{u-R_f}{u-d}\,H_d\right] && \text{common denominator, collect }H_u\text{ and }H_d\\
&= \frac{1}{R_f}\Bigl[q\,H_u + (1-q)\,H_d\Bigr], \qquad q=\frac{R_f-d}{u-d} && \text{so }1-q=\frac{u-R_f}{u-d}
\end{align*}
$$
This is the lecture's risk-neutral pricing rule. Over one step, $1/R_f$ is the discount factor, and the bracket is the expected payoff under $\mathbb{Q}$, where the share price moves up with probability $q$. Three points follow:
* __The real-world up probability never appears.__ The portfolio matches the claim in both states, whatever their probabilities, so the price does not depend on them.
* __No arbitrage requires $d<R_f<u$.__ If $R_f\leq d$, borrowing to buy the share loses nothing in the down state and gains in the up state. If $R_f\geq u$, selling the share short and lending the proceeds loses nothing in the up state and gains in the down state. With $d<R_f<u$, the weight $q$ lies strictly between $0$ and $1$.
* __The share is priced by the same rule.__ With $H_u=uS(0)$ and $H_d=dS(0)$, the rule returns $S(0)$, because $qu+(1-q)d=R_f$.

### State Prices and Event Contracts
The price is a weighted sum of the two payoffs. The weights are the __state prices__:
$$
\begin{align*}
\psi_u = \frac{q}{R_f}, \qquad \psi_d = \frac{1-q}{R_f}
\end{align*}
$$
With $\boldsymbol{\psi}=(\psi_u,\psi_d)$ and $\mathbf{H}=(H_u,H_d)$, the price of any claim is their inner product:
$$
\begin{align*}
V_0 = \psi_u H_u + \psi_d H_d = \bigl\langle \boldsymbol{\psi},\mathbf{H}\bigr\rangle
\end{align*}
$$
NPV is an inner product over dates, with each cash flow multiplied by its discount factor. A state price plays the same role across states: $\psi_u$ is the price today of one dollar paid only in the up state.

__Event contracts are the simplest claims.__ A YES contract on the event "the share price moves up" pays one dollar in the up state and nothing otherwise, so its price is $\psi_u$. The NO contract pays one dollar in the down state, so its price is $\psi_d$. Holding both pays one dollar in every state, which is a zero-coupon bond:
$$
\begin{align*}
\psi_u + \psi_d = \frac{q + (1-q)}{R_f} = \frac{1}{R_f}
\end{align*}
$$
This is the same idea as put–call parity: when two claims' payoffs add to a known amount, their prices add to that amount's present value.

## The Black–Scholes–Merton Formulas
The one-step rule extends to a lattice with $n$ steps of length $\Delta t=T/n$. We replicate the claim one step at a time, backward from expiration. The holdings $\Delta$ and $B$ change from node to node, because fixed holdings pay a straight line in $S(T)$, while the call's payoff bends at the strike. Later trades only swap shares and cash inside the portfolio, so the first purchase is the whole cost. That cost is the discounted expected payoff under $\mathbb{Q}$, with the share price moving up with probability $q$ at every step.

Previously, an optional notebook showed that as the steps shrink, a binomial lattice approaches geometric Brownian motion. We state the result under $\mathbb{Q}$ without proof.

> __Proposition (lattice limit under $\mathbb{Q}$):__ Let $u=e^{\sigma\sqrt{\Delta t}}$, where $\sigma$ is the volatility, $d=1/u$, and $R_f=e^{g_{y}\Delta t}$. As $n\to\infty$, the distribution of the share price at expiration under $\mathbb{Q}$ approaches:
> $$
> S(T) = S(0)\exp\Bigl(\bigl(g_{y}-\tfrac{\sigma^{2}}{2}\bigr)T + \sigma\sqrt{T}\,Z\Bigr), \qquad Z\sim\mathcal{N}(0,1)
> $$
> and the price of each European claim in this notebook, with payoff $H(S(T))$, approaches $\mathcal{D}^{-1}_{T,0}(g_{y})\,\mathbb{E}_{\mathbb{Q}}\bigl(H(S(T))\bigr)$.

Under $\mathbb{P}$, the GBM model has the same form, with the real-world mean growth rate $\mu_g$ in place of $g_{y}-\sigma^{2}/2$.

### The Black–Scholes–Merton Equation
In continuous time, we can rebalance the replicating portfolio at every instant. Write the premium at time $t$, when the share price is $S$, as $\mathcal{P}(S,t)$. At expiration, the premium equals the payoff, $\mathcal{P}(S,T)=V(K,S)$. In the real world, the share price follows geometric Brownian motion:
$$
\begin{align*}
dS(t)=\mu S(t)\,dt+\sigma S(t)\,dW(t)
\end{align*}
$$
where $\mu$ is the drift (the expected rate of return), so the mean growth rate is $\mu_{g}=\mu-\sigma^{2}/2$, and $dW(t)$ is a Wiener-process increment. To find how the premium changes over a short interval $dt$, we use Itô's lemma, the chain rule for random processes that we used to solve the GBM equation:

> __Itô's lemma:__ If $dX=a\,dt+b\,dW(t)$ and $Y(t)=f(t,X(t))$, then:
> $$
> dY=\left(\frac{\partial f}{\partial t}+a\frac{\partial f}{\partial x}+\frac{b^{2}}{2}\frac{\partial^{2} f}{\partial x^{2}}\right)dt+b\frac{\partial f}{\partial x}\,dW(t)
> $$

Let's hold one option and sell $\Delta$ shares, so the portfolio is worth $\Pi=\mathcal{P}-\Delta S$. We hold $\Delta$ fixed over the interval and apply Itô's lemma to $\mathcal{P}$ with $a=\mu S$ and $b=\sigma S$:
$$
\begin{align*}
d\Pi &= d\mathcal{P}-\Delta\,dS && \Delta\text{ held fixed over }dt\\
&= \Bigl(\frac{\partial\mathcal{P}}{\partial t}+\mu S\frac{\partial\mathcal{P}}{\partial S}+\frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}\mathcal{P}}{\partial S^{2}}-\mu S\,\Delta\Bigr)dt+\sigma S\Bigl(\frac{\partial\mathcal{P}}{\partial S}-\Delta\Bigr)dW(t) && \text{Itô's lemma, substitute }dS\\
&= \Bigl(\frac{\partial\mathcal{P}}{\partial t}+\frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}\mathcal{P}}{\partial S^{2}}\Bigr)dt && \text{choose }\Delta=\frac{\partial\mathcal{P}}{\partial S}
\end{align*}
$$
This $\Delta$ is the continuous version of the one-step share holding. It removes the $dW(t)$ term, so the hedged portfolio is risk-free over the interval. The terms with $\mu$ also cancel, so the expected growth rate of the share price does not enter the premium.

By the law of one price, a risk-free portfolio must grow at the risk-free rate $g_{y}$:
$$
\begin{align*}
\Bigl(\frac{\partial\mathcal{P}}{\partial t}+\frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}\mathcal{P}}{\partial S^{2}}\Bigr)dt &= g_{y}\,\Pi\,dt && \text{no arbitrage}\\
&= g_{y}\Bigl(\mathcal{P}-S\frac{\partial\mathcal{P}}{\partial S}\Bigr)dt && \Pi=\mathcal{P}-\Delta S\text{ with }\Delta=\frac{\partial\mathcal{P}}{\partial S}
\end{align*}
$$
Dividing by $dt$ and moving the $g_{y}S\,\partial\mathcal{P}/\partial S$ term to the left gives the __Black–Scholes–Merton equation__:
$$
\boxed{
\begin{align*}
\frac{\partial\mathcal{P}}{\partial t}+g_{y}S\frac{\partial\mathcal{P}}{\partial S}+\frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}\mathcal{P}}{\partial S^{2}}=g_{y}\,\mathcal{P}, \qquad \mathcal{P}(S,T)=V(K,S)
\end{align*}}
$$
It holds for every share price $S>0$ and time $0\leq t<T$. At expiration, $\mathcal{P}$ is $(S-K)^{+}$ for a call and $(K-S)^{+}$ for a put.

### From the Equation to an Expectation
We solve the equation backward in time from the payoff, and the solution at $t=0$ is the premium. Let's show that this solution is the discounted expected payoff under $\mathbb{Q}$. Under $\mathbb{Q}$, the expected share price grows at the risk-free rate:
$$
\begin{align*}
dS(t)=g_{y}S(t)\,dt+\sigma S(t)\,dW^{\mathbb{Q}}(t)
\end{align*}
$$
where $W^{\mathbb{Q}}(t)$ is a Wiener process under $\mathbb{Q}$. Let's discount the premium along a price path to today, $M(t)=\mathcal{D}^{-1}_{t,0}(g_{y})\cdot\mathcal{P}(S(t),t)$, with $\mathcal{D}^{-1}_{t,0}(g_{y})=e^{-g_{y}t}$. Itô's lemma with $f(t,S)=e^{-g_{y}t}\mathcal{P}(S,t)$, $a=g_{y}S$, and $b=\sigma S$ gives:
$$
\begin{align*}
dM &= e^{-g_{y}t}\Bigl(\frac{\partial\mathcal{P}}{\partial t}+g_{y}S\frac{\partial\mathcal{P}}{\partial S}+\frac{1}{2}\sigma^{2}S^{2}\frac{\partial^{2}\mathcal{P}}{\partial S^{2}}-g_{y}\mathcal{P}\Bigr)dt+e^{-g_{y}t}\,\sigma S\frac{\partial\mathcal{P}}{\partial S}\,dW^{\mathbb{Q}}(t) && \text{Itô's lemma, product rule in }t\\
&= e^{-g_{y}t}\,\sigma S\frac{\partial\mathcal{P}}{\partial S}\,dW^{\mathbb{Q}}(t) && \text{the bracket is zero by the BSM equation}
\end{align*}
$$
The discounted premium has no $dt$ term, so its expected change under $\mathbb{Q}$ is zero. At the two ends, $M(0)=\mathcal{P}(S(0),0)$ and $M(T)=\mathcal{D}^{-1}_{T,0}(g_{y})\cdot V(K,S(T))$, because the premium equals the payoff at expiration. Integrating from $0$ to $T$ and taking expectations gives:
$$
\begin{align*}
M(T)-M(0) &= \int_{0}^{T}e^{-g_{y}t}\,\sigma S(t)\frac{\partial\mathcal{P}}{\partial S}\,dW^{\mathbb{Q}}(t) && \text{integrate }dM\\
\mathbb{E}_{\mathbb{Q}}\bigl(M(T)\bigr)-M(0) &= 0 && \text{the stochastic integral has mean zero}\\
\mathcal{P}(S(0),0) &= \mathbb{E}_{\mathbb{Q}}\Bigl(\mathcal{D}^{-1}_{T,0}(g_{y})\cdot V(K,S(T))\Bigr) && \text{substitute }M(0)\text{ and }M(T)
\end{align*}
$$
This is the lecture's European premium. This link between the equation and the expectation is a case of the __[Feynman–Kac theorem](https://en.wikipedia.org/wiki/Feynman%E2%80%93Kac_formula)__. The stochastic integral has mean zero for calls and puts, because $\partial\mathcal{P}/\partial S$ stays between $-1$ and $1$ and the share price has a finite variance.

Let's compute this expectation for a call by splitting its payoff into two simple claims. As a check, the resulting formula with $T$ replaced by the time remaining $T-t$ solves the BSM equation and approaches $(S-K)^{+}$ as $t\to T$.

### Digital Claims
A __digital__ (cash-or-nothing) claim pays one dollar at expiration if $S(T)>K$ and nothing otherwise. Its payoff is $\mathbb{1}\{S(T)>K\}$, so it is an event contract. Taking logs of the lattice limit gives the probability of that event:
$$
\begin{align*}
\mathbb{Q}\bigl(S(T)>K\bigr)
&= \mathbb{Q}\Bigl(\bigl(g_{y}-\tfrac{\sigma^{2}}{2}\bigr)T + \sigma\sqrt{T}\,Z > \ln\tfrac{K}{S(0)}\Bigr) && \text{take logs}\\
&= \mathbb{Q}\bigl(Z > -d_{-}\bigr) && \text{solve for }Z\\
&= N(d_{-}) && \text{the standard normal is symmetric}
\end{align*}
$$
where $d_{-}=\bigl[\ln\frac{S(0)}{K}+\bigl(g_{y}-\frac{\sigma^{2}}{2}\bigr)T\bigr]/(\sigma\sqrt{T})$ is the lecture's $d_{-}$.

The same steps with the real-world mean growth rate $\mu_g$ in place of $g_{y}-\sigma^{2}/2$ give the terminal target probability $\mathbb{P}(S(T)>K)$. The price of the digital claim is the discounted $\mathbb{Q}$-probability:
$$
\begin{align*}
\mathcal{D}^{-1}_{T,0}(g_{y})\cdot N(d_{-})
\end{align*}
$$

### Asset-or-Nothing Claims
An __asset-or-nothing__ claim pays the share price $S(T)$ at expiration if $S(T)>K$ and nothing otherwise. Write $m=\bigl(g_{y}-\tfrac{\sigma^{2}}{2}\bigr)T$ and $s=\sigma\sqrt{T}$, so that $S(T)=S(0)e^{m+sZ}$ and $S(T)>K$ exactly when $Z>-d_{-}$. With $\varphi$ the standard normal density, its expected payoff under $\mathbb{Q}$ is:
$$
\begin{align*}
\mathbb{E}_{\mathbb{Q}}\bigl(S(T)\,\mathbb{1}\{S(T)>K\}\bigr)
&= S(0)\,e^{m}\int_{-d_{-}}^{\infty} e^{sz}\,\varphi(z)\,dz && \text{expectation over }Z\\
&= S(0)\,e^{m+s^{2}/2}\int_{-d_{-}}^{\infty}\varphi(z-s)\,dz && e^{sz}\varphi(z)=e^{s^{2}/2}\varphi(z-s)\text{, complete the square}\\
&= S(0)\,e^{g_{y}T}\,\mathbb{Q}\bigl(Z>-d_{-}-s\bigr) && m+\tfrac{s^{2}}{2}=g_{y}T\text{, shift the variable}\\
&= S(0)\,e^{g_{y}T}\,N(d_{+}) && d_{-}+\sigma\sqrt{T}=d_{+}\text{, symmetry}
\end{align*}
$$
Discounting by $\mathcal{D}^{-1}_{T,0}(g_{y})=e^{-g_{y}T}$, the price of the asset-or-nothing claim is $S(0)\,N(d_{+})$.

### The Call and the Put
The call payoff splits into the two claims we just priced:
$$
\begin{align*}
V_{c}(K,S(T)) = \max\bigl(S(T)-K,~0\bigr) = S(T)\,\mathbb{1}\{S(T)>K\} - K\cdot\mathbb{1}\{S(T)>K\}
\end{align*}
$$
The call is therefore one asset-or-nothing claim minus $K$ digital claims. Pricing each piece gives the call premium:
$$
\begin{align*}
\mathcal{P}_{c}(K,S(0))
&= \mathcal{D}^{-1}_{T,0}(g_{y})\Bigl[\mathbb{E}_{\mathbb{Q}}\bigl(S(T)\,\mathbb{1}\{S(T)>K\}\bigr) - K\,\mathbb{Q}\bigl(S(T)>K\bigr)\Bigr] && \text{expectation is linear}\\
&= e^{-g_{y}T}\Bigl[S(0)\,e^{g_{y}T}N(d_{+}) - K\,N(d_{-})\Bigr] && \text{asset-or-nothing and digital}\\
&= N(d_{+})\cdot S(0) - N(d_{-})\cdot K\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) && \text{the lecture's BSM call}
\end{align*}
$$
For the put, [put–call parity](../../CHEME-5660-L9b-InterestingFact-PutCallParity-Fall-2026.ipynb) gives:
$$
\begin{align*}
\mathcal{P}_{p}(K,S(0))
&= \mathcal{P}_{c}(K,S(0)) - S(0) + K\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) && \text{put–call parity}\\
&= S(0)\bigl[N(d_{+})-1\bigr] + K\cdot\mathcal{D}^{-1}_{T,0}(g_{y})\bigl[1-N(d_{-})\bigr] && \text{substitute the call}\\
&= N(-d_{-})\cdot K\cdot\mathcal{D}^{-1}_{T,0}(g_{y}) - N(-d_{+})\cdot S(0) && 1-N(x)=N(-x)
\end{align*}
$$
which is the lecture's BSM put.

### Reading $N(d_{-})$ and $N(d_{+})$
In the call premium, $N(d_{-})$ is the $\mathbb{Q}$-probability that the call is exercised. It can differ from the real-world probability $\mathbb{P}(S(T)>K)$ that the terminal target formula computes with $\mu_g$. Because $S(0)\,N(d_{+})$ is the price of receiving the share only when the call is exercised, $N(d_{+})$ is the fraction of today's share price that comes from those states. It is larger than $N(d_{-})$, because the share price is higher on average in those states than overall.

## Certainty Equivalents and Market Prices
Previously, we priced a bet that pays one dollar at expiration if $S(T)\geq K$ and nothing otherwise. Under geometric Brownian motion, the share price finishes exactly at $K$ with probability zero, so this bet is the digital claim priced above. A buyer can also value the bet with a certainty equivalent. Do the buyer's price and the market price agree?

Let's work one case with $S(0)=K=100$ USD/share, $T=1$ year, volatility $\sigma=0.30$, and $g_{y}=0.04$ per year. In the real world, the mean growth rate of the share price is $\mu_{g}=0.06$ per year, while under $\mathbb{Q}$ it is $g_{y}-\sigma^{2}/2=-0.005$ per year. Repeating the digital-claim calculation with each mean growth rate gives the two probabilities of the event:

$$
\begin{align*}
\mathbb{P}\bigl(S(T)\geq K\bigr) &= N\!\left(\frac{\ln\frac{S(0)}{K}+\mu_{g}T}{\sigma\sqrt{T}}\right) = N(0.200) \approx 0.579 && \text{real-world mean growth rate}\\
\mathbb{Q}\bigl(S(T)\geq K\bigr) &= N\!\left(\frac{\ln\frac{S(0)}{K}+\bigl(g_{y}-\frac{\sigma^{2}}{2}\bigr)T}{\sigma\sqrt{T}}\right) \approx N(-0.017) \approx 0.493 && \text{risk-neutral mean growth rate, }N(d_{-})
\end{align*}
$$

Now consider a client with relative risk aversion $\bar{r}=3$ whose other wealth at expiration is a sure $w=50{,}000$ USD. Owning the bet with payoff $V$ makes the client's wealth $W=w+V$, with $\mathbb{E}_{\mathbb{P}}(V)=\mathbb{P}\bigl(S(T)\geq K\bigr)$. Pratt's small-risk result for the client's utility $U$ gives the client's risk premium for the bet:

$$
\begin{align*}
\mathrm{RP} &\approx -\frac{1}{2}\,\frac{U''(w)}{U'(w)}\,\mathrm{Var}(W) && \text{Pratt (1964), small risk}\\
&\approx \frac{\bar{r}}{2w}\,\mathrm{Var}(V) && \bar{r}=-w\,U''(w)/U'(w)\text{, and }w\text{ is sure}\\
&\approx \frac{\bar{r}}{2w}\,\mathbb{P}\bigl(S(T)\geq K\bigr)\Bigl[1-\mathbb{P}\bigl(S(T)\geq K\bigr)\Bigr] && \text{variance of a one-dollar bet}\\
&\approx \frac{3}{2\times 50{,}000}\times 0.579\times 0.421 \approx 7.3\times10^{-6}\ \text{USD} && \text{substitute the case}
\end{align*}
$$

The bet is worth a sure $\mathbb{E}_{\mathbb{P}}(V)-\mathrm{RP}$ to the client at expiration, and discounting it gives about the most the client would pay today. The table compares this value with the market price for the up-bet (the bet above) and the down-bet, which pays one dollar if $S(T)<K$:

| Bet | $\mathbb{P}$ (real world) | $\mathbb{Q}$ (risk neutral) | Client's RP (USD) | Client's value today (USD) | Market price (USD) |
|:--|--:|--:|--:|--:|--:|
| Up: pays if $S(T)\geq K$ | 0.579 | 0.493 | 0.000007 | 0.557 | 0.474 |
| Down: pays if $S(T)<K$ | 0.421 | 0.507 | 0.000007 | 0.404 | 0.487 |

The client values the up-bet at almost its full discounted expected payoff, because one dollar is tiny next to 50,000 USD. The market sells the up-bet for 8.3 cents less than the client's value, and it sells the down-bet for 8.3 cents more. The client's risk premium is less than a thousandth of a cent, so it cannot explain either gap.

### Marginal Utility and the Risk-Neutral Probability
Where do the gaps come from? Let's return to the one-step lattice, where the up and down states have real-world probabilities $\mathbb{P}(u)$ and $\mathbb{P}(d)=1-\mathbb{P}(u)$. An investor holds the mix of shares and the risk-free asset that maximizes expected utility, and ends the step with wealth $W$, equal to $W_{u}$ or $W_{d}$.

Now let the investor move a small amount $\varepsilon$ (USD) from the risk-free asset into a claim with price $V_{0}$ and payoff $H$, equal to $H_{u}$ or $H_{d}$. In each state, the claim adds $\varepsilon H/V_{0}$ to wealth, and the risk-free asset adds $\varepsilon R_{f}$ less. At the best holdings, no small move can raise expected utility, so the derivative at $\varepsilon=0$ is zero:

$$
\begin{align*}
0 &= \frac{d}{d\varepsilon}\,\mathbb{E}_{\mathbb{P}}\Bigl[U\Bigl(W+\varepsilon\bigl(\tfrac{H}{V_{0}}-R_{f}\bigr)\Bigr)\Bigr]_{\varepsilon=0} && \text{best holdings}\\
&= \mathbb{P}(u)\,U'(W_{u})\Bigl(\frac{H_{u}}{V_{0}}-R_{f}\Bigr)+\mathbb{P}(d)\,U'(W_{d})\Bigl(\frac{H_{d}}{V_{0}}-R_{f}\Bigr) && \text{chain rule in each state}\\
V_{0}\,R_{f}\,\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr) &= \mathbb{P}(u)\,U'(W_{u})\,H_{u}+\mathbb{P}(d)\,U'(W_{d})\,H_{d} && \text{multiply by }V_{0}\text{, rearrange}\\
V_{0} &= \frac{1}{R_{f}}\left[\frac{\mathbb{P}(u)\,U'(W_{u})}{\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr)}\,H_{u}+\frac{\mathbb{P}(d)\,U'(W_{d})}{\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr)}\,H_{d}\right] && \text{divide}
\end{align*}
$$

where $U'$ is the investor's marginal utility and $\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr)=\mathbb{P}(u)\,U'(W_{u})+\mathbb{P}(d)\,U'(W_{d})$ is its real-world average. Replication priced every claim on the lattice with the weights $q$ and $1-q$. Both rules hold for every claim, so the weights on $H_{u}$ must match:

$$
\boxed{
\begin{align*}
q = \mathbb{P}(u)\,\frac{U'(W_{u})}{\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr)}
\end{align*}}
$$

The risk-neutral probability is the real-world probability reweighted by marginal utility, and the state price $\psi_{u}=q/R_{f}$ is that probability discounted. Three points follow:
* __A bet that pays when wealth is high costs less.__ An investor who holds shares is richer in the up state, $W_{u}>W_{d}$. Because $U$ is concave ($U''<0$), $U'(W_{u})$ is below the average $\mathbb{E}_{\mathbb{P}}\bigl(U'(W)\bigr)$, so $q<\mathbb{P}(u)$ and the up-bet sells below its discounted expected payoff. The up-bet pays its dollar in the state where the investor has more wealth, so that dollar adds less utility.
* __Insurance costs more than its discounted expected payoff.__ The same argument puts $U'(W_{d})$ above the average, so $1-q>\mathbb{P}(d)$ and the down-bet sells above its discounted expected payoff. The down-bet pays in the state where the investor has less wealth, so its dollar adds more utility.
* __The certainty equivalent values the bet on its own.__ With sure other wealth, the client ends with $w$ or $w+1$ USD, so the client's marginal utility is almost the same in both states. In this model, prices weight the states by the marginal utility of investors whose wealth rises and falls with the share price. So the gap between the client's value and the market price comes from the state in which the bet pays, not from the client's risk premium.

Replication fixed $q$ from $u$, $d$, and $R_{f}$ alone, so the BSM formulas never need $U$. [Rubinstein (1976)](https://doi.org/10.2307/3003264) derived the same formulas from the utility rule. He assumed that one investor with constant relative risk aversion sets all prices, with the investor's log wealth and the log share price jointly normal.

The market price of the bet does not depend on one buyer's risk aversion. A buyer's risk aversion sets how much of the bet to hold, in the same way that it sets $w_{f}$ on the capital allocation line.

## Summary
We derived the Black–Scholes–Merton premiums from one rule, the law of one price. We then used marginal utility to explain why the market price of a bet differs from a client's certainty equivalent.

> __Key Takeaways:__
>
> * __Replication and state prices:__ We matched a claim's payoff in both lattice states with shares and the risk-free asset, and the portfolio's cost defined the risk-neutral probability. We then wrote every price as state prices times payoffs, so a YES and NO pair on one event costs the same as a zero-coupon bond.
> * __The BSM equation and two simple claims:__ We hedged an option with shares in continuous time and obtained the BSM equation, whose solution is the risk-neutral expectation of the discounted payoff. We then priced a digital claim and an asset-or-nothing claim, built the call from them, and used put–call parity for the put.
> * __Marginal utility and the market price:__ We found a client's certainty equivalent for a one-dollar bet almost equal to its expected payoff, while the market price differed by 8 cents. We showed that the risk-neutral probability is the real-world probability reweighted by marginal utility, so a bet that pays when investors' wealth is high costs less, and insurance costs more.

__Assumptions.__ We assume European exercise, no arbitrage, no dividends, and a constant risk-free rate $g_{y}$ and volatility $\sigma$. Trading has no costs and no limits on borrowing or short selling. The BSM equation also assumes geometric Brownian motion for the share price, continuous rebalancing, and a premium smooth enough for Itô's lemma. The marginal-utility argument assumes investors with increasing, concave utility who hold their best portfolio and can trade small amounts of any claim.
