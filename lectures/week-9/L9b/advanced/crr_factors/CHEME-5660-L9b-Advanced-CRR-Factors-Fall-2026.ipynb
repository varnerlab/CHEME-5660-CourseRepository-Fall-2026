# L9b Advanced: Where Do the CRR Up and Down Factors Come From?
The [L9b lecture](../../CHEME-5660-L9b-Lecture-American-Derivatives-Intrinsic-Extrinsic-Value-Fall-2026.ipynb) prices American options on the Cox, Ross, Rubinstein (CRR) lattice. But where do the expressions for its up and down factors come from?

## Derivation
In the CRR model the $u$ and $d$ are based on the volatility $\sigma$ of the underlying asset, the time step $\Delta{t}$ and the symmetric condition $ud = 1$:
$$\begin{align*}
u & = e^{\sigma\sqrt{\Delta{t}}} \\
d & = e^{-\sigma\sqrt{\Delta{t}}}
\end{align*}$$
where $\sigma$ is the same volatility as in geometric Brownian motion and the Black–Scholes–Merton formula (units: $\mathrm{yr}^{-1/2}$).

The CRR model matches the volatility of the binomial model to that of geometric Brownian motion (GBM). In the binomial model, if the stock moves up by a factor $u$ or down by a factor $d$, the log-return over one step is:
$$
\ln\frac{S(t+\Delta t)}{S(t)} = 
\begin{cases}
\ln(u) = a & \text{with probability } p,\\
\ln(d) = -a & \text{with probability } 1-p,
\end{cases}
$$
where we define $a = \ln(u)$ and use the symmetry condition $d = 1/u$, which gives $\ln(d) = -a$. Let $X$ denote this log-return random variable:
$$
\begin{align*}
X & =
\begin{cases}
+a & \text{with probability } p,\\
-a & \text{with probability } 1-p.
\end{cases}
\end{align*}
$$
We want to match the variance of this two-point random variable $X$ to the variance of the GBM log-return. The mean of $X$ is:
$$
\mathbb{E}[X] = p\cdot(+a) + (1-p)\cdot(-a) = (2p-1)a.
$$
The second moment is:
$$
\mathbb{E}[X^2] = p\cdot a^2 + (1-p)\cdot a^2 = a^2.
$$
Using the variance formula $\mathrm{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2$, we have:
$$
\begin{align*}
\mathrm{Var}(X) & = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 && \text{variance formula}\\
& = a^2 - \big[(2p-1)a\big]^2 && \text{substitute the two moments}\\
& = a^2\Big[1 - (2p-1)^2\Big] && \text{factor out } a^{2}\\
& = 4p(1-p)a^2 && 1-(2p-1)^{2} = 4p(1-p)
\end{align*}
$$
From the geometric Brownian motion model, the variance of the log return over time $\Delta t$ is:
$$
\mathrm{Var}\Bigg[\ln\frac{S(t+\Delta t)}{S(t)}\Bigg]
= \sigma^2 \Delta t.
$$
We set the binomial variance equal to the GBM variance:
$$
4p(1-p)a^2 = \sigma^2 \Delta t.
$$
At this point, CRR makes a simplifying choice: at the microscopic time scale $\Delta t$, take the up/down moves to be balanced, i.e., $p = \tfrac{1}{2}$. Plugging $p = \tfrac{1}{2}$ gives:
$$
4\cdot \tfrac{1}{2}\cdot\tfrac{1}{2}a^2 = a^2 = \sigma^2 \Delta t.
$$
which yields:
$$
a^2 = \sigma^2 \Delta t
\quad\Longrightarrow\quad
a = \sigma\sqrt{\Delta t}.
$$
Recall that we defined $u = e^{a}$ and $d = e^{-a}$ from the symmetry condition $ud = 1$. Therefore:
$$
\boxed{
u = e^{+\sigma\sqrt{\Delta t}},
\qquad
d = e^{-\sigma\sqrt{\Delta t}}.
}
$$
That is exactly the Cox–Ross–Rubinstein form!
___

## The Lattice Is Priced With $q$
The derivation took $p=\tfrac{1}{2}$, but the lecture prices options with the risk-neutral probability $q=(e^{g_{y}\Delta t}-d)/(u-d)$. How well do the CRR factors work with $q$?

* __The probability $q$ is close to one half:__ For a small time step, $q=\tfrac{1}{2}+\tfrac{g_{y}-\sigma^{2}/2}{2\sigma}\sqrt{\Delta t}+\mathcal{O}(\Delta t^{3/2})$. As the step shrinks, $q$ approaches the value $\tfrac{1}{2}$ that the derivation assumed.
* __The variance match holds to leading order:__ With probability $q$, the one-step variance is $4q(1-q)a^{2}=\sigma^{2}\Delta t-(g_{y}-\sigma^{2}/2)^{2}\Delta t^{2}+\mathcal{O}(\Delta t^{3})$. The error shrinks faster than the variance itself.
* __The time step must be small enough:__ The probability $q$ lies between zero and one only when $d<e^{g_{y}\Delta t}<u$, which for the CRR factors is $\lvert g_{y}\rvert\sqrt{\Delta t}<\sigma$. The [CRR pricing example](../../CHEME-5660-L9b-Example-AmericanContractsPremium-Fall-2026.ipynb) checks this volatility floor, `σ_min`, for every contract.

The CRR factors carry over to the pricing lattice when the time step is small.

## Summary
We derived the CRR up and down factors by matching the variance of the lattice's one-step log return to the geometric Brownian motion variance. We then checked the match on the pricing lattice.

> __Key Takeaways:__
>
> * __Variance matching sets the size of a move:__ The lattice's one-step log return takes two values, so its variance depends on the size of the move. Matching it to geometric Brownian motion ties that size to the volatility.
> * __Two choices give the CRR form:__ The symmetric condition makes a down move undo an up move. Equally likely moves then set the size of a move to the volatility times the square root of the time step.
> * __The factors work for small steps:__ As the step shrinks, the risk-neutral probability approaches one half and the variance match holds to leading order. The probability lies between zero and one only for a small enough step.

The CRR lattice assumes a constant volatility and risk-free rate, and no dividends.
